```python
try:
    trees = pd.read_csv("../data/p09_train.csv")
except FileNotFoundError:
    trees = pd.read_csv("mocktests/r1-001/data/p09_train.csv")

TARGET = "outcome"
FEATURES = [column for column in trees.columns if column != TARGET]
X = trees[FEATURES]
y = trees[TARGET]

print("shape:", trees.shape)
print("features:", FEATURES)
print("class counts:", y.value_counts().to_dict())
```

## Required prediction-function contract

Define `predict_labels(X_test)`. The contract has five clauses. The grader checks them mechanically, and **any violation scores zero**:

| # | Clause |
|---|---|
| R1 | a callable named exactly `predict_labels` exists after the notebook runs |
| R2 | it accepts a feature `DataFrame` of **any** length |
| R3 | it returns a **`pd.Series`** |
| R4 | the result has `len(result) == len(X_test)` **and** `result.index.equals(X_test.index)` — one prediction per input row, in the input's order |
| R5 | every value is a label from the training vocabulary |

The input has exactly the training feature columns in the same order. Fit your final recipe on all labeled rows before submission; `predict_labels` should perform prediction without refitting.

```python
# Runnable legal starter. Replace or extend this kNN recipe after honest validation.
final_model = Pipeline(
    [
        ("scaler", StandardScaler()),
        ("knn", KNeighborsClassifier(n_neighbors=5)),
    ]
)
final_model.fit(X, y)


def predict_labels(X_test):
    return pd.Series(final_model.predict(X_test[FEATURES]), index=X_test.index)
```

```python
# Contract self-check on an any-length probe with a non-default index.
probe = X.iloc[10:27].copy()
probe.index = np.arange(1000, 1000 + len(probe))
probe_predictions = predict_labels(probe)

assert callable(predict_labels)
assert isinstance(probe_predictions, pd.Series)
assert len(probe_predictions) == len(probe)
assert probe_predictions.index.equals(probe.index)
assert set(probe_predictions.unique()) <= set(y.unique())
print("contract self-check: passed")
```
