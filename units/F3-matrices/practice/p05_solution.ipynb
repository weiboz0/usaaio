# F3-matrices — Practice p05 — Solution

Broadcasting forms every matrix-entry/point-coordinate product at once, and summing the input-coordinate axis produces each mapped output coordinate. The quarter-turn changes direction but preserves each row's Euclidean length, so the maximum length change should be at floating-point roundoff scale.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
points = rng.normal(0, 2, (200, 2))
M = np.array([[0., -1.], [1., 0.]])

def apply_map(M, points):
    return (M[None, :, :] * points[:, None, :]).sum(axis=2)

turned = apply_map(M, points)
lengths_before = np.sqrt((points * points).sum(axis=1))
lengths_after = np.sqrt((turned * turned).sum(axis=1))
max_length_change = np.abs(lengths_after - lengths_before).max()
```

```python
# ### Answer check
assert turned.shape == (200, 2)
assert np.allclose(turned[:, 0], -points[:, 1])
assert np.allclose(turned[:, 1], points[:, 0])
assert max_length_change < 1e-12
```
