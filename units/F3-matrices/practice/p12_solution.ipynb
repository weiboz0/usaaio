# F3-matrices — Practice p12 — Solution

Substitution gives $c_1r_1+c_2r_2+c_3r_3=(c_1+c_3)r_1+(c_2-2c_3)r_2$, so all row combinations lie in the span of two rows and $\operatorname{rank}(A)\le2$. Such a map cannot cover all of $\mathbb R^3$ and therefore cannot have an inverse that sends every target back to an input. In the concrete case, the first two row equations are $x_0+2x_2=0$ and $x_1-x_2=0$; taking $x_2=1$ gives $x=(-2,1,1)$, and the third equation vanishes automatically because its row is $r_1-2r_2$. Thus the distinct inputs $0$ and $x$ collide at the zero output, which an inverse could not undo.

```python
import numpy as np

r1 = np.array([1., 0., 2.])
r2 = np.array([0., 1., -1.])
r3 = r1 - 2 * r2
A = np.stack((r1, r2, r3))
x_collide = np.array([-2., 1., 1.])
collision_output = (A * x_collide[None, :]).sum(axis=1)
collide_norm = np.sqrt((collision_output * collision_output).sum())
```

### Answer check

```python
assert np.any(x_collide != 0)
assert np.allclose(r3, r1 - 2 * r2)
assert np.allclose(collision_output, np.zeros(3), atol=1e-12)
assert collide_norm < 1e-12
```
