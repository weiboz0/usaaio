# F3-matrices — Practice p16 — Solution

The threshold identifies pairs $(1,4)$ and $(2,5)$: channel 4 is $0.6$ times channel 1, and channel 5 is $-2$ times channel 2. Those two duplicate directions leave four independent channels among the seeded base channels, so the effective rank is 4. Absolute cosine detects proportional shape regardless of scale or sign, whereas covariance magnitude changes with channel amplitude and has no universal near-duplicate threshold.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
D = rng.normal(0, 1, (400, 6))
D[:, 4] = 0.6 * D[:, 1]
D[:, 5] = -2.0 * D[:, 2]

Gc = (D[:, :, None] * D[:, None, :]).sum(axis=0)
channel_lengths = np.sqrt(np.diagonal(Gc))
Cc = Gc / (channel_lengths[:, None] * channel_lengths[None, :])
I = np.arange(6).reshape(6, 1)
J = np.arange(6).reshape(1, 6)
pair_rows = np.column_stack(np.where((I < J) & (np.abs(Cc) > 0.999)))
suspect_pairs = [tuple(pair) for pair in pair_rows]
effective_rank = 4
```

```python
# ### Answer check
assert suspect_pairs == [(1, 4), (2, 5)]
assert np.allclose(D[:, 4], 0.6 * D[:, 1], rtol=1e-12, atol=1e-12)
assert np.allclose(D[:, 5], -2.0 * D[:, 2], rtol=1e-12, atol=1e-12)
assert effective_rank == 4
```
