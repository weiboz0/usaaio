# F3-matrices — Practice p18 — Solution

In (a), $p_2=2p_1$, so the probes span at most the plane generated by $p_1,p_3$; changing the map on a direction outside that plane leaves all observations unchanged, so $Z$ cannot be determined. For (b), if $ap_1+bp_2+cp_3=0$, the coordinate equations are $a+c=0$, $a+b=0$, $b+c=0$, which force $a=b=c=0$, proving independence. Solving the same equations gives $e_0=(p_1-p_2+p_3)/2$, $e_1=(p_1+p_2-p_3)/2$, and $e_2=(-p_1+p_2+p_3)/2$; applying linearity gives the columns $(2,0,1)$, $(0,0,0)$, and $(1,1,0)$. The zero middle column and two non-proportional remaining columns imply $\operatorname{rank}(Z)=2$, so $Z$ is not invertible.

```python
import numpy as np

p1 = np.array([1., 1., 0.])
p2 = np.array([0., 1., 1.])
p3 = np.array([1., 0., 1.])
q1 = np.array([2., 0., 1.])
q2 = np.array([1., 1., 0.])
q3 = np.array([3., 1., 1.])

z_e0 = (q1 - q2 + q3) / 2
z_e1 = (q1 + q2 - q3) / 2
z_e2 = (-q1 + q2 + q3) / 2
Z = np.column_stack((z_e0, z_e1, z_e2))
rank_Z = 2
is_invertible = False
probes = np.stack((p1, p2, p3))
outputs = np.stack((q1, q2, q3))
reconstructed_outputs = (Z[None, :, :] * probes[:, None, :]).sum(axis=2)
recon_gap = np.abs(reconstructed_outputs - outputs).max()
```

```python
# ### Answer check
expected_Z = np.array([[2., 0., 1.], [0., 0., 1.], [1., 0., 0.]])
assert np.array_equal(Z, expected_Z)
assert rank_Z == 2
assert is_invertible is False
assert recon_gap < 1e-12
```
