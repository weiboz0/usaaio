# F3-matrices — Practice p04 — Solution

Reading the coefficient of each input coordinate gives $B=[[-2,0,1],[0,3,1]]$. Its entries sum to $s=3$, so the nonnegative normal-form rule gives $m+2=3+2=5$, which is choice **C**.

```python
import numpy as np

B = np.array([[-2, 0, 1], [0, 3, 1]])
s = int(B.sum())
m = abs(s)
normal_form_value = m + 2 if s >= 0 else m + 1
answer_p04 = "C"
```

```python
# ### Answer check
assert s == 3
assert normal_form_value == 5
assert answer_p04 == "C"
```
