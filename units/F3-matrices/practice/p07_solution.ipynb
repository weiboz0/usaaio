# F3-matrices — Practice p07 — Solution

Broadcasting pairs every row with every other row, and summing coordinatewise products gives the Gram entries. Diagonal entries are squared row lengths, and comparing swapped integer-grid indices verifies symmetry without changing array axes.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (6, 4))

def gram(W):
    return (W[:, None, :] * W[None, :, :]).sum(axis=2)

G = gram(W)
row_length_squares = (W * W).sum(axis=1)
diag_check = np.abs(np.diagonal(G) - row_length_squares).max()
I = np.arange(6).reshape(6, 1)
J = np.arange(6).reshape(1, 6)
sym_check = np.abs(G[I, J] - G[J, I]).max()
```

```python
# ### Answer check
assert G.shape == (6, 6)
assert diag_check < 1e-12
assert sym_check < 1e-12
```
