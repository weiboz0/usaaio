# F3-matrices — Practice p09 — Solution

Reshaping $u$ as a column and $v$ as a row lets broadcasting create the full product grid. At the zero entry of $v$, division is undefined, so the prefilled row scalar is retained there; all four displayed ratios then equal the corresponding component of $u$.

```python
import numpy as np

def outer(u, v):
    return u[:, None] * v[None, :]

u = np.array([1., 2., -1.])
v = np.array([3., 0., 1., -2.])
P = outer(u, v)
row_ratios = np.broadcast_to(u[:, None], P.shape).copy()
np.divide(P, v[None, :], out=row_ratios, where=v[None, :] != 0)
col_of_ratios = row_ratios[:, 0]
ratio_gap = np.abs(col_of_ratios - u).max()
```

### Answer check

```python
expected_P = np.array([[3., 0., 1., -2.], [6., 0., 2., -4.], [-3., 0., -1., 2.]])
assert np.array_equal(P, expected_P)
assert row_ratios.shape == (3, 4)
assert np.allclose(row_ratios, u[:, None], atol=1e-9, rtol=0)
assert ratio_gap < 1e-12
```
