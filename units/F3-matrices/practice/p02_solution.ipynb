# F3-matrices — Practice p02 — Solution

For a product, the inner dimensions must match and the outer dimensions determine the result. Thus $(2\times3)(3\times4)$ produces a $2\times4$ matrix, while $(3\times4)(2\times3)$ is undefined because 4 does not equal 2; the answer is **B**.

```python
import numpy as np

A_shape = (2, 3)
B_shape = (3, 4)
AB_shape = (A_shape[0], B_shape[1])
BA_is_defined = B_shape[1] == A_shape[0]
answer_p02 = "B"
```

### Answer check

```python
assert AB_shape == (2, 4)
assert BA_is_defined is False
assert answer_p02 == "B"
```
