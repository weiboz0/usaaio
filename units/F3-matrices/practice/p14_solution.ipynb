# F3-matrices — Practice p14 — Solution

The largest off-diagonal absolute cosine occurs for rows 1 and 3, revealing that they share a direction. The Gram ratio gives $\alpha=G_{1,3}/G_{1,1}=-1.7$, and the residual length is at roundoff scale. Part (c) certifies the rows are NOT all independent: row 3 is a multiple of row 1, so at most 4 of the 5 rows can be independent — and since the rows live in 3-D, the independent-direction count is at most 3 regardless.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (5, 3))
W[3] = -1.7 * W[1]

def gram(W):
    return (W[:, None, :] * W[None, :, :]).sum(axis=2)

G = gram(W)
lengths = np.sqrt(np.diagonal(G))
C = G / (lengths[:, None] * lengths[None, :])
I = np.arange(5).reshape(5, 1)
J = np.arange(5).reshape(1, 5)
upper_scores = np.where(I < J, np.abs(C), -np.inf)
dependent_pair = tuple(np.unravel_index(np.argmax(upper_scores), upper_scores.shape))
i, j = dependent_pair
alpha = G[i, j] / G[i, i]
residual = W[j] - alpha * W[i]
dependence_gap = np.sqrt((residual * residual).sum())
# (c) certifies dependence; no rank literal is asserted — see (d)
```

### Answer check

```python
assert dependent_pair == (1, 3)
assert np.isclose(alpha, -1.7, rtol=1e-12, atol=1e-12)
assert dependence_gap < 1e-12
```
