# F3-matrices — Practice p11 — Solution

For row $i$, $(P_{i1},\ldots,P_{im})=(u_i v_1,\ldots,u_i v_m)=u_i v$, so its scalar is $u_i$ and every row lies in $\operatorname{span}\{v\}$. Hence $\operatorname{rank}(P)\le1$; because $u\ne0$, some $u_i\ne0$, and because $v\ne0$, that row $u_i v$ is nonzero, so the rank is not zero and is exactly 1. For $Q$, choose $u=(2,3)$ and $v=(1,2,-1)$, giving $u_i v$ as its two rows. This factorization has a nonzero row and confines every row to one nonzero direction, certifying rank 1.

```python
import numpy as np

Q = np.array([[2., 4., -2.], [3., 6., -3.]])
u = np.array([2., 3.])
v = np.array([1., 2., -1.])
factored_Q = u[:, None] * v[None, :]
factor_gap = np.abs(Q - factored_Q).max()
```

```python
# ### Answer check
assert np.array_equal(factored_Q, Q)
assert np.any(u != 0) and np.any(v != 0)
assert factor_gap < 1e-12
```
