# F3-matrices — Practice p17 — Solution

Reading coefficients gives columns $c_0=(1,2,0,3)$, $c_1=(0,1,1,0)$, $c_2=-c_0$, and $c_3=0$. The first two columns are independent because the third component of $c_1$ is 1 while that of $c_0$ is 0, so the column rank and the minimum count are $I_{min}=2$. Accordingly, $S=c_0(1,0,-1,0)+c_1(0,1,0,0)$ is a two-outer-product decomposition. Fewer cannot work because a sum of $k$ rank-1 matrices has rank at most $k$, while $S$ has rank 2.

```python
import numpy as np

S = np.array([[1., 0., -1., 0.], [2., 1., -2., 0.], [0., 1., 0., 0.], [3., 0., -3., 0.]])
I_min = 2
us = np.array([[1., 2., 0., 3.], [0., 1., 1., 0.]])
vs = np.array([[1., 0., -1., 0.], [0., 1., 0., 0.]])
pieces = us[:, :, None] * vs[:, None, :]
reconstructed_S = pieces.sum(axis=0)
decomp_gap = np.abs(S - reconstructed_S).max()
```

```python
# ### Answer check
assert I_min == 2
assert us.shape == (I_min, 4) and vs.shape == (I_min, 4)
assert np.array_equal(S[:, 2], -S[:, 0])
assert np.array_equal(S[:, 3], np.zeros(4))
assert decomp_gap < 1e-12
```
