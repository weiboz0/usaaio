# F3-matrices — Practice p13 — Solution

Reading the coefficients of $(x_0,x_1,x_2)$ from each output gives $T=[[1,2,0],[3,6,0],[-1,-2,0]]$. Its second column is twice its first and its third is zero, while the first is nonzero, so the column span has exactly one independent direction and $\operatorname{rank}(T)=1$. Taking $u_3=(1,3,-1)$ and $v_3=(1,2,0)$ gives $T_{ij}=(u_3)_i(v_3)_j$, a single outer product. A nonzero single outer product has exactly the one-dimensional column and row spans just identified.

```python
import numpy as np

T = np.array([[1., 2., 0.], [3., 6., 0.], [-1., -2., 0.]])
rank_T = 1
u3 = np.array([1., 3., -1.])
v3 = np.array([1., 2., 0.])
outer_T = u3[:, None] * v3[None, :]
decomp_gap = np.abs(T - outer_T).max()
```

```python
# ### Answer check
assert rank_T == 1
assert np.array_equal(T[:, 1], 2 * T[:, 0])
assert np.array_equal(T[:, 2], np.zeros(3))
assert decomp_gap < 1e-12
```
