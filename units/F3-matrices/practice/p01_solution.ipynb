# F3-matrices — Practice p01 — Solution

A linear map sends a linear combination of inputs to the same linear combination of their outputs. Since $(2,1)=2(1,0)+(0,1)$, the result is $2(3,1)+(-2,4)=(4,6)$, so the answer is **A**.

```python
import numpy as np

M = np.array([[3, -2], [1, 4]])
x = np.array([2, 1])
mapped = M @ x
answer_p01 = "A"
mapped
```

### Answer check

```python
assert np.array_equal(mapped, np.array([4, 6]))
assert answer_p01 == "A"
```
