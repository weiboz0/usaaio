## Unit goals

By the end of this unit you can:

- read a matrix as a **machine acting on vectors** — compute actions by
  hand (row·input dot products) and in NumPy, and use the column view:
  column $j$ is the image of the basis vector $e_j$;
- state and use the **linearity contract** (scaling and addition
  respected), including checking whether a documented set of
  input→output examples is even *consistent* with a linear map;
- **reconstruct a matrix from its action** — the exam's signature
  pattern — both by reading coefficients off a formula (silent
  variables → zero columns) and by probing a black-box function with
  basis vectors;
- multiply matrices by hand (row·column, the shape rule
  $(n,k)(k,m) \to (n,m)$), use `@` in NumPy, and explain products as
  **composition of machines** (order matters, association is free);
- work the **banned-`@` register**: rebuild products, actions, and Gram
  matrices from elementwise multiplies, broadcasting, and axis sums
  when the exam bans `@`, `np.matmul`, `np.dot`, `.T`, and loops;
- compute and read **Gram matrices** $G = WW^{\mathsf T}$ (all pairwise
  row dot products; symmetric; diagonal = squared lengths) and their
  **cosine-normalized** versions for direction-only comparison;
- decide **linear independence** operationally, describe **spans** with
  membership tests, read **rank** as the independent-direction count
  (rows or columns — same number), and decide **invertibility via
  rank** (full rank ⇔ an undo exists; collisions and unreachable
  targets otherwise) — without ever computing an inverse;
- use **outer products** as rank-1 atoms: recognize and factor rank-1
  matrices, and write a rank-$r$ matrix as a *minimal* sum of $r$
  atoms, justified by the rank bound.
