## Concept summary

| Concept | One-line summary | Key fact to retain |
|---|---|---|
| Matrices as linear maps | A matrix is a machine: $y_i = (\text{row } i) \cdot x$; linearity ($f(ax+by) = af(x)+bf(y)$) is the contract | $f(0) = 0$; knowing $f$ on a basis determines it everywhere |
| Matrix from action | Coefficients of $x_j$ in output $i$'s formula fill row $i$; or probe: $M e_j = $ column $j$ | silent variable ⇒ **zero column**, never a missing one |
| Matrix multiplication | $C_{ij} = \text{row}_i(A) \cdot \text{col}_j(B)$; $(n,k)(k,m) \to (n,m)$ | $(AB)x = A(Bx)$: composition, right to left; $AB \ne BA$ in general |
| Banned-`@` register | Rebuild any product from elementwise `*` + broadcasting + axis sums | give every index its own axis; sum the shared one |
| Gram matrices | $G = WW^{\mathsf T}$: all pairwise row dots | symmetric; diagonal $= \lVert w_i \rVert^2$; cosine-normalize before calling entries "similarity" |
| Linear independence & span | Dependent ⇔ some listed vector is buildable from the rest; span = the reachable set | more vectors than dimensions ⇒ dependent; adding a dependent vector never grows the span |
| Rank | The independent-direction count — rows or columns, same number | $\operatorname{rank} \le \min(n, m)$; $\operatorname{rank}(AB) \le \min(\operatorname{rank}A, \operatorname{rank}B)$ |
| Invertibility via rank | Square: full rank ⇔ an undo exists | rank $< n$ ⇒ collisions ($Ax^* = 0$, $x^* \ne 0$) **and** unreachable targets; decide by rank, never by computing an inverse |
| Outer products | $(u \otimes v)_{ij} = u_i v_j$: the rank-1 atom | rank-$r$ matrix = sum of $r$ atoms, never fewer (sum of $k$ atoms has rank $\le k$) |

## The two-register multiplication idiom sheet

**Normal register (nothing banned):**

```python
C = A @ B                    # matrix product, shape rule enforced
y = M @ x                    # action on a vector, shape (n,)
G = W @ W.T                  # Gram matrix, (n, d)(d, n) -> (n, n)
```

**Banned-`@` register** (zero-points clauses naming `@`, `np.matmul`,
`np.dot`, `.T`, loops — elementwise `*` + broadcasting + axis sums
only):

