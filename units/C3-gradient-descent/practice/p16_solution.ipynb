# C3-gradient-descent — Practice p16 — Solution

A current-batch loss estimates full-data loss from only five randomly sampled examples, so it varies across batches even when the parameters are identical; here the fixed-parameter batch losses range from about $0.0207$ to $0.2634$. Although 8 of the 30 sampled batches have higher loss after this step, the full-data loss decreases from about $0.12882$ to $0.12517$. Progress should therefore be judged with a full-data or stable held-out loss, or with a sufficiently smoothed batch estimate, rather than two unrelated noisy batch readings.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (150, 2))
w_true = np.array([2.0, -1.0])
b_true = 0.5
y = (X * w_true).sum(axis=1) + b_true + rng.normal(0, 0.4, 150)
batch_indices = rng.integers(0, 150, size=(30, 5))
wA = np.array([1.90, -0.95])
bA = 0.45
wB = np.array([1.92, -0.96])
bB = 0.46

def mse_loss(X, y, w, b):
    residuals = (X * w).sum(axis=1) + b - y
    return (residuals ** 2).mean()

full_A = mse_loss(X, y, wA, bA)
full_B = mse_loss(X, y, wB, bB)
full_decreased = bool(full_B < full_A)
batch_losses_A = np.empty(30)
batch_losses_B = np.empty(30)
for position, idx in enumerate(batch_indices):
    batch_losses_A[position] = mse_loss(X[idx], y[idx], wA, bA)
    batch_losses_B[position] = mse_loss(X[idx], y[idx], wB, bB)
spread_min = batch_losses_A.min()
spread_max = batch_losses_A.max()
n_worse = int((batch_losses_B > batch_losses_A).sum())
full_A, full_B, full_decreased, spread_min, spread_max, n_worse
```

### Answer check

```python
assert np.isclose(full_A, 0.128820580246, atol=1e-12, rtol=0)
assert np.isclose(full_B, 0.125174615548, atol=1e-12, rtol=0)
assert full_decreased is True
assert np.isclose(spread_min, 0.020724666095, atol=1e-12, rtol=0)
assert np.isclose(spread_max, 0.263444280103, atol=1e-12, rtol=0)
assert n_worse == 8
```
