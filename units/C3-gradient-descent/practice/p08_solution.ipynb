# C3-gradient-descent — Practice p08 — Solution

A size-8 batch is one random sample of the data, so its gradient can differ substantially from the full-data gradient. Averaging many independently sampled batch gradients cancels much of that sampling noise, making the average close to the full gradient.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (200, 3))
w_true = np.array([2.0, -1.0, 0.5])
b_true = 0.7
y = (X * w_true).sum(axis=1) + b_true + rng.normal(0, 0.2, 200)
w0 = np.array([1.0, -0.5, 0.0])
b0 = 0.5
idx = rng.integers(0, 200, 8)
batch_indices = rng.integers(0, 200, size=(2000, 8))

def mse_gradients(X, y, w, b):
    residuals = (X * w).sum(axis=1) + b - y
    grad_w = 2 * (residuals[:, None] * X).mean(axis=0)
    grad_b = 2 * residuals.mean()
    return grad_w, grad_b

g_full_w, g_full_b = mse_gradients(X, y, w0, b0)
g_batch_w, g_batch_b = mse_gradients(X[idx], y[idx], w0, b0)
gap = np.abs(np.concatenate((g_batch_w - g_full_w, [g_batch_b - g_full_b]))).max()
sum_w = np.zeros(3)
sum_b = 0.0
for row in batch_indices:
    current_w, current_b = mse_gradients(X[row], y[row], w0, b0)
    sum_w += current_w
    sum_b += current_b
avg_w = sum_w / len(batch_indices)
avg_b = sum_b / len(batch_indices)
avg_gap = np.abs(np.concatenate((avg_w - g_full_w, [avg_b - g_full_b]))).max()
gap, avg_gap
```

### Answer check

```python
assert np.allclose(g_full_w, [-1.970039232699, 0.801465454423, -0.950485210480], atol=1e-12)
assert np.isclose(g_full_b, -0.143996216047, atol=1e-12)
assert np.isclose(gap, 1.462334807201, atol=1e-12)
assert np.isclose(avg_gap, 0.029938086585, atol=1e-12)
assert avg_gap < gap
```
