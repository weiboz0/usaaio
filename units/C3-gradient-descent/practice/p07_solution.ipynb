# C3-gradient-descent — Practice p07 — Solution

The $w_2$ coordinate sets the upper limit because its factor $1-20\eta$ has the larger curvature and must have magnitude below one, requiring $0<\eta<0.1$. At $\eta=0.09$ the factors are $0.82$ and $-0.8$, so both coordinates shrink, whereas at $\eta=0.11$ the $w_2$ factor is $-1.2$ and its magnitude grows while alternating signs.

```python
import numpy as np

def run_descent(w0, eta, steps):
    w = np.array(w0, dtype=float).copy()
    history = [w.copy()]
    for _ in range(steps):
        gradient = np.array([2.0, 20.0]) * w
        w = w - eta * gradient
        history.append(w.copy())
    return np.array(history)

w0 = np.array([8.0, 3.0])
hist_a = run_descent(w0, 0.09, 40)
final_a = hist_a[-1]
hist_b = run_descent(w0, 0.11, 40)
final_b = hist_b[-1]
diverged_b = bool(np.abs(final_b).max() > 100)
final_a, final_b, diverged_b
```
