# C3-gradient-descent — Practice p17 — Solution

Solving $L'(w)=4w(w^2-1)=0$ gives $w=-1,0,1$; since $L''(\pm1)=8>0$ and $L(\pm1)=0$, the points $\pm1$ are minima, while $L''(0)=-4<0$ and $L(0)=1$, so $0$ is a local maximum. The four $\eta=0.05$ runs finish at $(-1,-1,1,1)$, showing that for these nonzero starts the starting sign selects the valley. Near either minimum an error is multiplied by $1-8\eta$, so local convergence requires $|1-8\eta|<1$, equivalently $0<\eta<1/4$; the numeric runs confirm that $\eta=0.2$ settles at $1$ whereas $\eta=0.3$ does not. Unlike a single quadratic bowl, this surface makes the attained minimum depend on the starting basin, and its usable step size is controlled by the local curvature $L''(\pm1)=8$.

```python
import numpy as np

starts = np.array([-2.0, -0.5, 0.5, 2.0])

def descend(w0, eta, steps, keep_history=False):
    w = float(w0)
    history = [w]
    for _ in range(steps):
        w = w - eta * (4 * w ** 3 - 4 * w)
        if keep_history:
            history.append(w)
    return w, np.array(history)

finals = np.empty(4)
for position, start in enumerate(starts):
    finals[position], unused = descend(start, 0.05, 300)
w60_at_02, history_02 = descend(0.5, 0.2, 60, keep_history=True)
w60_at_03, history_03 = descend(0.5, 0.3, 60, keep_history=True)
settled_02 = bool(abs(w60_at_02 - 1) < 1e-6)
settled_03 = bool(abs(w60_at_03 - 1) < 1e-6)
first6_03 = history_03[:6]
finals, w60_at_02, w60_at_03, settled_02, settled_03, first6_03
```

### Answer check

```python
assert np.allclose(finals, [-1.0, -1.0, 1.0, 1.0], atol=1e-12)
assert np.isclose(w60_at_02, 1.0, atol=1e-12) and settled_02 is True
assert np.isclose(w60_at_03, 1.139538356503, atol=1e-12) and settled_03 is False
assert np.allclose(first6_03, [0.5, 0.95, 1.06115, 0.900654047515, 1.104730315642, 0.812512703952], atol=1e-12)
assert 0.2 < 0.25 < 0.3
```
