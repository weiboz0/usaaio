# C3-gradient-descent — Practice p03 — Solution

Writing $e_t=w_t-5$, the update gives $e_{t+1}=(1-2\eta)e_t$. The factors for $\eta=(0.001,0.2,1.2)$ are $(0.998,0.6,-1.4)$, matching the slow crawl, non-overshooting shrinkage, and alternating growth respectively; therefore the assignment is B.

```python
import numpy as np

etas = np.array([0.001, 0.2, 1.2])
factors = 1 - 2 * etas
answer_p03 = "B"
factors
```

### Answer check

```python
assert np.allclose(factors, [0.998, 0.6, -1.4], atol=1e-12)
assert answer_p03 == "B"
```
