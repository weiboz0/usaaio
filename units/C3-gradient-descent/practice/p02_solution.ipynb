# C3-gradient-descent — Practice p02 — Solution

Differentiating gives $L'(w)=2w+2$, which equals $4$ at $w=1$. The update is $1-0.25(4)=0$, so the derived choice is A.

```python
import numpy as np

w0 = 1.0
eta = 0.25
gradient = 2 * w0 + 2
w_new = w0 - eta * gradient
answer_p02 = "A"
w_new
```

### Answer check

```python
assert np.isclose(w_new, 0.0, atol=1e-12, rtol=0)
assert answer_p02 == "A"
```
