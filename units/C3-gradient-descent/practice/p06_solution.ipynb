# C3-gradient-descent — Practice p06 — Solution

Broadcasting the bias, weight, and data axes constructs all $81\times81\times3$ residuals at once, and averaging their squares over the data axis yields the requested grid orientation. The data lie exactly on $y=2x+1$, so the grid contains the zero-loss point $(w,b)=(2,1)$.

```python
import numpy as np

x = np.array([0.0, 1.0, 2.0])
y = np.array([1.0, 3.0, 5.0])
w_vals = np.linspace(0, 4, 81)
b_vals = np.linspace(-1, 3, 81)

residuals = b_vals[:, None, None] + w_vals[None, :, None] * x[None, None, :] - y[None, None, :]
L_grid = (residuals ** 2).mean(axis=2)
min_loss = L_grid.min()
k = L_grid.argmin()
best_w = w_vals[k % 81]
best_b = b_vals[k // 81]
min_loss, best_w, best_b
```

### Answer check

```python
assert L_grid.shape == (81, 81)
assert np.isclose(min_loss, 0.0, atol=1e-12)
assert np.isclose(best_w, 2.0, atol=1e-12)
assert np.isclose(best_b, 1.0, atol=1e-12)
```
