# C3-gradient-descent — Practice p10 — Solution

With error $e_t=w_t-2$, the recurrence is $e_{t+1}=(1-6\eta)e_t$, so convergence requires $|1-6\eta|<1$, equivalently $0<\eta<1/3$. Thus the exact boundary is $\eta_{crit}=1/3$; the sampled mask agrees because only the first six rates lie strictly below it. At the boundary the factor is $-1$, so the run alternates forever between $w=0$ and $w=4$ and never converges.

```python
import numpy as np
from fractions import Fraction

etas = np.array([0.01, 0.02, 0.05, 0.1, 0.2, 0.3, 1 / 3, 0.34, 0.4, 0.5])
final_w = np.empty(etas.size)
for index, eta in enumerate(etas):
    w = 0.0
    for _ in range(200):
        w = w - eta * 6 * (w - 2)
    final_w[index] = w
converged = np.abs(final_w - 2) < 1e-3
n_converged = int(converged.sum())
eta_crit = Fraction(1, 3)
w = 0.0
first_five = [w]
for _ in range(4):
    w = w - float(eta_crit) * 6 * (w - 2)
    first_five.append(w)
first_five = np.array(first_five)
final_w, converged, n_converged, first_five
```

### Answer check

```python
assert eta_crit == Fraction(1, 3)
assert np.array_equal(converged, [True, True, True, True, True, True, False, False, False, False])
assert n_converged == 6
assert np.allclose(first_five, [0.0, 4.0, 0.0, 4.0, 0.0], atol=1e-12, rtol=0)
```
