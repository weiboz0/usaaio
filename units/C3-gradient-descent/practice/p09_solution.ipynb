# C3-gradient-descent — Practice p09 — Solution

Each iteration computes predictions, MSE, and gradients using broadcasting and reductions across examples, then records the post-update loss. After 500 steps the fitted parameters are within about $0.022$ of the generating values and the final MSE is about $0.00682$, consistent with noise of standard deviation $0.1$.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (120, 3))
w_true = np.array([3.0, -2.0, 1.0])
b_true = -0.5
y = (X * w_true).sum(axis=1) + b_true + rng.normal(0, 0.1, 120)

def mse_loss(X, y, w, b):
    residuals = (X * w).sum(axis=1) + b - y
    return (residuals ** 2).mean()

def mse_gradients(X, y, w, b):
    residuals = (X * w).sum(axis=1) + b - y
    return 2 * (residuals[:, None] * X).mean(axis=0), 2 * residuals.mean()

def fit(X, y, eta, steps):
    w = np.zeros(X.shape[1])
    b = 0.0
    losses = np.empty(steps)
    for step in range(steps):
        grad_w, grad_b = mse_gradients(X, y, w, b)
        w = w - eta * grad_w
        b = b - eta * grad_b
        losses[step] = mse_loss(X, y, w, b)
    return w, b, losses

w_fit, b_fit, losses = fit(X, y, eta=0.1, steps=500)
final_loss = losses[-1]
param_gap = np.abs(np.concatenate((w_fit - w_true, [b_fit - b_true]))).max()
w_fit, b_fit, final_loss, param_gap
```

### Answer check

```python
assert losses.shape == (500,)
assert np.allclose(w_fit, [3.004962977651, -2.012688043528, 1.021916528303], atol=1e-12)
assert np.isclose(b_fit, -0.490767808587, atol=1e-12)
assert np.isclose(final_loss, 0.006817958925, atol=1e-12)
assert np.isclose(param_gap, 0.021916528303, atol=1e-12)
```
