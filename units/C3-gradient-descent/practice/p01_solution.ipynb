# C3-gradient-descent — Practice p01 — Solution

The squared terms are both minimized at zero, so the unique minimum is $(w_1,w_2)=(2,-1)$. Because the $w_2$ term has coefficient $4$, equal-loss contours are ellipses and the surface is steeper in the $w_2$ direction, giving choice B.

```python
import numpy as np

answer_p01 = "B"
minimum = np.array([2.0, -1.0])
curvatures = np.array([2.0, 8.0])
answer_p01, minimum, curvatures
```

### Answer check

```python
assert answer_p01 == "B"
assert np.allclose(minimum, [2.0, -1.0], atol=1e-12)
assert curvatures[1] > curvatures[0]
```
