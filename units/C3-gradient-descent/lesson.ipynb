## Unit goals

By the end of this unit you can:

- read a loss function as a **landscape** — contour plots over parameter
  space, minima as basin bottoms, curvature as steepness — and build such
  surfaces on a grid with broadcasting;
- run the **descent step** $w \leftarrow w - \eta \nabla L(w)$ by hand and
  in code, with the explicit-bias C2 forms
  $\hat y_i = \sum_k X_{ik} w_k + b$ and
  $L = \frac1n \sum_i (\hat y_i - y_i)^2$;
- predict descent on quadratic bowls **exactly**: the error factor
  $1 - 2a\eta$, geometric convergence, the overshoot regime, and the
  divergence threshold — and prove the 1-D contraction
  $w_{t+1} = (1 - 2\eta) w_t$ with its convergence condition $0 < \eta < 1$;
- diagnose the **learning rate** from behavior (crawl / smooth descent /
  damped oscillation / explosion), retune $\eta$ across mean-vs-sum loss
  scalings, and map convergence boundaries empirically;
- fit the C2 linear model **end to end** by gradient descent — loss curve,
  stopping judgment, noise floor — the unit's payoff;
- use **stochastic (mini-batch) gradients** as noisy-but-cheap estimates:
  prove the batch gradient is right on average (F5 expectation), read
  jittery loss curves, account for cost per step, and manage the
  $\eta$-dependent noise floor, including with a decaying schedule.

## Prerequisites (recap in one breath each)

- **F4-multivar-calculus** — gradients point steepest-uphill, so
  $-\nabla L$ is the local downhill compass; sum-of-squares gradients in
  component form; central-difference checking.
- **C2-linear-models** — the model $\hat y_i = \sum_k X_{ik} w_k + b$
  (explicit bias $b$), the MSE loss $L = \frac1n\sum_i(\hat y_i - y_i)^2$
  and its gradients at given parameters, L2 penalties.
  C2 evaluates and differentiates; this unit *iterates*.
- Through them: F1's broadcasting/axis sums and seeded randomness, F3's
  matrix view, F5's expectation and sampling language (for stochastic
  gradients), C1's train-vocabulary.
