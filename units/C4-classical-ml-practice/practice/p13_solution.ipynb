# C4-classical-ml-practice — Practice p13 — Solution

With five equal folds on 30 rows, each fold has $m=30/5=6$ validation rows and its model trains on the other $30-6=24$.  By definition the folds are a disjoint partition whose union is the dataset, so each row belongs to one and only one validation fold.

For equal size $m$, the unweighted mean is
$$\frac1K\sum_{f=1}^K a_f=\frac1K\sum_f\frac{c_f}{m}=\frac{\sum_f c_f}{Km}=\frac{\sum_f c_f}{n},$$
because $n=Km$; this is exactly pooled accuracy.  Equal sizes are necessary for that unweighted identity: for fold sizes 10 and 20 with correct counts 10 and 10, the mean of fold accuracies is $(1+1/2)/2=3/4$, while pooled accuracy is $20/30=2/3$.

```python
import numpy as np

correct = np.array([5, 4, 6, 5, 4])
fold_sizes = np.full(5, 6)
fold_accs = correct / fold_sizes
mean_acc = float(fold_accs.mean())
pooled_acc = float(correct.sum() / fold_sizes.sum())
unequal_sizes = np.array([10, 20])
unequal_correct = np.array([10, 10])
mean_uneq = float((unequal_correct / unequal_sizes).mean())
pooled_uneq = float(unequal_correct.sum() / unequal_sizes.sum())

fold_accs, mean_acc, pooled_acc, mean_uneq, pooled_uneq
```

The five concrete fold accuracies average to 0.8, matching 24 correct predictions pooled over 30 validation appearances. In the unequal counterexample, 0.75 differs from 2/3 because an unweighted fold mean gives the 10-row fold the same influence as the 20-row fold.

### Answer check

```python
assert np.allclose(fold_accs, np.array([5/6, 4/6, 1.0, 5/6, 4/6]))
assert np.isclose(mean_acc, 0.8)
assert np.isclose(pooled_acc, 0.8) and np.isclose(mean_acc, pooled_acc)
assert np.isclose(mean_uneq, 0.75)
assert np.isclose(pooled_uneq, 2/3) and not np.isclose(mean_uneq, pooled_uneq)
assert 30 - 6 == 24
```
