# C4-classical-ml-practice — Practice p04 — Solution

```python
import numpy as np
from fractions import Fraction

X_train = np.array([[1, 1], [2, 3], [4, 1], [3, 2], [0, 2]], dtype=float)
y_train = np.array([0, 0, 1, 1, 0])
q_point = np.array([2.0, 2.0])
squared_distances = ((X_train - q_point) ** 2).sum(axis=1)
nearest_rows = np.argsort(squared_distances)[:3]
majority_label = int(np.mean(y_train[nearest_rows]) >= 0.5)
mean_nearest_d2 = Fraction(int(squared_distances[nearest_rows].sum()), 3)
S = mean_nearest_d2.numerator + mean_nearest_d2.denominator - (2 if majority_label == 1 else 0)
letter_by_value = {3: "A", 4: "B", 5: "C", 7: "D", 9: "E"}
answer_p04 = letter_by_value[S]
nearest_rows, majority_label, mean_nearest_d2, S, answer_p04
```

The three nearest rows are 1, 3, and 0, with squared distances 1, 1, and 2 and labels 0, 1, and 0. Thus the majority class is 0, the mean squared distance is 4/3, and the normal-form rule gives S = 4 + 3 = 7, which is choice D.

### Answer check

```python
assert np.array_equal(nearest_rows, np.array([1, 3, 0]))
assert majority_label == 0
assert mean_nearest_d2 == Fraction(4, 3)
assert S == 7
expected_letter = "D"
assert answer_p04 == expected_letter
```
