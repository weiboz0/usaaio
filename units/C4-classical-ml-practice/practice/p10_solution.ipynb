# C4-classical-ml-practice — Practice p10 — Solution

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier

SEED = 20260804
X, y = load_iris(return_X_y=True)
perm = np.random.default_rng(SEED).permutation(150)
X, y = X[perm], y[perm]
folds = np.array_split(np.arange(150), 5)
scores = []
for fold_number in range(5):
    validation_idx = folds[fold_number]
    training_idx = np.concatenate(folds[:fold_number] + folds[fold_number + 1:])
    model = KNeighborsClassifier(n_neighbors=5)
    model.fit(X[training_idx], y[training_idx])
    scores.append(model.score(X[validation_idx], y[validation_idx]))

fold_scores = np.array(scores, dtype=float)
cv_mean = float(fold_scores.mean())
n_train_per_fold = int(150 - folds[0].size)

fold_scores, cv_mean, n_train_per_fold
```

The seeded permutation breaks iris’s original class ordering before the five disjoint 30-row validation folds are formed. Each fresh model trains on the concatenated other four folds, so every score uses 120 training rows and the mean held-out accuracy is about 0.9533.
