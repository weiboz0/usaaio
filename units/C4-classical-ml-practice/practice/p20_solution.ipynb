# C4-classical-ml-practice — Practice p20 — Solution

```python
import numpy as np
import pandas as pd

beans = pd.read_csv("data/beans.csv")
n_rows = int(beans.shape[0])
n_cols = int(beans.shape[1])
class_counts = beans["species"].value_counts()
mean_mass = float(beans["mass_g"].mean())
mass_by_species = beans.groupby("species")["mass_g"].mean()
n_heavy = int((beans["mass_g"] > 0.7).sum())
n_heavy_cava = int(((beans["mass_g"] > 0.7) & (beans["species"] == "cava")).sum())
max_moisture = float(beans["moisture_pct"].max())
X = beans[["length_mm", "width_mm", "mass_g", "moisture_pct"]].to_numpy()
y = beans["species"].to_numpy()

n_rows, n_cols, class_counts, mean_mass, mass_by_species, n_heavy, n_heavy_cava, max_moisture
```

Direct pandas expressions show a 90-by-5 table with 30 examples per species, overall mean mass 0.610433 g, 33 heavy beans, and 28 heavy `cava` beans. Grouping gives species means of about 0.388667, 0.625167, and 0.817467 g for `alba`, `brio`, and `cava`, while the maximum moisture is 13.57.

### Answer check

```python
assert (n_rows, n_cols) == (90, 5)
assert class_counts.to_dict() == {"brio": 30, "cava": 30, "alba": 30}
assert np.isclose(mean_mass, 0.6104333333333334)
assert np.allclose(mass_by_species.loc[["alba", "brio", "cava"]], np.array([0.38866666666666666, 0.6251666666666666, 0.8174666666666667]))
assert n_heavy == 33 and n_heavy_cava == 28
assert np.isclose(max_moisture, 13.57)
assert X.shape == (90, 4) and y.shape == (90,)
```
