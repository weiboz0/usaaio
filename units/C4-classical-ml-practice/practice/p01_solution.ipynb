# C4-classical-ml-practice — Practice p01 — Solution

```python
import pandas as pd

df = pd.DataFrame({
    "species": ["alba", "cava", "alba", "brio", "cava"],
    "mass_g": [0.42, 0.85, 0.39, 0.61, 0.79],
})
matching_rows = ((df["species"] == "cava") & (df["mass_g"] > 0.7)).sum()
answer_p01 = "B"
matching_rows, answer_p01
```

Each parenthesized comparison creates a Boolean Series aligned to the DataFrame rows, and `&` combines those masks elementwise. Both `cava` rows exceed 0.7 g, so summing the combined mask gives 2 and identifies choice B.

### Answer check

```python
assert int(matching_rows) == 2
expected_letter = "B"
assert answer_p01 == expected_letter
```
