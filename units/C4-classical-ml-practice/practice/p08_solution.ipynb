# C4-classical-ml-practice — Practice p08 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X_train = np.column_stack([rng.normal(20.0, 5.0, 12),
                           rng.normal(3000.0, 400.0, 12)])
X_test = np.column_stack([rng.normal(20.0, 5.0, 4),
                          rng.normal(3000.0, 400.0, 4)])

def standardize(X_tr, X_te):
    mu = X_tr.mean(axis=0)
    sd = X_tr.std(axis=0)
    return (X_tr - mu) / sd, (X_te - mu) / sd, mu, sd

Z_train, Z_test, mu, sd = standardize(X_train, X_test)
train_mean_max = float(np.abs(Z_train.mean(axis=0)).max())
train_std_gap = float(np.abs(Z_train.std(axis=0) - 1).max())
z_test_first = Z_test[0]

mu, sd, train_mean_max, train_std_gap, z_test_first
```

Both parameter vectors come exclusively from the training matrix and are then reused unchanged for the test transform, preventing test-data leakage. The standardized training columns have mean zero and population standard deviation one up to floating-point rounding, whereas the first test row maps to approximately (0.9682, 0.5729).
