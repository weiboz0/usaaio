# C4-classical-ml-practice — Practice p12 — Solution

Let $r_A=(q_1-A_1)^2$ and $r_B=(q_1-B_1)^2$.  Since all three first coordinates lie in $[0,1]$, both $r_A$ and $r_B$ lie in $[0,1]$, so $r_A-r_B\geq -1$.  Writing $h=(q_2-A_2)^2-(q_2-B_2)^2>1$, the complete squared-distance difference is
$$d_A^2-d_B^2=(r_A-r_B)+h>-1+1=0.$$
Thus $d_A^2>d_B^2$ and B is the unique raw 1-NN, independently of the first-coordinate choices.

For the two training rows, standardization divides each squared coordinate difference by that coordinate's training variance $\sigma_j^2$.  A feature with large variance therefore receives less distance weight, while a low-variance feature receives more; this variance-derived normalization makes contributions comparable instead of imposing an arbitrary weight.

```python
import numpy as np

P = np.array([[0.0, 440.0],
              [1.0, 405.0]])
labels = np.array(["red", "blue"])
q = np.array([0.1, 410.0])

d2_raw = ((P - q) ** 2).sum(axis=1)
nn_raw = labels[np.argmin(d2_raw)]
mu = P.mean(axis=0)
sigma = P.std(axis=0)
P_scaled = (P - mu) / sigma
q_scaled = (q - mu) / sigma
d2_scaled = ((P_scaled - q_scaled) ** 2).sum(axis=1)
nn_scaled = labels[np.argmin(d2_scaled)]
raw_hypothesis_lhs = (q[1] - P[0, 1]) ** 2 - (q[1] - P[1, 1]) ** 2

d2_raw, nn_raw, d2_scaled, nn_scaled, raw_hypothesis_lhs
```

Numerically, the raw squared distances are 900.01 and 25.81, and the second-feature difference in the proof hypothesis is 875 > 1, so raw 1-NN chooses blue. Training-only standardization changes the distances to about 2.9788 and 3.3216, making red the nearer label and concretely demonstrating the flip.

### Answer check

```python
assert np.allclose(d2_raw, np.array([900.01, 25.81]), atol=1e-9, rtol=0)
assert nn_raw == "blue"
assert raw_hypothesis_lhs > 1 and np.isclose(raw_hypothesis_lhs, 875.0, atol=1e-9, rtol=0)
assert np.allclose(d2_scaled, np.array([2.9787755102040823, 3.321632653061225]), atol=1e-9, rtol=0)
assert nn_scaled == "red" and nn_scaled != nn_raw
```
