# C4-classical-ml-practice — Practice p05 — Solution

```python
import numpy as np
import pandas as pd

FEATURES = ["length_mm", "width_mm", "mass_g", "moisture_pct"]

df = pd.read_csv("data/beans.csv")
n_rows = int(df.shape[0])
class_counts = df["species"].value_counts()
X = df[FEATURES].to_numpy(dtype=float)
y = df["species"].to_numpy()

n_rows, class_counts, X.shape, y.shape
```

A single ordered column-list selection preserves the required feature order before converting the table to a floating-point NumPy matrix. The dataset has 90 rows, and `value_counts()` shows 30 rows for each of the three species.

### Answer check

```python
assert n_rows == 90 and isinstance(n_rows, int)
assert class_counts.to_dict() == {"brio": 30, "cava": 30, "alba": 30}
assert X.shape == (90, 4) and np.issubdtype(X.dtype, np.floating)
assert y.shape == (90,)
assert np.array_equal(X, df[FEATURES].to_numpy(dtype=float))
```
