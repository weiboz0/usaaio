# C4-classical-ml-practice — Practice p07 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X_train = np.vstack([rng.normal((2.0, 3.0), 0.9, (20, 2)),
                     rng.normal((5.0, 1.0), 0.9, (20, 2))])
y_train = np.array([0] * 20 + [1] * 20)
X_query = np.array([[2.5, 2.5], [4.5, 1.5], [3.5, 2.0], [1.5, 4.0]])

def knn_predict(X_tr, y_tr, X_q, k):
    distances = ((X_tr[None, :, :] - X_q[:, None, :]) ** 2).sum(axis=2)
    neighbor_labels = y_tr[np.argsort(distances, axis=1)[:, :k]]
    return (neighbor_labels.mean(axis=1) >= 0.5).astype(int)

d2 = ((X_train[None, :, :] - X_query[:, None, :]) ** 2).sum(axis=2)
preds = knn_predict(X_train, y_train, X_query, 5)
train_acc_k1 = float((knn_predict(X_train, y_train, X_train, 1) == y_train).mean())

preds, train_acc_k1
```

Broadcasting constructs all query–training differences at once, and sorting each distance row selects the five labels whose mean determines the binary majority. Training accuracy for 1-NN is 1.0 by construction because every training row is its own zero-distance nearest neighbor (the seeded data contain no duplicate conflict).
