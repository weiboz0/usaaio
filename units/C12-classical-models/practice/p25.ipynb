# C12-classical-models — Practice p25

**Type:** scenario analysis · **Difficulty:** core · **Concepts:** k-means, classical-model-comparison

**Time budget:** 45 minutes

A retail team clusters customers using annual spending in dollars, visit count, and return rate.
With raw features, `k=4` gives inertia 8.1e9; after standardization, inertia is 920. Another run
with the same `k` gives different cluster numbers and 875 inertia. The team claims 875 proves the
second run is correct and plans to call cluster 0 “high value.”

Audit the claim:

1. explain why raw and standardized inertia values are not directly comparable;
2. explain arbitrary cluster-label permutations and how to compare partitions via co-clustering;
3. distinguish lower objective from global optimality, stability, and business validity;
4. propose a pinned `k`, initialization, `n_init`, seed, scaling, and empty-cluster protocol;
5. give an evidence plan across several `k` values using inertia shape, repeated-seed/resample
   stability, centroid interpretation in original units, and domain review;
6. contrast this unlabeled task with supervised logistic/SVM/tree selection and state what labels
   would be needed before using prediction metrics.

Do not invent customer labels or declare any run globally optimal.
