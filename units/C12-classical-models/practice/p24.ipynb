# C12-classical-models — Practice p24

**Type:** scenario analysis · **Difficulty:** core · **Concepts:** ensemble-learning, bagging-and-boosting, classical-model-comparison

**Time budget:** 45 minutes

Four teams face different binary-classification settings:

- Team A has three already-fitted, well-calibrated but architecturally different models.
- Team B has one unstable deep tree and enough CPU to fit independent resamples.
- Team C has many correlated tree predictors and wants additional diversity at each split.
- Team D has shallow stumps with systematic underfit, but labels contain a small suspected noisy subset.

For each team choose first among hard/soft voting, bagging, random forest, or AdaBoost. Then:

1. state the construction and aggregation rule;
2. identify the intended bias/variance effect;
3. name the principal failure mode in that team's setting;
4. pin seed/state evidence and a fair cross-validation comparison;
5. explain why the four names are not interchangeable;
6. for Team D, write the exact AdaBoost error, learner coefficient, row-weight update, and the
   warning caused by repeatedly misclassified noisy rows.

There may be a justified second choice, but every choice must be tied to the scenario rather than
to a universal ranking.


## Your four-team decision table

Write one evidence-based row per team, followed by the AdaBoost ledger and comparison protocol.
