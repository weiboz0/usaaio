# C12-classical-models — Practice p23

**Type:** scenario analysis · **Difficulty:** core · **Concepts:** decision-trees, tree-split-criteria

**Time budget:** 45 minutes

A decision tree has training accuracy 1.00 and validation accuracy 0.69. It has depth 18, 73
leaves, and many leaves with one row. Repeating the train/validation split changes validation
accuracy by up to 0.12. A teammate proposes standardizing every feature and increasing depth.

Diagnose the failure and propose a defensible response:

1. connect the observed instability to tree variance and overfitting;
2. explain why standardization does not repair axis-aligned order-based splits;
3. compare `max_depth`, `min_samples_split`, `min_samples_leaf`, and cost-complexity `ccp_alpha`,
   including how each changes candidate trees;
4. give a leakage-safe cross-validation protocol and a tie rule favoring the smaller tree when
   mean scores tie;
5. list the train/CV/held-out metrics and structural state to report;
6. explain when bagging is a reasonable next experiment and why it does not erase the need for a
   controlled base tree and held-out test.

Your conclusion must select an initial control and justify it from the supplied evidence.
