# C12-classical-models — Practice p17 — Solution


With centroids fixed, each row contributes one nonnegative squared distance.
Nearest-centroid assignment minimizes every term independently, so
$J(a^{new},\mu)\le J(a^{old},\mu)$. Equality holds exactly when every old assigned centroid is
also nearest for its row; deterministic relabeling among equal-distance nearest centroids can
change labels without changing WCSS.

For a nonempty cluster $C$ with mean $\bar x_C$,
\[
\sum_{i\in C}\|x_i-\mu\|^2
=\sum_{i\in C}\|x_i-\bar x_C\|^2+|C|\|\bar x_C-\mu\|^2,
\]
because $\sum_{i\in C}(x_i-\bar x_C)=0$. The mean update cannot increase WCSS, and equality
holds exactly when the old centroid already equals $\bar x_C$. Chaining the assignment and mean
steps proves nonincrease, not strict decrease or global optimality.

An empty cluster has no mean, so repair must be included in the comparison. Process each empty
index in order and move the farthest eligible row, retaining at least one donor row. Give the
moved row its own singleton centroid: its new contribution is zero, no larger than its removed
donor contribution. Remeaning the donor's remaining nonempty rows cannot increase their cost;
remeaning every other nonempty cluster also minimizes its fixed-membership cost. Applying this
argument after each deterministic move proves the repaired assignment-plus-update WCSS is no
larger than the pre-repair assignment WCSS. The policy therefore restores a well-defined update
while preserving nonincrease.


```python
import numpy as np

ATOL = 1e-12
RTOL = 1e-10
cluster_p17 = np.array([[0.,0.],[2.,0.],[4.,0.]])
mean_p17 = cluster_p17.mean(axis=0)
old_centroid_p17 = np.array([0.,0.])
old_wcss_p17 = float(((cluster_p17-old_centroid_p17)**2).sum())
new_wcss_p17 = float(((cluster_p17-mean_p17)**2).sum())

repair_X_p17 = np.array([[0.0], [2.0], [10.0]])
repair_old_centroids_p17 = np.array([[0.0], [0.0], [10.0]])
repair_assignment_labels_p17 = np.array([0, 0, 2], dtype=np.int64)
repair_assignment_wcss_p17 = float(np.sum(
    (repair_X_p17 - repair_old_centroids_p17[repair_assignment_labels_p17]) ** 2
))
repair_repaired_labels_p17 = np.array([0, 1, 2], dtype=np.int64)
repair_updated_centroids_p17 = np.array([[0.0], [2.0], [10.0]])
repair_repaired_wcss_p17 = float(np.sum(
    (repair_X_p17 - repair_updated_centroids_p17[repair_repaired_labels_p17]) ** 2
))
repair_witness_p17 = {
    "assignment_labels": repair_assignment_labels_p17,
    "repaired_labels": repair_repaired_labels_p17,
    "assignment_wcss": repair_assignment_wcss_p17,
    "repaired_wcss": repair_repaired_wcss_p17,
}
```

### Answer check


```python
assert np.allclose(mean_p17, [2.0,0.0], atol=ATOL, rtol=RTOL)
assert np.isclose(old_wcss_p17, 20.0, atol=ATOL, rtol=RTOL)
assert np.isclose(new_wcss_p17, 8.0, atol=ATOL, rtol=RTOL)
assert new_wcss_p17 <= old_wcss_p17
assert np.array_equal(repair_witness_p17["assignment_labels"], [0,0,2])
assert np.array_equal(repair_witness_p17["repaired_labels"], [0,1,2])
assert np.isclose(repair_witness_p17["assignment_wcss"], 4.0, atol=ATOL, rtol=RTOL)
assert np.isclose(repair_witness_p17["repaired_wcss"], 0.0, atol=ATOL, rtol=RTOL)
assert repair_witness_p17["repaired_wcss"] <= repair_witness_p17["assignment_wcss"]
```
