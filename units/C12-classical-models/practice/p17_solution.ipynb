# C12-classical-models — Practice p17 — Solution


Let $a$ be old assignments and $\mu$ fixed. Assigning each $x_i$ to a nearest
centroid minimizes its individual squared-distance term, so summing gives
$J(a^{new},\mu)\le J(a,\mu)$. A deterministic tie selects one equal minimizer and leaves the
objective unchanged.

For a nonempty cluster $C$, with $\bar x=|C|^{-1}\sum_{i\in C}x_i$,
\[
\sum_{i\in C}\|x_i-\mu\|^2
=\sum_{i\in C}\|x_i-\bar x\|^2+|C|\|\bar x-\mu\|^2,
\]
because the cross term contains $\sum_{i\in C}(x_i-\bar x)=0$. Therefore the mean is the
centroid minimizer and $J(a^{new},\mu^{new})\le J(a^{new},\mu)$. Chaining proves
nonincrease. Equality can occur at ties or a fixed point, and block-coordinate descent can stop
at a local rather than global minimum. An empty cluster has no mean, so the identity cannot
define its update; the deterministic farthest-eligible-row repair makes every cluster nonempty,
after which the mean step and its proof apply to the repaired assignment.


```python
import numpy as np
ATOL = 1e-12
RTOL = 1e-10
cluster_p17 = np.array([[0.,0.],[2.,0.],[4.,0.]])
mean_p17 = cluster_p17.mean(axis=0)
old_centroid_p17 = np.array([0.,0.])
old_wcss_p17 = float(((cluster_p17-old_centroid_p17)**2).sum())
new_wcss_p17 = float(((cluster_p17-mean_p17)**2).sum())
```

### Answer check


```python
assert np.allclose(mean_p17, [2.0,0.0], atol=ATOL, rtol=RTOL)
assert np.isclose(old_wcss_p17, 20.0, atol=ATOL, rtol=RTOL)
assert np.isclose(new_wcss_p17, 8.0, atol=ATOL, rtol=RTOL)
assert new_wcss_p17 <= old_wcss_p17
```
