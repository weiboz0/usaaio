# C12-classical-models — Practice p16

**Type:** proof · **Difficulty:** core · **Concepts:** decision-trees, tree-split-criteria

**Time budget:** 45 minutes

A parent contains labels `(0,0,0,0,1,1,1,1)`. Split S gives children
`(0,0,0,1)` and `(0,1,1,1)`. Split T gives `(0,0)` and `(0,0,1,1,1,1)`.

1. Derive binary Gini $2p(1-p)$ from $1-p^2-(1-p)^2$.
2. Compute the parent impurity and each child impurity exactly.
3. Compute each sample-count-weighted child impurity and information gain.
4. Prove algebraically that maximizing gain is equivalent to minimizing weighted child impurity
   for candidates at the same parent.
5. State which split is selected and why an unweighted average of child impurities can reverse or
   distort a decision when child sizes differ.

Reasoning is required. Use exact fractions and show every weight; decimal-only conclusions receive
no credit.
