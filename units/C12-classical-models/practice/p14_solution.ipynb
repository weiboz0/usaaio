# C12-classical-models — Practice p14 — Solution


From $p/(1-p)=e^z$, $p=e^z/(1+e^z)=\sigma(z)$ and quotient differentiation
gives $p'=p(1-p)$.  The observed-event likelihood is $p^y(1-p)^{1-y}$, so
$\ell=-y\log p-(1-y)\log(1-p)$.  Hence
\[
\frac{d\ell}{dz}=\left(-\frac yp+\frac{1-y}{1-p}\right)p(1-p)=p-y;
\]
for $y=1$ it is $p-1$, and for $y=0$ it is $p$.

For $X:(N,D)$, $w:(D)$, $b:()$, $z=Xw+b\mathbf1:(N)$, and $p-y:(N)$,
coordinatewise chain rule gives $\nabla_wL=X^T(p-y)/N:(D)$ and
$\partial L/\partial b=\mathbf1^T(p-y)/N:()$.  The single $1/N$ comes from the single
outer mean.

If $z\ge0$, $-yz+\log(1+e^z)=(1-y)z+\log(1+e^{-z})$; if $z<0$, it is
$-yz+\log(1+e^z)$.  These combine as
$\max(z,0)-yz+\log(1+e^{-|z|})$.


```python
import numpy as np
ATOL = 1e-12
RTOL = 1e-10
z_p14 = np.array([-3.0, 0.0, 2.0])
y_p14 = np.array([0.0, 1.0, 1.0])
p_p14 = 1.0 / (1.0 + np.exp(-z_p14))
stable_loss_p14 = np.maximum(z_p14, 0.0) - y_p14*z_p14 + np.log1p(np.exp(-np.abs(z_p14)))
gradient_p14 = p_p14 - y_p14
```

### Answer check


```python
expected_loss_p14 = np.array([np.log1p(np.exp(-3.0)), np.log(2.0), np.log1p(np.exp(-2.0))])
assert np.allclose(stable_loss_p14, expected_loss_p14, atol=ATOL, rtol=RTOL)
assert np.allclose(gradient_p14, [1/(1+np.exp(3.0)), -0.5, -1/(1+np.exp(2.0))], atol=ATOL, rtol=RTOL)
```
