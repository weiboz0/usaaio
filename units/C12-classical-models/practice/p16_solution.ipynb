# C12-classical-models — Practice p16 — Solution


Binary Gini is $1-p^2-(1-p)^2=2p(1-p)$. The parent has $p=1/2$, so
$G_P=1/2$. For S, both children have Gini $3/8$, giving weighted impurity $3/8$ and gain
$1/8$. For T, the left child is pure and the right has $p=4/6=2/3$, Gini $4/9$; hence
$G_T=(2/8)0+(6/8)(4/9)=1/3$ and gain $1/6$. T wins. In general
$\mathrm{gain}=G_P-G_{children}$, so for a fixed parent maximizing gain is exactly minimizing
weighted child impurity. An unweighted average treats a two-row child like a six-row child and
can distort the choice.


```python
from fractions import Fraction
parent_p16 = Fraction(1, 2)
s_weighted_p16 = Fraction(4,8)*Fraction(3,8) + Fraction(4,8)*Fraction(3,8)
t_weighted_p16 = Fraction(2,8)*0 + Fraction(6,8)*Fraction(4,9)
s_gain_p16 = parent_p16 - s_weighted_p16
t_gain_p16 = parent_p16 - t_weighted_p16
```

### Answer check


```python
assert s_weighted_p16 == Fraction(3,8) and s_gain_p16 == Fraction(1,8)
assert t_weighted_p16 == Fraction(1,3) and t_gain_p16 == Fraction(1,6)
assert t_weighted_p16 < s_weighted_p16 and t_gain_p16 > s_gain_p16
```
