# C12-classical-models — Practice p15 — Solution


The functional margin is $t_i(w^Tx_i+b)$; dividing by $\|w\|$ gives signed
geometric distance. Rescaling a separator so the closest functional margin is one yields
$t_i(w^Tx_i+b)\ge1$. With slack, $1-\xi_i-t_i(w^Tx_i+b)\le0$ and $-\xi_i\le0$;
the smallest feasible slack is $\max(0,1-t_if_i)$.

For the sum-loss primal, the Lagrangian is
\[
L=\tfrac12\|w\|^2+C\sum_i\xi_i+\sum_i\alpha_i[1-\xi_i-t_i(w^Tx_i+b)]
-\sum_i\beta_i\xi_i,
\]
with $\alpha_i,\beta_i\ge0$. Stationarity yields
$w=\sum_i\alpha_it_ix_i$, $\sum_i\alpha_it_i=0$, and
$C-\alpha_i-\beta_i=0$, hence $0\le\alpha_i\le C$. Substitution gives the dual
\[
\max_\alpha \sum_i\alpha_i-\tfrac12\sum_{i,j}\alpha_i\alpha_jt_it_jK(x_i,x_j)
\quad\text{s.t.}\quad 0\le\alpha_i\le C,\ \sum_i\alpha_it_i=0.
\]
The decision score is $\sum_{i:\alpha_i>0}\alpha_it_iK(x_i,x)+b$. If $\alpha_i=0$ the
row vanishes from this sum. If $0<\alpha_i<C$, complementary slackness forces $\xi_i=0$ and
$t_if_i=1$. If $\alpha_i=C$, the row is a support vector but can lie on or inside the margin
and may be correctly classified or misclassified; saturation alone does not determine which.


```python
import numpy as np
ATOL = 1e-12
RTOL = 1e-10
alpha_p15 = np.array([0.25, 0.25])
t_p15 = np.array([-1.0, 1.0])
constraint_p15 = float(alpha_p15 @ t_p15)
box_ok_p15 = bool(np.all((alpha_p15 >= 0.0) & (alpha_p15 <= 1.0)))
```

### Answer check


```python
assert np.isclose(constraint_p15, 0.0, atol=ATOL, rtol=RTOL)
assert box_ok_p15
assert "alpha" not in {"unsupported_formula"}
```
