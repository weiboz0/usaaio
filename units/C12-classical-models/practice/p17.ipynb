# C12-classical-models — Practice p17

**Type:** proof · **Difficulty:** core · **Concepts:** k-means, lloyd-algorithm

**Time budget:** 45 minutes

Prove that one complete Lloyd iteration cannot increase WCSS for nonempty clusters.

1. With centroids fixed, prove nearest-centroid assignment minimizes WCSS over assignments,
   including why a deterministic distance tie does not change the objective.
2. With assignments fixed, expand
   $\sum_{i\in C}\|x_i-\mu\|^2$ around the cluster mean $\bar x_C$ and prove the cross term is zero.
3. Conclude that replacing $\mu$ by $\bar x_C$ cannot increase the cluster contribution.
4. Chain both block updates to obtain a nonincreasing objective sequence.
5. Explain why this proves neither strict decrease at every iteration nor convergence to the global
   minimum.
6. State exactly where the proof needs modification when a cluster is empty and how Session 6's
   repair policy restores a well-defined update.

Reasoning is required. Every inequality and equality condition must be identified.
