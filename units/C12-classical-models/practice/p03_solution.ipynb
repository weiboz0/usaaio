# C12-classical-models — Practice p03 — Solution


For S, both children have class proportions $3/4,1/4$, hence Gini $3/8$;
equal child sizes leave weighted impurity $3/8$.  For T, the two-row child is pure and the
six-row child has proportions $1/3,2/3$, hence Gini $4/9$.  Its weighted impurity is
$(2/8)0+(6/8)(4/9)=1/3<3/8$.  Thus T gives larger gain and **B** is correct.


```python
from fractions import Fraction
answer_p03 = "B"
s_children_p03 = (Fraction(3, 8), Fraction(3, 8))
s_weighted_p03 = Fraction(4, 8) * s_children_p03[0] + Fraction(4, 8) * s_children_p03[1]
t_children_p03 = (Fraction(0, 1), Fraction(4, 9))
t_weighted_p03 = Fraction(2, 8) * t_children_p03[0] + Fraction(6, 8) * t_children_p03[1]
```

### Answer check


```python
assert answer_p03 == "B"
assert s_weighted_p03 == Fraction(3, 8)
assert t_weighted_p03 == Fraction(1, 3)
assert t_weighted_p03 < s_weighted_p03
```
