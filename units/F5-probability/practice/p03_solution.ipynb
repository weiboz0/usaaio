# F5-probability — Practice p03 — Solution

Independence makes $\operatorname{Cov}(X,Y)=0$, so $\operatorname{Var}(X\pm Y)=\operatorname{Var}(X)+\operatorname{Var}(Y)$. Both variances therefore equal $4+9=13$, which is choice C.

```python
import numpy as np

var_sum = 4 + 9
var_difference = 4 + 9
answer_p03 = "C"
```

### Answer check

```python
assert var_sum == 13 and var_difference == 13
derived_choice = "C" if (var_sum, var_difference) == (13, 13) else None
assert answer_p03 == derived_choice
```
