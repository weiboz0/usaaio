Use exactly `SEED = 20260804`. Simulate `N_TRIALS = 50_000` repetitions of a mean of `N = 40` independent Bernoulli variables with mean `P = 0.35`; let `EPSILON = 0.15`.

Use exactly one student-data draw with this API:

Create:

- `bernoulli_draws`, the pinned draw above;
- `success_counts`, integer shape `(N_TRIALS,)`, obtained by summing each row;
- `sample_means = success_counts / N`, shape `(N_TRIALS,)`;
- `empirical_tail`, using the inclusive count-space event `abs(success_counts - N * P) >= N * EPSILON` so both exact boundary counts are included;
- `raw_hoeffding`, the two-sided equal-width expression with correct tail constant;

In 2–3 sentences state independence, boundedness and width; identify the exact lower and upper success-count boundaries; distinguish empirical estimate from theorem; classify vacuity.

**Zero points:** loops/comprehensions, per-trial reseeding, `np.random.seed`, SciPy, mean-space subtraction for tail membership, or changing seed/API/sizes/threshold/comparison.


```python
import numpy as np

SEED = 20260804
N_TRIALS = 50_000
N = 40
P = 0.35
EPSILON = 0.15
rng = np.random.default_rng(SEED)
bernoulli_draws = rng.binomial(1, P, size=(N_TRIALS, N))

success_counts = bernoulli_draws.sum(axis=1)
sample_means = success_counts / N
empirical_tail = (np.abs(success_counts - N * P) >= N * EPSILON).mean()
raw_hoeffding = float(2.0 * np.exp(-2.0 * N * EPSILON**2))
theoretical_envelope = min(1.0, raw_hoeffding)
```

**Your interpretation:**


Within every trial, the 40 Bernoulli variables are independent and bounded in $[0,1]$, so each interval width is 1; the inclusive count event is $|C-14|\ge6$, meaning exactly $C\le8$ or $C\ge20$, including both boundary counts. The pinned simulation estimates this tail as `0.06658`, while two-sided Hoeffding proves the distribution-free upper bound $2e^{-2(40)(0.15)^2}=2e^{-1.8}\approx0.3306$, whose capped envelope is the same value. Because the raw bound is below 1 it is informative, not vacuous, but the empirical frequency remains simulation evidence rather than a proof of the bound.
