# F5-probability — Practice p16 — Solution

The data are reproduced in the stated draw order so the seeded samples match exactly. Definition-based covariance centers both arrays, multiplies paired deviations, and averages those products.

```python
import numpy as np

rng = np.random.default_rng(20260804)
temp = rng.normal(15, 8, 2000)
drinks = np.clip(80 - 2.0 * temp + rng.normal(0, 6, 2000), 0, None)
pastries = rng.poisson(30, 2000).astype(float)

def my_cov(a, b):
    return np.mean((a - np.mean(a)) * (b - np.mean(b)))

cov_dt = my_cov(drinks, temp)
cov_pt = my_cov(pastries, temp)
```

Pastries and temperature behave as independent, with covariance near zero, while drinks and temperature have a large negative covariance: hotter hours correspond to fewer hot drinks sold. Covariance magnitude depends on both variables' units and scales, so magnitudes from differently measured pairs should not be compared directly.
