**Type:** numeric normal-form MC · **Difficulty:** core · **Concept:** hoeffding-inequality

*Reasoning is required.* Independent $X_1,\ldots,X_{12}$ are bounded: six have interval width 1 and six have width 2. Which is the correct raw Hoeffding bound for

$$P\left(|\bar X-E\bar X|\ge\frac12\right)?$$

- **A.** $\exp(-12/5)$
- **B.** $2\exp(-12/5)$
- **C.** $2\exp(-6/5)$
- **D.** $2\exp(-1/24)$
- **E.** $\min\{1,2\exp(-12/5)\}$ is the theorem's raw normal form

Record `choice`, `squared_width_sum`, reduced positive integers `exponent_num`, `exponent_den` for $2e^{-p/q}$, and floating `raw_bound`.

Your derivation must list independence, boundedness and widths; compute $\sum_i(b_i-a_i)^2$; define upper event $U$ and lower event $L$; use $P(U\cup L)\le P(U)+P(L)$ immediately before adding the one-sided bounds; and classify the raw bound as informative or vacuous. Common-width shortcuts, a missing factor 2, or clipping before identifying the raw form earn no credit.


**Your derivation:**


The variables are independent and bounded, with six interval widths equal to 1 and six equal to 2. Thus

$$
\sum_{i=1}^{12}(b_i-a_i)^2=6(1^2)+6(2^2)=30.
$$

For (t=1/2), define the upper and lower events

$$
U=\{\bar X-E\bar X\ge 1/2\},
\qquad
L=\{\bar X-E\bar X\le -1/2\}.
$$

Each one-sided Hoeffding exponent is

$$
\frac{2n^2t^2}{\sum_i(b_i-a_i)^2}
=\frac{2(12)^2(1/2)^2}{30}=\frac{12}{5}.
$$

Applying the union bound before adding the one-sided bounds,

$$
P(U\cup L)\le P(U)+P(L)
\le e^{-12/5}+e^{-12/5}
=2e^{-12/5}.
$$

Therefore the correct raw-form letter is **B**, not E: clipping is a separate envelope operation, not the theorem's raw expression. Here (2e^{-12/5}\approx0.1814<1), so the raw bound is informative rather than vacuous.

```python
from fractions import Fraction
from math import gcd
import numpy as np

choice: str = "B"
squared_width_sum: int = 30
exponent_num: int = 12
exponent_den: int = 5
raw_bound: float = float(2.0 * np.exp(-exponent_num / exponent_den))

# Immutable structural and theorem-semantic checks; the written option justification is reviewed separately.
assert choice in tuple("ABCDE")
assert type(squared_width_sum) is int and squared_width_sum > 0
assert type(exponent_num) is int and type(exponent_den) is int
assert exponent_num > 0 and exponent_den > 0
assert gcd(exponent_num, exponent_den) == 1
_widths = np.array([1] * 6 + [2] * 6, dtype=int)
_expected_squared_width_sum = int((_widths**2).sum())
_expected_exponent = Fraction(2 * 12**2, 4 * _expected_squared_width_sum)
assert squared_width_sum == _expected_squared_width_sum
assert Fraction(exponent_num, exponent_den) == _expected_exponent
_expected_raw = 2.0 * np.exp(-float(_expected_exponent))
assert np.isclose(raw_bound, _expected_raw, atol=1e-15, rtol=0.0)
assert 0.0 < raw_bound < 1.0
print("p23 checks passed")
```
