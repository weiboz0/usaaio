# F5-probability — Practice p17 — Solution

For (a), scaling gives $\operatorname{Var}(w_cx_c)=w_c^2\sigma^2$, and independence makes all cross-covariances zero, so $\operatorname{Var}(y)=\sigma^2\sum_c w_c^2$. For (b), equal weights give $\sum_cw_c^2=C(1/C)^2=1/C$, hence $\operatorname{Var}(y)=\sigma^2/C$, which decreases to zero as $C$ grows.

For (c), nonnegativity gives $0\leq\sum_c(w_c-1/C)^2=\sum_cw_c^2-(2/C)\sum_cw_c+C/C^2=\sum_cw_c^2-1/C$, using $\sum_cw_c=1$. Therefore $\sum_cw_c^2\geq1/C$, with equality exactly when every $w_c=1/C$, proving the equal split minimizes variance.

For (d), with $C=5$ and $\sigma^2=4$, the equal split has variance $4[5(0.2)^2]=0.8$. The skewed weights have variance $4[0.6^2+4(0.1)^2]=4(0.40)=1.6$.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
sigma_squared = 4
equal_weights = np.full(5, 0.2)
skewed_weights = np.array([0.6, 0.1, 0.1, 0.1, 0.1])
var_equal = sigma_squared * (equal_weights**2).sum()
var_skewed = sigma_squared * (skewed_weights**2).sum()
venture_returns = rng.normal(0, np.sqrt(sigma_squared), size=(300_000, 5))
equal_returns = (venture_returns * equal_weights).sum(axis=1)
skewed_returns = (venture_returns * skewed_weights).sum(axis=1)
sim_var_equal = np.var(equal_returns)
sim_var_skewed = np.var(skewed_returns)
```

The simulations instantiate the analytic comparison with independent Gaussian venture returns. Their empirical variances should be close to 0.8 and 1.6, while the proof applies beyond the normal anchor.

### Answer check

```python
assert abs(var_equal - 0.8) < 1e-12
assert abs(var_skewed - 1.6) < 1e-12
assert var_equal < var_skewed
assert abs(sim_var_equal - var_equal) < 0.01
assert abs(sim_var_skewed - var_skewed) < 0.02
```
