# F5-probability — Practice p05 — Solution

Each row contains two independent integer draws from 1 through 6, and summing across columns produces one dice sum per trial. The seeded empirical fraction estimates the exact probability $6/36=1/6$.

```python
import numpy as np

def simulate_sums(n_trials, seed):
    rng = np.random.default_rng(seed)
    rolls = rng.integers(1, 7, size=(n_trials, 2))
    return rolls.sum(axis=1)

sums = simulate_sums(100_000, 20260804)
frac_seven = (sums == 7).sum() / sums.size
```

### Answer check

```python
assert sums.shape == (100_000,)
assert abs(frac_seven - 1 / 6) < 0.005
assert abs(frac_seven - 0.16461) < 1e-12
```
