# F5-probability — Practice p04 — Solution

The mean is $E[T]=1/4+3/2+7/4=7/2$, while $E[T^2]=1/4+9/2+49/4=17$. Thus $\operatorname{Var}(T)=17-(7/2)^2=19/4$, so $p+q=23$ and the decoded choice is E.

```python
import numpy as np
from fractions import Fraction

mean_t = Fraction(1, 4) * 1 + Fraction(1, 2) * 3 + Fraction(1, 4) * 7
second_moment = Fraction(1, 4) * 1**2 + Fraction(1, 2) * 3**2 + Fraction(1, 4) * 7**2
variance_t = second_moment - mean_t**2
p_plus_q = variance_t.numerator + variance_t.denominator
choice_by_value = {6: "A", 9: "B", 11: "C", 21: "D", 23: "E"}
answer_p04 = choice_by_value[p_plus_q]
```

### Answer check

```python
recomputed = Fraction(1, 4) * (1 - mean_t)**2 + Fraction(1, 2) * (3 - mean_t)**2 + Fraction(1, 4) * (7 - mean_t)**2
assert recomputed == variance_t == Fraction(19, 4)
assert p_plus_q == 23
assert answer_p04 == "E"
```
