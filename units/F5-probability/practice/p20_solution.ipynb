A quality audit records 200 devices. Use the joint table to produce the requested joint-table normal forms.

| | Alarm | No alarm | Total |
| --- | ---: | ---: | ---: |
| Failed | 42 | 38 | 80 |
| Passed | 18 | 102 | 120 |
| Total | 60 | 140 | 200 |

**(a)** Compute $P(\text{Failed}\mid\text{Alarm})$.

**(b)** Compute $P(\text{Alarm}\mid\text{Failed})$.

Give each in normal form $p/q$: integer numerator, positive denominator, and $\gcd(|p|,q)=1$. Show the intersection numerator and correct conditioning denominator in markdown; explain why the directions differ. State what happens if a conditioning row or column has total zero.


**Your derivation:**


Both conditionals use the same intersection count, 42, but condition on different totals:

$$
P(\text{Failed}\mid\text{Alarm})
=\frac{\#(\text{Failed}\cap\text{Alarm})}{\#(\text{Alarm})}
=\frac{42}{60}=\frac{7}{10},
$$

whereas

$$
P(\text{Alarm}\mid\text{Failed})
=\frac{\#(\text{Alarm}\cap\text{Failed})}{\#(\text{Failed})}
=\frac{42}{80}=\frac{21}{40}.
$$

The directions differ because the first denominator is the 60 devices in the Alarm column, while the second is the 80 devices in the Failed row. If the relevant conditioning row or column total is zero, the conditional probability is undefined because its denominator is zero; it must not be reported as zero.

```python
from math import gcd

failed_given_alarm_num: int = 7
failed_given_alarm_den: int = 10
alarm_given_failed_num: int = 21
alarm_given_failed_den: int = 40

# Immutable structural and table-semantic checks; written reasoning is reviewed separately.
assert type(failed_given_alarm_num) is int and type(failed_given_alarm_den) is int
assert type(alarm_given_failed_num) is int and type(alarm_given_failed_den) is int
assert failed_given_alarm_den > 0 and gcd(abs(failed_given_alarm_num), failed_given_alarm_den) == 1
assert alarm_given_failed_den > 0 and gcd(abs(alarm_given_failed_num), alarm_given_failed_den) == 1
assert 0 <= failed_given_alarm_num <= failed_given_alarm_den
assert 0 <= alarm_given_failed_num <= alarm_given_failed_den
assert failed_given_alarm_num * 60 == failed_given_alarm_den * 42
assert alarm_given_failed_num * 80 == alarm_given_failed_den * 42
print("p20 checks passed")
```
