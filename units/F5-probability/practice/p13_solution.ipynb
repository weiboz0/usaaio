# F5-probability — Practice p13 — Solution

For (a), probability-weighted moments give $\mu=1.05$ and $\operatorname{Var}(P)=E[P^2]-\mu^2=0.9475$. For (b), linearity multiplies the mean by 900, and independence permits variance additivity, so $E[T]=945$ and $\operatorname{Var}(T)=852.75$.

For (c), a 20-entry lookup table represents probabilities 7/20, 7/20, 4/20, and 2/20 exactly, and integer indices produce the requested nonuniform draws. The simulated totals can then be compared directly with the exact results from (b).

```python
import numpy as np

values = np.array([0., 1., 2., 3.])
probs = np.array([0.35, 0.35, 0.2, 0.1])
mu = (values * probs).sum()
second_moment = (values**2 * probs).sum()
var_p = second_moment - mu**2
mu_T = 900 * mu
var_T = 900 * var_p

SEED = 20260804
rng = np.random.default_rng(SEED)
sampling_table = np.repeat(values, np.array([7, 7, 4, 2]))
parcel_counts = sampling_table[rng.integers(0, 20, size=(50_000, 900))]
totals = parcel_counts.sum(axis=1)
sim_mu_T = np.mean(totals)
sim_var_T = np.mean((totals - sim_mu_T)**2)
mu_gap = abs(mu_T - sim_mu_T)
var_gap = abs(var_T - sim_var_T)
```

The simulated mean 945.168 and variance 849.042 are close to 945 and 852.75, so the seeded experiment confirms the exact sum rules within ordinary Monte Carlo error. The modest gaps are small relative to the scales of the total mean and variance.
