# F5-probability — Practice p11 — Solution

For (a), independence implies $E[s_cx_c]=E[s_c]E[x_c]=0$. For (b), $\operatorname{Var}(s_cx_c)=E[s_c^2x_c^2]-E[s_cx_c]^2$; independence factors the first term as $E[s_c^2]E[x_c^2]=\sigma_s^2\sigma^2$, since both means are zero.

For (c), the products $s_cx_c$ are independent across $c$ because they are functions of disjoint mutually independent pairs, so variance additivity gives $\operatorname{Var}(y)=\sum_c\operatorname{Var}(s_cx_c)=C\sigma_s^2\sigma^2$. Requiring this to equal $\sigma^2$ gives $\sigma_s^2=1/C$, hence the nonnegative standard deviation is $\sigma_s=1/\sqrt C$.

For the anchor, $C=4$ gives $\sigma_s=1/2$. Drawing independent normal arrays with standard deviations $1/2$ and $3$, multiplying corresponding entries, and summing each row instantiates the derived construction; the target variance is 9.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
C = 4
sigma_squared = 9
sigma_s = 1 / np.sqrt(C)
s_samples = rng.normal(0, sigma_s, size=(400_000, C))
x_samples = rng.normal(0, np.sqrt(sigma_squared), size=(400_000, C))
y_samples = (s_samples * x_samples).sum(axis=1)
sim_var_y = np.var(y_samples)
```

### Answer check

```python
assert abs(sigma_s - 0.5) < 1e-12
assert abs(C * sigma_s**2 * sigma_squared - 9) < 1e-12
assert abs(sim_var_y - 9) < 0.08
```
