## Concept summary

| Concept | One-line summary | Key formula / API | Session |
| --- | --- | --- | --- |
| `random-variables` | A number produced by a random process; capital letter, table of values | `rng.choice(values, size, p=probs)` | 01 |
| `expectation` | Probability-weighted sum over the table = long-run average of runs | $E[X] = \sum_x x P(X{=}x)$; `(values * probs).sum()` | 01 |
| `sampling-simulation` | Seeded, loop-free repetition; frequencies → probabilities, means → expectations as $n$ grows | `SEED`, mask `.mean()`, $n \ge 10^5$ | 01–03 |
| `variance` | Expected squared deviation; shortcut $E[X^2] - E[X]^2$; shift-proof, scale by $w^2$ | `(((v - mu) ** 2) * p).sum()` | 02 |
| `independence` | Joint table factors: $P(x, y) = P(x)P(y)$ for every pair; gives $E[XY] = E[X]E[Y]$ | separate `rng` draws | 02 |
| `variance-of-sums` | Independent ⇒ variances add; scaled sums: $\operatorname{Var}[\sum w_i x_i] = \sigma^2 \sum w_i^2$ (zero-expectation factors) | dependent case adds $2\operatorname{Cov}$ | 02–03 |
| `gaussian-distribution` | Limiting bell of summed independent contributions; $\mathcal N(\mu, \sigma^2)$; 68–95–99.7; stated facts, sim-verified | `rng.normal(mu, sigma, size)` — takes $\sigma$! | 03 |
| `covariance` | How two variables move together; $E[XY] - E[X]E[Y]$; independent ⇒ 0, converse FALSE | `((x - x.mean()) * (y - y.mean())).mean()` | 03 |

## Formula sheet

**Proved (discrete register):**

- $E[X] = \sum_x x\,P(X{=}x)$, $\quad E[g(X)] = \sum_x g(x)\,P(X{=}x)$
- Linearity (no assumptions): $E[aX + b] = aE[X] + b$, $\quad E[X + Y] = E[X] + E[Y]$
- $\operatorname{Var}[X] = E[(X - \mu)^2] = E[X^2] - E[X]^2 \ge 0$; $\quad \sigma = \sqrt{\operatorname{Var}}$
- $\operatorname{Var}[X + b] = \operatorname{Var}[X]$, $\quad \operatorname{Var}[wX] = w^2 \operatorname{Var}[X]$
- Independence: $P(X{=}x, Y{=}y) = P(X{=}x)P(Y{=}y)$ for **every** pair $\Rightarrow E[XY] = E[X]E[Y]$
- Independent sum: $\operatorname{Var}[X + Y] = \operatorname{Var}[X] + \operatorname{Var}[Y]$
- Scaled sums (independent, $E[x_i] = 0$, variances $\sigma_i^2$):
  $\operatorname{Var}\big[\sum_i w_i x_i\big] = \sum_i w_i^2 \sigma_i^2$
- General sum: $\operatorname{Var}[X + Y] = \operatorname{Var}[X] + \operatorname{Var}[Y] + 2\operatorname{Cov}[X, Y]$
- $\operatorname{Cov}[X, Y] = E[(X - \mu_X)(Y - \mu_Y)] = E[XY] - E[X]E[Y]$; $\operatorname{Cov}[X, X] = \operatorname{Var}[X]$

**Stated facts (continuous register, simulation-verified):**

- $X \sim \mathcal N(\mu, \sigma^2)$: $E[X] = \mu$, $\operatorname{Var}[X] = \sigma^2$; 68–95–99.7 within $1/2/3\,\sigma$
- $aX + b$ and independent sums of Gaussians are Gaussian (parameters from the proved laws)
- $(X - \mu)/\sigma$ is standard $\mathcal N(0, 1)$
- Functions of independent variables are independent

**Simulation idioms:**

| Goal | Idiom |
| --- | --- |
| exact table expectation | `(values * probs).sum()` |
| exact table variance | `((values ** 2) * probs).sum() - ((values * probs).sum()) ** 2` |
| simulate a table variable | `rng.choice(values, size=n, p=probs)` |
| estimate $P(\text{event})$ | `(event_mask).mean()` |
| estimate $E[X]$, $E[g(X)]$ | `x.mean()`, `g(x).mean()` |
| estimate Var from samples | `((x - x.mean()) ** 2).mean()` |
| estimate Cov from samples | `((x - x.mean()) * (y - y.mean())).mean()` |
| $n$ independent runs of a $k$-part process | `rng.integers(..., size=(n, k))` + `.sum(axis=1)` |

1. Values $2, 8$ with $P = 0.6, 0.4$. The event is a yes/no occurrence, not
   a number; its 0/1 *indicator* would be a random variable.
2. **(D)** — one value with probability 1 is legitimate (a constant). A sums
   to 1.1; B has a negative entry; C sums to 0.6; E sums to 1.1.
3. $E[X] = 2(0.6) + 8(0.4) = 4.4$;
   $E[X^2] = 4(0.6) + 64(0.4) = 28$.
4. $E[aX + b] = aE[X] + b$ and $E[X + Y] = E[X] + E[Y]$ — **no assumption**
   about the relation between $X$ and $Y$ (linearity holds even for
   dependent pairs).
5. $\operatorname{Var}[X] = 28 - 4.4^2 = 8.64$;
   $\operatorname{Var}[5X - 3] = 25 \cdot 8.64 = 216$ (shift ignored,
   scale squared).
6. **(B)** — variance is an average of squares, so a negative result means
   the shortcut was subtracted in reverse order.
7. Independent iff $P(X{=}x, Y{=}y) = P(X{=}x)P(Y{=}y)$ for *every* pair.
   Failing pair: $P(X{=}1, Y{=}1) = 0$ but the product is $\tfrac1{36}$.
8. $2 + 7 = 9$; $\operatorname{Var}[X - Y] = 2 + 7 = 9$ too ($-Y$ has
   $Y$'s variance); $\operatorname{Var}[X + X] = \operatorname{Var}[2X]
   = 8$.
9. $\operatorname{Var}[S] = E[S^2] = \sum_i \sum_j c^2 E[x_i x_j]$;
   for $i \ne j$ **independence** factors $E[x_i x_j] = E[x_i]E[x_j] = 0$
   (this is the step), leaving the $n$ diagonal terms $c^2\sigma^2$ each:
   $c^2 n \sigma^2$. With $n = 64$, $\sigma^2 = 4$: $256 c^2 = 1$, so
   $c = \tfrac1{16}$.
10. $\sigma = 2$: about 95% within $[26, 34]$ ($\mu \pm 2\sigma$).
    36 mm is $3\sigma$ above: about $0.15\%$ exceed it (half of the 0.3%
    outside $\pm 3\sigma$).
11. Proved: expectation $3(10) + 1 = 31$ (linearity) and variance
    $9 \cdot 4 = 36$ (scale law). Stated fact: the result is still
    Gaussian-shaped.
12. Seed it (one `SEED`, one generator); size it ($n \ge 10^5$, more for
    tails); mask it (no loops). Idiom: `(S > 180).mean()`.
13. $\operatorname{Cov}[X, Y] = E[XY] - E[X]E[Y]$. Theorem: independent
    $\Rightarrow$ covariance 0. False converse: $X$ uniform on
    $\{-1, 0, 1\}$, $Y = X^2$ has covariance 0 but is completely
    dependent.

```python
# Answer verification (run AFTER attempting the quiz — this cell contains answers)
values = np.array([2.0, 8.0]); probs = np.array([0.6, 0.4])
e_x = (values * probs).sum(); e_x2 = ((values ** 2) * probs).sum()
assert np.isclose(e_x, 4.4, atol=1e-12, rtol=0) and np.isclose(e_x2, 28.0, atol=1e-12, rtol=0)                      # Q3
var_x = e_x2 - e_x ** 2
assert np.isclose(var_x, 8.64, atol=1e-12, rtol=0) and np.isclose(25 * var_x, 216.0, atol=1e-12, rtol=0)            # Q5
assert np.isclose(2 + 7, 9, atol=1e-12, rtol=0) and np.isclose(4 * 2, 8, atol=1e-12, rtol=0)                        # Q8
assert np.isclose((1 / 16) ** 2 * 64 * 4, 1.0, atol=1e-12, rtol=0)                              # Q9

SEED = 20260804
rng = np.random.default_rng(SEED)
x = rng.choice(np.array([-1.0, 0.0, 1.0]), size=300_000)                    # Q13
y = x ** 2
assert abs(((x - x.mean()) * (y - y.mean())).mean()) < 0.01
assert abs(((x == 1) & (y == 0)).mean() - (x == 1).mean() * (y == 0).mean()) > 0.05

g = rng.normal(30.0, 2.0, size=300_000)                                     # Q10
assert 0.94 < ((g >= 26) & (g <= 34)).mean() < 0.96
assert (g > 36).mean() < 0.005
print("all quiz checks passed")
```
