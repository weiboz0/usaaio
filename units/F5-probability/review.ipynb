## Formula and assumption sheet

**Discrete algebra**

- $E[g(X)]=\sum_xg(x)P(X=x)$; linearity needs no independence.
- $\operatorname{Var}(X)=E[(X-EX)^2]=E[X^2]-(EX)^2$.
- Independent sums add variances; general sums include covariance.
- $P(A\mid B)=P(A\cap B)/P(B)$ only when $P(B)>0$.
- $P(B)=P(B\mid A)P(A)+P(B\mid A^c)P(A^c)$.
- Bayes reverses direction; its denominator includes every route to evidence.

**Hoeffding**

For independent $a_i\le X_i\le b_i$, widths $w_i=b_i-a_i$, and $\varepsilon>0$:

$$P(\bar X-E\bar X\ge\varepsilon)\le
e^{-2n^2\varepsilon^2/\sum_iw_i^2}.$$

The lower tail has the same bound. Since
$\{|\bar X-E\bar X|\ge\varepsilon\}=U\cup L$ and
$P(U\cup L)\le P(U)+P(L)$,

$$P(|\bar X-E\bar X|\ge\varepsilon)\le
2e^{-2n^2\varepsilon^2/\sum_iw_i^2}.$$

For common width $w$, replace $n^2/\sum_iw_i^2$ by $n/w^2$. Preserve the raw theorem expression; use `min(1.0, raw)` for a useful envelope. Above 1 means vacuous, not false.

## Self-quiz answers

<details><summary><b>Reveal after attempting all items</b></summary>

1. A numerical outcome of a random process; probabilities are nonnegative and sum to 1.
2. $E[X]=1/2$, $E[X^2]=1$, so variance is $3/4$.
3. $E[aX+b]=aE[X]+b$ and $E[X+Y]=E[X]+E[Y]$; no independence.
4. $4(3)+5=17$.
5. $P(X=x,Y=y)=P(X=x)P(Y=y)$ for every pair.
6. Covariance sees only a linear product average; e.g. symmetric nonconstant $X$ and $X^2$ can be dependent with covariance zero.
7. $\sigma=2$, so about $[6,14]$. Mean/variance transformations are proved; preservation of Gaussian shape is stated.
8. One fixed seed/generator, many vectorized repetitions, Boolean-mask aggregation; `(S > 7).mean()`.
9. $P(A\mid B)=12/30=2/5$; $P(B\mid A)=12/20=3/5$.
10. The ordinary conditional is undefined; empirical code raises an error.
11. Denominator $P(B\mid A)P(A)+P(B\mid A^c)P(A^c)$; numerator $P(B\mid A)P(A)$.
12. The large non-disease population can contribute many false positives; sensitivity is the reverse conditional.
13. Independent variables, known almost-sure bounds $a_i\le X_i\le b_i$, positive $\varepsilon$; width $b_i-a_i$.
14. With upper $U$ and lower $L$, the absolute event is $U\cup L$; union bound adds two one-sided $e^{-K}$ bounds to get $2e^{-K}$.
15. Raw $2e^{-2(50)(0.05)^2}=2e^{-1/4}>1$; cap at 1, so it is vacuous.
</details>


```python
# Computational spot checks for quiz items 2, 4, 9, and 15.
values = np.array([0.0, 2.0])
probs = np.array([0.75, 0.25])
mu = (values * probs).sum()
var = ((values - mu) ** 2 * probs).sum()
assert np.isclose(mu, 0.5, atol=1e-12, rtol=0)
assert np.isclose(var, 0.75, atol=1e-12, rtol=0)
assert np.isclose(4 * 3 + 5, 17, atol=1e-12, rtol=0)
assert np.isclose(12 / 30, 2 / 5, atol=1e-12, rtol=0)
raw = 2.0 * np.exp(-2.0 * 50 * 0.05**2)
assert raw > 1 and min(1.0, raw) == 1.0
print("review checks passed")
```
