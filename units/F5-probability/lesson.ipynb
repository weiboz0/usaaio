# F5-probability — Unit Overview

**Probability and Statistics Essentials.** This unit builds the language of
randomness — random variables, expectation, variance, independence — up to
the two results the rest of the course leans on hardest: the variance of a
scaled sum of independent factors, and the Gaussian as the universal shape
of accumulated randomness. Everything is developed twice: proved in the
discrete (table) register, and verified by seeded NumPy simulation.

## What you are assumed to know

- **F1-scientific-python** — arrays, elementwise operations, broadcasting,
  masks, axis aggregations, seeded random generation, matplotlib basics.
  Every simulation in this unit is built from those parts, loop-free.
- **Math:** the Calculus AB baseline. No integral is ever evaluated over a
  continuous density in this unit: continuous facts (the Gaussian family)
  enter as clearly labeled *stated facts* and are verified by simulation.

## Concepts taught

`random-variables`, `expectation`, `variance`, `independence`,
`variance-of-sums`, `gaussian-distribution`, `sampling-simulation`,
`covariance`.

## Practice sets

Eighteen problems in `practice/`, organized as three sets (the manifest
carries the same grouping):

**Set A — fundamentals** (do these right after the session that covers
them):

- `p01` (MC, intro) — what makes a station's daily lightning-strike count
  a random variable.
- `p02` (MC, intro) — $E[W]$ of a five-sector game wheel, and what that
  number does (and does not) mean.
- `p05` (coding, intro) — seeded `simulate_sums` for two dice, loop-free,
  plus the empirical frequency of sevens.
- `p06` (coding, intro) — the wheel's exact expectation versus a seeded
  simulation estimate, with `rng.choice` banned (index-lookup sampling).
- `p07` (coding, intro) — `my_variance(values, probs)` from
  probability-weighted sums; `np.var`, `np.average`, `np.mean` banned.

**Set B — exam register** (start after all three sessions):

- `p03` (MC, core) — variance of a sum and of a difference of independent
  variables (the difference does not subtract).
- `p04` (MC, numeric normal form, core) — $\operatorname{Var}[T]$ for a
  three-value table, decoded as $p + q$ in lowest terms.
- `p08` (coding, core) — draw two independent samples, measure the
  variance-additivity gap, and name the property that licenses the rule.
- `p09` (coding, core) — standardized sums of 40 coin flips: bell
  histogram plus the within-1σ/2σ fractions against the stated ≈0.68/≈0.95.
- `p10` (coding, advanced) — covariance implemented twice (definition and
  shortcut) and read on a dependent pair versus an independent pair.
- `p11` (proof, advanced) — the unit's centerpiece:
  $\operatorname{Var}[\sum_c s_c x_c] = C\,\sigma_s^2\sigma^2$ for
  independent zero-expectation factors, solved for the variance-preserving
  $\sigma_s$, with a seeded anchor.
- `p12` (proof, core) — derive the shortcut
  $\operatorname{Var}[X] = E[X^2] - (E[X])^2$ from the definition, then
  compute a spinner's variance both ways.

**Set C — integration and challenge** (closest to real exam texture —
save for last, ideally under time pressure):

- `p13` (integrative, core) — parcel-arrival table → exact $E$ and
  $\operatorname{Var}$ of a 900-minute total → 50,000-run simulation
  check, each part consuming the last.
- `p14` (integrative, advanced) — a sensor reading summing 25 independent
  error sources: exact moments, the 95% Gaussian interval, simulated
  fraction inside.
- `p15` (scenario, core) — how much averaging four noisy weighings really
  shrinks the spread, and what a fourfold shrink would actually take.
- `p16` (scenario, core) — café data: which pair behaves as independent,
  what the covariance sign says about hot-drink sales, and a caution on
  comparing magnitudes across units.
- `p17` (challenge, advanced) — portfolio variance
  $\sigma^2 \sum_c w_c^2$; prove the equal split minimizes it; verify
  two splits by seeded simulation.
- `p18` (challenge, advanced) — shuffle one partner of a correlated pair:
  covariance collapses while both variances survive — pairing, not value
  collections, is what covariance measures.

Every problem has a paired `pNN_solution.ipynb` with worked reasoning and
an answer-check cell — solve first, then compare; do not read solutions
as a first resort.

## How to study this unit

1. Work the sessions in order — each proof uses the previous session's
   results by name, and every checkpoint has answers at its session's end.
   Do every checkpoint *before* looking.
2. Keep the two registers straight as you read: **proved** claims (discrete
   tables, linearity, variance algebra) versus **stated facts** (Gaussian
   shape claims), the latter always paired with a simulation check you can
   rerun.
3. Then the practice sets in `practice/`, in the order mapped above:
   Set A drills, then Set B's exam register, then Set C's integration and
   challenge arcs. Student notebooks contain no answers; solutions live in
   the paired `*_solution.ipynb`.
4. Finish with `review.ipynb`: summary table, formula sheet, self-quiz,
   and redo pointers.

The unit's centerpiece derivation — why
$\operatorname{Var}[\sum_i w_i x_i] = \sigma^2 \sum_i w_i^2$ for
independent zero-expectation factors — is Session 02, Section 7. If you can
reproduce that derivation cold, naming the exact step where independence
acts, you have this unit.
