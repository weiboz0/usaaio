## Concept summary

| Concept | One-line summary | Key fact to retain |
|---|---|---|
| Vectors & norms | A 1-D array is a vector; norms measure its size | $\|v\| = \sqrt{\sum v_i^2}$, $\;\|v\|_1 = \sum |v_i|$ |
| Distance | Norm of the difference | Euclidean = straight line; Manhattan = city block; Manhattan ≥ Euclidean |
| Dot product | $\sum_i a_i b_i$, also $\|a\|\|b\|\cos\theta$ | sign ⇔ acute / right / obtuse; $a\cdot a = \|a\|^2$ |
| Cosine similarity | Direction agreement in $[-1, 1]$ | scale-free: $\cos(a, cb) = \cos(a, b)$ for $c>0$; undefined for the zero vector |
| Unit vectors | Direction with the length standardized to 1 | $\hat v = v / \|v\|$; cosine similarity $= \hat a \cdot \hat b$ |
| Orthogonality / orthonormality | Perpendicular = zero dot product; orthonormal = orthogonal + unit | coordinates read off by dot products, pieces rebuild the vector |
| Projection | The shadow along a direction | $(a\cdot\hat b)\hat b$, or $\frac{a\cdot b}{b\cdot b}b$ for general $b$ |
| Residuals | What is left after the shadow | $r = a - \text{proj}$; always $r \perp \hat b$; $\|a\|^2 = \|\text{proj}\|^2 + \|r\|^2$ |

## Formula and idiom sheet

**By hand (exact forms the exam's normal forms expect):**

- Norm: $\|v\| = \sqrt{\sum_i v_i^2}$ — reduce roots: $\sqrt{52} = 2\sqrt{13}$.
- Unit vector: $\frac{1}{\sqrt a}(b, c)$ with $\gcd(b, c) = 1$.
- Projection ratio: $\dfrac{a\cdot b}{b\cdot b}$ (a plain number, can be negative).
- Residual orthogonality (the free self-check): $r \cdot b = 0$, always.

**NumPy idioms (single vector → batch of rows):**
