# F2-vectors — Practice p18 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

$a\cdot b = 4 + 2 = 6$ and $b\cdot b = 2$, so
$p = \tfrac62\,(1,1) = (3, 3)$ and $r = (4, 2) - (3, 3) = (1, -1)$.

Squared norms, all integers:
$$\|a\|^2 = 16 + 4 = 20, \qquad \|p\|^2 = 9 + 9 = 18, \qquad \|r\|^2 = 1 + 1 = 2,$$
and indeed $20 = 18 + 2$. $\checkmark$

## Part (b)

Expand the squared norm as a dot product of a sum with itself:
$$\|p + r\|^2 = (p + r)\cdot(p + r) = p\cdot p + p\cdot r + r\cdot p + r\cdot r,$$
using distributivity ($x\cdot(y+z) = x\cdot y + x\cdot z$) twice.
By symmetry ($x\cdot y = y\cdot x$) the two middle terms are equal, and by
the hypothesis $r\cdot p = 0$ they both vanish, leaving
$\|p+r\|^2 = p\cdot p + r\cdot r = \|p\|^2 + \|r\|^2$.

## Part (c)

The residual property from the lesson: the residual of a projection is
orthogonal to the projection direction, and since $p$ is a scalar multiple
of that direction, $r\cdot p = 0$ follows by the scalar-pull-out rule —
so the Pythagorean split holds for *every* projection/residual pair.

## Numeric check

```python
a = np.array([4.0, 2.0])
b = np.array([1.0, 1.0])

p = (a @ b) / (b @ b) * b
r = a - p

print("p =", p, " r =", r)
print(a @ a, p @ p + r @ r)
```

The run reproduces $p = (3, 3)$, $r = (1, -1)$, and the split
$20 = 18 + 2$ exactly.
The exact-integer arithmetic in part (a) is what makes this a proof-backed
check rather than a float coincidence.

### Answer check

```python
assert np.allclose(p, np.array([3.0, 3.0]), atol=1e-9, rtol=0)
assert np.allclose(r, np.array([1.0, -1.0]), atol=1e-9, rtol=0)
assert np.isclose(r @ p, 0.0, atol=1e-9, rtol=0)
assert np.isclose(a @ a, 20.0, atol=1e-9, rtol=0)
assert np.isclose(p @ p, 18.0, atol=1e-9, rtol=0) and np.isclose(r @ r, 2.0, atol=1e-9, rtol=0)
assert np.isclose(a @ a, p @ p + r @ r, atol=1e-9, rtol=0)
print("All checks passed.")
```
