$\|v\| = \sqrt{81 + 9} = \sqrt{90} = 3\sqrt{10}$, so
$$\hat v = \frac{(9, 3)}{3\sqrt{10}} = \frac{1}{\sqrt{10}}\,(3,\ 1),$$
with $\gcd(3, 1) = 1$ as required.
So $a = 10$, $b = 3$, $c = 1$ and $S = 14$: choice **B**.
Choice **E** (102) is the unreduced-form trap $90 + 9 + 3$ — writing
$\frac{1}{\sqrt{90}}(9, 3)$ without pulling the common factor 3 out of the
entries first.

```python
answer_p09 = "B"

v = np.array([9.0, 3.0])
v_hat = v / np.linalg.norm(v)
print(v_hat, (1 / np.sqrt(10)) * np.array([3.0, 1.0]))
```
