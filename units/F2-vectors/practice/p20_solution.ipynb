```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Scenario 1

```python
target = np.array([6.0, 8.0])
actual = np.array([3.0, 4.0])

s1_error = float(np.sqrt(np.sum((target - actual) ** 2)))
scenario_1 = "euclidean"
print(s1_error, "->", scenario_1)
```

Physical position error is a straight-line quantity: the tip is simply 5 cm
from where it should be (a 3-4-5 triangle).
Neither city blocks nor angles are the right frame for a gap in physical
space.

## Scenario 2

```python
stock = np.array([10.0, 4.0, 7.0])
target_stock = np.array([8.0, 7.0, 7.0])

s2_work = float(np.sum(np.abs(stock - target_stock)))
scenario_2 = "manhattan"
print(s2_work, "->", scenario_2)
```

The work happens one item at a time, per product: remove 2, add 3, touch
nothing — 5 units of work.
That per-entry accumulation is exactly the Manhattan distance; the Euclidean
$\sqrt{13} \approx 3.6$ would understate the labor.

## Scenario 3

```python
m1 = np.array([300.0, 150.0, 50.0])
m2 = np.array([60.0, 30.0, 10.0])

s3_cosine = float((m1 @ m2) / (np.linalg.norm(m1) * np.linalg.norm(m2)))
s3_euclid = float(np.sqrt(np.sum((m1 - m2) ** 2)))
scenario_3 = "cosine"
print(s3_cosine, s3_euclid, "->", scenario_3)
```

$m_1 = 5\,m_2$, so the logs point in the same direction: cosine similarity
$1.0$ — identical practice mix.
The Euclidean distance ($\approx 271$ minutes) only reports that one
musician practices far more in total, which is a different question.

**Discussion:** the rulers disagree in general because Manhattan sums
per-entry gaps while Euclidean squares them first — a difference spread as
$(3, 3)$ and one concentrated as $(0.5, 4.24)$ can have equal Euclidean
sizes but different Manhattan sizes (6 versus about 4.71), so candidates can
swap order between the two rulers.
