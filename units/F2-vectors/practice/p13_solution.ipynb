# F2-vectors — Practice p13 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
v1 = np.array([3.0, 4.0])
v2 = np.array([2.0, 1.0])

u1 = v1 / np.sqrt(np.sum(v1**2))          # (0.6, 0.8)

reach = v2 @ u1                           # 1.2 + 0.8 = 2.0
resid = v2 - reach * u1                   # (2, 1) - (1.2, 1.6) = (0.8, -0.6)
u2 = resid / np.sqrt(np.sum(resid**2))    # already unit length!

print(u1, reach, resid, u2)
```

By hand: $\|v_1\| = 5$ gives $u_1 = (0.6, 0.8)$; the reach of $v_2$ along
$u_1$ is $2\cdot0.6 + 1\cdot0.8 = 2.0$, so the residual is
$(2, 1) - 2\,(0.6, 0.8) = (0.8, -0.6)$.
This residual happens to have norm $\sqrt{0.64 + 0.36} = 1$ already, so
normalizing leaves it unchanged: $u_2 = (0.8, -0.6)$.

## Part (b)

```python
print("u1 . u1 =", u1 @ u1)
print("u2 . u2 =", u2 @ u2)
print("u1 . u2 =", u1 @ u2)
```

**Why step 2 guarantees orthogonality:** the residual of any vector after
projecting onto the direction $u_1$ is orthogonal to $u_1$ — that is the
residual property proved in the lesson, which holds before any numbers are
plugged in.
Normalizing the residual only rescales it by a positive number, which cannot
change a zero dot product, so $u_1 \cdot u_2 = 0$ is automatic.

## Part (c)

```python
w = np.array([7.0, 1.0])

t1 = w @ u1        # 4.2 + 0.8 = 5.0
t2 = w @ u2        # 5.6 - 0.6 = 5.0
rebuilt = t1 * u1 + t2 * u2
print(t1, t2, rebuilt)
```

Both coordinates come out exactly 5, and
$5\,(0.6, 0.8) + 5\,(0.8, -0.6) = (3, 4) + (4, -3) = (7, 1) = w$.
Because $\{u_1, u_2\}$ is orthonormal, two dot products recover the
coordinates and the pieces reassemble $w$ with nothing lost — the
"good axes" property from the lesson, now for axes we built ourselves.
