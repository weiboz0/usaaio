# F2-vectors — Practice p17 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
X = rng.normal(size=(80, 3))
q = rng.normal(size=3)
print(X.shape, q)
```

## Part (a)

```python
def cosine_to_query(X, q):
    dots = np.sum(X * q, axis=1)                 # (80,): every row . q
    row_norms = np.sqrt(np.sum(X**2, axis=1))    # (80,)
    q_norm = np.sqrt(np.sum(q**2))               # scalar
    return dots / (row_norms * q_norm)


sims = cosine_to_query(X, q)
print(sims[:3])
```

One axis sum produces all 80 numerators, a second produces all 80 row
norms, and the scalar norm of `q` completes the denominator — the full
cosine-similarity formula applied to the whole batch with pure broadcasting.

## Part (b)

```python
def most_similar(X, q):
    return int(np.argmax(cosine_to_query(X, q)))


best_row = most_similar(X, q)
print(best_row, sims[best_row])
```

`np.argmax` picks the index of the highest direction agreement.
With this seed, row 59 wins with a cosine similarity of about 0.988 — an
angle of under $9^\circ$ from the query's direction.

## Part (c)

```python
check_range = bool(np.all((sims >= -1.0) & (sims <= 1.0)))
print(check_range)
```

Cosine similarity is the cosine of a real angle, so the whole batch must
land in $[-1, 1]$; `np.all` verifies it in one expression.
