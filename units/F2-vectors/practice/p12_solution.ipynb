## Scenario 1

```python
a = np.array([2.0, 3.0])
b = np.array([7.0, 9.0])

s1_euclid = float(np.sqrt(np.sum((a - b) ** 2)))   # sqrt(25 + 36)
s1_manhattan = float(np.sum(np.abs(a - b)))        # 5 + 6
scenario_1 = "manhattan"
print(s1_euclid, s1_manhattan, "->", scenario_1)
```

The courier must travel along streets: 5 blocks one way and 6 the other, so
the 11-block Manhattan distance is the ride length.
The Euclidean $\sqrt{61} \approx 7.81$ would cut through buildings.

## Scenario 2

```python
scenario_2 = "euclidean"
print(s1_euclid, "->", scenario_2)
```

The drone is not bound to the grid, so the straight-line Euclidean distance
$\sqrt{61} \approx 7.81$ blocks is the honest measure of its flight.

## Scenario 3

```python
h1 = np.array([6.0, 2.0, 4.0])
h2 = np.array([12.0, 4.0, 8.0])

s3_euclid = float(np.sqrt(np.sum((h1 - h2) ** 2)))
s3_cosine = float((h1 @ h2) / (np.linalg.norm(h1) * np.linalg.norm(h2)))
scenario_3 = "cosine"
print(s3_euclid, s3_cosine, "->", scenario_3)
```

$h_2 = 2\,h_1$ exactly, so the two count vectors point in the same
direction and the cosine similarity is $1.0$: identical proportions.
The Euclidean distance is large ($\approx 7.48$) only because the bigger
household buys *more of everything* — it measures the size gap, not the
proportion difference, which is exactly the wrong question here.
