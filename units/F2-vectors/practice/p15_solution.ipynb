# F2-vectors — Practice p15 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

The reach along $\hat b$ is
$a\cdot\hat b = 2\cdot\tfrac35 + 7\cdot\tfrac45 = \tfrac{34}{5}$, so
$$p = \tfrac{34}{5}\left(\tfrac35, \tfrac45\right)
= \left(\tfrac{102}{25},\ \tfrac{136}{25}\right),
\qquad
r = \left(\tfrac{50}{25} - \tfrac{102}{25},\ \tfrac{175}{25} - \tfrac{136}{25}\right)
= \left(-\tfrac{52}{25},\ \tfrac{39}{25}\right).$$

Factor the residual: $r = \tfrac{13}{25}\,(-4, 3)$, and $\|(-4, 3)\| = 5$, so
$$\|r\| = \tfrac{13}{25}\cdot 5 = \tfrac{13}{5}, \qquad \gcd(13, 5) = 1.$$

Hence $S = 13 + 5 = 18$: choice **B**.
Trap **E** (194) is $169 + 25$, the sum from the *squared* norm
$\|r\|^2 = \tfrac{169}{25}$ — the norm-versus-squared-norm confusion; trap
**C** (36) comes from failing to reduce an unsimplified $\tfrac{26}{10}$.

```python
answer_p15 = "B"

a = np.array([2.0, 7.0])
b_hat = np.array([3.0, 4.0]) / 5.0
p_vec = (a @ b_hat) * b_hat
r_vec = a - p_vec
print(r_vec, np.linalg.norm(r_vec), 13 / 5)
```

### Answer check

```python
assert answer_p15 == "B"
assert np.allclose(r_vec, np.array([-52 / 25, 39 / 25]), atol=1e-9, rtol=0)
assert np.isclose(np.linalg.norm(r_vec), 13 / 5, atol=1e-9, rtol=0)
assert np.isclose(r_vec @ b_hat, 0.0, atol=1e-9, rtol=0)
assert 13 + 5 == 18
print("All checks passed.")
```
