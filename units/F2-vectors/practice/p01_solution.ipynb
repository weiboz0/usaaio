# F2-vectors — Practice p01 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
norm_v1_by_hand = 3.0    # sqrt(4 + 4 + 1) = sqrt(9)  = 3
norm_v2_by_hand = 10.0   # sqrt(36 + 64)   = sqrt(100) = 10
print(norm_v1_by_hand, norm_v2_by_hand)
```

Square each entry, add, take the square root.
Both vectors were chosen so the sum of squares is a perfect square
(9 and 100), which makes the by-hand roots exact.

## Part (b)

```python
def euclidean_norm(v):
    return np.sqrt(np.sum(v**2))


print(euclidean_norm(np.array([2.0, -2.0, 1.0])))   # 3.0
print(euclidean_norm(np.array([6.0, 8.0])))         # 10.0
```

`v**2` squares every entry elementwise and `np.sum` adds them, both as
covered in F1-scientific-python; `np.sqrt` finishes the formula.
The function reproduces both hand answers exactly.

## Part (c)

```python
v = rng.normal(size=5)
print(v)
```

```python
mine = euclidean_norm(v)
builtin = np.linalg.norm(v)
print(mine, builtin, np.isclose(mine, builtin))
```

The two computations follow the same formula, so they agree to floating-point
precision.
`np.isclose` is the right comparison here — tiny rounding differences between
two float computations are normal.
