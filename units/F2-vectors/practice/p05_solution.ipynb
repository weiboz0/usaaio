# F2-vectors — Practice p05 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
# a . b = 4 + 2 = 6 ;  b . b = 1 + 1 = 2 ;  ratio = 3
proj_by_hand = np.array([3.0, 3.0])     # 3 * (1, 1)
resid_by_hand = np.array([1.0, -1.0])   # (4, 2) - (3, 3)
print(proj_by_hand, resid_by_hand)
```

The ratio $\dfrac{a\cdot b}{b\cdot b} = \dfrac62 = 3$ says the shadow
reaches 3 copies of $b$ along its line, giving $(3, 3)$.
The residual $(1, -1)$ satisfies $r\cdot b = 1 - 1 = 0$, confirming by hand
that the leftover is orthogonal to $b$.

## Part (b)

```python
def project(a, b):
    return (a @ b) / (b @ b) * b


def residual(a, b):
    return a - project(a, b)


a2 = np.array([4.0, 2.0])
b2 = np.array([1.0, 1.0])
print(project(a2, b2), residual(a2, b2))
```

This is the general-direction form of the projection formula: dividing by
$b\cdot b = \|b\|^2$ does the normalizing, so `b` need not be a unit vector.
Both functions reproduce the part (a) arrays exactly.

## Part (c)

```python
a = rng.normal(size=4)
b = rng.normal(size=4)
print(a)
print(b)
```

```python
proj_c = project(a, b)
resid_c = residual(a, b)

check_orth = bool(np.isclose(resid_c @ b, 0.0))
check_pyth = bool(np.isclose(a @ a, proj_c @ proj_c + resid_c @ resid_c))

print(proj_c)
print(check_orth, check_pyth)
```

Both checks pass on 4-entry vectors, where no picture is available — the
algebra alone guarantees them.
The projection and residual are orthogonal pieces of $a$, so their squared
norms add up to $\|a\|^2$ exactly as in the right-triangle picture from the
lesson.
