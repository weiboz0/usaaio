# F2-vectors — Practice p04 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
def normalize(v):
    return v / np.linalg.norm(v)


# ||(3, 4)|| = 5, so the unit vector is (3/5, 4/5):
u_by_hand = np.array([0.6, 0.8])

print(normalize(np.array([3.0, 4.0])), u_by_hand)
print(np.linalg.norm(normalize(np.array([2.0, -1.0, 2.0]))))
```

Dividing every entry by the norm keeps the direction and forces the length
to 1.
For $(3, 4)$ the norm is 5, giving $(0.6, 0.8)$ by hand, and the function
agrees; the second vector has norm 3, and its normalized version has norm 1
as required.

## Part (b)

```python
a = np.array([2.0, 1.0, -2.0])
b = np.array([1.0, 2.0, 2.0])
c = np.array([1.0, 1.0, 0.0])
d = np.array([1.0, 0.0, 1.0])

orth_ab = bool(np.isclose(a @ b, 0.0, atol=1e-9, rtol=0))   # 2 + 2 - 4 = 0
orth_cd = bool(np.isclose(c @ d, 0.0, atol=1e-9, rtol=0))   # 1 + 0 + 0 = 1

print(orth_ab, orth_cd)
```

One dot product settles each question.
$a\cdot b = 2 + 2 - 4 = 0$, so that pair is orthogonal; $c\cdot d = 1 \neq 0$,
so that pair is not.
Wrapping the comparison in `np.isclose` is a good habit for float inputs.

## Part (c)

```python
u1 = np.array([0.6, 0.8])

# Swap the entries and flip one sign: (0.6, 0.8) -> (-0.8, 0.6).
# (The choice (0.8, -0.6) is equally correct.)
u2 = np.array([-0.8, 0.6])

print("u1 . u1 =", u1 @ u1)
print("u2 . u2 =", u2 @ u2)
print("u1 . u2 =", u1 @ u2)
```

In 2-D, swapping the entries and negating one always produces an orthogonal
vector: $(x, y)\cdot(-y, x) = -xy + yx = 0$, and the swap leaves the norm
unchanged, so `u2` is still a unit vector.
The three dot products come out 1, 1, 0 — exactly the orthonormal
conditions: both unit length, mutually perpendicular.
