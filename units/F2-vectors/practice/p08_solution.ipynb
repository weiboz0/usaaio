# F2-vectors — Practice p08 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

Cosine similarity is $\dfrac{a\cdot b}{\|a\|\|b\|}$; with nonzero norms the
fraction is $0$ exactly when the numerator is, so **I** holds.
Expanding with the dot-product algebra rules,
$\|a-b\|^2 = a\cdot a - 2\,a\cdot b + b\cdot b = \|a\|^2 + \|b\|^2$ once the
middle term vanishes, so **II** holds — this is the Pythagorean theorem for
the orthogonal pair.
**III** is the same zero numerator divided by positive norms, so it holds.
**IV** fails: $(1, 0)$ and $(0, 5)$ have cosine similarity $0$ and different
norms — orthogonality says nothing about lengths.

```python
answer_p08 = "C"

# Numeric spot check of I-III on an orthogonal pair with unequal norms (IV fails):
a = np.array([1.0, 0.0])
b = np.array([0.0, 5.0])
print(a @ b)
print(np.sum((a - b) ** 2), a @ a + b @ b)
print((a / np.linalg.norm(a)) @ (b / np.linalg.norm(b)))
print(np.linalg.norm(a) == np.linalg.norm(b))
```
