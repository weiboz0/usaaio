# F2-vectors — Practice p07 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

$$\hat b \cdot \hat b = \left(\tfrac35\right)^2 + \left(\tfrac45\right)^2
= \tfrac{9}{25} + \tfrac{16}{25} = \tfrac{25}{25} = 1 .$$

Since $\|\hat b\|^2 = \hat b\cdot\hat b = 1$ and norms are nonnegative,
$\|\hat b\| = 1$: a unit vector.
(This is the 3-4-5 right triangle scaled down by 5.)

## Part (b)

First the reach of the shadow:
$$a \cdot \hat b = 5\cdot\tfrac35 + 1\cdot\tfrac45 = \tfrac{15}{5} + \tfrac{4}{5} = \tfrac{19}{5}.$$

Then the projection and residual:
$$p = \tfrac{19}{5}\left(\tfrac35, \tfrac45\right) = \left(\tfrac{57}{25},\ \tfrac{76}{25}\right),
\qquad
r = a - p = \left(\tfrac{125}{25} - \tfrac{57}{25},\ \tfrac{25}{25} - \tfrac{76}{25}\right)
= \left(\tfrac{68}{25},\ -\tfrac{51}{25}\right).$$

Finally, exactly:
$$r \cdot \hat b = \tfrac{68}{25}\cdot\tfrac35 + \left(-\tfrac{51}{25}\right)\cdot\tfrac45
= \tfrac{204}{125} - \tfrac{204}{125} = 0 . \;\checkmark$$

Every step stays in exact fractions, so the final zero is exact — not a
rounding artifact.
The two products $\tfrac{204}{125}$ canceling to the digit is the whole
point: the leftover $r$ carries no trace of the $\hat b$ direction.

## Part (c)

Expand with the dot product's algebra rules:
$$r \cdot \hat b = \left(a - (a\cdot\hat b)\,\hat b\right)\cdot\hat b
= a\cdot\hat b - (a\cdot\hat b)\,(\hat b\cdot\hat b),$$
using distributivity, $x\cdot(y+z) = x\cdot y + x\cdot z$, and the
scalar-pull-out rule, $(c\,x)\cdot y = c\,(x\cdot y)$.
Because $\hat b$ is a unit vector, $\hat b\cdot\hat b = \|\hat b\|^2 = 1$, so
the expression collapses to $a\cdot\hat b - a\cdot\hat b = 0$.
Nothing in the argument used the specific numbers — only the algebra rules
and unit length — so the residual is orthogonal to $\hat b$ for every choice
of $a$ and every unit vector $\hat b$.

## Numeric check

```python
a = np.array([5.0, 1.0])
b_hat = np.array([3.0, 4.0]) / 5.0

reach = a @ b_hat
p = reach * b_hat
r = a - p

print("b_hat . b_hat =", b_hat @ b_hat)
print("a . b_hat     =", reach, "(= 19/5)")
print("p =", p, "(= 57/25, 76/25)")
print("r =", r, "(= 68/25, -51/25)")
print("r . b_hat     =", r @ b_hat)
```

The floating-point run reproduces every fraction from part (b) and returns a
residual dot product at the level of rounding error — numerically zero.
The exact-fraction argument is what certifies it as truly zero.
