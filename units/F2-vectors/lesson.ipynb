## Unit goals

By the end of this unit you can:

- treat a 1-D NumPy array as a vector, and picture it as a point or an
  arrow;
- measure vector size two ways (Euclidean and Manhattan norms) and compute
  distances between data points — including for a whole batch of vectors at
  once, without loops;
- compute the dot product algebraically and geometrically, and explain why
  the two definitions agree;
- compare directions with cosine similarity and know when it is the right
  tool (and when a distance is);
- normalize vectors to unit length and work with orthonormal sets as custom
  axes;
- split any vector into a projection along a direction plus an orthogonal
  residual, prove the orthogonality, and verify it numerically.
