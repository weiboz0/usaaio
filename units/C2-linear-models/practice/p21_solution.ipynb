The intercept is the first column:
\[
X=\begin{pmatrix}
1&0&1\\1&1&0\\1&2&1\\1&3&2
\end{pmatrix},\qquad
y=\begin{pmatrix}1\\2\\2\\5\end{pmatrix}.
\]

**(a)** Starting from
$L(\beta)=\lVert X\beta-y\rVert^2/4$, derive the gradient and normal
equations. Multiply out exact integer $G=X^TX$ and $c=X^Ty$.

**(b)** Solve by elimination. Give $\hat\beta$ as reduced fractions and
store a common-denominator form in integer \`beta_num_p21\` and positive
integer \`beta_den_p21\`. Show the elimination; decimals alone earn no
credit.

**(c)** Consume those results: build \`gram_p21\`, \`rhs_p21\`, solve
with \`np.linalg.solve\`, then compute predictions, residuals, MSE, and
maximum residual-orthogonality gap.

**(d)** Use \`ATOL = 1e-10\`, \`RTOL = 0.0\` to check the hand-derived
system, reduced fraction, solve result, and orthogonality. Explain why a
nonzero residual does not invalidate the normal equations.

*Your derivation, elimination, and explanation here.*

Because there are four observations,

$$
\nabla_\beta L=\frac{2}{4}X^T(X\beta-y)
=\frac12X^T(X\beta-y).
$$

Setting the gradient to zero gives $X^TX\hat\beta=X^Ty$. Direct multiplication yields

$$
G=\begin{pmatrix}4&6&4\\6&14&8\\4&8&6\end{pmatrix},
\qquad
c=\begin{pmatrix}10\\21\\13\end{pmatrix}.
$$

Thus

$$
\begin{aligned}
4\beta_0+6\beta_1+4\beta_2&=10,\\
6\beta_0+14\beta_1+8\beta_2&=21,\\
4\beta_0+8\beta_1+6\beta_2&=13.
\end{aligned}
$$

Subtracting the first equation from the third gives $2\beta_1+2\beta_2=3$. Twice the second equation minus three times the first gives $10\beta_1+4\beta_2=12$, or $5\beta_1+2\beta_2=6$. Subtracting gives $3\beta_1=3$, so

$$
\hat\beta_1=1,
\qquad
\hat\beta_2=\frac12,
\qquad
\hat\beta_0=\frac12.
$$

Hence the reduced common-denominator form is $\hat\beta=(1,2,1)^T/2$. The predictions are $(1,3/2,3,9/2)^T$, the residual is $(0,-1/2,1,-1/2)^T$, and the MSE is $3/8$. A nonzero residual does not invalidate the normal equations: least squares requires $X^Tr=0$, meaning the residual is orthogonal to the column space, not that every observation is interpolated exactly.

```python
import numpy as np
from math import gcd

ATOL = 1e-10
RTOL = 0.0

X = np.array([[1.0, 0.0, 1.0],
              [1.0, 1.0, 0.0],
              [1.0, 2.0, 1.0],
              [1.0, 3.0, 2.0]])
y = np.array([1.0, 2.0, 2.0, 5.0])

gram_hand_p21 = np.array([[4, 6, 4],
                          [6, 14, 8],
                          [4, 8, 6]], dtype=int)
rhs_hand_p21 = np.array([10, 21, 13], dtype=int)
beta_num_p21 = np.array([1, 2, 1], dtype=int)
beta_den_p21 = 2

gram_p21 = X.T @ X
rhs_p21 = X.T @ y
beta_solve_p21 = np.linalg.solve(gram_p21, rhs_p21)
beta_exact_p21 = beta_num_p21 / beta_den_p21
pred_p21 = X @ beta_solve_p21
resid_p21 = pred_p21 - y
mse_p21 = float(np.mean(resid_p21**2))
orthogonality_gap_p21 = float(np.max(np.abs(X.T @ resid_p21)))

fraction_is_reduced_p21 = (
    gcd(
        gcd(gcd(abs(int(beta_num_p21[0])), abs(int(beta_num_p21[1]))),
            abs(int(beta_num_p21[2]))),
        beta_den_p21,
    ) == 1
)
gram_check_p21 = np.allclose(gram_p21, gram_hand_p21, atol=ATOL, rtol=RTOL)
rhs_check_p21 = np.allclose(rhs_p21, rhs_hand_p21, atol=ATOL, rtol=RTOL)
beta_check_p21 = np.allclose(beta_solve_p21, beta_exact_p21, atol=ATOL, rtol=RTOL)
orthogonality_check_p21 = np.allclose(X.T @ resid_p21, 0.0, atol=ATOL, rtol=RTOL)
```

```python
assert np.array_equal(gram_hand_p21, np.array([[4, 6, 4], [6, 14, 8], [4, 8, 6]]))
assert np.array_equal(rhs_hand_p21, np.array([10, 21, 13]))
assert np.array_equal(beta_num_p21, np.array([1, 2, 1]))
assert beta_den_p21 == 2 and fraction_is_reduced_p21
assert gram_check_p21 and rhs_check_p21 and beta_check_p21
assert np.allclose(beta_exact_p21, np.array([0.5, 1.0, 0.5]), atol=ATOL, rtol=RTOL)
assert np.allclose(pred_p21, np.array([1.0, 1.5, 3.0, 4.5]), atol=ATOL, rtol=RTOL)
assert np.allclose(resid_p21, np.array([0.0, -0.5, 1.0, -0.5]), atol=ATOL, rtol=RTOL)
assert np.allclose(mse_p21, 3 / 8, atol=ATOL, rtol=RTOL)
assert orthogonality_check_p21 and orthogonality_gap_p21 <= ATOL
```
