# C2-linear-models — Practice p07 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (20, 3))
y = rng.normal(0, 1, 20)
w0 = np.array([1.0, -1.0, 0.5])
b0 = 0.5
lam = 0.4

def ridge_loss(X, y, w, b, lam):
    residuals = np.sum(X * w, axis=1) + b - y
    return float(np.mean(residuals ** 2) + lam * np.sum(w ** 2))

def ridge_grad_w(X, y, w, b, lam):
    residuals = np.sum(X * w, axis=1) + b - y
    return 2 * np.mean(residuals[:, None] * X, axis=0) + 2 * lam * w

def ridge_grad_b(X, y, w, b, lam):
    residuals = np.sum(X * w, axis=1) + b - y
    return float(2 * np.mean(residuals))

loss_val = ridge_loss(X, y, w0, b0, lam)
grad_w = ridge_grad_w(X, y, w0, b0, lam)
grad_b = ridge_grad_b(X, y, w0, b0, lam)
h = 1e-6
w_plus = w0 + h * np.eye(3)
w_minus = w0 - h * np.eye(3)
resid_plus = np.sum(X[None, :, :] * w_plus[:, None, :], axis=2) + b0 - y[None, :]
resid_minus = np.sum(X[None, :, :] * w_minus[:, None, :], axis=2) + b0 - y[None, :]
loss_plus = np.mean(resid_plus ** 2, axis=1) + lam * np.sum(w_plus ** 2, axis=1)
loss_minus = np.mean(resid_minus ** 2, axis=1) + lam * np.sum(w_minus ** 2, axis=1)
fd_w = (loss_plus - loss_minus) / (2 * h)
fd_b = (ridge_loss(X, y, w0, b0 + h, lam) - ridge_loss(X, y, w0, b0 - h, lam)) / (2 * h)
check_gap = float(np.max(np.abs(np.append(grad_w, grad_b) - np.append(fd_w, fd_b))))
```

The residual-based gradient adds the ridge pull $2\lambda w$ to each weight derivative while leaving the bias derivative unpenalized. Central differences agree with all four analytic partials to numerical precision.
