# C2-linear-models — Practice p17 — Solution

```python
import numpy as np

a = np.array([3.0, -1.0, 0.5, -4.0, 1.2])
lam = 2.5
w_l1 = np.where(np.abs(a) > lam / 2, np.sign(a) * (np.abs(a) - lam / 2), 0.0)
w_l2 = a / (1 + lam)
grid = np.linspace(-5, 5, 1000001)
grid_losses = (grid[:, None] - a[None, :]) ** 2 + lam * np.abs(grid[:, None])
grid_minimizers = grid[np.argmin(grid_losses, axis=0)]
max_anchor_gap = float(np.max(np.abs(w_l1 - grid_minimizers)))
zeros_l1 = int(np.count_nonzero(w_l1 == 0.0))
zeros_l2 = int(np.count_nonzero(w_l2 == 0.0))
shrink_check = float(np.max(np.abs(w_l2 - a / 3.5)))
```

L1 kills coordinates 1, 2, and 4 because each satisfies $|a_k|\le\lambda/2=1.25$. The remaining coordinates exceed the threshold and retain their signs after shrinking by $1.25$.

### Answer check

```python
assert np.allclose(w_l1, np.array([1.75, 0.0, 0.0, -2.75, 0.0]))
assert np.allclose(w_l2, np.array([6/7, -2/7, 1/7, -8/7, 12/35]))
assert max_anchor_gap <= 1e-5
assert zeros_l1 == 3
assert zeros_l2 == 0
assert shrink_check == 0.0
```
