# C2-linear-models — Practice p04 — Solution

```python
import numpy as np

X = np.array([[1.0, 2.0], [2.0, 0.0]])
y = np.array([2.0, 1.0])
w = np.array([2.0, -1.0])
b = 1.0
residuals = np.sum(X * w, axis=1) + b - y
feature_index = 1
derivative_w1 = float(2.0 * np.mean(residuals * X[:, feature_index]))
sign_part = float(np.sign(derivative_w1))
magnitude_part = abs(derivative_w1)
normal_form_value = sign_part + magnitude_part
choice_values = np.array([-1.0, 0.0, 1.0, 3.0, 5.0])
answer_p04 = str(np.array(list("ABCDE"))[np.flatnonzero(np.isclose(choice_values, normal_form_value))[0]])
```

Using zero-based feature indexing, $w_1$ multiplies the second column, so its derivative is $-2$. Its normal-form value is $-1+2=1$, which maps to choice C.

### Answer check

```python
recomputed_derivative = float(2.0 * np.mean((np.sum(X * w, axis=1) + b - y) * X[:, 1]))
assert np.isclose(recomputed_derivative, -2.0)
assert np.isclose(normal_form_value, 1.0)
assert answer_p04 == "C"
```
