# C2-linear-models — Practice p18 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (25, 3))
y = 2 * X[:, 0] - X[:, 1] + 1 + rng.normal(0, 0.5, 25)
w_base = np.array([2.0, -1.0, 0.5])
u = np.array([1.0, 0.5, -1.0])
b = 1.0
t_grid = np.linspace(-3, 3, 121)
weight_family = w_base + t_grid[:, None] * u
predictions = np.sum(X[None, :, :] * weight_family[:, None, :], axis=2) + b
mse_vals = np.mean((predictions - y[None, :]) ** 2, axis=1)
best_index = int(np.argmin(mse_vals))
t_best = float(t_grid[best_index])
mse_best = float(mse_vals[best_index])
second_diffs = np.diff(mse_vals, 2)
parabola_gap = float(second_diffs.max() - second_diffs.min())
```

Along the line, every prediction is affine in $t$, so every squared residual is quadratic and their average has a nonnegative leading coefficient. The resulting one-dimensional landscape is a convex parabola, promising no spurious local minima for a downhill walker along this direction.
