# C2-linear-models — Practice p09 — Solution

```python
import numpy as np

X = np.array([[1.0, 2.0], [3.0, 0.0], [2.0, 1.0], [0.0, 4.0]])
y = np.array([4.0, 2.0, 3.0, 6.0])
w = np.array([1.0, 1.0])
b = 0.0
preds = np.sum(X * w, axis=1) + b
resid = preds - y
mse_val = float(np.mean(resid ** 2))
worst_i = int(np.argmax(np.abs(resid)))
```

The predictions are $(3,3,3,4)$, giving prediction-minus-target residuals $(-1,1,0,-2)$. Their squared mean is $1.5$, and index $3$ has the largest absolute residual.

### Answer check

```python
assert np.array_equal(preds, np.array([3.0, 3.0, 3.0, 4.0]))
assert np.array_equal(resid, np.array([-1.0, 1.0, 0.0, -2.0]))
assert np.isclose(mse_val, 1.5)
assert worst_i == 3
```
