# C2-linear-models — Practice p08 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (15, 2))
y = X[:, 0] - 2 * X[:, 1] + 0.5 + rng.normal(0, 0.4, 15)
lam = 0.8

def lasso_loss(X, y, w, b, lam):
    residuals = np.sum(X * w, axis=1) + b - y
    return float(np.mean(residuals ** 2) + lam * np.sum(np.abs(w)))

w_a = np.array([1.0, -2.0])
w_b = np.array([0.0, -2.5])
w_c = np.array([0.0, 0.0])
pen_a, loss_a = float(lam * np.sum(np.abs(w_a))), lasso_loss(X, y, w_a, 0.5, lam)
pen_b, loss_b = float(lam * np.sum(np.abs(w_b))), lasso_loss(X, y, w_b, 1.0, lam)
pen_c, loss_c = float(lam * np.sum(np.abs(w_c))), lasso_loss(X, y, w_c, 3.0, lam)
lowest = str(np.array(["a", "b", "c"])[np.argmin(np.array([loss_a, loss_b, loss_c]))])
h = 1e-6
slope_right = float((abs(h) - abs(0.0)) / h)
slope_left = float((abs(0.0) - abs(-h)) / h)
kink_jump = slope_right - slope_left
```

Candidate a has the smallest full loss even though its penalty is larger than those of b and c, because its data fit is much better. The one-sided slopes $1$ and $-1$ produce a kink jump of $2$ at zero.

### Answer check

```python
assert np.allclose(np.array([pen_a, pen_b, pen_c]), np.array([2.4, 2.0, 0.0]))
assert np.allclose(np.array([loss_a, loss_b, loss_c]), np.array([2.512828270575275, 4.50522463602382, 7.224042297705222]))
assert lowest == "a"
assert np.allclose(np.array([slope_right, slope_left, kink_jump]), np.array([1.0, -1.0, 2.0]))
```
