# C2-linear-models — Practice p05 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (30, 3))
w = np.array([1.5, -2.0, 0.5])
b = 4.0

def predict(X, w, b):
    return np.sum(X * w, axis=1) + b

preds = predict(X, w, b)
pred_first = float(preds[0])
```

Broadcasting multiplies every feature column by its corresponding weight. Summing across columns produces one score per row, after which the scalar bias shifts every prediction.
