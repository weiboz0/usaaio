# C2-linear-models — Practice p13 — Solution

```python
import numpy as np

X = np.array([[1.0, 0.0], [2.0, 1.0], [0.0, 1.0], [1.0, 2.0]])
y = np.array([3.0, 3.0, 1.0, 4.0])
w = np.array([1.0, 1.0])
b = 1.0
lam = 0.5
preds = np.sum(X * w, axis=1) + b
resid = preds - y
mse_hand = float(np.mean(resid ** 2))

def mse_grad_w(X, y, w, b):
    residuals = np.sum(X * w, axis=1) + b - y
    return 2 * np.mean(residuals[:, None] * X, axis=0)

def mse_grad_b(X, y, w, b):
    residuals = np.sum(X * w, axis=1) + b - y
    return float(2 * np.mean(residuals))

grad_w = mse_grad_w(X, y, w, b)
grad_b = mse_grad_b(X, y, w, b)
h = 1e-6
w_plus = w + h * np.eye(2)
w_minus = w - h * np.eye(2)
mse_plus = np.mean((np.sum(X[None, :, :] * w_plus[:, None, :], axis=2) + b - y[None, :]) ** 2, axis=1)
mse_minus = np.mean((np.sum(X[None, :, :] * w_minus[:, None, :], axis=2) + b - y[None, :]) ** 2, axis=1)
fd_w = (mse_plus - mse_minus) / (2 * h)
fd_b = (np.mean((preds + h - y) ** 2) - np.mean((preds - h - y) ** 2)) / (2 * h)
check_gap = float(np.max(np.abs(np.append(grad_w, grad_b) - np.append(fd_w, fd_b))))
ridge_grad = grad_w + 2 * lam * w
diff_vec = ridge_grad - grad_w
formula_gap = float(np.max(np.abs(diff_vec - 2 * lam * w)))
```

The hand calculation gives predictions $(2,4,2,4)$, residuals $(-1,1,1,0)$, and MSE $0.75$. These residuals give weight gradient $(0.5,1)$ and bias gradient $0.5$, matching central differences.

Both features feel the same ridge penalty pull because their weights have the same magnitude and $2\lambda w=(1,1)$. The computed difference vector confirms that prediction exactly.

### Answer check

```python
assert np.array_equal(preds, np.array([2.0, 4.0, 2.0, 4.0]))
assert np.array_equal(resid, np.array([-1.0, 1.0, 1.0, 0.0]))
assert np.isclose(mse_hand, 0.75)
assert np.allclose(grad_w, np.array([0.5, 1.0]))
assert np.isclose(grad_b, 0.5)
assert check_gap < 1e-6
assert np.allclose(ridge_grad, np.array([1.5, 2.0]))
assert formula_gap < 1e-12
```
