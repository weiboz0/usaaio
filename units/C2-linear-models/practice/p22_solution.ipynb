**Reasoning is required.** Let
$X\in\mathbb R^{n\times p}$ and $G=X^TX$.

**(a)** Prove
$v^TGv=\lVert Xv\rVert_2^2$. Use it for both directions of
\[
X\text{ full column rank}
\Longleftrightarrow G\text{ positive definite}
\Longleftrightarrow G\text{ invertible}.
\]
Justify the symmetric positive-definite/invertible step.

**(b) Refute $\boldsymbol{n>p}$.** Construct \`X_counter_p22\` with
shape \`(5, 3)\`, no all-zero column, and dependent columns. Supply a
nonzero \`v_counter_p22 (3,)\` with $Xv=0$. Explain why row count does
not prove identifiability.

**(c)** Compute Gram, rank, witness image/quadratic form, and smallest
singular value. Check the two zero identities with
\`ATOL = 1e-10\`, \`RTOL = 0.0\`. The exact column relation and null
vector are the proof; the numeric singular value is only evidence.

*Your proof, exact column relation, and explanation here.*

For every $v\in\mathbb R^p$, associativity and the Euclidean inner product give

$$
v^TGv=v^TX^TXv=(Xv)^T(Xv)=\lVert Xv\rVert_2^2.
$$

If $X$ has full column rank, then every nonzero $v$ has $Xv\ne0$, so $v^TGv>0$; hence $G$ is positive definite. Conversely, if $G$ is positive definite and $Xv=0$, then $v^TGv=0$, forcing $v=0$, so $X$ has full column rank. Because $G=X^TX$ is symmetric, positive definiteness implies invertibility: a singular $G$ would have nonzero $v$ with $Gv=0$, contradicting $v^TGv>0$. Finally, if $G$ is invertible but $X$ were not full column rank, some nonzero $v$ would satisfy $Xv=0$, hence $Gv=X^TXv=0$, contradicting invertibility. Thus

$$
X\text{ full column rank}
\Longleftrightarrow G\text{ positive definite}
\Longleftrightarrow G\text{ invertible}.
$$

For the counterexample, write the columns as $c_1,c_2,c_3$. The exact relation is $c_3=c_1+c_2$, so

$$
X\begin{pmatrix}1\\1\\-1\end{pmatrix}=0.
$$

All three columns are nonzero, and $c_1,c_2$ are independent because their first two entries are $(1,0)^T$ and $(0,1)^T$; therefore the rank is exactly 2. Although $n=5>3=p$, extra rows do not prevent a dependence among columns, so row count alone does not establish identifiability. The exact relation and null vector prove dependence; the near-zero smallest singular value is only numerical evidence.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

X_counter_p22 = np.array([[1.0, 0.0, 1.0],
                          [0.0, 1.0, 1.0],
                          [1.0, 1.0, 2.0],
                          [2.0, -1.0, 1.0],
                          [-1.0, 2.0, 1.0]])
v_counter_p22 = np.array([1.0, 1.0, -1.0])

gram_counter_p22 = X_counter_p22.T @ X_counter_p22
rank_counter_p22 = int(np.linalg.matrix_rank(X_counter_p22))
null_image_p22 = X_counter_p22 @ v_counter_p22
quadratic_form_p22 = float(v_counter_p22.T @ gram_counter_p22 @ v_counter_p22)
smallest_singular_p22 = float(np.linalg.svd(X_counter_p22, compute_uv=False)[-1])

shape_check_p22 = X_counter_p22.shape == (5, 3) and v_counter_p22.shape == (3,) and 5 > 3
nonzero_columns_check_p22 = bool(np.all(np.linalg.norm(X_counter_p22, axis=0) > 0.0))
null_check_p22 = np.allclose(null_image_p22, 0.0, atol=ATOL, rtol=RTOL)
quadratic_check_p22 = np.allclose(quadratic_form_p22, 0.0, atol=ATOL, rtol=RTOL)
```

```python
assert shape_check_p22 and nonzero_columns_check_p22
assert np.array_equal(X_counter_p22[:, 2], X_counter_p22[:, 0] + X_counter_p22[:, 1])
assert np.array_equal(v_counter_p22, np.array([1.0, 1.0, -1.0]))
assert rank_counter_p22 == 2
assert null_check_p22 and quadratic_check_p22
assert np.allclose(null_image_p22, 0.0, atol=ATOL, rtol=RTOL)
assert np.allclose(quadratic_form_p22, 0.0, atol=ATOL, rtol=RTOL)
assert smallest_singular_p22 <= ATOL
```
