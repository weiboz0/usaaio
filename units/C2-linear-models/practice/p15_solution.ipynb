# C2-linear-models — Practice p15 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
mags = rng.uniform(0.01, 0.15, 40)
signs = np.where(rng.uniform(0, 1, 40) < 0.5, -1.0, 1.0)
w_l2 = np.round(mags * signs, 4)
w_l1 = np.zeros(40)
w_l1[np.array([2, 7, 11, 19, 26, 33])] = np.array([1.42, -0.85, 0.60, -1.10, 0.95, 0.30])
COST = 120
n_sensors_l2 = int(np.count_nonzero(w_l2 != 0.0))
n_sensors_l1 = int(np.count_nonzero(w_l1 != 0.0))
cost_l2 = n_sensors_l2 * COST
cost_l1 = n_sensors_l1 * COST
saved = cost_l2 - cost_l1
```

Near zero, L1 still exerts a constant-magnitude pull while L2's pull is proportional to the weight and fades to zero, so only L1 naturally pins coefficients exactly at zero. Manually discarding small nonzero L2 coefficients changes its predictions without retraining, so all 40 readings remain inputs to the supplied L2 model.
