# C2-linear-models — Practice p11 — Solution

For $w>0$, $f'(w)=2(w-a)+\lambda$, so the only candidate is $w=a-\lambda/2$ and it is admissible exactly when $a>\lambda/2$. For $w<0$, $f'(w)=2(w-a)-\lambda$, so the only candidate is $w=a+\lambda/2$ and it is admissible exactly when $a<-\lambda/2$. At the kink, the right slope is $-2a+\lambda$ and the left slope is $-2a-\lambda$, hence both one-sided minimum conditions hold exactly when $-\lambda/2\le a\le\lambda/2$. Because each smooth region has second derivative $2>0$, these candidates are minima, and together they give $w^*=a-\lambda/2$ for $a>\lambda/2$, $w^*=0$ for $|a|\le\lambda/2$, and $w^*=a+\lambda/2$ for $a<-\lambda/2$; at either boundary the neighboring formula also equals zero.

Thus $w^*=0$ exactly when $|a|\le\lambda/2$. In that interval the L1 term supplies opposing nonzero one-sided pulls that bracket zero, so the kink can be the exact minimum. A squared penalty has derivative $2\lambda w$, which vanishes at zero and therefore cannot counter a nonzero data-term slope there.

```python
import numpy as np

a = 3.0
lam = 4.0
w_star_hand = float(np.where(abs(a) <= lam / 2, 0.0, np.sign(a) * (abs(a) - lam / 2)))
grid = np.linspace(-2, 4, 600001)
values = (grid - a) ** 2 + lam * np.abs(grid)
w_star_grid = float(grid[np.argmin(values)])
anchor_gap = abs(w_star_hand - w_star_grid)
a_zero = 1.5
w_star_zero = float(np.where(abs(a_zero) <= lam / 2, 0.0, np.sign(a_zero) * (abs(a_zero) - lam / 2)))
zero_values = (grid - a_zero) ** 2 + lam * np.abs(grid)
w_star_zero_grid = float(grid[np.argmin(zero_values)])
```

For $(a,\lambda)=(3,4)$ the formula gives $w^*=1$, and the dense-grid anchor agrees. For $(1.5,4)$ the threshold condition holds, so both the formula and grid give exactly zero.

### Answer check

```python
assert np.isclose(w_star_hand, 1.0, atol=1e-9, rtol=0)
assert anchor_gap <= 1e-5
assert w_star_zero == 0.0
assert abs(w_star_zero_grid - w_star_zero) <= 1e-5
```
