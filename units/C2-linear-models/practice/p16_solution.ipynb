# C2-linear-models — Practice p16 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X_test = rng.normal(0, 1, (40, 4))
y_test = 1.2 * X_test[:, 0] + 0.8 * X_test[:, 1] + 2 + rng.normal(0, 0.5, 40)
w_S = np.array([1.9, 0.2, -1.1, 0.9])
b_S = 1.4
w_B = np.array([1.0, 0.7, 0.0, 0.1])
b_B = 1.9
pred_S = np.sum(X_test * w_S, axis=1) + b_S
pred_B = np.sum(X_test * w_B, axis=1) + b_B
mse_S = float(np.mean((pred_S - y_test) ** 2))
mse_B = float(np.mean((pred_B - y_test) ** 2))
better_model = str(np.array(["S", "B"])[np.argmin(np.array([mse_S, mse_B]))])
```

The teammate conflates penalized training loss with plain held-out MSE, and totals at different strengths are not comparable because model B is mechanically charged the much higher $\lambda=5$ tax. The fair scoreboard applies the same plain test MSE to both models: S scores about $2.2083$, whereas B scores about $0.3249$. Model B is therefore better on the held-out data despite its larger penalized training total.
