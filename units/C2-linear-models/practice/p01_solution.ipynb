# C2-linear-models — Practice p01 — Solution

```python
import numpy as np

x = np.array([4.0, 5.0])
w = np.array([2.0, -1.0])
b = 3.0
predicted_value = float(np.sum(x * w) + b)
choice_values = np.array([-2.0, 3.0, 6.0, 8.0, 13.0])
choice_letters = np.array(list("ABCDE"))
answer_p01 = str(choice_letters[np.flatnonzero(np.isclose(choice_values, predicted_value))[0]])
```

The feature contributions are $4(2)=8$ and $5(-1)=-5$, and the bias contributes $3$. Their sum is $6$, so the matching choice is C.

### Answer check

```python
assert np.isclose(predicted_value, 6.0)
assert answer_p01 == "C"
```
