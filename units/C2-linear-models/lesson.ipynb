## Unit goals

By the end of this unit you can:

- write and compute the linear model in this course's fixed component
  form, $\hat y_i = \sum_k X_{ik} w_k + b$ — weights `w (d,)`, explicit
  bias `b`, data `X (n, d)` — and see the same model as F3's
  matrix-map / column-blend picture;
- define residuals $r_i = \hat y_i - y_i$ and the mean squared error
  $\mathrm{MSE} = \frac1n\sum_i(\hat y_i - y_i)^2$, and connect it to
  F2's squared distance and F4's sum-of-squares score;
- **evaluate** losses at given weights (candidate scoreboards, train vs
  test columns) and **differentiate** them at given weights —
  $\partial\,\mathrm{MSE}/\partial w_j = \frac{2}{n}\sum_i r_i X_{ij}$,
  $\partial\,\mathrm{MSE}/\partial b = \frac{2}{n}\sum_i r_i$, plus the
  penalty terms — verified with central differences, implemented
  broadcasting-only under the exam's API bans;
- read the gradient-zero condition geometrically: at the best weights
  the residual is orthogonal to every feature column (F2's projection
  picture through F3's reachable set);
- define the L2 (ridge) and L1 (lasso) penalties, keep the bias out of
  both, and evaluate/differentiate the penalized losses at given
  weights;
- **derive**, by Calc AB case analysis at the kink, the exact minimizer
  of $(w - a)^2 + \lambda|w|$ — the soft threshold, zero exactly when
  $|a| \le \lambda/2$ — and the ridge counterpart $a/(1+\lambda)$, and
  explain from them why **L1 zeroes while L2 shrinks**;
- define **sparsity**, count exact zeros honestly, and read
  regularization paths given as supplied weight tables (train/test MSE
  along the path, zero counts, choosing $\lambda$ from the held-out
  column).

**What this unit deliberately does not do:** find weights.
No iterative updates, no training loops, no normal equations, no matrix
inversion, no `np.linalg` — every weight in this unit is given, supplied
in the statement, or obtained in closed form by calculus.
Iteration is the entire subject of `C3-gradient-descent`, which fits
this unit's model with this unit's losses, notation included.
