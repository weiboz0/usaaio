## Unit goals

By the end of this unit you can:

- compute a **convolution** by hand and in NumPy component form —
  1-D and 2-D, valid mode, the no-flip (cross-correlation) convention
  — and rebuild the same arithmetic as `nn.Conv2d` with hand-set
  kernels (`nn.Parameter(..., requires_grad=False)`, C6's register);
- design kernels as **detectors** (edge, blur), read their response
  maps, and command the output-size formula
  $\lfloor (n + 2p - K)/s \rfloor + 1$ for every stride/padding
  combination;
- read a multi-kernel layer's output as stacked **feature maps** —
  `(N, F, H', W')`, one map per detector — and track channel meaning
  from colors to pattern responses;
- compute **receptive fields** by the growth rule
  ($r \leftarrow r + (K-1)J$, $J \leftarrow Js$), prove the stride-1
  closed form, and verify either empirically with an impulse;
- articulate the **feature hierarchy** — edges → textures → parts →
  objects — and back it with measurable statistics (activation
  fraction, roughness) on synthetic maps, plus the shape signature
  (grids shrink, channels grow);
- load **ResNet-50** reproducibly
  (`weights=ResNet50_Weights.IMAGENET1K_V1`, `model.eval()`, forwards
  under `torch.inference_mode()`, the float32 artifact register with
  `atol=1e-6`/`rtol=1e-5`) from the course's local cache;
- read its **architecture**: stem / `layer1..4` / head via
  `named_children`, block counts `[3, 4, 6, 3]`, hand-traced and
  code-verified stage shapes for any input size;
- open a **bottleneck block** — 1×1 reduce, 3×3, 1×1 expand,
  expansion 4, skip connection, downsample openers — and hand-count
  its parameters (convs $\text{out}\cdot\text{in}\cdot K^2$, BN $2C$)
  in the exam's count-without-`numel` register;
- **truncate** the network with `nn.Sequential` slices over
  `list(model.children())`, including mid-stage cuts (`layerN[:k]`),
  and predict every trunk's output shape;
- **freeze at scale** — `requires_grad=False` loops, name-prefix
  filters — and audit frozen/trainable in both currencies (tensors
  and scalars);
- build **transfer constructions**: frozen backbone + pool + flatten
  + fresh `nn.Linear` head (slice-build or `model.fc` surgery), with
  the full audit signature — output shape, flag membership, hand-
  predicted counts ($f \cdot k + k$; the $2049k$ full-depth form).

**What this unit deliberately does not do: train.**
No losses, no gradients, no optimizers — the fence from C5/C6 carries
over, sharpened in Session 3's scope note: *training the head needs
machinery beyond this course; the exam grades the construction.*
Every network here is loaded or hand-built, run forward under
`inference_mode`, and audited.

manifest carries the same grouping).

frozen, counted, and audited — never trained.

**Set A — fundamentals** (start after Session 1; `p12` after
Session 2):

- `p02` (MC, intro) — conv output shapes with stride and padding.

  count the top level.
- `p21` (drill, intro) — rapid-fire feature-map shapes and counts on

  tiny map statistics.

**Set B — exam register** (after Session 2; `p09`–`p11` after
Session 3):

- `p03` (MC, core) — the feature hierarchy, in the depth-ordering
  register.
- `p04` (MC, numeric normal form, core) — bottleneck conv arithmetic
  as a reduced fraction.
- `p05` (MC, core) — receptive field of a small stack.
- `p07` (coding, core) — a two-kernel detector bank reconciled
  against component form.
- `p08` (coding, core) — architecture reading: blocks, channels,
  strides off the loaded model.
- `p09` (coding, core) — three trunks by slicing, shapes verified.

   bedrock.
   Then do `p01`, `p02`, `p06`, `p21`, `p22` the same day.

3. Work Session 3; rebuild the §6 `TransferNet` from memory before

   consume earlier results; write `p13`'s and `p14`'s derivations as
   if a grader who has not read your mind will mark them.

   and the self-quiz will tell you which problems to redo.
