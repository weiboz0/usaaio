# C7-cnn-transfer — Practice p21 — Solution

The layer creates twelve maps. Same padding keeps 30×30 at stride one; the stride-two formula gives 15.

```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)   # course convention (no pretrained weights here)
SEED = 20260804
torch.manual_seed(SEED)
conv = nn.Conv2d(3, 12, kernel_size=5, stride=1, padding=2)
x = torch.randn(4, 3, 30, 30)

w_shape = tuple(conv.weight.shape)
b_shape = tuple(conv.bias.shape)
n_maps = 12
with torch.inference_mode():
    y_shape = tuple(conv(x).shape)
n_params = 12 * 3 * 5 * 5 + 12
out_h_s2 = (30 + 2 * 2 - 5) // 2 + 1
n_positions = 30 * 30
```

### Answer check

```python
assert w_shape == (12, 3, 5, 5)
assert b_shape == (12,)
assert n_maps == 12
assert y_shape == (4, 12, 30, 30)
assert n_params == 912
assert out_h_s2 == 15
assert n_positions == 900
```
