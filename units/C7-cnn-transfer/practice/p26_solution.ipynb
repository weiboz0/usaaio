# C7-cnn-transfer — Practice p26 — Solution

Each iteration carries the previous convolution's height and width forward, applies integer floor division independently on the two axes, and replaces the channel count with `c_out`. The three derived spatial pairs are `(96, 227)`, `(32, 114)`, and `(32, 36)`; a size check occurs before appending so an invalid convolution cannot enter the returned trace.

```python
def trace_conv_stack(input_shape, specs):
    batch, channels, height, width = input_shape
    trace = []
    for c_out, k_h, k_w, s_h, s_w, p_h, p_w in specs:
        next_height = (height + 2 * p_h - k_h) // s_h + 1
        next_width = (width + 2 * p_w - k_w) // s_w + 1
        if next_height < 1 or next_width < 1:
            raise ValueError("computed spatial size must be at least 1")
        height, width, channels = next_height, next_width, c_out
        trace.append((batch, channels, height, width))
    return trace


input_shape = (3, 17, 191, 227)
specs = [
    (31, 7, 3, 2, 1, 3, 1),
    (43, 3, 5, 3, 2, 0, 2),
    (59, 1, 7, 1, 3, 0, 0),
]
trace = trace_conv_stack(input_shape, specs)
```

### Answer check

```python
assert trace == [
    (3, 31, 96, 227),
    (3, 43, 32, 114),
    (3, 59, 32, 36),
]

try:
    trace_conv_stack((1, 2, 2, 2), [(4, 5, 5, 1, 1, 0, 0)])
except ValueError:
    invalid_size_rejected = True
else:
    invalid_size_rejected = False

assert invalid_size_rejected is True
```
