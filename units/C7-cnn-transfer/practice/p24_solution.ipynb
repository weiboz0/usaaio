# C7-cnn-transfer — Practice p24 — Solution

Hand tracing each axis gives
\[
(2,7,211,229)\to(2,13,106,115)\to(2,13,53,58)\to(2,23,53,27).
\]
For example, conv_a height is $\lfloor(211+4-5)/2\rfloor+1=106$, the pool width is $\lfloor(115+2-3)/2\rfloor+1=58$, and conv_b uses stride 2 only on width, giving $\lfloor(58-5)/2\rfloor+1=27$.

```python
shape_after_a = (2, 13, 106, 115)
shape_after_pool = (2, 13, 53, 58)
shape_after_b = (2, 23, 53, 27)
shape_trace = [shape_after_a, shape_after_pool, shape_after_b]
```

### MARKED VERIFICATION CELL

This is the only cell that runs the stack, after the hand trace above has been fixed.

```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)
SEED = 20260804
torch.manual_seed(SEED)

stack = nn.Sequential(
    nn.Conv2d(7, 13, kernel_size=5, stride=2, padding=2),
    nn.MaxPool2d(kernel_size=3, stride=2, padding=1),
    nn.Conv2d(13, 23, kernel_size=(3, 5), stride=(1, 2), padding=(1, 0)),
)
x_verify = torch.zeros(2, 7, 211, 229)
observed_shapes = []
with torch.inference_mode():
    for layer in stack:
        x_verify = layer(x_verify)
        observed_shapes.append(tuple(x_verify.shape))
```

### Answer check

```python
assert shape_after_a == (2, 13, 106, 115)
assert shape_after_pool == (2, 13, 53, 58)
assert shape_after_b == (2, 23, 53, 27)
assert shape_trace == [
    (2, 13, 106, 115),
    (2, 13, 53, 58),
    (2, 23, 53, 27),
]
assert observed_shapes == shape_trace
```
