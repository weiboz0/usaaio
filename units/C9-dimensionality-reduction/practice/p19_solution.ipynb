A team projects a fixed training split onto the first $r$ principal components, fits the same downstream classifier at every $r$, and evaluates once on a held-out validation split. The cumulative variance-explained ratio (`cum_evr`) measures how much **feature variance** the projection keeps; it is not itself a generalization score.

| retained rank $r$ | `cum_evr` | train accuracy | validation accuracy |
|---:|---:|---:|---:|
| 1 | 0.47 | 0.66 | 0.64 |
| 2 | 0.72 | 0.79 | 0.76 |
| 4 | 0.90 | 0.91 | 0.83 |
| 7 | 0.98 | 0.98 | 0.71 |

1. Set `selected_r: int` to the rank you would deploy from this table and justify it using the held-out score rather than `cum_evr` alone.
2. Set `capacity_trend: str` to either `"increases-with-r"` or `"decreases-with-r"`. Explain why retaining more components gives the downstream rule more capacity.
3. Set `r1_diagnosis: str` and `r7_diagnosis: str` to `"underfitting"`, `"balanced"`, or `"overfitting"`. Cite both the score level and train–validation gap.
4. In 3–5 sentences, distinguish the feature variance measured by `cum_evr` from the across-training-samples variance in the bias–variance tradeoff, and explain why high EVR does not guarantee a small train–validation gap.


```python
selected_r: int = 4
capacity_trend: str = "increases-with-r"
r1_diagnosis: str = "underfitting"
r7_diagnosis: str = "overfitting"

```

From the supplied table, $r=4$ is the deployment choice because its held-out accuracy, $0.83$, is the largest; the larger $r=7$ keeps more feature variance but falls to $0.71$ held-out accuracy. Retaining more components increases capacity because the classifier receives more coordinates on which it can form sample-specific decisions. At $r=1$, the low $0.66/0.64$ train/validation scores and tiny $0.02$ gap indicate underfitting, whereas $r=7$ combines $0.98$ training accuracy with a $0.27$ gap and therefore overfits. PCA's `cum_evr` measures feature spread retained in this dataset, while bias–variance variance describes how the fitted rule changes across training samples, so $0.98$ cumulative EVR cannot guarantee generalization or a small gap. Methodologically, the held-out split here is a *validation* split precisely because it selected $r$; an unbiased final estimate would require a further test split that no selection step has touched.
