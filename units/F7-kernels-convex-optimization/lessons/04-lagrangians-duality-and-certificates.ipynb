## 1. Fix the minimization sign convention first

We write a constrained minimization problem as

$$\text{minimize }f(x)\quad\text{subject to }g_i(x)\le0,\quad h_j(x)=0.$$

Its Lagrangian is

$$L(x,\lambda,\nu)=f(x)+\sum_i\lambda_i g_i(x)+\sum_j\nu_j h_j(x),$$

with inequality multipliers $\lambda_i\ge0$ and equality multipliers $\nu_j\in\mathbb{R}$. For feasible $x$, each $\lambda_i g_i(x)\le0$ and each equality term is zero, so

$$L(x,\lambda,\nu)\le f(x).$$

This direction is the anchor for everything that follows. If a source writes constraints as $g_i(x)\ge0$, its multiplier signs must change; never mix the two conventions.

## 2. The dual function and weak-duality direction

Define the dual function by minimizing the Lagrangian over the primal variable:

$$q(\lambda,\nu)=\inf_x L(x,\lambda,\nu).$$

For any dual-feasible $\lambda\ge0$ and any primal-feasible $x$,

$$q(\lambda,\nu)\le L(x,\lambda,\nu)\le f(x).$$

Therefore every dual value is a lower bound on every primal feasible value. Maximizing the lower bound gives the dual optimum $d^*$, while the primal minimum is $p^*$, and **weak duality** says

$$d^*\le p^*.$$

This proof does not require convexity. Convexity and regularity enter later when we want equality and multiplier existence.

**Worked exam-style example 1.** Minimize $x^2$ subject to $x\ge1$. Write $g(x)=1-x\le0$. Then

$$L(x,\lambda)=x^2+\lambda(1-x)=(x-\lambda/2)^2+\lambda-\lambda^2/4,$$

so $q(\lambda)=\lambda-\lambda^2/4$ for $\lambda\ge0$. Completing the square again gives $q(\lambda)=1-(\lambda-2)^2/4\le1$. The best lower bound occurs at $\lambda=2$ and equals the primal value $1$.

### Checkpoint 2

1. Which inequality direction is weak duality for this minimization convention?
2. Does observing $d^*<p^*$ contradict weak duality?

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

def q_first_example(lam):
    return lam - 0.25 * lam**2

lam_grid = np.array([0.0, 1.0, 2.0, 3.0, 4.0])
q_values = q_first_example(lam_grid)
assert np.all(q_values <= 1.0 + ATOL)
assert np.isclose(q_first_example(2.0), 1.0, atol=ATOL, rtol=RTOL)
```

## 3. Active constraints and complementary slackness

At a feasible point $x_*$, inequality $g_i(x)\le0$ is **active** when $g_i(x_*)=0$ and **inactive** when $g_i(x_*)<0$. Complementary slackness requires

$$\lambda_i g_i(x_*)=0\quad\text{for every }i.$$

Thus an inactive constraint must have $\lambda_i=0$, and a positive multiplier implies activity. The converses are not automatic: an active constraint can still have zero multiplier.

The multiplier measures sensitivity only within the convention and assumptions of the problem. Complementary slackness is not permission to set every active multiplier positive, nor does it replace primal feasibility.

## 4. The four KKT conditions and their assumptions

For differentiable data, a candidate tuple $(x_*,\lambda_*,\nu_*)$ satisfies the Karush–Kuhn–Tucker conditions when:

1. **primal feasibility:** $g_i(x_*)\le0$ and $h_j(x_*)=0$;
2. **dual feasibility:** $\lambda_{*,i}\ge0$;
3. **stationarity:** $\nabla f(x_*)+\sum_i\lambda_{*,i}\nabla g_i(x_*)+\sum_j\nu_{*,j}\nabla h_j(x_*)=0$;
4. **complementary slackness:** $\lambda_{*,i}g_i(x_*)=0$.

For a differentiable convex problem — convex $f$ and $g_i$, affine $h_j$ — any tuple satisfying all four KKT conditions certifies global optimality. In the finite-dimensional, everywhere-differentiable setting taught here, assume the primal optimum is finite and attained. If a Slater point exists (it satisfies the affine equalities and all inequality constraints strictly), then strong duality holds, a dual optimum is attained, and each primal optimum admits KKT multipliers. Without those structural, attainment, and regularity assumptions, do not claim that every optimum has such a tuple or that the duality gap is zero.

## 5. Worked exam-style example 2: a boundary certificate

Minimize $f(x)=(x-3)^2$ subject to $x\le1$. Use $g(x)=x-1\le0$ and

$$L(x,\lambda)=(x-3)^2+\lambda(x-1),\qquad\lambda\ge0.$$

At the visible boundary candidate $x_*=1$:

- primal feasibility: $g(1)=0$;
- stationarity: $2(x_*-3)+\lambda_*=0$, so $\lambda_*=4$;
- dual feasibility: $4\ge0$;
- complementary slackness: $4(1-1)=0$.

The objective and inequality function are convex, and $x=0$ is a Slater point because $g(0)<0$. Hence this KKT tuple certifies the global solution. The dual function comes from $x=3-\lambda/2$:

$$q(\lambda)=2\lambda-\lambda^2/4=4-(\lambda-4)^2/4,$$

so its maximum $4$ equals $f(1)=4$.

### Checkpoint 5

1. Why would $\lambda_*=-4$ fail even though its magnitude matches the boundary slope?

```python
x_star = 1.0
lam_star = 4.0
primal_residual = x_star - 1.0
stationarity_residual = 2.0 * (x_star - 3.0) + lam_star
complementarity_residual = lam_star * primal_residual
primal_value = (x_star - 3.0) ** 2
dual_value = 2.0 * lam_star - 0.25 * lam_star**2

assert primal_residual <= ATOL
assert lam_star >= 0.0
assert np.isclose(stationarity_residual, 0.0, atol=ATOL, rtol=RTOL)
assert np.isclose(complementarity_residual, 0.0, atol=ATOL, rtol=RTOL)
assert np.isclose(primal_value, dual_value, atol=ATOL, rtol=RTOL)
```

## 6. Equalities, free multipliers, and certificate reading

An affine equality $h(x)=0$ contributes $\nu h(x)$ with a **free** multiplier $\nu$: both signs are allowed because the term vanishes at every feasible point. For example, minimize $(x-1)^2$ subject to $h(x)=x-2=0$. Feasibility fixes $x_*=2$, and stationarity gives

$$2(x_*-1)+\nu_*=0,\qquad\nu_*=-2.$$

Rejecting this certificate because $\nu_*<0$ would confuse equality and inequality multipliers.

A complete certificate reader checks assumptions first, then the four residual families, then compares primal and dual values when a zero-gap claim is made. A small numerical residual uses a declared tolerance; it does not erase a wrong sign or an unstated regularity assumption.

## 7. Common pitfalls, exam connections, and forward links

**Pitfall — reverse weak duality.** For minimization with $g_i\le0$ and $\lambda_i\ge0$, the dual supplies lower bounds: $d^*\le p^*$. Fix: re-derive $q\le L\le f$ instead of memorizing an arrow.

**Pitfall — complementary slackness means every constraint is tight.** Inactive constraints simply receive zero multiplier. Fix: test activity and the product separately.

**Pitfall — KKT always means optimum.** The global certificate uses convex objective/inequalities and affine equalities; multiplier existence and zero gap use regularity such as Slater's condition. Fix: state assumptions before the conclusion.

**Pitfall — equality multipliers must be nonnegative.** Only inequality multipliers have that sign restriction in this convention.

**Exam connection.** Expect a candidate tuple and a request to audit feasibility, stationarity, activity, multiplier signs, and the duality direction. Write the constraint convention before doing algebra.

**Going deeper.** Later SVM material produces a convex quadratic program whose dual uses kernel Gram matrices. This unit supplies the certificates; the future unit supplies the classifier.
