### Worked solution

**Part A — convex set and convex objective.**  Feasibility is $x\ge2$, so if $x,y$ are feasible and $0\le\theta\le1$, then $(1-\theta)x+\theta y\ge2$; hence the feasible set is convex.  For $f(x)=(x+1)^2$,
\[
(1-\theta)f(x)+\theta f(y)-f((1-\theta)x+\theta y)
=\theta(1-\theta)(x-y)^2\ge0.
\]
Thus the objective is convex without a Hessian.  The constraint $g(x)=2-x$ is affine.

**Part B — KKT derivation and certificate.**  With the required convention,
\[
L(x,\lambda)=(x+1)^2+\lambda(2-x),\qquad \lambda\ge0.
\]
The four KKT conditions are
\[
2-x\le0,\quad \lambda\ge0,\quad 2(x+1)-\lambda=0,
\quad \lambda(2-x)=0.
\]
The unconstrained minimizer $-1$ is infeasible, so the feasible boundary gives $x^*=2$.  Stationarity then gives $\lambda^*=6$, and $f(x^*)=9$.  This tuple has all four residuals zero.  Part A supplies the missing structural premise: for a convex differentiable objective and affine constraint, a feasible KKT tuple is globally sufficient.  Concretely, stationarity gives $f'(x^*)=\lambda^*=6$, so for any feasible $y\ge2$ the supporting inequality gives $f(y)\ge f(x^*)+6(y-2)\ge f(x^*)$.

**Part C — dual function, weak-duality chain, Slater, and zero gap.**  Completing the square (or minimizing the scalar quadratic) gives the unconstrained minimizer of $L$ as $x=\lambda/2-1$, so for every dual-feasible $\lambda\ge0$,
\[
q(\lambda)=\inf_xL(x,\lambda)=3\lambda-\frac{\lambda^2}{4}.
\]
For any feasible $x$ and $\lambda\ge0$,
\[
q(\lambda)\le L(x,\lambda)=f(x)+\lambda g(x)\le f(x),
\]
because $g(x)\le0$.  Taking the supremum over dual-feasible $\lambda$ and infimum over feasible $x$ proves $d^*\le p^*$.  Consuming the KKT multiplier, $q(6)=18-9=9=f(2)$.

For the strong-duality conclusion, the objective is convex, the constraint is affine, both optima are finite and attained, and $x=3$ is a strict Slater point because $g(3)=-1<0$.  Under these stated assumptions, strong duality explains the zero gap; the KKT equations alone are not being used as a theorem of zero gap.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

x_star_p14 = 2.0
lambda_star_p14 = 6.0
primal_value_p14 = float((x_star_p14 + 1.0) ** 2)
_g_p14 = 2.0 - x_star_p14
residuals_p14 = np.array([
    max(_g_p14, 0.0),
    max(-lambda_star_p14, 0.0),
    abs(2.0 * (x_star_p14 + 1.0) - lambda_star_p14),
    abs(lambda_star_p14 * _g_p14),
], dtype=float)
dual_value_p14 = float(3.0 * lambda_star_p14 - 0.25 * lambda_star_p14 ** 2)
slater_x_p14 = 3.0
```
