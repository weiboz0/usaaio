### Worked solution

**Part A — matrix computation.**  For the ordered inputs $(-1,0,2)$, direct substitution gives
\[
K=\begin{pmatrix}4&1&1\\1&1&1\\1&1&25\end{pmatrix}.
\]
The code computes each entry from the polynomial formula rather than inserting this matrix as a literal.

**Part B — universal feature-map proof consuming Part A.**  Define
\[
\phi(t)=(1,\sqrt2\,t,t^2).
\]
Then $\phi(x)^T\phi(z)=1+2xz+x^2z^2=(1+xz)^2$.  On the Part A inputs, the rows of `Phi_p13` are these feature vectors, so `Phi_p13 @ Phi_p13.T` reproduces the already-computed `K_p13`; `feature_gap_p13` measures that dependency.  Universally, $k(x,z)=\phi(x)^T\phi(z)=\phi(z)^T\phi(x)=k(z,x)$.  For every $n\ge1$, every inputs $t_1,\ldots,t_n$, and every $a\in\mathbb R^n$,
\[
a^TKa=\left\|\sum_i a_i\phi(t_i)\right\|_2^2\ge0.
\]
Thus symmetry and universal finite-Gram PSD—not the sampled eigenvalues—prove validity.

**Part C — finite refutation consuming Parts A and B.**  The perturbed matrix on the same ordered inputs is
\[
K_{\rm bad}=K_{\rm p13}-2\mathbf1\mathbf1^T.
\]
For $a=(0,1,0)$, the quadratic form selects the center diagonal entry: $a^TK_{\rm bad}a=1-2=-1<0$.  Part B proves the original $k$ is valid; subtracting the positive constant kernel is not a validity-preserving closure operation, and this single negative witness decisively refutes the perturbed candidate.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

x_p13 = np.array([-1.0, 0.0, 2.0])
K_p13 = (1.0 + x_p13[:, None] * x_p13[None, :]) ** 2
Phi_p13 = np.column_stack((np.ones_like(x_p13), np.sqrt(2.0) * x_p13, x_p13 ** 2))
feature_gap_p13 = float(np.max(np.abs(K_p13 - Phi_p13 @ Phi_p13.T)))
K_bad_p13 = K_p13 - 2.0 * np.ones_like(K_p13)
a_p13 = np.array([0.0, 1.0, 0.0])
bad_energy_p13 = float(a_p13.T @ K_bad_p13 @ a_p13)
```
