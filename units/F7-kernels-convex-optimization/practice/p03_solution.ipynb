Consider $C=\{(u,v)\in\mathbb R^2: uv\ge1\}$. Which option gives a valid finite witness that $C$ is not convex?

A. $(1,1)$ and $(2,2)$ are in $C$, and their midpoint is also in $C$.

B. $(1,1)$ and $(-1,-1)$ are in $C$, but their midpoint $(0,0)$ is not in $C$.

C. $(1,1)$ is in $C$ while $(1,0)$ is not in $C$.

D. $(2,1)$ and $(1,2)$ are in $C$, and their midpoint has product greater than one.

E. The boundary curve $uv=1$ is curved, so no pair of points or segment calculation is needed.

Reasoning is not required. Record your choice below.

```python
ATOL = 1e-10
RTOL = 0.0

answer_p03 = "B"
```

## Answer-form check — do not edit

### Solution reasoning

Both \((1,1)\) and \((-1,-1)\) belong to \(C\) because each coordinate product equals \(1\). Their midpoint is \((0,0)\), whose product is \(0<1\), so the segment between two points of \(C\) leaves \(C\). This finite witness proves nonconvexity, making **B** correct.
