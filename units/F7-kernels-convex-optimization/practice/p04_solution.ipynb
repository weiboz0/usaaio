# F7-kernels-convex-optimization — Practice p04

**Type:** multiple choice (numeric, normal form) · **Difficulty:** core · **Concepts:** convex-functions, first-order-optimality

Let

$$f(x)=\frac32x^2-5x+7.$$

The positive quadratic coefficient makes $f$ convex. Its stationary point is therefore its global minimizer. Write the minimizer in reduced-fraction normal form $p/q$, where $p,q$ are integers, $q>0$, and $\gcd(|p|,q)=1$. Which option is $p/q$?

A. $5/3$

B. $3/5$

C. $-5/3$

D. $5/6$

E. $8/3$

Reasoning is not required. Record the option and the same fraction below.

```python
answer_p04 = "A"
p_p04: int = 5
q_p04: int = 3
```

## Normal-form check — do not edit

This public check validates the option form, integer types, positive denominator, reduced fraction, and agreement between your chosen option and submitted fraction. It does not reveal or assert the correct option.

```python
from math import gcd

ATOL = 1e-10
RTOL = 0.0
_OPTION_VALUES_P04 = {
    "A": (5, 3),
    "B": (3, 5),
    "C": (-5, 3),
    "D": (5, 6),
    "E": (8, 3),
}
assert type(answer_p04) is str and answer_p04 in _OPTION_VALUES_P04
assert type(p_p04) is int and type(q_p04) is int
assert q_p04 > 0
assert gcd(abs(p_p04), q_p04) == 1
_option_num_p04, _option_den_p04 = _OPTION_VALUES_P04[answer_p04]
assert p_p04 * _option_den_p04 == q_p04 * _option_num_p04
```

### Solution reasoning

Differentiate:
\[
f'(x)=3x-5.
\]
The stationary point satisfies \(3x-5=0\), hence \(x=5/3\). The problem states that the positive quadratic coefficient makes \(f\) convex, so this stationary point is its global minimizer. The fraction is already reduced with positive denominator, so the matching option is **A**.
