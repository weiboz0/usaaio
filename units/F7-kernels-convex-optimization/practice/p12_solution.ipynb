### Worked solution

**Claim 1 — line restriction/Jensen.**  Let $d=y-x$, $r=Bx-c$, and $v=Bd$.  Along the arbitrary segment,
\[
h(t)=f(x+td)=\|r+tv\|_2^2
=\|r\|_2^2+2t\,r^Tv+t^2\|v\|_2^2.
\]
The coefficient of $t^2$ is $\|B(y-x)\|_2^2=d^TB^TBd\ge0$, a PSD Gram form.  More explicitly, for $0\le\theta\le1$,
\[
(1-\theta)h(0)+\theta h(1)-h(\theta)
=\theta(1-\theta)\|v\|_2^2\ge0.
\]
Because $x,y,$ and $\theta$ were arbitrary, this is the convexity inequality for $f$.

**Claim 2 — gradient.**  For an arbitrary increment $u$,
\[
f(x+u)-f(x)=\|Bx-c+Bu\|^2-\|Bx-c\|^2
=2(Bx-c)^TBu+\|Bu\|^2.
\]
The term linear in $u$ is $[2B^T(Bx-c)]^Tu$, while the last term is quadratic in $u$.  Hence $\nabla f(x)=2B^T(Bx-c)$.

**Claim 3 — supporting-gradient identity.**  Put $u=y-x$ in the same expansion.  Subtracting the linear term gives
\[
f(y)-f(x)-\nabla f(x)^T(y-x)=\|B(y-x)\|_2^2\ge0.
\]
Thus $f(y)\ge f(x)+\nabla f(x)^T(y-x)$ for every $x,y$.

**Claim 4 — global optimality and possible nonuniqueness.**  If $\nabla f(x_*)=0$, Claim 3 gives $f(y)\ge f(x_*)$ for every $y$, so $x_*$ is global.  Convexity was established before making this inference.  If $B$ has a nonzero null vector $w$, then $f(x_*+w)=\|Bx_*-c+Bw\|^2=f(x_*)$.  Therefore a stationary minimizer need not be unique.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

B_p12 = np.array([[1.0, -2.0], [0.5, 1.0], [2.0, 0.0]])
c_p12 = np.array([1.0, -1.0, 0.5])
x_p12 = np.array([-1.0, 2.0])
y_p12 = np.array([3.0, -0.5])
theta_p12 = 0.3

def f_p12(z):
    return float(np.linalg.norm(B_p12 @ z - c_p12) ** 2)

gradient_p12 = 2.0 * B_p12.T @ (B_p12 @ x_p12 - c_p12)
_mix_p12 = (1.0 - theta_p12) * x_p12 + theta_p12 * y_p12
jensen_gap_p12 = float((1.0 - theta_p12) * f_p12(x_p12) + theta_p12 * f_p12(y_p12) - f_p12(_mix_p12))
support_gap_p12 = float(f_p12(y_p12) - f_p12(x_p12) - gradient_p12.T @ (y_p12 - x_p12))
```
