### Worked solution

**Claim 1 — symmetry.**  For every $x,z\in\mathcal X$,
\[
k(x,z)=\phi(x)^T\phi(z)=\phi(z)^T\phi(x)=k(z,x),
\]
because a real scalar equals its transpose.  This establishes symmetry independently of any quadratic-form calculation.

**Claim 2 — arbitrary finite-Gram quadratic form.**  Fix an arbitrary $n\ge 1$, arbitrary $x_1,\ldots,x_n$, and arbitrary $a\in\mathbb R^n$.  Then
\[
\begin{aligned}
a^TKa
&=\sum_{i=1}^n\sum_{j=1}^n a_i a_j K_{ij}
 =\sum_{i,j}a_i a_j\phi(x_i)^T\phi(x_j)\\
&=\left(\sum_i a_i\phi(x_i)\right)^T
  \left(\sum_j a_j\phi(x_j)\right)
 =\left\|\sum_i a_i\phi(x_i)\right\|_2^2\ge 0.
\end{aligned}
\]
The dimensions are consistent: each $\phi(x_i)$ lies in $\mathbb R^d$, so the sum is a vector in $\mathbb R^d$.

**Claim 3 — universal conclusion.**  Claim 1 gives symmetry for every pair.  Claim 2 gives a positive-semidefinite Gram matrix for every finite list (of every length $n\ge1$) and every real coefficient vector.  Together these are the two required duties, so $k$ is a valid kernel.  The calculation below checks just one list and is evidence, not a replacement for these universal quantifiers.

**Claim 4 — repeated inputs and semidefiniteness.**  If $x_r=x_s$ for $r\ne s$, rows (and columns) $r$ and $s$ of $K$ coincide.  Thus $K(e_r-e_s)=0$, so $K$ can be singular.  Kernel validity requires $a^TKa\ge0$, not $>0$ for every nonzero $a$, so this does not invalidate the kernel.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

Phi_p11 = np.array([[1.0, -1.0], [2.0, 0.5], [-1.0, 3.0]])
a_p11 = np.array([2.0, -1.0, 0.5])

K_p11 = Phi_p11 @ Phi_p11.T
energy_p11 = float(a_p11.T @ K_p11 @ a_p11)
```
