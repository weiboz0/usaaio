### Worked solution

**Convex report.**  For arbitrary $u,v\in\mathbb R$ and $0\le\theta\le1$,
\[
(1-\theta)g(u)+\theta g(v)-g((1-\theta)u+\theta v)
=\theta(1-\theta)(u-v)^2\ge0,
\]
so $g(t)=(t-2)^2$ is convex without using a Hessian.  Also $g'(t)=2(t-2)$, hence $g'(2)=0$.  Direct expansion gives the global supporting identity
\[
g(s)-g(t)-g'(t)(s-t)=(s-t)^2\ge0
\]
for every $s,t$.  At $t=2$, this says $g(s)\ge g(2)=0$ for every real $s$, so the global conclusion is justified.

**Nonconvex report.**  Since $f'(t)=4t^3-2t$, $f'(0)=0$.  But
\[
f(1/2)=\frac1{16}-\frac14=-\frac3{16}<0=f(0).
\]
Thus the concrete nearby point $1/2$ already refutes the claim that $0$ is a minimum, global or local.

**Local versus global conclusion.**  Stationarity is only a necessary condition in many smooth unconstrained problems; it is not sufficient by itself.  Likewise, failing to find descent among finitely many local probes would not quantify over the whole domain and could not establish a global result.  The first report succeeds because the supporting inequality is proved for every point after convexity is established.  The second report is missing that convexity premise (and in fact $f$ is nonconvex), so its stationary-point inference is invalid.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

def g_p16(t):
    return float((t - 2.0) ** 2)

def f_p16(t):
    return float(t ** 4 - t ** 2)

t_star_p16 = 2.0
s_stationary_p16 = 0.0
s_nearby_p16 = 0.5
probe_p16 = -1.0

grad_g_p16 = float(2.0 * (t_star_p16 - 2.0))
grad_f_p16 = float(4.0 * s_stationary_p16 ** 3 - 2.0 * s_stationary_p16)
nearby_drop_p16 = float(f_p16(s_nearby_p16) - f_p16(s_stationary_p16))
support_gap_p16 = float(g_p16(probe_p16) - g_p16(t_star_p16) - grad_g_p16 * (probe_p16 - t_star_p16))
```
