Let $k$ be a real-valued function on an input domain. Which statement is exactly the validity requirement for a real PSD kernel?

A. It is enough that $k(x,x)\ge0$ for every single input $x$.

B. It is enough that one large randomly sampled Gram matrix is symmetric and has no negative computed eigenvalue.

C. Both conditions are required separately: $k(x,z)=k(z,x)$ for every input pair, and for every $n\ge1$, every finite list $x_1,\ldots,x_n$, and every $a\in\mathbb R^n$, the matrix $K_{ij}=k(x_i,x_j)$ satisfies $a^TKa\ge0$.

D. For every finite list, $a^TKa\ge0$ is sufficient even when $k$ is not symmetric.

E. Every finite Gram matrix must be strictly positive definite, including lists with repeated inputs.

Reasoning is not required. Record your choice below.

```python
ATOL = 1e-10
RTOL = 0.0

answer_p02 = "C"
```

## Answer-form check — do not edit

This public check validates only the required response form. It does not reveal or assert the correct option.

```python
assert type(answer_p02) is str
assert answer_p02 in ("A", "B", "C", "D", "E")
```

### Solution reasoning

A real PSD kernel must be symmetric on every pair and must produce a positive-semidefinite Gram matrix for every finite input list. Equivalently, every coefficient vector \(a\) must satisfy \(a^TKa\ge 0\). Diagonal tests or random samples are only necessary spot checks, and repeated inputs prevent requiring strict positive definiteness. Therefore **C** is correct.
