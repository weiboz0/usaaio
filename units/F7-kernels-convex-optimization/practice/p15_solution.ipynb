### Worked solution

**Candidate A — sampled evidence versus universal proof.**  The 1,000 tests establish only that the particular finite input lists and numerical computations tested produced symmetric PSD matrices (up to the numerical test's tolerances).  They do not quantify over every list size and every real list, so they cannot prove kernel validity.  A universal proof uses
\[
\phi_A(t)=(1,\sqrt2\,t,t^2),\qquad
\phi_A(x)^T\phi_A(z)=1+2xz+x^2z^2=(1+xz)^2.
\]
This feature representation proves symmetry, and for arbitrary $n$, arbitrary $x_1,\ldots,x_n$, and arbitrary $a\in\mathbb R^n$ it gives
\[
a^TKa=\left\|\sum_i a_i\phi_A(x_i)\right\|^2\ge0.
\]
Thus $k_A$ is universally valid.

**Candidate B — symmetry boundary.**  Here $k_B(z,x)=z-x=-(x-z)$, so $k_B$ is not symmetric except on the diagonal.  On the supplied points,
\[
K_B=\begin{pmatrix}0&-2\\2&0\end{pmatrix},
\]
whose symmetry gap is $4$.  Every real skew-symmetric matrix has $a^TKa=0$ because this scalar equals its transpose $a^TK^Ta=-a^TKa$.  Therefore the observed zero quadratic forms say nothing that repairs the separate symmetry failure; $k_B$ is not a valid kernel.

**Candidate C — finite refutation.**  On $x=(-1,1)$,
\[
K_C=\begin{pmatrix}0&2\\2&0\end{pmatrix}.
\]
For $a=(1,-1)$, $a^TK_Ca=-4<0$.  Kernel validity universally requires every such finite energy to be nonnegative, so this one exact witness decisively refutes $k_C$; more passing samples cannot undo it.

**Repeated inputs.**  If two inputs coincide, the corresponding Gram rows coincide and their difference supplies a nonzero null vector.  Valid kernels therefore require PSD, not strict PD, on arbitrary finite lists.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

x_p15 = np.array([-1.0, 1.0])
a_skew_p15 = np.array([2.0, -3.0])
a_bad_p15 = np.array([1.0, -1.0])

K_B_p15 = x_p15[:, None] - x_p15[None, :]
symmetry_gap_p15 = float(np.max(np.abs(K_B_p15 - K_B_p15.T)))
skew_energy_p15 = float(a_skew_p15.T @ K_B_p15 @ a_skew_p15)
K_C_p15 = 1.0 - x_p15[:, None] * x_p15[None, :]
bad_energy_p15 = float(a_bad_p15.T @ K_C_p15 @ a_bad_p15)
```

## Scenario-anchor check — do not edit

This checks the two finite audits. The universal proof/evidence distinctions are reviewed against the memo rubric.

```python
_K_B_ref_p15 = x_p15[:,None] - x_p15[None,:]
_K_C_ref_p15 = 1.0 - x_p15[:,None] * x_p15[None,:]
assert isinstance(K_B_p15, np.ndarray) and K_B_p15.shape == (2,2)
assert isinstance(K_C_p15, np.ndarray) and K_C_p15.shape == (2,2)
assert np.allclose(K_B_p15, _K_B_ref_p15, atol=ATOL, rtol=RTOL)
assert np.allclose(K_C_p15, _K_C_ref_p15, atol=ATOL, rtol=RTOL)
assert type(symmetry_gap_p15) is float and type(skew_energy_p15) is float and type(bad_energy_p15) is float
assert np.isclose(symmetry_gap_p15, np.max(np.abs(K_B_p15-K_B_p15.T)), atol=ATOL, rtol=RTOL)
assert symmetry_gap_p15 > ATOL
assert np.isclose(skew_energy_p15, float(a_skew_p15@K_B_p15@a_skew_p15), atol=ATOL, rtol=RTOL)
assert np.isclose(skew_energy_p15, 0.0, atol=ATOL, rtol=RTOL)
assert np.isclose(bad_energy_p15, float(a_bad_p15@K_C_p15@a_bad_p15), atol=ATOL, rtol=RTOL)
assert bad_energy_p15 < -ATOL
```
