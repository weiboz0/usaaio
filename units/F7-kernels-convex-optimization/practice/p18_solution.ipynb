### Worked solution

**Dual derivation and domain.**  With $g(x,y)=2-x-y\le0$ and $\lambda\ge0$,
\[
L=x^2+y^2+\lambda(2-x-y)
=\left(x-\frac\lambda2\right)^2+\left(y-\frac\lambda2\right)^2
+2\lambda-\frac{\lambda^2}{2}.
\]
Therefore the infimum over unrestricted $x,y$ is attained at $x=y=\lambda/2$, and
\[
q(\lambda)=2\lambda-\frac{\lambda^2}{2}.
\]
This formula is finite for every real $\lambda$, while the dual problem's feasible domain under the stated inequality convention is $\lambda\ge0$.

**Primal and dual optima.**  For a primal-feasible point, $x+y\ge2$.  Since
\[
x^2+y^2=\frac{(x+y)^2+(x-y)^2}{2}\ge\frac{(x+y)^2}{2}\ge2,
\]
the primal optimum is $(x^*,y^*)=(1,1)$ with $p^*=2$.  On $\lambda\ge0$,
\[
q(\lambda)=2-\frac12(\lambda-2)^2,
\]
so $\lambda^*=2$ and $d^*=q(2)=2$.  Weak duality has the direction $d^*\le p^*$: for any dual-feasible $\lambda$ and primal-feasible $(x,y)$,
\[
q(\lambda)\le L(x,y,\lambda)=f(x,y)+\lambda g(x,y)\le f(x,y).
\]

**KKT audit.**  At $(1,1)$ and $\lambda=2$, primal feasibility gives $g=0$, dual feasibility gives $\lambda\ge0$, stationarity gives
\[
(2x-\lambda,2y-\lambda)=(0,0),
\]
and complementary slackness gives $\lambda g=0$.  Thus all four recorded residuals vanish.

**Weak/strong duality assumptions.**  The objective $x^2+y^2$ is convex, the constraint is affine, $(2,2)$ is a strict Slater point because $2-2-2=-2<0$, and both finite optima above are attained.  These facts justify strong duality and hence the observed zero gap.  The KKT equalities are a certificate consistent with that theorem; they are not, by themselves and without its assumptions, a proof that every problem has zero gap.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

x_star_p18 = np.array([1.0, 1.0])
lambda_star_p18 = 2.0
primal_value_p18 = float(x_star_p18 @ x_star_p18)
dual_value_p18 = float(2.0 * lambda_star_p18 - 0.5 * lambda_star_p18 ** 2)
_g_p18 = 2.0 - float(np.sum(x_star_p18))
kkt_residuals_p18 = np.array([
    max(_g_p18, 0.0),
    max(-lambda_star_p18, 0.0),
    np.max(np.abs(2.0 * x_star_p18 - lambda_star_p18)),
    abs(lambda_star_p18 * _g_p18),
], dtype=float)
slater_point_p18 = np.array([2.0, 2.0])
```
