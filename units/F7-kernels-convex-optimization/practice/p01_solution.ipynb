**Type:** multiple choice · **Difficulty:** intro · **Concepts:** positive-semidefinite-matrices

Consider the symmetric matrix

$$A=\begin{bmatrix}2&-2\\-2&2\end{bmatrix}.$$

Which classification and certificate are correct?

A. $A$ is positive definite because both diagonal entries are positive.

B. $A$ is positive semidefinite but not positive definite: $v^TAv=2(v_1-v_2)^2\ge0$, while the nonzero vector $(1,1)$ has zero energy.

C. $A$ is not positive semidefinite because it has negative off-diagonal entries.

D. $A$ is both positive semidefinite and positive definite because all of its eigenvalues are nonnegative.

E. $A$ is not eligible for PSD classification because it is not symmetric.

Reasoning is not required. Record your choice below.

```python
ATOL = 1e-10
RTOL = 0.0

answer_p01 = "B"
```

### Solution reasoning

For any vector \(v=(v_1,v_2)\),
\[
v^TAv=2v_1^2-4v_1v_2+2v_2^2=2(v_1-v_2)^2\ge 0.
\]
Thus \(A\) is PSD. It is not positive definite because the nonzero vector \((1,1)\) gives zero quadratic-form value, so **B** is the unique correct option.
