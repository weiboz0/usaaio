# C8-embeddings — Practice p12 — Solution

### (a) Upper bound

Expanding with bilinearity and symmetry gives
$$\lVert u-v\rVert^2=(u-v)\cdot(u-v)=u\cdot u-2u\cdot v+v\cdot v=2-2u\cdot v.$$
Because a squared norm is nonnegative, $2-2u\cdot v\ge0$, hence $u\cdot v\le1$.

### (b) Lower bound

Similarly,
$$\lVert u+v\rVert^2=u\cdot u+2u\cdot v+v\cdot v=2+2u\cdot v\ge0,$$
so $u\cdot v\ge-1$. Together these are the unit-vector form of Cauchy–Schwarz.

### (c) Equality cases

If $u\cdot v=1$, then $\lVert u-v\rVert^2=0$, so $u=v$. If $u\cdot v=-1$, then $\lVert u+v\rVert^2=0$, so $u=-v$.

### (d) Arbitrary nonzero vectors

Let $u=x/\lVert x\rVert$ and $v=y/\lVert y\rVert$. Applying the unit-vector bounds yields
$$-1\le u\cdot v=\frac{x\cdot y}{\lVert x\rVert\lVert y\rVert}=\cos(x,y)\le1.$$
Thus every cosine lies in the fixed interval $[-1,1]$, which licenses the same endpoints on every cosine heatmap.

```python
import os, pathlib
_root = next(p for p in [pathlib.Path.cwd(), *pathlib.Path.cwd().parents]
             if (p / "pyproject.toml").exists())
os.environ["GENSIM_DATA_DIR"] = str(_root / "reference" / "cache" / "gensim")

import numpy as np
import gensim.downloader
```

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

WORDS = ["glacier", "iceberg", "desert", "piano"]
SEED = 20260804

V = np.asarray(kv[WORDS], dtype=np.float64)
W = V / np.sqrt((V * V).sum(axis=1, keepdims=True))
S = W @ W.T
diagonal = np.eye(len(WORDS), dtype=bool)
off_min = float(np.min(np.where(diagonal, np.inf, S)))
off_max = float(np.max(np.where(diagonal, -np.inf, S)))
lhs = float(((W[0] - W[1]) ** 2).sum())
rhs = float(2 - 2 * (W[0] * W[1]).sum())

rng = np.random.default_rng(SEED)
u = rng.normal(0, 1, 100)
v = rng.normal(0, 1, 100)
cos_rand = float((u * v).sum() /
                 (np.sqrt((u * u).sum()) * np.sqrt((v * v).sum())))
```
