```python
import os, pathlib
_root = next(p for p in [pathlib.Path.cwd(), *pathlib.Path.cwd().parents]
             if (p / "pyproject.toml").exists())
os.environ["GENSIM_DATA_DIR"] = str(_root / "reference" / "cache" / "gensim")

import numpy as np
import gensim.downloader
```

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

WORDS = ["falcon", "heron", "otter", "badger", "salmon"]
V = np.asarray(kv[WORDS], dtype=np.float64)
norms = np.sqrt((V * V).sum(axis=1, keepdims=True))
W = V / norms
unit_gap = float(np.max(np.abs((W * W).sum(axis=1) - 1.0)))
norm_falcon = float(norms[0, 0])
```

The rows of `V` follow token order. Keeping the norm axis produces shape `(5, 1)`, so broadcasting divides every coordinate in a row by that row's own norm.

### Answer check

```python
assert V.shape == (5, 100) and V.dtype == np.float64
assert norms.shape == (5, 1)
assert W.shape == (5, 100)
assert unit_gap <= 1e-12
assert np.isclose(norm_falcon, 4.764169599066711, atol=1e-12)
```
