```python
import os, pathlib
_root = next(p for p in [pathlib.Path.cwd(), *pathlib.Path.cwd().parents]
             if (p / "pyproject.toml").exists())
os.environ["GENSIM_DATA_DIR"] = str(_root / "reference" / "cache" / "gensim")

import numpy as np
import gensim.downloader
```

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

dim = int(kv.vector_size)
has_trumpet = bool("trumpet" in kv.key_to_index)
has_trumpetish = bool("trumpetish" in kv.key_to_index)
sim_cello = float(kv.similarity("violin", "cello"))
sim_garlic = float(kv.similarity("violin", "garlic"))
closer_word = "cello" if sim_cello > sim_garlic else "garlic"
top1 = str(kv.most_similar("trumpet", topn=1)[0][0])
```

The model has 100-dimensional vectors, contains `trumpet` but not the invented form `trumpetish`, and places `violin` far closer to `cello` than to `garlic`. Its nearest vocabulary neighbor to `trumpet` is `saxophone`.

### Answer check

```python
assert dim == 100
assert has_trumpet is True and has_trumpetish is False
assert np.isclose(sim_cello, 0.9249897599220276, atol=1e-5, rtol=0)
assert np.isclose(sim_garlic, 0.15266083180904388, atol=1e-5, rtol=0)
assert closer_word == "cello"
assert top1 == "saxophone"
```
