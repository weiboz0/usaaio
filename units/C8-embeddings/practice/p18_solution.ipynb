```python
import os, pathlib
_root = next(p for p in [pathlib.Path.cwd(), *pathlib.Path.cwd().parents]
             if (p / "pyproject.toml").exists())
os.environ["GENSIM_DATA_DIR"] = str(_root / "reference" / "cache" / "gensim")

import numpy as np
import gensim.downloader
```

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

sim_gale = float(kv.similarity("breeze", "gale"))
sim_hammer = float(kv.similarity("breeze", "hammer"))
more_related = "gale" if sim_gale > sim_hammer else "hammer"

narrow_hits = kv.most_similar("lantern", topn=5, restrict_vocab=10000)
wide_hits = kv.most_similar("lantern", topn=5, restrict_vocab=100000)
top5_narrow = [word for word, _ in narrow_hits]
top5_wide = [word for word, _ in wide_hits]
# Only shared membership is being counted, so discarding order is appropriate here.
n_common = len(set(top5_narrow) & set(top5_wide))

lantern_rank = int(kv.key_to_index["lantern"])

pair_hits = kv.most_similar(positive=["rain", "wind"], topn=3,
                            restrict_vocab=20000)
top3_pair = [word for word, _ in pair_hits]
sim_first = float(pair_hits[0][1])
```

Widening the frequency-ordered universe admits 90,000 additional, rarer candidates, and higher-scoring newcomers can push most narrow top-five hits out of the displayed list. It never removes any of the original 10,000 candidates from the search universe—the earlier words remain eligible even when their displayed ranks fall.

Here `lantern` has zero-based vocabulary index 18,755, so it lies outside the narrow universe. The two top-five lists share only `dragon`.

### Answer check

```python
assert np.isclose(sim_gale, 0.42755770683288574, atol=1e-5, rtol=0)
assert np.isclose(sim_hammer, 0.1442744880914688, atol=1e-5, rtol=0)
assert more_related == "gale"
assert top5_narrow == ["dragon", "flame", "magic", "shadow", "lights"]
assert top5_wide == ["lanterns", "dragon", "lamp", "sinestro", "candle"]
assert n_common == 1
assert lantern_rank == 18755
assert top3_pair == ["winds", "rains", "snow"]
assert np.isclose(sim_first, 0.823852002620697, atol=1e-5, rtol=0)
```
