# C8-embeddings — Practice p06 — Solution

```python
from gensim.utils import simple_preprocess

TEXT = ("At 7 a.m. the market opened: crates of bread, wheels of cheese, and jars of "
        "honey lined the stalls. The baker praised the bread; the baker's rival praised "
        "the cheese. By 9 a.m. the honey was gone.")

toks = simple_preprocess(TEXT)
n_tokens = len(toks)
n_types = len(dict.fromkeys(toks))
uniq_ordered = list(dict.fromkeys(toks))
n_bread = toks.count("bread")
first_cheese = toks.index("cheese")
```

Tokenization removes the numerals and the one-letter fragments of the time abbreviations. `dict.fromkeys` deduplicates while retaining the position of each word's first occurrence.

### Answer check

```python
assert n_tokens == 33
assert n_types == 20
assert uniq_ordered == ["at", "the", "market", "opened", "crates", "of", "bread",
                        "wheels", "cheese", "and", "jars", "honey", "lined", "stalls",
                        "baker", "praised", "rival", "by", "was", "gone"]
assert n_bread == 2
assert first_cheese == 9
```
