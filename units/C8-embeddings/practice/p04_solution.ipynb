# C8-embeddings — Practice p04 — Solution

```python
from math import gcd

dot = 3 * 4 + 4 * 3
norm_product = 5 * 5
common = gcd(dot, norm_product)
p = dot // common
q = norm_product // common
p_plus_q = p + q
answer_p04 = "D"
```

Each vector has norm 5 and their dot product is 24, so the cosine is the already-reduced fraction $24/25$. Thus $p+q=49$ and the answer is **D**.

- A ($2$) is the "same numbers, so parallel" trap: assuming $\cos = 1 = 1/1$
  gives $p+q = 2$ — but reordered entries are NOT parallel vectors.
- B ($25$) reports the raw dot product $24$ in normal form ($24/1 \to 25$),
  forgetting to divide by the norms.
- C ($37$) drops one cross term: $3\cdot4 = 12$ alone gives $12/25 \to 37$.
- D is $24+25$ — correct.
- E ($51$) is $26/25$, the off-by-two arithmetic slip in the cross terms;
  no correct partial computation lands there.

### Answer check

```python
assert (p, q) == (24, 25)
assert p_plus_q == 49
assert answer_p04 == "D"
```
