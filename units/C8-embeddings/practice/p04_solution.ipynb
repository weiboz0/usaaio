# C8-embeddings — Practice p04 — Solution

```python
from math import gcd

dot = 3 * 4 + 4 * 3
norm_product = 5 * 5
common = gcd(dot, norm_product)
p = dot // common
q = norm_product // common
p_plus_q = p + q
answer_p04 = "D"
```

Each vector has norm 5 and their dot product is 24, so the cosine is the already-reduced fraction $24/25$. Thus $p+q=49$ and the answer is **D**.

- A would correspond to an unreached endpoint-style shortcut, not the computed fraction.
- B is only the denominator.
- C does not equal the reduced numerator-plus-denominator.
- D is $24+25$.
- E is two too large and has no basis in the dot or norm calculation.

### Answer check

```python
assert (p, q) == (24, 25)
assert p_plus_q == 49
assert answer_p04 == "D"
```
