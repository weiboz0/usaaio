```python
import os, pathlib
_root = next(p for p in [pathlib.Path.cwd(), *pathlib.Path.cwd().parents]
             if (p / "pyproject.toml").exists())
os.environ["GENSIM_DATA_DIR"] = str(_root / "reference" / "cache" / "gensim")

import numpy as np
import gensim.downloader
```

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

WORDS = ["dawn", "dusk", "noon", "midnight"]
V = np.asarray(kv[WORDS], dtype=np.float64)
shape_ok = bool(V.shape == (4, 100))
dtype_ok = bool(V.dtype == np.float64)
row0_is_dawn = bool(np.allclose(
    V[0], np.asarray(kv["dawn"], dtype=np.float64), atol=1e-9, rtol=0
))
norms = np.sqrt((V * V).sum(axis=1))
norm_dawn = float(norms[0])
```

The lookup stacks one token per row in the given order. The resulting matrix is float64 with shape `(4, 100)`, its first row reproduces the `dawn` lookup, and the raw `dawn` norm is about 5.1855.

### Answer check

```python
assert shape_ok is True and dtype_ok is True and row0_is_dawn is True
assert V.shape == (4, 100) and V.dtype == np.float64
assert norms.shape == (4,)
assert np.allclose(norms,
                   np.array([5.185521549747855, 5.179510764523977,
                             5.349181216471941, 5.158055015834607]),
                   atol=1e-12, rtol=0)
assert np.isclose(norm_dawn, 5.185521549747855, atol=1e-12, rtol=0)
```
