## Unit goals

By the end of this unit you can:

- **tokenize** raw text with `gensim.utils.simple_preprocess` and
  predict its output by rule (lowercase; punctuation and digits
  stripped; only 2–15-letter pieces survive); count **tokens** versus
  **types**, and state precisely what Python's `set` destroys
  (first-occurrence order and counts) and what to use when that matters
  (`list(dict.fromkeys(...))`);
- state the three governing facts about **word embeddings** — similar
  use → nearby vectors; individual coordinates mean nothing; the GloVe
  artifact is fixed — and use them to reason about what vector geometry
  can and cannot tell you;
- drive **gensim KeyedVectors**: load `glove-wiki-gigaword-100` through
  the repo's shared cache, test membership (`w in kv.key_to_index`),
  look up vectors, and apply this course's **float64 boundary cast**
  (`np.asarray(..., dtype=np.float64)`, same cell as the lookup) every
  time vectors leave gensim;
- build the **embedding matrix** in the pinned convention — rows are
  tokens, shape $(N, 100)$ — and **row-normalize it in the exam's
  broadcasting register**: `np.sqrt((V * V).sum(axis=1, keepdims=True))`,
  no `np.linalg`, no loops, producing the unit-row stack `W`;
- form the **similarity matrix** $S = W W^{\mathsf T}$, derive
  entrywise why $S_{ij}$ is the cosine of tokens $i$ and $j$, verify
  its symmetry and unit diagonal in code, and prove its range
  ($[-1, 1]$, via Cauchy–Schwarz on unit vectors);
- rank **nearest neighbors** from a similarity row — `np.argsort`
  descending, self-exclusion by mask, top-$k$ — and cross-check the
  manual ranking against `kv.most_similar` at the right tolerance;
- assemble the five-stage **retrieval mini-pipeline** (tokenize →
  dedup → filter → embed → rank) on fresh text, and read its output
  with calibrated skepticism (frequency effects, hubness).

**What this unit deliberately does not do:** train anything.
The vectors are a fixed published artifact; no gradient ever flows.
And the geometry stops at the Gram matrix —
factoring `W` (SVD, low-rank structure) is
`C9-dimensionality-reduction`'s whole subject, which consumes this
unit's `W` unchanged.

## Prerequisites

**F2-vectors** — dot products, norms, unit vectors, cosine similarity:
this unit computes little else.
**F3-matrices** — matrices as stacked rows and the entrywise matrix
product, which is exactly how $S = W W^{\mathsf T}$ is derived and
read.
**F1-scientific-python** — arrays, broadcasting, axis sums, seeded
randomness, matplotlib.
No calculus is used in this unit; the proofs are algebraic.
If any of that feels shaky, redo those units' practice first.
