### Claim-by-claim proof

**(a) Independence of the summands.** Each product $W_kx_k$ is a function of the pair $(W_k,x_k)$.  The pairs are disjoint sub-collections of the mutually independent collection $(W_1,\ldots,W_C,x_1,\ldots,x_C)$, so the group-form independence fact makes the products mutually independent.

**(b) One product.** By the product rule for independent variables, $E[W_kx_k]=E[W_k]E[x_k]=0$.  The zero-mean variance shortcut then gives

$$\operatorname{Var}(W_kx_k)=E[(W_kx_k)^2]=E[W_k^2x_k^2].$$

Because $W_k^2$ and $x_k^2$ are functions of the independent variables $W_k$ and $x_k$, they are independent; the product rule gives $E[W_k^2x_k^2]=E[W_k^2]E[x_k^2]$.  Applying the zero-mean variance shortcut separately to $W_k$ and $x_k$ yields $E[W_k^2]=\sigma^2$ and $E[x_k^2]=1$.  Therefore $\operatorname{Var}(W_kx_k)=\sigma^2$.

**(c) The sum and scale.** The products are mutually independent by (a), so variance-of-sums and (b) give

$$\operatorname{Var}(z)=\sum_{k=1}^C\operatorname{Var}(W_kx_k)=C\sigma^2.$$

Demanding unit variance gives $C\sigma^2=1$.  Since $\sigma>0$, the unique solution is $\sigma=1/\sqrt C$.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
n, C = 50_000, 100
W_mat = rng.normal(0, 0.1, (n, C))
X_mat = rng.normal(0, 1, (n, C))
z_samples = (W_mat * X_mat).sum(axis=1)
var_est = float(z_samples.var())
anchor_gap = float(abs(var_est - 1.0))
var_est, anchor_gap
```

```python
# Keep this concrete-anchor assertion from the statement.
assert anchor_gap < 0.05
```

### Answer check

```python
assert np.isclose(var_est, 0.9975040868065064, atol=1e-12)
assert np.isclose(anchor_gap, 0.0024959131934936263, atol=1e-12)
assert anchor_gap < 0.05
```
