# C5-neural-networks — Practice p16 — Solution

The variance formula is $\operatorname{Var}(z)=C\sigma^2\operatorname{Var}(x)$, so standard deviation is multiplied by $\sigma\sqrt C$.  At width $100$, $\sigma=0.5$ gives factor $5$ and predicted spreads $5,25,125,625$.  Width $4$ masked the bug because $0.5\sqrt4=1$; a scale copied to another width cannot preserve variance.  The fix is $\sigma=1/\sqrt{100}=0.1$.

```python
import numpy as np

SEED = 20260804
factor = float(0.5 * np.sqrt(100))
pred_stds = factor ** np.arange(1, 5)


def run_stack(sigma):
    rng = np.random.default_rng(SEED)
    h = rng.normal(0, 1, (200, 100))
    spreads = []
    for _ in range(4):
        W = rng.normal(0, sigma, (100, 100))
        h = (h[:, None, :] * W[None, :, :]).sum(axis=2)
        spreads.append(float(h.std()))
    return np.array(spreads)


sim_stds = run_stack(0.5)
sigma_fix = float(1 / np.sqrt(100))
fixed_final = float(run_stack(sigma_fix)[-1])
factor, pred_stds, sim_stds, sigma_fix, fixed_final
```

### Answer check

```python
assert np.isclose(factor, 5.0, atol=1e-12)
assert np.allclose(pred_stds, [5.0, 25.0, 125.0, 625.0], atol=1e-12)
assert np.allclose(sim_stds, [5.008000859148938, 24.99415654259038, 120.91735325264992, 595.5926924373416], atol=1e-10)
assert np.isclose(sigma_fix, 0.1, atol=1e-12)
assert np.isclose(fixed_final, 0.9529483078997465, atol=1e-12)
```
