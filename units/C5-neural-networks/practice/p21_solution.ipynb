# C5-neural-networks — Practice p21 — Solution

The segment direction is $B-A=(11,7)$, so the primitive perpendicular vector $(7,-11)$ is a normal because $11\cdot7+7\cdot(-11)=0$. The bias $9$ makes both boundary scores zero, and this sign makes $P$'s score $72>0$ while $N$'s score is $-69<0$.

```python
import numpy as np

A = np.array([5, 4])
B = np.array([16, 11])
P = np.array([20, 7])
N = np.array([3, 9])


def step_activation(z):
    return (z >= 0).astype(int)


w = np.array([7, -11], dtype=int)
b = 9
boundary_scores = (np.stack([A, B]) * w).sum(axis=1) + b
probe_scores = (np.stack([P, N]) * w).sum(axis=1) + b
probe_outputs = step_activation(probe_scores)
```

### Answer check

```python
assert w.dtype.kind in "iu"
assert tuple(w) == (7, -11)
assert b == 9 and isinstance(b, (int, np.integer))
assert tuple(B - A) == (11, 7)
assert int(((B - A) * w).sum()) == 0
assert tuple(boundary_scores) == (0, 0)
assert tuple(probe_scores) == (72, -69)
assert tuple(probe_outputs) == (1, 0)
```
