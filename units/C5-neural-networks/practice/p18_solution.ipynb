# C5-neural-networks — Practice p18 — Solution

A hidden unit at kink $a$ computes $\operatorname{relu}(x-a)$, and its output weight is the change in target slope at $a$.

For $f_1$, the slopes are $0,2,0$, so the changes at $0$ and $3$ are $+2,-2$.  Thus $f_1(x)=2\operatorname{relu}(x)-2\operatorname{relu}(x-3)$.

For $f_2$, the slopes are $0,1,-2,0$, so the changes at $-1,1,2$ are $+1,-3,+2$.  Thus $f_2(x)=\operatorname{relu}(x+1)-3\operatorname{relu}(x-1)+2\operatorname{relu}(x-2)$.  Both targets start at zero, so both output biases are zero.

```python
import numpy as np


def affine_layer(x, W, b):
    return (x[:, None, :] * W[None, :, :]).sum(axis=2) + b


def relu(z):
    return np.maximum(z, 0.0)


SEED = 20260804
grid = np.linspace(-4.0, 6.0, 501)
target1 = np.interp(grid, [0.0, 3.0], [0.0, 6.0])
target2 = np.interp(grid, [-1.0, 1.0, 2.0], [0.0, 2.0, 0.0])
W1_f1 = np.array([[1.0], [1.0]])
b1_f1 = np.array([0.0, -3.0])
w2_f1 = np.array([[2.0, -2.0]])
b2_f1 = np.array([0.0])
W1_f2 = np.array([[1.0], [1.0], [1.0]])
b1_f2 = np.array([1.0, -1.0, -2.0])
w2_f2 = np.array([[1.0, -3.0, 2.0]])
b2_f2 = np.array([0.0])


def mlp_relu(x, W1, b1, w2, b2):
    hidden = relu(affine_layer(x, W1, b1))
    return affine_layer(hidden, w2, b2)


pred1 = mlp_relu(grid[:, None], W1_f1, b1_f1, w2_f1, b2_f1)[:, 0]
pred2 = mlp_relu(grid[:, None], W1_f2, b1_f2, w2_f2, b2_f2)[:, 0]
gap1 = float(np.abs(pred1 - target1).max())
gap2 = float(np.abs(pred2 - target2).max())
gap1, gap2
```
