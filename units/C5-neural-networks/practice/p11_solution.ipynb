### Claim-by-claim proof

**(a) Two-layer collapse.** Substitute the first layer into the second:

$$z_i^{(2)}=\sum_j W^{(2)}_{ij}\left(\sum_k W^{(1)}_{jk}x_k+b^{(1)}_j\right)+b^{(2)}_i.$$

Distributing and swapping the finite sums gives

$$z_i^{(2)}=\sum_k\left(\sum_jW^{(2)}_{ij}W^{(1)}_{jk}\right)x_k+\left(\sum_jW^{(2)}_{ij}b^{(1)}_j+b^{(2)}_i\right).$$

Hence $W^{\mathrm{eff}}_{ik}=\sum_jW^{(2)}_{ij}W^{(1)}_{jk}$ and $b^{\mathrm{eff}}_i=\sum_jW^{(2)}_{ij}b^{(1)}_j+b^{(2)}_i$.

**(b) Any depth.** Induction hypothesis: the first $L$ activation-free affine layers equal one affine map $A_Lx+c_L$.  Composing one more affine layer gives $A_{L+1}(A_Lx+c_L)+c_{L+1}=(A_{L+1}A_L)x+(A_{L+1}c_L+c_{L+1})$, again affine.  The two-layer calculation is the base case, so every finite depth collapses.

After dropping the hidden step, a final step can only threshold one affine function, producing one half-plane.  A bounded quadrilateral is an intersection of several differently oriented half-planes and cannot be represented by a single half-plane.  With a step between layers, substitution produces $W^{(2)}\operatorname{step}(W^{(1)}x+b^{(1)})$; the nonlinear step cannot be distributed into coefficients of $x$, so the collapse stops there.

```python
import numpy as np


def affine_layer(x, W, b):
    return (x[:, None, :] * W[None, :, :]).sum(axis=2) + b


W1 = np.array([[2.0, -1.0], [0.0, 1.0]])
b1 = np.array([1.0, -1.0])
W2 = np.array([[1.0, 3.0]])
b2 = np.array([-2.0])
SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (50, 2))
W_eff_hand = np.array([[2.0, 2.0]])
b_eff_hand = np.array([-4.0])
stacked = affine_layer(affine_layer(X, W1, b1), W2, b2)
direct = affine_layer(X, W_eff_hand, b_eff_hand)
anchor_gap = float(np.abs(stacked - direct).max())
W_eff_hand, b_eff_hand, anchor_gap
```

```python
# Keep this concrete-anchor assertion from the statement.
assert anchor_gap < 1e-12
```

### Answer check

```python
assert np.allclose(W_eff_hand, [[2.0, 2.0]], atol=1e-12)
assert np.allclose(b_eff_hand, [-4.0], atol=1e-12)
assert np.isclose(anchor_gap, 1.7763568394002505e-15, rtol=1e-12, atol=1e-18)
assert anchor_gap < 1e-12
```
