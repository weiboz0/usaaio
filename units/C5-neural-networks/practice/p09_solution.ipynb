# C5-neural-networks — Practice p09 — Solution

The intercept form is $x_1/2+x_2/3=1$, or $3x_1+2x_2-6=0$.  Its value at $(4,4)$ is $14>0$, so the requested side uses $(w_1,w_2,b)=(3,2,-6)$.  These integers have gcd $1$ and $w_1>0$.

```python
import numpy as np


def affine_layer(x, W, b):
    return (x[:, None, :] * W[None, :, :]).sum(axis=2) + b


def step_activation(z):
    return (z >= 0).astype(float)


SEED = 20260804
test_points = np.array([[2.0, 0.0], [0.0, 3.0], [4.0, 4.0], [0.0, 0.0], [1.0, 1.0]])
rng = np.random.default_rng(SEED)
pts = rng.uniform(-5, 5, (2000, 2))
w_vec = np.array([3.0, 2.0])
b_val = -6.0
test_labels = step_activation(affine_layer(test_points, w_vec[None, :], np.array([b_val])))[:, 0]
frac_in = float(step_activation(affine_layer(pts, w_vec[None, :], np.array([b_val])))[:, 0].mean())
w_vec, b_val, test_labels, frac_in
```

### Answer check

```python
assert np.allclose(w_vec, [3.0, 2.0], atol=1e-12)
assert np.isclose(b_val, -6.0, atol=1e-12)
assert np.allclose(test_labels, [1.0, 1.0, 1.0, 0.0, 0.0], atol=1e-12)
assert np.isclose(frac_in, 0.3015, atol=1e-12)
```
