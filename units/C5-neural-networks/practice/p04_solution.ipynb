# C5-neural-networks — Practice p04 — Solution

For the first hidden unit, $z_1=2(1)-1(2)-1=-1$, so it outputs $0$.  For the second, $z_2=-1(1)+1(2)+1=2$, so it outputs $1$.  Thus $y=4(0)+5(1)-5/2=5/2$.  In lowest terms $p=5$, $q=2$, and $p+q=7$, giving **B**.

- A proposes $6$, but $5+2$ is not $6$.
- C proposes $9$, which would require a different numerator or denominator.
- D proposes $15$, confusing neither the output $5/2$ nor its normal-form sum.
- E proposes $17$, likewise unrelated to the reduced numerator and denominator.

```python
import numpy as np

SEED = 20260804
answer_p04 = "B"
hidden_z = np.array([-1.0, 2.0])
hidden_h = (hidden_z >= 0).astype(float)
y = float((hidden_h * np.array([4.0, 5.0])).sum() - 2.5)
p, q = 5, 2
normal_form_sum = p + q
answer_p04, hidden_z, hidden_h, y, normal_form_sum
```

### Answer check

```python
assert answer_p04 == "B"
assert np.allclose(hidden_z, [-1.0, 2.0], atol=1e-12)
assert np.allclose(hidden_h, [0.0, 1.0], atol=1e-12)
assert np.isclose(y, 2.5, atol=1e-12)
assert normal_form_sum == 7
```
