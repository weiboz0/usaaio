# C5-neural-networks — Practice p07 — Solution

Each ramp adds slope at its first kink and removes the same slope at its second.  Therefore the kinks are $0,2,4,5$.  The slopes on $(-\infty,0)$, $(0,2)$, $(2,4)$, $(4,5)$, and $(5,\infty)$ are respectively $0,1,0,2,0$.  At the kinks the adjacent formulas agree, so endpoint ownership does not change the values.

```python
import numpy as np

SEED = 20260804
grid = np.linspace(-2.0, 8.0, 501)
probes = np.array([-1.0, 1.0, 3.0, 4.5, 6.0])


def relu(z):
    return np.maximum(z, 0.0)


def ramp(z, a, b):
    return relu(z - a) - relu(z - b)


g_grid = ramp(grid, 0.0, 2.0) + 2.0 * ramp(grid, 4.0, 5.0)
g_probes = ramp(probes, 0.0, 2.0) + 2.0 * ramp(probes, 4.0, 5.0)
plateau_val = float(ramp(10.0, 0.0, 2.0) + 2.0 * ramp(10.0, 4.0, 5.0))
g_probes, plateau_val
```
