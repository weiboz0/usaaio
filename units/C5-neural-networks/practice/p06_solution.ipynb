# C5-neural-networks — Practice p06 — Solution

The step is a vectorized comparison followed by a float cast.  ReLU is the elementwise maximum with zero.  Both operations preserve the input shape.

```python
import numpy as np

SEED = 20260804
z_vec = np.array([-2.0, -0.5, 0.0, 0.7, 3.0])
rng = np.random.default_rng(SEED)
Z = rng.normal(0, 1, (4, 5))


def step_activation(z):
    return (z >= 0).astype(float)


def relu(z):
    return np.maximum(z, 0.0)


step_vals = step_activation(z_vec)
relu_vals = relu(z_vec)
tanh_vals = np.tanh(z_vec)
n_fired = int(step_activation(Z).sum())
relu_total = float(relu(Z).sum())
step_vals, relu_vals, tanh_vals, n_fired, relu_total
```

### Answer check

```python
assert np.allclose(step_vals, [0.0, 0.0, 1.0, 1.0, 1.0], atol=1e-12, rtol=0)
assert np.allclose(relu_vals, [0.0, 0.0, 0.0, 0.7, 3.0], atol=1e-12, rtol=0)
assert np.allclose(tanh_vals, np.array([-0.9640275800758169, -0.46211715726000974, 0.0, 0.6043677771171636, 0.9950547536867305]), atol=1e-12, rtol=0)
assert n_fired == 9
assert np.isclose(relu_total, 6.8340928370840865, atol=1e-12, rtol=0)
```
