# C5-neural-networks — Practice p20 — Solution

The hand table is organized with points as rows and units as columns.  The exact-zero entry is point $(0,0)$ with unit 2: $2(0)-1(0)+0=0$.  The inclusive convention scores that entry as firing, so the table contains `1.0` there.

```python
import numpy as np


def affine_layer(x, W, b):
    return (x[:, None, :] * W[None, :, :]).sum(axis=2) + b


def step_activation(z):
    return (z >= 0).astype(float)


SEED = 20260804
X = np.array([[0.0, 0.0], [2.0, 1.0], [1.0, 3.0]])
W = np.array([[1.0, 1.0], [2.0, -1.0], [-1.0, 3.0]])
b = np.array([-2.0, 0.0, -3.0])
outputs_by_hand = [[0.0, 1.0, 0.0], [1.0, 1.0, 0.0], [1.0, 0.0, 1.0]]
preactivations = affine_layer(X, W, b)
outputs = step_activation(preactivations)
print("network:", outputs, "hand:", np.array(outputs_by_hand))
```

### Answer check

```python
assert np.allclose(preactivations, [[-2.0, 0.0, -3.0], [1.0, 3.0, -2.0], [2.0, -1.0, 5.0]], atol=1e-12)
assert np.allclose(outputs, outputs_by_hand, atol=1e-12)
assert np.isclose(preactivations[0, 1], 0.0, atol=1e-12)
assert np.isclose(outputs[0, 1], 1.0, atol=1e-12)
```
