# C5-neural-networks — Practice p19 — Solution

Applying the inclusive sign test by hand gives three firing entries in `z_list`.  In the matrix, the zero at row 1, column 3 fires, just like the positive entries.

```python
import numpy as np

SEED = 20260804
z_list = [-3.5, -1.0, 0.0, 0.25, 2.0, -0.001]
Z = np.array([[1.5, -2.0, 0.0], [-0.5, 3.0, -0.25]])
step_by_hand = [0.0, 0.0, 1.0, 1.0, 1.0, 0.0]
n_fire_by_hand = 3
H_by_hand = [[1.0, 0.0, 1.0], [0.0, 1.0, 0.0]]


def step_activation(z):
    return (z >= 0).astype(float)


step_vals = step_activation(np.array(z_list))
H = step_activation(Z)
print("vectorized:", step_vals, "hand:", step_by_hand)
print("vectorized matrix:", H, "hand matrix:", np.array(H_by_hand))
```

### Answer check

```python
assert n_fire_by_hand == 3
assert np.allclose(step_vals, step_by_hand, atol=1e-12, rtol=0)
assert np.allclose(H, H_by_hand, atol=1e-12, rtol=0)
```
