# C1-ml-fundamentals — Practice p05 — Solution

A factory screens 200 parts; only 8 are actually defective (class `1`). Two
detectors are proposed. Detector A never flags anything. Detector B flags 10
parts: 5 of the 8 real defects, plus 5 good parts.

1. Compute detector A's accuracy `acc_a` and recall `recall_a`, and
   detector B's accuracy `acc_b`, precision `precision_b`, recall
   `recall_b`, and F1 `f1_b` (exact names).
2. **Short answer** (markdown cell): the two accuracies come out identical.
   Which detector should the factory deploy, and which metric(s) justify the
   choice?

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
y_true = np.zeros(200, dtype=int)
y_true[:8] = 1                     # the 8 real defects

pred_a = np.zeros(200, dtype=int)  # detector A: never flags

pred_b = np.zeros(200, dtype=int)  # detector B:
pred_b[:5] = 1                     #   catches 5 of the 8 defects
pred_b[8:13] = 1                   #   plus 5 false alarms
```

```python
def counts(y_t, y_p):
    TP = np.sum((y_t == 1) & (y_p == 1))
    FP = np.sum((y_t == 0) & (y_p == 1))
    FN = np.sum((y_t == 1) & (y_p == 0))
    TN = np.sum((y_t == 0) & (y_p == 0))
    return TP, FP, FN, TN

TP_a, FP_a, FN_a, TN_a = counts(y_true, pred_a)
acc_a    = (TP_a + TN_a) / 200
recall_a = TP_a / (TP_a + FN_a)
print(f"A: accuracy = {acc_a}  recall = {recall_a}")

TP_b, FP_b, FN_b, TN_b = counts(y_true, pred_b)
acc_b       = (TP_b + TN_b) / 200
precision_b = TP_b / (TP_b + FP_b)
recall_b    = TP_b / (TP_b + FN_b)
f1_b        = 2 * precision_b * recall_b / (precision_b + recall_b)
print(f"B: accuracy = {acc_b}  precision = {precision_b}  recall = {recall_b}  F1 = {f1_b:.4f}")
```

**Task 2 answer.** Both detectors score accuracy 0.96 — A's 8 misses
and B's 3 misses + 5 false alarms happen to cost the same number of correct
answers. But accuracy is the wrong lens on a 96%-negative dataset: A has
recall 0 (it ships every defect), while B catches 62.5% of defects
with precision 50% and F1 ≈ 0.56. The factory should
deploy B; recall (and F1, which folds it in) is the metric that justifies the
choice, because a shipped defect is the costly mistake.

### Answer check

```python
assert np.isclose(acc_a, 0.96)
assert np.isclose(recall_a, 0.0)
assert np.isclose(acc_b, 0.96)
assert np.isclose(precision_b, 0.5)
assert np.isclose(recall_b, 0.625)
assert np.isclose(f1_b, 0.5555555555555556)
assert np.isclose(acc_a, acc_b)   # identical accuracies -- the trap
print("p05 OK")
```
