A dataset has N examples of which exactly M are positive, with M < N/2. A
**constant classifier** outputs the same label for every example.

**(a)** Show that the always-negative constant classifier has accuracy
(N − M)/N and the always-positive one has accuracy M/N. Conclude which
constant classifier achieves the best possible accuracy among constant
classifiers, and how large that accuracy is when M is small.

**(b)** Show that the always-negative classifier has recall 0, and explain
why its precision is not even defined.

**(c)** Let a classifier have precision P > 0 and recall R > 0, and let
s = min(P, R). Prove that

s ≤ F1 ≤ 2s,

where F1 = 2PR/(P + R). (Hint: assume P ≤ R without losing generality, and
bound the denominator P + R from both sides.) Conclude in one sentence why a
respectable F1 is impossible when either P or R is tiny.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

**(a)** The always-negative classifier is correct on precisely the negative
examples, of which there are N − M, so its accuracy is (N − M)/N. The
always-positive classifier is correct on precisely the M positives, giving
accuracy M/N. Since M < N/2 we have N − M > M, so (N − M)/N > M/N and
**always-negative is the best constant classifier**. When M is small its
accuracy 1 − M/N is close to 1 — a high score obtained while ignoring the
input entirely, which is why accuracy must always be compared against this
constant baseline under class imbalance.

**(b)** The always-negative classifier makes no positive calls, so TP = 0
while FN = M > 0; recall = TP/(TP + FN) = 0/M = **0**. Precision is
TP/(TP + FP) = 0/0 — its denominator counts positive calls and there are
none, so the ratio is **undefined**: "how often are its positive calls
right?" is a question about calls that were never made.

**(c)** Assume P ≤ R (otherwise swap the letters throughout; F1 is symmetric
in P and R). Then s = P.

*Upper bound.* The denominator satisfies P + R ≥ R, so

F1 = 2PR/(P + R) ≤ 2PR/R = 2P = 2s.

*Lower bound.* Since P ≤ R, the denominator satisfies P + R ≤ 2R, so

F1 = 2PR/(P + R) ≥ 2PR/(2R) = P = s.

Together, s ≤ F1 ≤ 2s. **Conclusion:** F1 is trapped within a factor of two
of the *smaller* of precision and recall, so if either one is tiny, F1 is at
most twice that tiny number — no strength in the other metric can rescue it.

Optional numeric spot-check of all three parts:

```python
# (a)/(b): N = 500, M = 40
y = np.zeros(500, dtype=int)
y[:40] = 1
acc_neg = np.mean(y == 0)          # always-negative accuracy
acc_pos = np.mean(y == 1)          # always-positive accuracy
recall_neg = 0 / 40                # TP = 0, FN = 40
print("always-negative accuracy:", acc_neg, " always-positive:", acc_pos,
      " recall:", recall_neg)

# (c): the inequality on 10,000 random (P, R) pairs
P = rng.uniform(0.01, 1.0, size=10000)
R = rng.uniform(0.01, 1.0, size=10000)
F1 = 2 * P * R / (P + R)
s = np.minimum(P, R)
print("s <= F1 always:", bool(np.all(s <= F1 + 1e-12)))
print("F1 <= 2s always:", bool(np.all(F1 <= 2 * s + 1e-12)))
```

### Answer check

```python
assert np.isclose(acc_neg, 0.92)
assert np.isclose(acc_pos, 0.08)
assert recall_neg == 0.0
assert acc_neg > acc_pos                       # best constant rule
assert np.all(s <= F1 + 1e-12) and np.all(F1 <= 2 * s + 1e-12)
print("p12 OK")
```
