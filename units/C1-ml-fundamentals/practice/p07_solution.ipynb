A classifier for handwritten letters scores **100% accuracy on its training
set** and **55% on the held-out test set**. Always predicting the most common
letter would score 60% on that same test set.

Which statement best explains the situation?

- **A.** The rule is too rigid; it should be made flexible enough to absorb
  more detail from the training examples.
- **B.** The rule is too flexible: it latched onto specifics of its training
  examples, so it performs far better on them than on new data — and it even
  fails to beat the always-majority baseline.
- **C.** The test set must be defective, because training accuracy is the
  truer measure of a rule's quality.
- **D.** Nothing is wrong: 100% training accuracy shows the pattern has been
  learned, and test accuracy will catch up with more evaluation runs.
- **E.** The rule is exactly balanced between rigid and flexible, since the
  average of 100% and 55% is close to 60%.

Record your choice as a single capital letter in the code cell below.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
answer_p07 = "B"
```

**Explanation.** A huge train-to-test gap (100% vs 55%) is the signature of
overfitting, which sits at the *flexible* end of the rigid-flexible spectrum —
the rule bent itself around the particular training examples instead of the
pattern behind them. Scoring below the 60% always-majority baseline confirms
the flexibility bought nothing transferable. A is backwards (more flexibility
worsens this failure), C and D trust the wrong number, and E invents a
meaningless average.
