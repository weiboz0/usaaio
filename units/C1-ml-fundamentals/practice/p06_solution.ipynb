# C1-ml-fundamentals — Practice p06 — Solution

The 3×3 confusion table below summarizes a 150-photo animal classifier
(classes `0` = cat, `1` = dog, `2` = rabbit; rows = actual class, columns =
predicted class). Using NumPy axis-wise sums and `np.diag`:

1. Compute the per-class precision, recall, and F1 as arrays
   `prec_per_class`, `rec_per_class`, `f1_per_class` (exact names).
2. Compute the macro-F1 as `macro_f1` (exact name) and print everything.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
C = np.array([[42,  5,  3],
              [ 6, 30,  4],
              [ 2,  8, 50]])
```

```python
diag = np.diag(C).astype(float)          # correct predictions per class

prec_per_class = diag / C.sum(axis=0)    # column sums: everything predicted k
rec_per_class  = diag / C.sum(axis=1)    # row sums: everything actually k
f1_per_class   = 2 * prec_per_class * rec_per_class / (prec_per_class + rec_per_class)
macro_f1 = f1_per_class.mean()

for k, name in enumerate(["cat", "dog", "rabbit"]):
    print(f"{name:>6}: precision={prec_per_class[k]:.4f}  "
          f"recall={rec_per_class[k]:.4f}  F1={f1_per_class[k]:.4f}")
print(f"macro-F1 = {macro_f1:.4f}")
```

**Explanation.** Class *k* is scored one-vs-rest: its precision divides the
diagonal entry by column *k*'s sum (all photos *predicted* class *k*), and its
recall divides the same diagonal entry by row *k*'s sum (all photos *actually*
class *k*). The per-class F1 values come out to about
0.840 (cat), 0.723 (dog), and 0.855 (rabbit), and
macro-F1 is their plain average, ≈ 0.8059 — each class counts
equally, so the weaker dog class pulls the average down.
