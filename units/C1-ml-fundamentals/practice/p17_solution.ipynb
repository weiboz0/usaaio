The lesson clustered unlabeled 1-D data by sorting it and cutting at the
biggest jump between neighbours. Package that idea:

```python
def gap_split(values):
    ...
```

takes a 1-D NumPy array (length ≥ 2, no labels) and returns the tuple
`(group_a, group_b)`: the sorted values below the largest neighbour gap and
those above it. `group_a` holds the smaller values; together the two groups
contain every input value.

**Banned inside the function: Python `for` and `while` loops and list
comprehensions. Any use of a banned construct scores zero points for this
problem.** `np.sort`, `np.diff`, and `np.argmax` (as covered in
F1-scientific-python) are all you need.

Apply it to both datasets below and print the groups. Then answer in one
markdown sentence: what did the function find in dataset A, and what
information about the two groups can it *not* provide?

````python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
````

````python
# Dataset A: heights (cm) at a family picnic -- no labels
heights = np.array([131.0, 174.0, 128.0, 180.0, 136.0, 169.0, 125.0, 177.0, 133.0])

# Dataset B: reaction times (s) from two experimental conditions -- no labels
times = np.array([0.9, 5.6, 1.3, 6.1, 0.7, 5.9, 1.1, 6.4])
````

````python
def gap_split(values):
    s = np.sort(values)
    k = int(np.argmax(np.diff(s)))     # index of the largest neighbour gap
    return s[:k + 1], s[k + 1:]

ga, gb = gap_split(heights)
print("A:", ga, "|", gb)
ta, tb = gap_split(times)
print("B:", ta, "|", tb)
````

**Answer.** In dataset A the function found the two height clusters hiding in
the unlabeled data — one group in the 125–136 cm range and one in the
169–180 cm range — but it cannot say *what* the groups are (children vs.
adults) or which one matters: clustering delivers the grouping, and the
interpretation and naming of the groups remain a human's job.

### Answer check

````python
ga, gb = gap_split(heights)
assert np.allclose(ga, [125.0, 128.0, 131.0, 133.0, 136.0], atol=1e-9, rtol=0)
assert np.allclose(gb, [169.0, 174.0, 177.0, 180.0], atol=1e-9, rtol=0)
ta, tb = gap_split(times)
assert np.allclose(ta, [0.7, 0.9, 1.1, 1.3], atol=1e-9, rtol=0)
assert np.allclose(tb, [5.6, 5.9, 6.1, 6.4], atol=1e-9, rtol=0)
import inspect
src = inspect.getsource(gap_split)
assert "for " not in src and "while " not in src
print("p17 OK")
````
