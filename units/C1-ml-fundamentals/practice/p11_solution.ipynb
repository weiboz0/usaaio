Write code to implement

taking a K×K NumPy integer confusion table `cm` (rows = actual class,
columns = predicted class, any K ≥ 2) and returning the macro-F1 as a single
`float`. The same function body must work for **every** K without
modification. You may assume every class has at least one actual example and
at least one predicted example.

**Banned inside the function: Python `for` and `while` loops and list
comprehensions. Any use of a banned construct scores zero points for this
problem.** Use `np.diag`, axis-wise sums, and elementwise arithmetic, as
covered in F1-scientific-python.

Apply your function to the three tables below (3×3, 4×4, and 2×2) and print
the three results.

````python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
````

````python
C3 = np.array([[4, 1, 0],
               [1, 3, 1],
               [0, 1, 4]])

C4 = np.array([[10,  2,  1,  0],
               [ 1, 12,  2,  1],
               [ 0,  3,  9,  2],
               [ 1,  0,  2, 14]])

C2 = np.array([[45,  5],
               [10, 40]])
````

````python
def my_macro_f1(cm):
    diag = np.diag(cm).astype(float)      # correct predictions, shape (K,)
    precision = diag / cm.sum(axis=0)     # column sums: predicted-as-k
    recall    = diag / cm.sum(axis=1)     # row sums: actually-k
    f1        = 2 * precision * recall / (precision + recall)
    return float(f1.mean())

for name, C in [("3x3", C3), ("4x4", C4), ("2x2", C2)]:
    print(name, "macro-F1:", round(my_macro_f1(C), 4))
````

**Explanation.** The whole computation is shape-(K,) arithmetic: the diagonal
holds each class's correct predictions, `sum(axis=0)` gives the K
predicted-class totals and `sum(axis=1)` the K actual-class totals, so
per-class precision, recall, and F1 all fall out of three elementwise
divisions — the size of K never appears in the code. The three tables give
macro-F1 ≈ 0.7333, 0.7484, and 0.8496; the 4×4 case working
unchanged is the point of the shape contract.
