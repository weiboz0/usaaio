# F6-svd-spectral — Practice p09 — Solution


The `eigh` outputs are reversed on both the value axis and the eigenvector-column axis. The second reconstruction explicitly sums the six rank-1 atoms.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
B = rng.normal(0, 1, (6, 6))
S = (B + B.T) / 2

vals_asc, vecs_asc = np.linalg.eigh(S)
vals_desc = vals_asc[::-1]
vecs_desc = vecs_asc[:, ::-1]
orth_gap = np.abs(vecs_desc.T @ vecs_desc - np.eye(6)).max()
recon_gap = np.abs(vecs_desc @ np.diag(vals_desc) @ vecs_desc.T - S).max()
atoms = (vals_desc[None, None, :] * vecs_desc[:, None, :]
         * vecs_desc[None, :, :])
S_atom_sum = atoms.sum(axis=2)
sum_gap = np.abs(S_atom_sum - S).max()
vals_desc, orth_gap, recon_gap, sum_gap
```

### Answer check


```python
assert vals_desc.shape == (6,) and vecs_desc.shape == (6, 6)
assert np.all(np.diff(vals_desc) <= 0)
assert orth_gap < 1e-10
assert recon_gap < 1e-10
assert sum_gap < 1e-10
```
