# F6-svd-spectral — Practice p11 — Solution


The thin bridge matches the three squared singular values to the top three eigenvalues; the remaining three eigenvalues are machine zero.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (6, 3))
S = W @ W.T

lam_asc, Q_asc = np.linalg.eigh(S)
lam_desc, Q_desc = lam_asc[::-1], Q_asc[:, ::-1]
s = np.linalg.svd(W, full_matrices=False, compute_uv=False)
bridge_gap = np.abs(lam_desc[:3] - s**2).max()
tail_max = np.abs(lam_desc[3:]).max()
lam_desc, s, bridge_gap, tail_max
```

### Answer check


```python
assert lam_desc.shape == (6,) and s.shape == (3,)
assert bridge_gap < 1e-8
assert tail_max < 1e-10
```
