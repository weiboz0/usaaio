# F6-svd-spectral — Practice p12 — Solution


Each direct reconstruction is checked against the tail-sum identity, including the two endpoint anchors.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (8, 5))


def truncate(U, s, Vt, r):
    return U[:, :r] @ np.diag(s[:r]) @ Vt[:r, :]


U, s, Vt = np.linalg.svd(W, full_matrices=False)
errs_direct = np.array([np.linalg.norm(W - truncate(U, s, Vt, r))
                        for r in range(6)])
errs_formula = np.array([np.sqrt((s[r:]**2).sum()) for r in range(6)])
identity_gap = np.abs(errs_direct - errs_formula).max()
err_r2 = errs_direct[2]
errs_direct, err_r2, identity_gap
```

### Answer check


```python
assert errs_direct.shape == (6,) and errs_formula.shape == (6,)
assert identity_gap < 1e-8
assert np.isclose(errs_direct[0], np.linalg.norm(W), rtol=0, atol=1e-10)
assert errs_direct[5] < 1e-10
assert np.isclose(err_r2, 3.438729577690459, rtol=0, atol=1e-12)
```
