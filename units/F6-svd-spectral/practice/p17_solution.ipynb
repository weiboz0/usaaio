# F6-svd-spectral — Practice p17 — Solution


The fixed $7\times4$ matrix has four linearly independent columns, so $W$ and $WW^{\mathsf T}$ have rank $4$; therefore exactly four eigenvalues are nonzero.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (7, 4))
S = W @ W.T

lam_asc, vecs_asc = np.linalg.eigh(S)
lam_desc, vecs_desc = lam_asc[::-1], vecs_asc[:, ::-1]
n_nonzero = int((np.abs(lam_desc) > 1e-10).sum())
recon_gap = np.abs(vecs_desc @ np.diag(lam_desc) @ vecs_desc.T - S).max()
frac = lam_desc / lam_desc.sum()
frac_top = frac[0]
qualifies = np.cumsum(frac) >= 0.90
assert qualifies.any()
k90 = int(np.argmax(qualifies)) + 1
content_gap = abs(lam_desc.sum() - (W * W).sum())
lam_desc, n_nonzero, frac_top, k90, recon_gap, content_gap
```

Because $S=WW^{\mathsf T}$ is positive semidefinite, its exact eigenvalues are non-negative, so dividing them by their sum produces non-negative shares whose sum is $1$.


### Answer check


```python
assert n_nonzero == 4
assert recon_gap < 1e-10 and content_gap < 1e-8
assert np.isclose(frac.sum(), 1.0, rtol=0, atol=1e-12)
assert np.isclose(frac_top, 0.6606752209424495, rtol=0, atol=1e-12)
assert k90 == 2
```
