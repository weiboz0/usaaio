## Unit goals

By the end of this unit you can:

- recognize **eigenvectors** as the directions a matrix does not turn
  and **eigenvalues** as their signed stretch factors; verify claimed
  eigenpairs by the eigen-equation residual;
- find $2 \times 2$ eigenvalues **by hand** via the dependent-rows
  route — $(S - \lambda I)$ singular ⟺ rows proportional ⟺
  $(a{-}\lambda)(d{-}\lambda) - bc = 0$ — and recover eigenvectors
  from the row-proportionality read;
- use `np.linalg.eig` and `np.linalg.eigh` safely: column convention,
  complex-dtype hygiene, and the **pinned reorder idiom**
  (`eigh` returns ascending → `vals[::-1]`, `vecs[:, ::-1]`);
- state and use the symmetric promises (real eigenvalues, orthonormal
  eigenvectors) and the **spectral decomposition**
  $S = Q\Lambda Q^{\mathsf T} = \sum_i \lambda_i q_i q_i^{\mathsf T}$,
  including the energy reading $x^{\mathsf T}Sx$ and why Gram matrices
  are PSD;
- apply the course **sign-fixing pin** (largest-$|$entry$|$ component
  made positive) when comparing eigenvectors across routes — and know
  its scope: distinct, well-separated eigenvalues only, with
  **invariant checks** (reconstruction, residuals, projectors) for
  degenerate spectra;
- compute and read the **SVD** $W = U\Sigma V^{\mathsf T}$ of any
  matrix: thin vs full shapes (the shape table), output conventions
  (`s` descending, `Vt` pre-transposed), and the per-problem
  `np.linalg` legality register;
- state, derive (component form), and use the **bridge in both
  forms**: thin — the columns of thin $U$ are eigenvectors of
  $S = WW^{\mathsf T}$ for the nonzero eigenvalues
  $\lambda_i = \sigma_i^2$; full — full $U$ plus zero-padded
  $\lambda$ is the complete spectral decomposition;
- compute **Frobenius norms** from entries, derive and use
  $\lVert W \rVert_F^2 = \sum_i \sigma_i^2$ and the truncation error
  identity $\lVert W - W_r \rVert_F^2 = \sum_{i>r} \sigma_i^2$, quote
  Eckart–Young, and answer storage/budget questions from the
  error-vs-$r$ curve.
