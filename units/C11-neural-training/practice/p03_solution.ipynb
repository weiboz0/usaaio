# C11-neural-training — Practice p03 — Solution


**Type:** multiple choice · **Difficulty:** intro · **Concepts:** manual-backpropagation


**C.** The branch through \(u=3h\) contributes \(3\), while the branch through
\(v=h^2\) contributes \(2h=-4\) at \(h=-2\). The shared node must accumulate
them, so \(dL/dh=3-4=-1\). In an array graph, that accumulated derivative has
exactly the shape of \(h\). Updating an activation or parameter before every
downstream path has contributed would make later derivatives use a different
forward snapshot and would discard or corrupt part of the chain rule sum.


```python
import numpy as np
h = np.array([-2.0, 1.5])
du_dh = np.full_like(h, 3.0)
dv_dh = 2.0 * h
total = du_dh + dv_dh
answer_p03 = "C"
```

### Answer check


```python
assert answer_p03 == "C"
assert du_dh[0] == 3.0 and dv_dh[0] == -4.0 and total[0] == -1.0
assert total.shape == h.shape
```
