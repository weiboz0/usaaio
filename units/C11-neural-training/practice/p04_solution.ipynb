# C11-neural-training — Practice p04 — Solution


**Type:** multiple choice (numeric, normal form) · **Difficulty:** intro · **Concepts:** dropout


**B.** If \(M_i\sim\operatorname{Bernoulli}(q)\), inverted dropout gives
\(\widetilde h_i=M_i h_i/q\), hence
\(E[\widetilde h_i]=E[M_i]h_i/q=h_i\). Therefore the requested ratio is
\(-2/(1+4)=-2/5\). Its denominator is positive and
\(\gcd(|-2|,5)=1\), so this is the required reduced signed fraction.


```python
from fractions import Fraction
from math import gcd
q = Fraction(3, 4)
h = [Fraction(-2), Fraction(1), Fraction(4)]
expectations = [q * value / q for value in h]
ratio = expectations[0] / (expectations[1] + expectations[2])
answer_p04 = "B"
```

### Answer check


```python
assert answer_p04 == "B"
assert expectations == h
assert ratio == Fraction(-2, 5)
assert ratio.denominator > 0 and gcd(abs(ratio.numerator), ratio.denominator) == 1
```
