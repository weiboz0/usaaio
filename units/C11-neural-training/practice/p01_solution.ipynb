# C11-neural-training — Practice p01 — Solution


**Type:** multiple choice · **Difficulty:** intro · **Concepts:** softmax


**B.** For any finite scalar \(a\),
\[
\frac{e^{z_c-a}}{\sum_j e^{z_j-a}}
=\frac{e^{-a}e^{z_c}}{e^{-a}\sum_j e^{z_j}}
=\frac{e^{z_c}}{\sum_j e^{z_j}}.
\]
Taking \(a=\max_j z_j=1001\) gives shifted logits \((-1,0,-2)\), so every
exponential is at most \(e^0=1\); this avoids overflow without changing the
probabilities.


```python
import numpy as np
z = np.array([1000.0, 1001.0, 999.0])
answer_p01 = "B"
shifted = z - z.max()
p_shifted = np.exp(shifted) / np.exp(shifted).sum()
p_reference = np.exp(z - 1000.0) / np.exp(z - 1000.0).sum()
```

### Answer check


```python
assert answer_p01 == "B"
assert np.all(shifted <= 0.0)
assert np.all(np.exp(shifted) <= 1.0)
assert np.allclose(p_shifted, p_reference, atol=1e-15, rtol=1e-15)
```
