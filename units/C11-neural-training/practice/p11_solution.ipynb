# C11-neural-training — Practice p11 — Solution


**Type:** proof / derivation · **Difficulty:** core · **Concepts:** softmax, cross-entropy-loss


Let \(S=\sum_j e^{z_j}\). If \(c=k\), quotient differentiation gives
\(\partial p_c/\partial z_c=p_c(1-p_c)\); if \(c\ne k\), it gives
\(-e^{z_c}e^{z_k}/S^2=-p_cp_k\). Thus
\(J_{ck}=p_c(\delta_{ck}-p_k)\), or
\(J=\operatorname{diag}(p)-pp^T\).

For row \(c\), \(\sum_kJ_{ck}=p_c(1-\sum_kp_k)=0\). For column \(k\),
\(\sum_cJ_{ck}=p_k-p_k\sum_cp_c=0\).
Since \(\partial L/\partial p_c=-y_c/p_c\),
\[
\frac{\partial L}{\partial z_k}=\sum_c\frac{-y_c}{p_c}
p_c(\delta_{ck}-p_k)=-y_k+p_k\sum_cy_c=p_k-y_k.
\]
For a batch mean, \(dZ=(P-Y)/N\), with shape \((N,C)\). Computing this
directly from stable rowwise probabilities costs \(O(NC)\), avoids every
\(C\times C\) Jacobian, and avoids unstable unshifted exponentials.


```python
import numpy as np
p_p11 = np.array([0.2, 0.3, 0.5])
J_p11 = np.diag(p_p11) - np.outer(p_p11, p_p11)
y_p11 = np.array([0., 0., 1.])
chain_p11 = J_p11.T @ (-y_p11 / p_p11)
```

### Answer check


```python
assert np.allclose(J_p11.sum(axis=0), 0.0, atol=1e-15, rtol=0.0)
assert np.allclose(J_p11.sum(axis=1), 0.0, atol=1e-15, rtol=0.0)
assert np.allclose(chain_p11, p_p11-y_p11, atol=1e-15, rtol=1e-15)
P_p11 = np.vstack([p_p11, p_p11])
Y_p11 = np.vstack([y_p11, y_p11])
assert ((P_p11-Y_p11)/2).shape == (2, 3)
```
