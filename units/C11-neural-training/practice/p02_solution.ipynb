# C11-neural-training — Practice p02 — Solution


**Type:** multiple choice · **Difficulty:** intro · **Concepts:** cross-entropy-loss


**B.** With one-hot targets \(Y_{ic}=1\) only when \(c=y_i\),
\[
-\frac1N\sum_i\sum_c Y_{ic}\log P_{ic}
=-\frac1N\sum_i\log P_{i,y_i}.
\]
The sum is over examples and the factor \(1/N\) averages the batch, not the
classes. Reducing the probability assigned to an observed class makes its
negative logarithm, and hence the mean loss, larger.


```python
import numpy as np
P = np.array([[0.8, 0.2], [0.25, 0.75]], dtype=np.float64)
y = np.array([0, 1])
Y = np.eye(2)[y]
answer_p02 = "B"
one_hot_loss = -np.sum(Y * np.log(P)) / len(y)
gather_loss = -np.mean(np.log(P[np.arange(len(y)), y]))
```

### Answer check


```python
assert answer_p02 == "B"
assert np.isclose(one_hot_loss, gather_loss, atol=1e-15, rtol=1e-15)
assert -np.log(0.2) > -np.log(0.8)
```
