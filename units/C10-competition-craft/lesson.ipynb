## Unit goals

By the end of this unit you can:

- state the **graded-notebook model** — a submission runs top to
  bottom on a fresh machine, defines one prediction function to a
  pinned signature, and is scored on a held-back split plus reasoning
  quality;
- write a **contract-honoring prediction function**,
  `predict_labels(X_test) -> pd.Series` — any-length feature
  DataFrame in; a Series out with matching length, matching index,
  original order, training-vocabulary values — and run the four
  self-checks that catch every packaging zero before the grader does;
- state the **hidden-test protocol** honestly: the held-back split is
  regenerated deterministically for grading; your notebook never
  reads it, and everything you report is computable without it;
- **derive macro-F1 from its confusion-matrix pieces** (per-class
  one-vs-rest precision/recall/F1, averaged with equal class weights)
  and explain, with numbers, why it disagrees with accuracy on an
  imbalanced task;
- run the **iteration loop** — baseline → error analysis → one change
  → re-validate — against a frozen, seeded, stratified validation
  carve, with a log and a cap, and state the overfitting-to-validation
  fact that motivates both;
- enforce **notebook discipline** (pinned seeds, cell order =
  execution order, no dead cells) and prove **deterministic re-run**
  with a two-run audit;
- write a **rubric-grade writeup** — approach / intuition /
  alternatives — where every claim is backed by a number the notebook
  actually printed, and score any writeup against the unit's
  six-point rubric;
- author a deliberate **mixed-cell Colab response**, keeping rendered Markdown
  claims separate from executable evidence;
- render matched fenced Python excerpts and inline/display mathematics, defining
  symbols and interpreting every scored formula;
- run the **coding-submission preflight**: Restart and Run All, exact identifiers,
  required outputs/files, then download and inspect the `.ipynb`;
- state the round boundary exactly: **Round 1 is CPU-only; Round 2 permits the
  Colab L4/GPU environment**, without assuming the later GPU workflow.

**What this unit deliberately does not do:** improve your models.
The model family is k-nearest neighbors throughout — deliberately, and
mirroring the exam's applied-problem constraint — because competition
craft is the discipline *around* the model: the contract, the metric,
the protocol, the notebook, the writeup.
Every problem enforces the constraint with the exam's own device: a
ban clause with a zero-points register.

## The harness

The unit runs on a mini-competition you can hold in your hands:

- `data/make_dataset.py` — the seeded, public generator (read it!);
- `data/train.csv` — 600 labeled colonies: predict which honeybee
  colonies **thrive** vs **struggle** through winter from 12 autumn
  inspection features (classes ~2:1 imbalanced, scales wildly mixed);
- a **held-back split** of 200 further colonies, regenerated
  deterministically at grading time by the course's grading
  notebooks — never read by yours (the honesty framing is stated in
  Session 1 §4);
- the graded metric: **macro-F1** of `predict_labels` on the
  held-back rows.
