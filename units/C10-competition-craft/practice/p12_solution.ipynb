# C10-competition-craft — Practice p12 — Solution

### (a) Claim 1 — each candidate estimate is unbiased

Write (A_j=\frac1{50}\sum_{i=1}^{50} I_{ji}), where (I_{ji}) is the fair-coin correctness indicator. Since (E[I_{ji}]=1/2), linearity gives

\[
E[A_j]=\frac1{50}\sum_{i=1}^{50}E[I_{ji}]
=\frac1{50}\cdot 50\cdot\frac12=\frac12.
\]

### (b) Claim 2 — selection cannot push the estimate down

For every realized set of ten scores, (M=\max_j A_j\ge A_1). Monotonicity of expectation therefore gives (E[M]\ge E[A_1]=1/2).

### (c) Claim 3 — selection pushes it strictly up here

The ten independent estimates have genuine spread, so their maximum exceeds a fixed candidate with positive probability; in this stated-fact register, that makes (E[M]>1/2). This is the **max of noise** mechanism: selection turns a lucky validation fluctuation into apparent skill.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)

flips = rng.integers(0, 2, size=(2000, 10, 50))
val_accs = flips.mean(axis=2)
mean_selected = float(val_accs.max(axis=1).mean())

fresh = rng.integers(0, 2, size=(2000, 50)).mean(axis=1)
mean_fresh = float(fresh.mean())

print("selected:", mean_selected, "| fresh:", mean_fresh)
```

In an iteration loop, the tested recipes are the ten candidates and the largest logged validation score is (M). The hidden-test protocol is load bearing because unlimited repeated evaluation on grading rows would turn them into another worn-out selection set, rewarding noise-fitting rather than generalization.
