# C10-competition-craft — Practice p04 — Solution

```python
import numpy as np

C = np.array([[3, 3], [1, 13]])
f1_struggles = 2 * 3 / (2 * 3 + 1 + 3)
f1_thrives = 2 * 13 / (2 * 13 + 3 + 1)
macro_f1 = (f1_struggles + f1_thrives) / 2
accuracy = np.trace(C) / C.sum()
fraction_p, fraction_q = 11, 15
S = fraction_p + fraction_q if macro_f1 < accuracy else fraction_p + fraction_q - 2
answer_p04 = "D"
(f1_struggles, f1_thrives, macro_f1, accuracy, S, answer_p04)
```

**D is correct.** The class F1 scores are (3/5) and (13/15), so macro-F1 is (11/15). Accuracy is (16/20=4/5), hence macro-F1 is lower and (S=11+15=26).

- A and B do not match either required branch of the reduced fraction.
- C is the (p+q-2=24) value from taking the wrong comparison branch.
- E does not arise from the reduced macro-F1 fraction.
