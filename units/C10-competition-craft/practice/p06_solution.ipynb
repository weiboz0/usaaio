# C10-competition-craft — Practice p06 — Solution

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

SEED = 20260804

df = pd.read_csv("../data/train.csv")
FEATURES = [c for c in df.columns if c != "outcome"]
X = df[FEATURES]
y = df["outcome"].to_numpy()


def make_carve():
    return train_test_split(
        X, y, test_size=150, random_state=SEED, stratify=y
    )


X_tr_a, X_val_a, y_tr_a, y_val_a = make_carve()
X_tr_b, X_val_b, y_tr_b, y_val_b = make_carve()
labels, counts = np.unique(y_val_a, return_counts=True)
val_counts = {label: int(count) for label, count in zip(labels, counts)}
carve_identical = bool(
    X_val_a.index.equals(X_val_b.index)
    and np.array_equal(y_val_a, y_val_b)
)

_, X_val_1, _, _ = train_test_split(
    X, y, test_size=150, random_state=1, stratify=y
)
_, X_val_2, _, _ = train_test_split(
    X, y, test_size=150, random_state=2, stratify=y
)
seed_matters = bool(not X_val_1.index.equals(X_val_2.index))
(val_counts, carve_identical, seed_matters)
```

The two-call equality habit checks D4, deterministic re-run, with D1's pinned seed making it possible. A grader's fresh Restart & Run All discards interactive state, so an unpinned carve changes the downstream results and makes the filed claims unreproducible.
