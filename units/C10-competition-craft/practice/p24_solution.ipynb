# C10-competition-craft — Practice p24 — Solution

Broadcasting forms one comparison row per target value: every row counts how many entries in the original array are strictly greater than that target. Adding one gives competition rank directly, so equal values receive equal counts and the ranks after a tie contain the intended gaps.

```python
import numpy as np

values = np.array([2.7, -4.3, 8.1, 2.7, 5.9, 8.1])


def competition_ranks(values):
    greater_counts = (values[None, :] > values[:, None]).sum(axis=1)
    return (greater_counts + 1).astype(int)


ranks = competition_ranks(values)
tie_ok = bool(ranks[0] == ranks[3] and ranks[2] == ranks[5])
```

The two maxima tie at rank 1, 5.9 has rank 3, the two 2.7 values tie at rank 4, and -4.3 has rank 6. Thus the full result is `[4, 6, 1, 4, 3, 1]`.
