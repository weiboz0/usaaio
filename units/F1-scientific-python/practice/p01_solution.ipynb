# F1-scientific-python — Practice p01 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
# Given: one temperature reading per hour, for 7 days (rows) x 24 hours (columns).
temps = 10 + (np.arange(7 * 24).reshape(7, 24) * 7) % 13    # degrees Celsius
print(temps.shape)
```

**Task A — building arrays.** Create three arrays:

- **`hours`** — the integers 0 through 23, using `np.arange`
- **`markers`** — exactly 5 evenly spaced values from 0 to 23 (both ends
  included), using `np.linspace`
- **`blank`** — a 7×24 array of zeros

```python
hours = np.arange(24)
markers = np.linspace(0, 23, 5)
blank = np.zeros((7, 24))
print(hours)
print(markers)
print(blank.shape)
```

`np.arange(24)` excludes its stop, giving 0..23. `np.linspace` includes both endpoints and takes a *count* of points, so 5 points from 0 to 23 land 5.75 apart. `np.zeros` takes the shape as a tuple.

**Task B — one row.** Extract all 24 of Tuesday's readings (day index 1) as
**`tuesday`**.

```python
tuesday = temps[1]
print(tuesday)
```

Indexing a 2-D array with a single index selects that whole row, shape `(24,)`.

**Task C — one column.** Extract the reading at noon (hour index 12) for every
day of the week as **`noon`**.

```python
noon = temps[:, 12]
print(noon)
```

A `:` in the row slot means every row; fixing the column to 12 gives one value per day, shape `(7,)`.

**Task D — a block.** Extract the weekend-morning block — days 5 and 6, hours
6 through 11 inclusive — as **`weekend_mornings`**. Predict its shape before
running, then print the shape to check.

```python
weekend_mornings = temps[5:, 6:12]
print(weekend_mornings.shape)
print(weekend_mornings)
```

Slicing both axes gives a rectangular block. Rows `5:` keeps days 5-6; columns `6:12` keeps hours 6-11 (the stop is excluded), so the shape is `(2, 6)`.

**Task E — boolean mask.** Build **`warm_mask`**, a True/False array marking
readings of 18 degrees or more. Then use it to build **`warm_values`** (the
readings themselves) and **`warm_count`** (how many there are).

```python
warm_mask = temps >= 18
warm_values = temps[warm_mask]
warm_count = warm_mask.sum()
print(warm_count, "warm readings")
print(warm_values[:10])
```

The comparison runs elementwise, producing a mask the same shape as `temps`. Indexing with it keeps the True positions as a 1-D array, and a mask's `.sum()` counts its True entries.

**Task F — integer-array indexing.** Create:

- **`mwf`** — the full rows for Monday, Wednesday, Friday (days 0, 2, 4), in
  that order
- **`picked`** — the two individual readings at (day 0, hour 8) and
  (day 3, hour 20), fetched in a single indexing expression

```python
mwf = temps[[0, 2, 4]]
picked = temps[[0, 3], [8, 20]]
print(mwf.shape)
print(picked)
```

A list of row indices selects those whole rows in the order listed. Two parallel lists pair up: the result is `[temps[0, 8], temps[3, 20]]`.
