```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

**Task A.** Implement exactly `def rescale(values, lo, hi):`.

- `values` is a 1-D float array containing at least two distinct values;
  `lo < hi`.
- Return a **new** array of the same shape in which the smallest element of
  `values` becomes `lo`, the largest becomes `hi`, and every other element
  keeps its relative position:
  $\text{out} = lo + (v - v_{min}) \cdot (hi - lo) / (v_{max} - v_{min})$.

```python
def rescale(values, lo, hi):
    vmin = values.min()
    vmax = values.max()
    return lo + (values - vmin) * (hi - lo) / (vmax - vmin)
```

`min`/`max` collapse the array to its two extremes; the rest is pure elementwise arithmetic, so the whole array is mapped in one expression and a new array is returned — `values` is never written to.

**Task B.** Implement exactly `def frac_above(values, cutoff):` — return the
fraction of elements **strictly** above `cutoff`, as a plain Python `float`.

```python
def frac_above(values, cutoff):
    return float((values > cutoff).mean())
```

The mask-`.mean()` idiom: True counts as 1, so the mean of the mask is exactly the fraction passing the test.

**Task C.** Test both. For `v = np.array([2.0, 4.0, 10.0])` check by hand what
`rescale(v, 0.0, 1.0)` must be, then verify; confirm `v` is unchanged; and
check `frac_above(v, 3.0)` against a hand count.

```python
v = np.array([2.0, 4.0, 10.0])
print(rescale(v, 0.0, 1.0))      # by hand: (0, 2/8, 8/8) = [0, 0.25, 1]
print("v unchanged:", np.array_equal(v, [2.0, 4.0, 10.0]))
print(frac_above(v, 3.0))        # 2 of 3 above -> 0.666...
```

### Answer check

```python
v = np.array([2.0, 4.0, 10.0])
out = rescale(v, 0.0, 1.0)
assert np.allclose(out, [0.0, 0.25, 1.0])
assert np.array_equal(v, np.array([2.0, 4.0, 10.0]))
out2 = rescale(v, 10.0, 20.0)
assert np.isclose(out2.min(), 10.0) and np.isclose(out2.max(), 20.0)
big = np.random.default_rng(SEED).random(500) * 30 - 10
scaled = rescale(big, 0.0, 1.0)
assert np.isclose(scaled.min(), 0.0) and np.isclose(scaled.max(), 1.0)
assert scaled.shape == big.shape
f = frac_above(v, 3.0)
assert isinstance(f, float) and np.isclose(f, 2 / 3)
assert np.isclose(frac_above(big, 100.0), 0.0)
print("all checks passed")
```
