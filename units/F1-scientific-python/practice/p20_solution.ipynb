**Part c — redo it properly.** With `SEED = 20260804` and a seeded generator,
redraw `before` and `after` exactly as above (`before` first, then `after` —
draw order matters). Then make ONE labeled figure comparing the two groups
with overlaid histograms: 10 bins each, a `label=` per group,
transparency (`alpha=0.6`) so both stay visible, title, axis labels, legend.

```python
SEED = 20260804
rng = np.random.default_rng(SEED)
before = rng.integers(60, 101, size=250)
after = rng.integers(55, 96, size=250)

plt.figure(figsize=(7, 4))
plt.hist(before, bins=10, alpha=0.6, label="before tweak")
plt.hist(after, bins=10, alpha=0.6, label="after tweak")
plt.title("Page load times before vs. after the tweak")
plt.xlabel("load time (ms)")
plt.ylabel("how many measurements")
plt.legend()
plt.show()
```

Histograms are the right form: each group is one array of measurements, and the question is where the values pile up. Overlaying with transparency puts the comparison in a single picture.

**Part d — the verdict, with numbers.** Compute **`mean_before`**,
**`mean_after`**, and **`shift`** (their difference, before minus after).
Then, in 1–2 sentences (edit this cell): does the evidence support "the tweak
worked", and what does the histogram add that the two means alone do not?

*(Write your verdict here.)*

```python
mean_before = before.mean()
mean_after = after.mean()
shift = mean_before - mean_after
print(f"before: {mean_before:.1f} ms | after: {mean_after:.1f} ms | shift: {shift:.1f} ms")
```

The averages differ by a few ms in the expected direction, consistent with the tweak helping — a modest, honest claim the raw zig-zag plot could never support. The histograms add what single numbers hide: the two groups overlap heavily, so *individual* loads are often slower after the tweak even though the typical load is faster.
