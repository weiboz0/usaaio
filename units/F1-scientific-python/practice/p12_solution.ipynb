`np.random.default_rng()` is built **without a seed**, so the generator starts from an unpredictable state and `data` differs on every run — nobody can reproduce or verify the printed averages. Fix: define a seed constant and build the generator from it, e.g. `rng = np.random.default_rng(20260804)`.

**Part b — why 4 numbers instead of 30?** Explain in 1–2 sentences (edit this

`axis=0` collapses the **student** axis, leaving one average per quiz — shape `(4,)`. One average per student needs the quiz axis to disappear: `axis=1`, giving shape `(30,)`. The rule: the axis you pass is the axis that disappears.

**Part c — the corrected cell.** Write the fixed version: `SEED = 20260804`,
a seeded `rng`, the same `(30, 4)` draw as **`data_fixed`**, and
**`student_avg_fixed`** with shape `(30,)`. Then demonstrate reproducibility:
build a second generator from the same seed, redraw, and show the two `data`
arrays are identical.

```python
SEED = 20260804
rng = np.random.default_rng(SEED)
data_fixed = rng.integers(50, 101, size=(30, 4))
student_avg_fixed = data_fixed.mean(axis=1)
print(student_avg_fixed.shape, student_avg_fixed[:5])

rng_again = np.random.default_rng(SEED)
data_again = rng_again.integers(50, 101, size=(30, 4))
print("reproducible:", np.array_equal(data_fixed, data_again))
```

With the seed fixed, the draw is identical on every run and every machine — which is what makes the averages checkable. And `axis=1` delivers the intended one-number-per-student shape.

**Part d — judgment call.** The classmate suggests: *"easier fix — I'll just

Hardcoded screenshots cannot be re-derived: the moment the code changes, the pasted numbers silently stop matching what the code produces, and no grader (or teammate) can verify where they came from. A seed keeps the numbers *live* — regenerated from code on every run — while still being identical for everyone. Cherry-picking a run that "looks reasonable" also quietly biases the reported results toward what you expected to see.
