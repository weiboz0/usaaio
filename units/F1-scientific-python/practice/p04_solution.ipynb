# F1-scientific-python — Practice p04 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
# Given: quiz scores for 6 students (rows) across 4 quizzes (columns).
scores = np.array([
    [78, 85, 90, 66],
    [92, 79, 84, 88],
    [55, 64, 72, 60],
    [81, 81, 95, 77],
    [69, 93, 58, 84],
    [88, 72, 76, 91],
])
print(scores.shape)
```

**Task A — per-quiz averages.** Set **`quiz_mean`** to the average score of
each quiz (one number per quiz). Which axis must disappear?

```python
quiz_mean = scores.mean(axis=0)
print(quiz_mean, quiz_mean.shape)
```

One number per quiz means the student axis (axis 0, the rows) must collapse, leaving shape `(4,)`.

**Task B — per-student results.** Set **`student_total`** to each student's
total across the 4 quizzes, and **`student_best`** to each student's single
highest score.

```python
student_total = scores.sum(axis=1)
student_best = scores.max(axis=1)
print(student_total)
print(student_best)
```

Both collapse the quiz axis (axis 1), leaving one value per student, shape `(6,)`.

**Task C — locating extremes.** Set **`hardest_quiz`** to the *index* of the
quiz with the lowest average, and **`top_student`** to the index of the student
with the highest total.

```python
hardest_quiz = quiz_mean.argmin()
top_student = student_total.argmax()
print("hardest quiz:", hardest_quiz, "| top student:", top_student)
```

`argmin`/`argmax` return positions instead of values, which is what "which quiz" and "which student" ask for.

**Task D — keepdims.** Set **`kept`** to `scores.sum(axis=1, keepdims=True)`
and **`share`** to each score as a fraction of that student's total. Print
`kept.shape`, and check that every row of `share` sums to 1. Why does this
division *need* `keepdims=True`?

```python
kept = scores.sum(axis=1, keepdims=True)
share = scores / kept
print(kept.shape)
print(share.round(3))
print("row sums:", share.sum(axis=1))
```

`keepdims=True` keeps the collapsed axis as size 1, so the totals have shape `(6, 1)` and broadcast row-by-row against `(6, 4)`. Without it the shape is `(6,)`, which lines its 6 against the 4 when compared from the right — an error.

### Answer check

```python
assert quiz_mean.shape == (4,)
assert np.allclose(quiz_mean, scores.mean(axis=0))
assert student_total.shape == (6,) and student_best.shape == (6,)
assert np.array_equal(student_total, scores.sum(axis=1))
assert np.array_equal(student_best, scores.max(axis=1))
assert hardest_quiz == int(scores.mean(axis=0).argmin())
assert top_student == int(scores.sum(axis=1).argmax())
assert kept.shape == (6, 1)
assert share.shape == (6, 4)
assert np.allclose(share.sum(axis=1), 1.0)
print("all checks passed")
```
