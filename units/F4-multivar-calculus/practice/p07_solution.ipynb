# F4-multivar-calculus — Practice p07 — Solution

Writing $r_n=y_n-\sum_kX_{n,k}w_k$, differentiation gives $\partial Q/\partial w_j=-(2/N)\sum_n r_nX_{n,j}$. Broadcasting forms every $X_{n,k}w_k$ and every $r_nX_{n,j}$ without matrix operators.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (50, 3))
y = rng.normal(0, 1, 50)
w = np.array([0.5, -1.0, 2.0])

def q_gradient(X, y, w):
    residual = y - (X * w).sum(axis=1)
    return (-2.0 / X.shape[0]) * (residual[:, None] * X).sum(axis=0)

def q_value(candidate):
    residual = y - (X * candidate).sum(axis=1)
    return (residual**2).mean()

g = q_gradient(X, y, w)
h = 1e-6
numeric = np.array([
    (q_value(w + np.array([h, 0.0, 0.0])) - q_value(w - np.array([h, 0.0, 0.0]))) / (2*h),
    (q_value(w + np.array([0.0, h, 0.0])) - q_value(w - np.array([0.0, h, 0.0]))) / (2*h),
    (q_value(w + np.array([0.0, 0.0, h])) - q_value(w - np.array([0.0, 0.0, h]))) / (2*h)
])
check_gap = np.abs(g - numeric).max()
```

### Answer check

```python
expected_g = np.array([0.2968880039364528, -2.328767947307388, 3.0541808484754216])
assert np.allclose(g, expected_g, rtol=0.0, atol=1e-12)
assert check_gap < 1e-8
```
