# F4-multivar-calculus — Practice p03 — Solution

Because $\tanh(0)=0$, the identity gives $\tanh\'(0)=1$. As $x\to\infty$, $\tanh(x)\to1$, so its derivative approaches $0$ and the correct choice is B.

```python
import numpy as np

value_at_zero = 1.0 - np.tanh(0.0)**2
large_x_value = 1.0 - np.tanh(20.0)**2
answer_p03 = "B"
```

### Answer check

```python
my_answer_letter = answer_p03
assert np.isclose(value_at_zero, 1.0, atol=1e-9, rtol=0)
assert abs(large_x_value) < 1e-12
assert my_answer_letter == "B"
```
