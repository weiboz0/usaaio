# F4-multivar-calculus — Practice p05 — Solution

The central difference changes only $v$ and divides the symmetric function difference by $2h$. Since $\partial_v(u^2v+3v)=u^2+3$, the exact value at $u=2$ is $7$.

```python
import numpy as np

def partial_v(f, u, v, h):
    return (f(u, v + h) - f(u, v - h)) / (2 * h)

f = lambda u, v: u**2 * v + 3*v
est = partial_v(f, 2.0, 1.0, 1e-6)
exact = 2.0**2 + 3.0
gap = abs(est - exact)
```

### Answer check

```python
assert np.isclose(exact, 7.0)
assert np.isclose(est, exact, rtol=0.0, atol=1e-8)
assert gap < 1e-8
```
