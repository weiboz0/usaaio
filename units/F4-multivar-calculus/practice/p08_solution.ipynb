# F4-multivar-calculus — Practice p08 — Solution

Substitution gives $f(r(t))=\cos^2t+3\sin t+t^2$. Differentiating term by term yields $-2\cos t\sin t+3\cos t+2t$, which is also the chain-rule sum $(2\cos t)(-\sin t)+3\cos t+2t$.

```python
import numpy as np

def chain_exact(t):
    return -2*np.cos(t)*np.sin(t) + 3*np.cos(t) + 2*t

def chain_numeric(t, h):
    composite_plus = np.cos(t + h)**2 + 3*np.sin(t + h) + (t + h)**2
    composite_minus = np.cos(t - h)**2 + 3*np.sin(t - h) + (t - h)**2
    return (composite_plus - composite_minus) / (2*h)

points = np.array([0.5, 1.0, 2.0])
vals = np.array([chain_exact(t) for t in (0.5, 1.0, 2.0)])
numeric_vals = chain_numeric(points, 1e-6)
gap = np.abs(vals - numeric_vals).max()
```

### Answer check

```python
expected_vals = np.array([2.7912767008632215, 2.7116094907787374, 3.508361985666501])
assert np.allclose(vals, expected_vals, rtol=0.0, atol=1e-12)
assert gap < 1e-8
```
