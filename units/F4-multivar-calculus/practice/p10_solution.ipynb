# F4-multivar-calculus — Practice p10 — Solution

The squared-error term contributes $-(2/N)\sum_n r_nX_{n,j}$, while the penalty contributes $2\lambda w_j$. Thus the regularized gradient is the unregularized gradient plus $2\lambda w$.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (50, 3))
y = rng.normal(0, 1, 50)
w = np.array([0.5, -1.0, 2.0])

def r_gradient(X, y, w, lam):
    residual = y - (X * w).sum(axis=1)
    data_part = (-2.0 / X.shape[0]) * (residual[:, None] * X).sum(axis=0)
    return data_part + 2.0 * lam * w

g_reg = r_gradient(X, y, w, 0.1)
g_unreg = r_gradient(X, y, w, 0.0)
reg_effect = np.abs(g_reg - g_unreg)
formula_gap = np.abs(reg_effect - np.abs(2.0 * 0.1 * w)).max()
```

The exact componentwise difference in magnitude is $|2\lambda w|$. For $\lambda=0.1$, it is therefore $(0.1,0.2,0.4)$.
