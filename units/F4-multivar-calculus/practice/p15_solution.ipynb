# F4-multivar-calculus — Practice p15 — Solution

The direction of fastest increase is the normalized gradient, and its maximum directional derivative is the gradient norm. The northward rate is the gradient's scalar projection onto $(0,1)$.

```python
import numpy as np

gradient = np.array([3.2, -2.4])
rate = np.linalg.norm(gradient)
direction = gradient / rate
rate_north = gradient[1]
```

Due east increases the signal at rate $3.2$, but the southeast unit direction $(0.8,-0.6)$ combines both favorable coordinate changes and reaches rate $4.0$. Due north actually decreases the signal at rate $-2.4$, so the largest single partial does not by itself identify the best unrestricted direction.

### Answer check

```python
assert np.allclose(direction, np.array([0.8, -0.6]), atol=1e-9, rtol=0)
assert np.isclose(np.linalg.norm(direction), 1.0, atol=1e-9, rtol=0)
assert np.isclose(rate, 4.0, atol=1e-9, rtol=0)
assert np.isclose(rate_north, -2.4, atol=1e-9, rtol=0)
```
