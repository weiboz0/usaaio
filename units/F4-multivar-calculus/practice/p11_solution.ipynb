# F4-multivar-calculus — Practice p11 — Solution

 Since tanh²(x) ≥ 0 for every x, we get tanh′(x) = 1 − tanh²(x) ≤ 1 directly; and since tanh²(x) < 1 for all real x (the numerator's magnitude is strictly below the denominator's), tanh′(x) > 0. Since $D^2-N^2=(D-N)(D+N)=(2e^{-x})(2e^x)=4$, this is $4/D^2$, while $1-\tanh^2x=(D^2-N^2)/D^2$, proving the identity. The denominator $D^2$ is positive, so $\tanh'(x)>0$, and the AM-GM inequality $D=e^x+e^{-x}\ge2$ gives $4/D^2\le1$. Equality in AM-GM requires $e^x=e^{-x}$, hence $x=0$, so equality occurs only there.

```python
import numpy as np

x = 0.7
h = 1e-6
direct = (np.tanh(x + h) - np.tanh(x - h)) / (2*h)
identity_val = 1.0 - np.tanh(x)**2
anchor_gap = abs(direct - identity_val)
```

At $x=0.7$, the identity predicts a positive derivative below one, consistent with the proof. The central difference provides an independent numerical sanity check of the algebra.
