# F4-multivar-calculus — Practice p13 — Solution

With residual $r_n=s_n-w_0-w_1h_n$, the chain rule gives $\partial Q/\partial w_0=-(2/5)\sum_nr_n$. Similarly, $\partial Q/\partial w_1=-(2/5)\sum_nr_nh_n$ because $\partial r_n/\partial w_1=-h_n$.

```python
import numpy as np

hours = np.array([1., 2., 3., 4., 5.])
scores = np.array([52., 55., 61., 64., 70.])

def q2_gradient(w0, w1):
    residual = scores - w0 - w1 * hours
    return np.array([-2.0 * residual.mean(), -2.0 * (residual * hours).mean()])

w_final = np.array([0.0, 0.0])
for _ in range(2000):
    w_final -= 0.01 * q2_gradient(w_final[0], w_final[1])
grad_norm_final = np.linalg.norm(q2_gradient(w_final[0], w_final[1]))
```

The fitted intercept predicts about $46.9$ points at zero study hours, while the slope predicts about $4.5$ additional points per hour. These values describe the least-squares linear relationship for the five observations.
