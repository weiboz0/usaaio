# F4-multivar-calculus — Practice p12 — Solution

The chain rule is $\frac{d}{dt}f(2t,t^2)=2f_u(2t,t^2)+2t f_v(2t,t^2)$. Both terms appear because a nudge in $t$ changes both $u$ and $v$, and each coordinate change contributes through its own partial derivative. For $f(u,v)=uv+v^2$, $f_u=v$ and $f_v=u+2v$, so the chain rule gives $2t^2+2t(2t+2t^2)=6t^2+4t^3$. Direct substitution gives $f(2t,t^2)=2t^3+t^4$, whose derivative is likewise $6t^2+4t^3$.

```python
import numpy as np

t = 1.5
via_chain = 2*t**2 + 2*t*(2*t + 2*t**2)
via_subst = 6*t**2 + 4*t**3
anchor_gap = abs(via_chain - via_subst)
```

At $t=1.5$, both derivations produce $27$. This numeric anchor confirms that neither coordinate route was omitted.

### Answer check

```python
assert np.isclose(via_chain, 27.0)
assert np.isclose(via_subst, 27.0)
assert anchor_gap < 1e-12
```
