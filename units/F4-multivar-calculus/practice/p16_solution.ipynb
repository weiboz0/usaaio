# F4-multivar-calculus — Practice p16 — Solution

The chain rule gives $r'(t)=[1-\tanh^2(0.5m(t))](0.5)m'(t)$. Thus the multiplier before the tanh-derivative factor is $3$ when $m'(t_0)=6$.

```python
import numpy as np

m_prime = 6.0
r_rate = (1.0 - np.tanh(0.5 * 4.0)**2) * 0.5 * m_prime
r_rate_zero = (1.0 - np.tanh(0.5 * 0.0)**2) * 0.5 * m_prime
```

At raw value $4$, saturation reduces the reported rate to about $0.212$, despite the raw rate of $6$; at raw value $0$, the reported rate is $3$. The pipeline is most responsive at raw-measurement level $m=0$, where the tanh derivative is maximal.

### Answer check

```python
assert np.isclose(r_rate, 0.21195247455949396, rtol=0.0, atol=1e-15)
assert np.isclose(r_rate_zero, 3.0, atol=1e-9, rtol=0)
```
