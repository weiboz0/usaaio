# F4-multivar-calculus — Practice p09 — Solution

Let $z=\sum_ka_kw_k$, so $g=\tanh z$. The chain rule gives $\partial g/\partial w_j=(1-\tanh^2z)a_j$, which can be evaluated with elementwise multiplication and a sum.

```python
import numpy as np

def tanh_gradient(a, w):
    z = (a * w).sum()
    return (1.0 - np.tanh(z)**2) * a

a = np.array([1.0, -2.0, 0.5])
w = np.array([0.3, 0.1, -0.4])
grad = tanh_gradient(a, w)
h = 1e-6

def g_value(candidate):
    return np.tanh((a * candidate).sum())

numeric = np.array([
    (g_value(w + np.array([h, 0.0, 0.0])) - g_value(w - np.array([h, 0.0, 0.0]))) / (2*h),
    (g_value(w + np.array([0.0, h, 0.0])) - g_value(w - np.array([0.0, h, 0.0]))) / (2*h),
    (g_value(w + np.array([0.0, 0.0, h])) - g_value(w - np.array([0.0, 0.0, h]))) / (2*h)
])
check_gap = np.abs(grad - numeric).max()
```

### Answer check

```python
expected_grad = np.array([0.9900662908474398, -1.9801325816948796, 0.4950331454237199])
assert np.allclose(grad, expected_grad, rtol=0.0, atol=1e-12)
assert check_gap < 1e-8
```
