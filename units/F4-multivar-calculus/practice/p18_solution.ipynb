# F4-multivar-calculus — Practice p18 — Solution

Differentiation gives $\nabla g=(1-\tanh^2x,\,y)$. Therefore its norm is $\sqrt{(1-\tanh^2x)^2+y^2}$, evaluated by broadcasting the row of $x$ values against the column of $y$ values.

```python
import numpy as np

xs = np.linspace(-3, 3, 301)
ys = np.linspace(-2, 2, 201)

def g_grad_norm(xx, yy):
    return np.sqrt((1.0 - np.tanh(xx)**2)**2 + yy**2)

xx = xs[None, :]
yy = ys[:, None]
norms = g_grad_norm(xx, yy)
flat_index = int(np.argmin(norms))
column_count = norms.shape[1]
flat_row = flat_index // column_count
flat_column = flat_index % column_count
flat_point = np.array([xs[flat_column], ys[flat_row]])
steep_index = int(np.argmax(norms))
steep_row = steep_index // column_count
steep_column = steep_index % column_count
steep_norm = norms[steep_row, steep_column]
```

The flat point has $y=0$, minimizing $|y|$, and $x=\pm3$, maximizing $|x|$ on the grid so the tanh derivative is smallest. The steepest locations lie on the edges $y=\pm2$ and at $x=0$, where both $|y|$ and the tanh derivative attain their grid maxima.

### Answer check

```python
assert np.allclose(flat_point, np.array([-3.0, 0.0]))
assert np.isclose(norms[flat_row, flat_column], 1.0 - np.tanh(3.0)**2)
assert np.isclose(steep_norm, np.sqrt(5.0))
```
