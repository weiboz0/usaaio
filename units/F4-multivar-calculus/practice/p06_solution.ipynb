# F4-multivar-calculus — Practice p06 — Solution

Each coordinate is perturbed independently, so every output entry is a central-difference partial derivative. Here $f(p)=p_0^2+p_1^2+2p_0p_1$, giving $\nabla f=(2p_0+2p_1,2p_1+2p_0)=(6,6)$.

```python
import numpy as np

def num_gradient(f, point, h):
    result = np.empty_like(point, dtype=float)
    for j in range(point.size):
        step = np.zeros_like(point, dtype=float)
        step[j] = h
        result[j] = (f(point + step) - f(point - step)) / (2 * h)
    return result

f = lambda p: (p**2).sum() + 2*p[0]*p[1]
point = np.array([1.0, 2.0])
grad_est = num_gradient(f, point, 1e-6)
grad_exact = np.array([6.0, 6.0])
max_gap = np.abs(grad_est - grad_exact).max()
```

### Answer check

```python
assert np.allclose(grad_exact, np.array([6.0, 6.0]))
assert np.allclose(grad_est, grad_exact, rtol=0.0, atol=1e-8)
assert max_gap < 1e-8
```
