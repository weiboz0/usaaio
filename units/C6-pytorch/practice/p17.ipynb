subclasses that only supply a score.

its own `score`.

- constructor takes `w` (shape `(2,)`) and `c` (shape `(1,)`);

- `score(x) = x @ w + c` for `x` of shape `(n, 2)` — so the gate fires

  points within score-distance `r` of the line, a region no single
  `PlaneGate` can express.
