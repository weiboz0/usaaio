# C6-pytorch — Practice p03 — Solution


`a` and `c` are registered `nn.Parameter` objects; the ordinary tensor
`b` is not returned by `parameters()`.  Thus `n = 2`.  The explicit
flag on `a` is false, while `c` uses `nn.Parameter`'s true default, so
the sorted flags are `[False, True]`: **C**.

A and D wrongly count the plain tensor `b`.  B wrongly treats `c` as
frozen, and E drops one of the two registered parameters.


```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)

class Probe(nn.Module):
    def __init__(self):
        super().__init__()
        self.a = nn.Parameter(torch.zeros(3), requires_grad=False)
        self.b = torch.ones(3)
        self.c = nn.Parameter(torch.ones(2))


m = Probe()
n = len(list(m.parameters()))
flags = sorted(p.requires_grad for p in m.parameters())
answer_p03 = "C"
answer_p03, n, flags
```

### Answer check


```python
assert answer_p03 == "C"
assert n == 2
assert flags == [False, True]
assert sorted(dict(m.named_parameters())) == ["a", "c"]
```
