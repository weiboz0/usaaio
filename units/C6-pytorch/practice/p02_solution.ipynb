# C6-pytorch — Practice p02 — Solution


`super().__init__()` first creates `tag = "base"`, after which the
child's constructor replaces that attribute with `"child"`.  `name`
is inherited from `Base`, but it reads the current attribute from the
child object, so the output is `Base-child`: **B**.

A ignores the later assignment.  C incorrectly assumes the defining
class changes the literal `"Base-"`.  D is impossible because `tag`
is assigned twice, and E ignores method inheritance.


```python
import torch

torch.set_default_dtype(torch.float64)

class Base:
    def __init__(self):
        self.tag = "base"

    def name(self):
        return "Base-" + self.tag


class Child(Base):
    def __init__(self):
        super().__init__()
        self.tag = "child"


observed = Child().name()
answer_p02 = "B"
answer_p02, observed
```

### Answer check


```python
assert answer_p02 == "B"
assert observed == "Base-child"
assert isinstance(Child(), Base)
```
