# C6-pytorch — Practice p20 — Solution


Module calls produce the forward result, while the inspection APIs
expose the two registered frozen tensors and their names.  The scalar
total is read directly from shapes `(3, 2)` and `(3,)` as $3\cdot2+3$.


```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)

class Blender(nn.Module):
    """A fixed hand-weighted layer for inspection drills."""

    def __init__(self):
        super().__init__()
        self.mix = nn.Parameter(torch.tensor([[2.0, -1.0],
                                              [0.0, 1.0],
                                              [1.0, 1.0]]), requires_grad=False)
        self.lift = nn.Parameter(torch.tensor([0.5, -0.5, 0.0]), requires_grad=False)

    def forward(self, x):
        return x @ self.mix.T + self.lift


blend = Blender()
x0 = torch.tensor([[1.0, 2.0]])

y = blend(x0)
n_tensors = sum(1 for _ in blend.parameters())
names = sorted(name for name, _ in blend.named_parameters())
sd_keys = sorted(blend.state_dict().keys())
mix_shape = tuple(blend.mix.shape)
all_frozen = all(not p.requires_grad for p in blend.parameters())
n_scalars = 3 * 2 + 3

y, n_tensors, names, sd_keys, mix_shape, all_frozen, n_scalars
```

### Answer check


```python
assert torch.equal(y, torch.tensor([[0.5, 1.5, 3.0]]))
assert n_tensors == 2
assert names == ["lift", "mix"]
assert sd_keys == ["lift", "mix"]
assert mix_shape == (3, 2)
assert all_frozen is True
assert n_scalars == 9
```
