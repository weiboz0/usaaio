# C6-pytorch — Practice p07 — Solution


The conjunction of the two inclusive comparisons is a boolean tensor.
Casting that mask to `x.dtype` produces the required zero-one output,
including both boundary values.


```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)

class Window(nn.Module):
    def __init__(self, lo, hi):
        super().__init__()
        self.lo = lo
        self.hi = hi

    def forward(self, x):
        return ((self.lo <= x) & (x <= self.hi)).to(x.dtype)


t = torch.tensor([-2.0, -0.5, 0.0, 0.7, 1.0, 3.0])
w = Window(-0.5, 1.0)
inside = w(t)
n_inside = int(inside.sum())

torch.manual_seed(20260804)
v = torch.randn(200)
frac_inside = float(w(v).mean())

inside, n_inside, frac_inside
```

### Answer check


```python
assert torch.equal(inside, torch.tensor([0.0, 1.0, 1.0, 1.0, 1.0, 0.0]))
assert n_inside == 4
assert frac_inside == 0.565
assert len(list(w.parameters())) == 0
```
