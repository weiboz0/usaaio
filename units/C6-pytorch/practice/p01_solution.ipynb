# C6-pytorch — Practice p01 — Solution


Reshaping gives rows `[0, 1, 2]` and `[3, 4, 5]`.  Summing with
`dim=0` collapses the row dimension, so the column sums are
`[3, 5, 7]`.  `arange(6)` contains integers, so its dtype—and the
sum's dtype—is `torch.int64`.  Therefore the answer is **D**.

A and C use row sums, not column sums.  B has the right values but
incorrectly changes an integer tensor to float32.  E is false because
torch supports summation of integer tensors.


```python
import torch

torch.set_default_dtype(torch.float64)

t = torch.arange(6).reshape(2, 3)
u = t.sum(dim=0)
answer_p01 = "D"
answer_p01, u, u.dtype
```

### Answer check


```python
assert answer_p01 == "D"
assert torch.equal(u, torch.tensor([3, 5, 7], dtype=torch.int64))
assert u.dtype == torch.int64
```
