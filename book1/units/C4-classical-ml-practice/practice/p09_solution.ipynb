# C4-classical-ml-practice — Practice p09 — Solution

```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

SEED = 20260804
X, y = load_wine(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=SEED, stratify=y
)
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5)),
])
pipe.fit(X_tr, y_tr)
test_acc = float(pipe.score(X_te, y_te))
bare_acc = float(KNeighborsClassifier(n_neighbors=5).fit(X_tr, y_tr).score(X_te, y_te))

test_acc, bare_acc
```

The pipeline learns scaling statistics only during its fit on the training rows and applies the same transform to held-out rows. Wine features have very different numerical scales, so raw distance is dominated by the largest-scale columns; standardization raises held-out accuracy from about 0.689 to 1.0 here.

### Answer check

```python
assert list(pipe.named_steps) == ["scaler", "knn"]
assert np.isclose(test_acc, 1.0, atol=1e-9, rtol=0)
assert np.isclose(bare_acc, 31 / 45, atol=1e-9, rtol=0)
assert test_acc > bare_acc
```
