# C4-classical-ml-practice — Practice p03 — Solution

```python
import numpy as np

n_rows = 40
n_folds = 5
validation_rows_per_fold = n_rows // n_folds
training_rows_per_fold = n_rows - validation_rows_per_fold
answer_p03 = "B"
training_rows_per_fold, validation_rows_per_fold, answer_p03
```

Five-fold cross-validation partitions the 40 rows into five disjoint validation folds of 8 rows each. Each of the five models trains on the other 32 rows, and the partition makes every row a validation row exactly once, so B is correct.

### Answer check

```python
assert training_rows_per_fold == 32
assert validation_rows_per_fold == 8
expected_letter = "B"
assert answer_p03 == expected_letter
```
