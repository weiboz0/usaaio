# C4-classical-ml-practice — Practice p06 — Solution

```python
import numpy as np
import pandas as pd

sensors = pd.read_csv("data/sensors.csv")
n_faulty = int((sensors["status"] == "faulty").sum())
pressure_mean = float(sensors["pressure_pa"].mean())
hot = sensors.loc[sensors["temp_c"] > 67.0]
hot_faulty_frac = float((hot["status"] == "faulty").mean())
X_sub = sensors[["temp_c", "vibration_mm_s"]].to_numpy()

n_faulty, pressure_mean, hot.shape[0], hot_faulty_frac, X_sub.shape
```

Boolean-mask aggregation finds 24 faulty machines, while filtering first ensures that the faulty fraction uses only the 18 hot rows as its denominator. Fourteen of those hot rows are faulty, giving 14/18 = 7/9, and the requested two-column matrix has shape (60, 2).

### Answer check

```python
assert n_faulty == 24 and isinstance(n_faulty, int)
assert np.isclose(pressure_mean, 100673.535, atol=1e-9, rtol=0)
assert hot.shape[0] == 18
assert np.isclose(hot_faulty_frac, 7 / 9, atol=1e-9, rtol=0)
assert X_sub.shape == (60, 2)
assert np.array_equal(X_sub, sensors[["temp_c", "vibration_mm_s"]].to_numpy())
```
