## Unit goals

By the end of this unit you can:

- load a CSV into a **pandas DataFrame** and interrogate it
  (`shape`, `head`, `dtypes`, `describe`, `value_counts`), select
  columns and rows (brackets, boolean masks with proper parentheses,
  `.loc`/`.iloc`), and compute **group summaries** with `groupby`;
- perform the **table → `(X, y)` bridge** — an explicit feature list,
  `.to_numpy()`, the label column kept out — with the shape/dtype
  ritual that catches silent bugs, in the exam starter code's idiom
  (paraphrased);
- build **k-nearest neighbors from first principles** — F2 distances
  via broadcasting, `argsort`, majority vote — and explain the knob
  $k$: $k=1$ memorizes (training accuracy 1.0 by construction, C1's
  overfitting), large $k$ smooths;
- demonstrate the **feature-scaling distortion** with concrete numbers
  (a big-scale feature carries its spread *squared* in every distance
  and dictates the neighbors) and fix it by **standardization**
  $z = (x - \mu)/\sigma$, with $\sigma$ from F5's variance —
  computing $\mu, \sigma$ from **training rows only**;
- use the sanctioned sklearn surface — `KNeighborsClassifier`,
  `StandardScaler`, `Pipeline`, `train_test_split`,
  `cross_val_score` — and *verify* the first two against your own
  implementations;
- run **honest evaluation**: seeded stratified splits, the test set
  touched exactly once, **cross-validation** for comparing candidates
  (fold arithmetic cold: $K$ models, each row validated once), and
  the full choose-$k$-honestly arc with the pipeline living *inside*
  the CV so nothing leaks.
