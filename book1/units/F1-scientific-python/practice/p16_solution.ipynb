(A) is false — same seed means same numbers on every machine; that is the point of seeding. (C) confuses a fixed *sequence* with a fixed *value*: successive calls continue along the sequence and differ. (D) is backwards — `shuffle` works in place and returns `None`. (E) — no seed means an unpredictable start, not zero. (B) is the definition of seeded reproducibility.

**Verify in code**: demonstrate that the true statement holds and that at
least one false statement fails (for example, show two calls to `rng.random()`
differ, refuting (C)).

```python
r1 = np.random.default_rng(5)
r2 = np.random.default_rng(5)
print("B holds:", np.array_equal(r1.random(4), r2.random(4)))

r3 = np.random.default_rng(5)
print("C fails:", r3.random() != r3.random())

a = np.arange(6)
result = np.random.default_rng(5).shuffle(a)
print("D fails: shuffle returned", result, "and a is now", a)
```
