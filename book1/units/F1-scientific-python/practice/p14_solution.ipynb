**Part 1.** Determine, with a rule-by-rule justification, the result shape of

```python
np.ones((6, 1, 4)) + np.ones((3, 1))
```

Set **`shape1`** to the resulting shape as a tuple (or the string `"ERROR"`).

*(Write your derivation here: pad which shape, compare which sizes, stretch
which axes?)*

````python
shape1 = (6, 3, 4)
````

Pad `(3, 1)` to `(1, 3, 1)`. Compare right-to-left: 4 vs 1 → stretch to 4; 1 vs 3 → stretch to 3; 6 vs 1 → stretch to 6. All pairs compatible, so the result is `(6, 3, 4)`.

**Part 2.** Same drill for

```python
np.arange(8).reshape(2, 4) + np.arange(4)
```

Set **`shape2`** to the resulting shape as a tuple (or `"ERROR"`), with
justification.

*(Write your derivation here.)*

````python
shape2 = (2, 4)
````

`(4,)` pads to `(1, 4)`; 4 vs 4 match, 2 vs 1 stretches. Result `(2, 4)` — the 1-D array is added to each row.

**Part 3.** Same drill for

```python
np.ones((2, 3)) + np.ones((3, 2))
```

Set **`shape3`** to the resulting shape as a tuple (or `"ERROR"`), with
justification.

*(Write your derivation here.)*

````python
shape3 = "ERROR"
````

Rightmost axes: 3 vs 2 — unequal and neither is 1. The rules fail at the first comparison; no padding or stretching can rescue it. ERROR.

**Part 4.** By hand, using the stretching picture: what **value** sits at
position `[2, 1]` of

```python
np.arange(3)[:, None] * np.arange(1, 5)[None, :]
```

Set **`value4`** to that number, with justification (which element of each
array meets at `[2, 1]`?).

*(Write your derivation here.)*

````python
value4 = 4
````

The `(3, 1)` column repeats across columns, the `(1, 4)` row repeats down rows, so entry `[i, j]` is `arange(3)[i] * arange(1, 5)[j]`. At `[2, 1]`: `2 * 2 = 4`.

**Part 5 — verify everything in code.** Confirm each shape (wrap the ERROR
case in `try`/`except`) and the value from Part 4.

````python
print((np.ones((6, 1, 4)) + np.ones((3, 1))).shape)
print((np.arange(8).reshape(2, 4) + np.arange(4)).shape)
try:
    np.ones((2, 3)) + np.ones((3, 2))
except ValueError as err:
    print("ERROR confirmed:", err)
print(np.arange(3)[:, None] * np.arange(1, 5)[None, :])
````
