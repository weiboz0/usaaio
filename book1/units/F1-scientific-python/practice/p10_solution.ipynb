**Task A.** Implement exactly `def column_shares(data):`.

- `data` is a 2-D float array of shape `(n, m)`; every column total is nonzero.
- Return a **new** array of shape `(n, m)` where entry `(i, j)` is
  `data[i, j]` divided by the total of column `j`.
- `data` itself must be left unchanged.

````python
def column_shares(data):
    return data / data.sum(axis=0, keepdims=True)
````

Column totals need the row axis to disappear (`axis=0`); `keepdims=True` keeps the result `(1, m)` so it broadcasts against `(n, m)` column-by-column. Division returns a new array, so `data` is untouched.

**Task B.** Implement exactly `def busiest_row(data):`.

- `data` is a 2-D float array of shape `(n, m)`.
- Return the **index** (an integer) of the row with the largest total.

````python
def busiest_row(data):
    return int(data.sum(axis=1).argmax())
````

Per-row totals collapse the column axis (`axis=1`), and `argmax` converts "which row" into an index.

**Task C.** Test both functions on the grid below. Check that every column of
`column_shares(grid)` sums to 1, that `grid` is unchanged afterwards, and that
`busiest_row(grid)` returns the index you expect from reading the grid.

```python
grid = np.array([[2.0, 1.0, 4.0],
                 [6.0, 1.0, 4.0],
                 [2.0, 8.0, 2.0]])
```

````python
grid = np.array([[2.0, 1.0, 4.0],
                 [6.0, 1.0, 4.0],
                 [2.0, 8.0, 2.0]])
shares = column_shares(grid)
print(shares.round(3))
print("column sums:", shares.sum(axis=0))
print("grid unchanged:", np.array_equal(grid, [[2, 1, 4], [6, 1, 4], [2, 8, 2]]))
print("busiest row:", busiest_row(grid))    # row 2 totals 12, the largest
````

### Answer check

````python
grid = np.array([[2.0, 1.0, 4.0],
                 [6.0, 1.0, 4.0],
                 [2.0, 8.0, 2.0]])
shares = column_shares(grid)
assert shares.shape == (3, 3)
assert np.allclose(shares.sum(axis=0), 1.0, atol=1e-9, rtol=0)
assert np.array_equal(grid, np.array([[2.0, 1.0, 4.0], [6.0, 1.0, 4.0], [2.0, 8.0, 2.0]]))
assert np.isclose(shares[1, 0], 0.6, atol=1e-9, rtol=0)
assert busiest_row(grid) == 2
assert busiest_row(np.array([[9.0, 9.0], [1.0, 1.0]])) == 0
big = np.random.default_rng(SEED).random((50, 7)) + 0.1
assert np.allclose(column_shares(big).sum(axis=0), 1.0, atol=1e-9, rtol=0)
assert isinstance(busiest_row(big), int)
print("all checks passed")
````
