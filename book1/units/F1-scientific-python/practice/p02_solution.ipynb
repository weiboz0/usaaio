# F1-scientific-python — Practice p02 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
# Given: a small chain of 4 stores selling the same 5 items.
base = np.array([4.00, 2.50, 10.00, 6.25, 3.75])    # base price of the 5 items
markup = np.array([1.00, 1.10, 1.25, 1.40])         # price multiplier of the 4 stores
fee = np.array([0.50, 0.25, 1.00, 0.75, 0.30])      # per-item packaging fee
celsius = np.array([-5.0, 0.0, 12.5, 21.0, 30.0])   # for the warm-up
```

**Task A — elementwise warm-up.** Convert the `celsius` readings to
**`fahrenheit`** with $F = \frac{9}{5}C + 32$, in one line.

```python
fahrenheit = celsius * 9 / 5 + 32
print(fahrenheit)
```

Each arithmetic step applies to every element at once: multiply all five readings, then add 32 to each.

**Task B — the price grid.** Build **`price_grid`** with shape `(4, 5)`:
row `i`, column `j` holds `markup[i] * base[j]`. Use broadcasting — reshape
`markup` to a `(4, 1)` column so it combines with the `(5,)` base prices.

```python
price_grid = markup[:, None] * base
print(price_grid.shape)
print(price_grid)
```

`markup[:, None]` has shape `(4, 1)` and `base` pads to `(1, 5)`; each size-1 axis stretches, so every store multiplier meets every base price, giving `(4, 5)` — a nested loop's worth of work in one expression.

**Task C — add the fee.** Every item's packaging fee is charged on top at every
store. Build **`with_fee`** = `price_grid` plus `fee`, using broadcasting.
Which axis does `fee` line up with, and why?

```python
with_fee = price_grid + fee
print(with_fee)
```

Shapes are compared from the right: `(4, 5)` against `(5,)` lines the 5s up, so `fee` pads to `(1, 5)` and is added to every row — one fee per item (column), the same at all stores.

**Task D — storewide sale.** Everything is 10% off. Build **`sale`** from
`with_fee` in one elementwise line.

```python
sale = with_fee * 0.9
print(sale.round(2))
```

Multiplying an array by a single number scales every element.

**Task E — shape drills.** Using `np.ones` to test:

- set **`shape_sum`** to the shape of `np.ones((3, 1)) + np.ones((1, 4))`
- set **`compat`** to `True` or `False`: can shapes `(2, 3)` and `(2,)` be
  combined? Decide from the rules first, then verify (wrap the attempt in
  `try`/`except` if you test it in code).

```python
shape_sum = (np.ones((3, 1)) + np.ones((1, 4))).shape
print(shape_sum)

compat = False
try:
    np.ones((2, 3)) + np.ones(2)
except ValueError as err:
    print("as predicted:", err)
```

`(3, 1)` with `(1, 4)`: each size-1 axis stretches, giving `(3, 4)`. `(2, 3)` with `(2,)` compares from the right: 3 vs 2, unequal and neither is 1, so the pair is incompatible.

### Answer check

```python
assert np.allclose(fahrenheit, [23.0, 32.0, 54.5, 69.8, 86.0], atol=1e-9, rtol=0)
assert price_grid.shape == (4, 5)
assert np.allclose(price_grid[0], base, atol=1e-9, rtol=0)                      # markup 1.00 row
assert np.isclose(price_grid[2, 0], 4.00 * 1.25, atol=1e-9, rtol=0)
assert with_fee.shape == (4, 5)
assert np.isclose(with_fee[2, 0], 4.00 * 1.25 + 0.50, atol=1e-9, rtol=0)
assert np.allclose(with_fee, markup[:, None] * base + fee, atol=1e-9, rtol=0)
assert np.allclose(sale, with_fee * 0.9, atol=1e-9, rtol=0)
assert shape_sum == (3, 4)
assert compat is False
print("all checks passed")
```
