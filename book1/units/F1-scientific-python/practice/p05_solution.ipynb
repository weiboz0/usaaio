```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

**Task A — seeded setup.** Set **`SEED`** to `20260804`, build the generator
**`rng`** with `np.random.default_rng`, and draw **`rolls`** — 1000 dice rolls
(integers 1 through 6).

```python
SEED = 20260804
rng = np.random.default_rng(SEED)
rolls = rng.integers(1, 7, size=1000)
print(rolls[:12])
```

`rng.integers(1, 7, ...)` excludes the upper end, so the faces are 1-6. Because the generator starts from `SEED`, these 1000 rolls are the same on every run and every machine.

**Task B — count without a loop.** Set **`six_fraction`** to the fraction of
the 1000 rolls that came up 6. It should land near 1/6 ≈ 0.167.

```python
six_fraction = (rolls == 6).mean()
print(six_fraction)
```

The comparison builds a True/False mask and `.mean()` gives the fraction of True entries — the standard loop-free counting idiom.

**Task C — same seed, same numbers.** Build a *second* generator from the same
seed, draw **`rolls2`** (again 1000 dice rolls), and set **`same`** to
`np.array_equal(rolls, rolls2)`. Predict `same` before running.

```python
rng2 = np.random.default_rng(SEED)
rolls2 = rng2.integers(1, 7, size=1000)
same = np.array_equal(rolls, rolls2)
print(same)
```

A fresh generator built from the same seed restarts the same deterministic recipe from the same starting point, so it reproduces the draws exactly: `same` is True.

**Task D — different seed, different numbers.** Build a third generator from
seed `7`, draw **`rolls3`** the same way, and set **`differs`** to `True` if
`rolls3` is *not* identical to `rolls`.

```python
rng3 = np.random.default_rng(7)
rolls3 = rng3.integers(1, 7, size=1000)
differs = not np.array_equal(rolls, rolls3)
print(differs)
```

A different seed is a different starting point, so the sequence of draws diverges immediately.

**Task E — reproducible shuffle.** Make **`deck`** = `np.arange(10)`, copy it
to **`shuffled`** with `.copy()`, and shuffle the copy with a *fresh*
generator built from `SEED`. Check that `deck` itself is untouched and that
`shuffled` still contains every value 0-9 exactly once (hint: `np.sort`).

```python
deck = np.arange(10)
shuffled = deck.copy()
rng4 = np.random.default_rng(SEED)
rng4.shuffle(shuffled)
print("deck:    ", deck)
print("shuffled:", shuffled)
print("same contents:", np.array_equal(np.sort(shuffled), deck))
```

`shuffle` mixes the array **in place**, which is why we shuffle a copy. Sorting the shuffled copy recovers 0..9, showing the shuffle only reordered — and with a fixed seed, the reordering is the same every run.
