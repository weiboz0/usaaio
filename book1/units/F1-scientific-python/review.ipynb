## Concept summary

| Concept | One-line summary | Key API | Session |
| --- | --- | --- | --- |
| `numpy-arrays` | One shared dtype, a shape tuple per axis; whole-array math is fast and readable | `np.array`, `zeros`, `ones`, `full`, `arange`, `linspace`, `.dtype`, `.shape`, `.reshape`, `.astype` | 01 |
| `array-indexing-slicing` | `[row, col]`; slices are **views**; masks filter; index lists pick positions | `a[i, j]`, `a[1:3, ::2]`, `a[mask]`, `a[[2, 0]]`, `a[[i...], [j...]]`, `.copy()` | 01 |
| `elementwise-ops` | Arithmetic, comparisons, and math functions apply per element | `+ - * / **`, `np.sqrt`, `np.abs`, `np.where`, `np.maximum` | 02 |
| `broadcasting` | Compare shapes from the right; pad left with 1s; size-1 axes stretch; else ERROR | `a[:, None]`, `b[None, :]`, `reshape` | 02 |
| `vectorization` | Replace loops with whole-array operations (exam tasks may ban loops outright) | masks + `.sum()`, `np.where`, broadcasting, shifted slices | 02 |
| `aggregation-axis` | The axis you pass is the axis that disappears; `keepdims` keeps it as size 1 | `.sum/mean/max/min(axis=...)`, `argmax`, `keepdims=True` | 02 |
| `random-seeding` | One `SEED`, one generator, every draw from it → identical results every run | `np.random.default_rng(SEED)`, `.random`, `.integers`, `.choice`, `.shuffle` | 03 |
| `matplotlib-basics` | Line = ordered, scatter = observations, hist = value spread; always label | `plt.plot`, `scatter`, `hist`, `title`, `xlabel`, `ylabel`, `legend`, `show` | 03 |
| `seaborn-programming` | Pass arrays through named arguments onto an explicit axes; pin bins and test axes semantics | `sns.histplot`, `sns.scatterplot`, `ax=`, `hue=`, `style=`, `ax.patches`, `get_offsets` | 04 |

## Idiom sheet

The fifteen expressions this unit wants in your fingers:

| Goal | Idiom |
| --- | --- |
| how many pass a test | `(a > t).sum()` |
| what fraction pass | `(a > t).mean()` |
| keep the passers | `a[a > t]` |
| per-element if/else | `np.where(cond, x, y)` |
| elementwise larger-of-two | `np.maximum(a, b)` |
| floor/cap values | `np.maximum(a, lo)` / `np.minimum(a, hi)` |
| every pairing of two 1-D arrays | `a[:, None] OP b[None, :]` |
| one summary per column / per row | `g.OP(axis=0)` / `g.OP(axis=1)` |
| which position, not which value | `.argmax()` / `.argmin()` |
| divide rows by their totals | `g / g.sum(axis=1, keepdims=True)` |
| consecutive differences | `x[1:] - x[:-1]` |
| reproducible randomness | `SEED = ...; rng = np.random.default_rng(SEED)` |
| array histogram on explicit axes | `sns.histplot(x=values, bins=edges, ax=ax)` |
| grouped scatter semantics | `sns.scatterplot(x=x, y=y, hue=groups, style=groups, ax=ax)` |
| check histogram meaning | `[patch.get_height() for patch in ax.patches]` |

A quick demo of a few, runnable:

```python
g = np.arange(12).reshape(3, 4).astype(np.float64)
print((g > 5).sum(), (g > 5).mean())
print(g / g.sum(axis=1, keepdims=True))
print(np.arange(1, 5)[:, None] * np.arange(1, 5)[None, :])
```

## Self-quiz answers

<details>
<summary><b>Click to reveal all answers — only after writing yours down.</b></summary>

1. Shape `(2, 2)`, dtype `float64` — the single `4.0` promotes every element
   to the more general shared type.
2. `np.linspace(0, 2, 9)` — you know the *count*, and linspace includes both
   endpoints; decimal-step `arange` risks rounding surprises at the stop.
3. Rows 0 and 2, last column → `[4, 14]`.
4. Row 0 of `g` is now all `-1` — `r` was a **view**. Prevention:
   `r = g[0].copy()`.
5. `["small", "big", "big"]` — elementwise test, elementwise pick.
6. **(B)** — elementwise larger-of-each-pair against 0; it works fine on 2-D
   arrays too.
7. `(4, 1) + (6,)` → `(4, 6)` (both stretch); `(2, 3) + (2,)` → ERROR (3 vs 2
   from the right); `(5, 2) + (1, 2)` → `(5, 2)`.
8. `np.abs(a[:, None] - b[None, :])`
9. `(n % 3 == 0).mean()`
10. `m.mean(axis=1)`, shape `(12,)`; without `keepdims` the totals have shape
    `(12,)`, which compares its 12 against the 31 from the right and fails —
    `keepdims=True` gives `(12, 1)`, which broadcasts row-by-row.
11. One `SEED` constant at the top; one `rng = np.random.default_rng(SEED)`;
    every draw from that `rng`. `shuffle` reorders `deck` **in place** and
    returns `None` — shuffle a `.copy()` if you need the original.
12. A histogram; title, x-axis label, y-axis label, and a legend whenever two
    or more things share the axes.
13. Create `fig, ax`; pass arrays through named arguments; pass `ax=ax`; label
    with axes methods; return or inspect the axes and show only at the boundary.
14. Shared edges make corresponding bars count the same intervals; inspect
    heights through `ax.patches`.
15. `hue` maps group values to colors and builds legend semantics, while fixed
    color carries no group meaning; `style` adds marker-shape redundancy so the
    grouping survives grayscale and color-vision differences.

</details>

A code check for the computable ones (run after committing to answers):

```python
g = np.arange(20).reshape(4, 5)
assert np.array_equal(g[::2, -1], np.array([4, 14]))                        # Q3
assert list(np.where(np.array([1, 6, 3]) > 2, "big", "small")) == ["small", "big", "big"]  # Q5
assert (np.ones((4, 1)) + np.ones(6)).shape == (4, 6)                       # Q7
assert (np.ones((5, 2)) + np.ones((1, 2))).shape == (5, 2)                  # Q7
a = np.array([1.0, 5.0, 9.0]); b = np.array([2.0, 4.0, 6.0, 8.0])
assert np.abs(a[:, None] - b[None, :]).shape == (3, 4)                      # Q8
n = np.arange(10)
assert np.isclose((n % 3 == 0).mean(), 0.4, atol=1e-12, rtol=0)                                 # Q9
m = np.ones((12, 31))
assert m.mean(axis=1).shape == (12,)                                        # Q10
print("all quiz checks passed")
```
