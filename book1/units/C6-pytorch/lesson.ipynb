## Unit goals

By the end of this unit you can:

- create and manipulate **torch tensors** with your F1 NumPy reflexes —
  shapes, dtypes, broadcasting, `dim=` aggregation, seeded randomness
  (`torch.manual_seed(20260804)`), and the `from_numpy`/`.numpy()`
  bridge with explicit dtype casts at every border;
- manage the **float32/float64 border**: torch's float32 default vs
  NumPy's float64, the course's `torch.set_default_dtype(torch.float64)`
  convention, and honest tolerances (`atol=1e-6`/`rtol=1e-5`) whenever
  float32 is deliberate;
- read and write **Python classes and inheritance**: `__init__`,
  `self`, methods, subclassing, method overriding, `super().__init__()`
  and what breaks without it — in plain Python first, then as the
  anatomy of `nn.Module`;
- build **custom layers** by subclassing `nn.Module`: the pinned
  `DenseLayer` (weight `(out, in)`, bias `(out,)`, forward
  `x @ weight.T + bias`) and `ThresholdGate`
  (`(x >= 0).to(x.dtype)`) — C5's `affine_layer` and `step_activation`
  reborn as modules;
- register numbers correctly with
  **`nn.Parameter(..., requires_grad=False)`**, explain what plain
  tensor attributes lose (visibility in `parameters()` and
  `state_dict`), and state what the flag signals;
- compose **inference-only MLPs with manual weights** — submodule
  registration, dotted parameter names, `nn.Sequential` vs custom
  classes — and design weights by hand so a network implements a
  written specification;
- **count parameters** by hand ($\text{out} \times (\text{in} + 1)$
  per dense layer, gates free), with `numel`, *without* `numel` (the
  exam's ban register), and backwards — from a parameter budget to a
  width via Calc AB algebra;
- **inspect** any module: `parameters()`, `named_parameters()`,
  `state_dict` keys and shapes, `load_state_dict` for swapping in
  hand-set weights.

**What this unit deliberately does not do: adjust weights.**
Every network here is built by hand and run *forward* only —
inference engineering, exactly the register of the real paper's torch
problems (see Session 3's Exam Connections).
C5's scope note carries over unchanged, and Session 2 §3 contains this
unit's one-sentence acknowledgement of what full PyTorch can
additionally do and why we construct every parameter with
`requires_grad=False`.

## Practice sets

Twenty problems in `practice/`, organized as three sets (the manifest
carries the same grouping).
Everything respects the unit's fence: weights are designed, registered,
composed, counted, and inspected — never adjusted.

**Set A — fundamentals** (start after Session 1; `p20` after
Session 3):

- `p01` (MC, intro) — `dim=` aggregation and integer dtypes.
- `p02` (MC, intro) — a plain-Python inheritance trace.
- `p05` (coding, intro) — seeded tensor statistics under loop/NumPy
  bans.
- `p06` (coding, intro) — subclassing a supplied base class,
  inheriting shared behavior via `super().__init__`.
- `p19` (drill, intro) — rapid-fire tensor one-liners, including the
  NumPy bridge.
- `p20` (drill, intro) — rapid-fire module inspection on a supplied
  layer (count *without* `numel`).

**Set B — exam register** (after Session 2; `p04`, `p11`, `p12` after
Session 3):

- `p03` (MC, core) — what registers, and with which `requires_grad`
  flags.
- `p04` (MC, numeric normal form, core) — a parameter count decoded
  as $2^a b$.
- `p07` (coding, core) — a parameter-free module from an exact
  contract, comparison-only tools.
- `p08` (coding, core) — the pinned `DenseLayer` built from its
  written spec, plus a frozen-flag audit.
- `p09` (coding, core) — manual weights that realize a stated affine
  map and a half-plane detector.
- `p10` (coding, advanced) — a submodule-registered region detector
  with hand-set weights, `nn.Sequential` banned.
- `p11` (proof, core) — derive $\text{out}\times(\text{in}+1)$ and
  the stack formula; numeric anchor $4 \to 7 \to 3 = 59$.
- `p12` (proof, advanced) — width from a 226-parameter budget via a
  strictly-increasing quadratic; torch anchor.

**Set C — integration and challenge** (closest to real exam texture —
save for last):

- `p13` (integrative, core) — the same network in NumPy and as
  modules, reconciled to zero gap.
- `p14` (integrative, advanced) — build a three-parameter layer to
  spec, then audit a storage-defective rewrite of it.
- `p15` (scenario, core) — a rule-specified alarm network: weights as
  the complete program, no data anywhere.
- `p16` (scenario, core) — the "will the weights drift?" worry,
  settled with evidence and brought up to the frozen-flag convention.
- `p17` (challenge, advanced) — an inheritance template inside torch:
  a base module owning `forward`, subclasses supplying scores.
- `p18` (challenge, advanced) — a square-region detector from four
  half-planes; then move the region by editing only a bias vector.

Every problem states its own contracts (identifiers, shapes, bans).
Solve first, then compare with the solution notebook; do not read
