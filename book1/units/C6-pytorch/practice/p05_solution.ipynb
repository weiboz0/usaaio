# C6-pytorch — Practice p05 — Solution


Each reduction names the dimension to collapse.  Column centering uses
the `(4,)` mean vector directly; broadcasting subtracts it from every
row without a loop.


```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)

torch.manual_seed(20260804)
M = torch.randn(6, 4)

col_means = M.mean(dim=0)
row_maxes = M.max(dim=1).values
n_negative = int((M < 0).sum())
centered = M - col_means
check_zero = float(centered.mean(dim=0).abs().max())

col_means, row_maxes, n_negative, check_zero
```

### Answer check


```python
expected_col_means = torch.tensor([0.4927029559216754, -0.5448468626470716,
                                   -0.19686409776395294, -0.26718179728830627])
expected_row_maxes = torch.tensor([1.3354632410873106, 1.3713398626679465,
                                   0.01640254071834947, 2.577352603481901,
                                   0.8570473048799313, 0.7964284212824997])
assert torch.allclose(col_means, expected_col_means, atol=1e-12, rtol=0.0)
assert torch.allclose(row_maxes, expected_row_maxes, atol=1e-12, rtol=0.0)
assert isinstance(n_negative, int) and n_negative == 13
assert centered.shape == (6, 4)
assert check_zero < 1e-12
assert torch.allclose(centered.mean(dim=0), torch.zeros(4), atol=1e-12, rtol=0.0)
```
