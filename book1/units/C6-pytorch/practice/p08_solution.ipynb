# C6-pytorch — Practice p08 — Solution


Registering the `(out, in)` weight and `(out,)` bias directly makes
them visible to module inspection.  The transpose aligns each input
row with each output unit's weight row, and broadcasting adds the bias.


```python
import torch
import torch.nn as nn

torch.set_default_dtype(torch.float64)

class DenseLayer(nn.Module):
    """Session 2's pinned dense layer: weight (out, in), bias (out,)."""

    def __init__(self, weight, bias):
        super().__init__()
        self.weight = nn.Parameter(torch.as_tensor(weight), requires_grad=False)
        self.bias = nn.Parameter(torch.as_tensor(bias), requires_grad=False)

    def forward(self, x):
        return x @ self.weight.T + self.bias


W = torch.tensor([[1.0, -1.0],
                  [0.5, 2.0],
                  [0.0, 1.0]])
b = torch.tensor([0.0, -1.0, 2.0])
x = torch.tensor([[2.0, 1.0],
                  [-1.0, 3.0]])

layer = DenseLayer(W, b)
y = layer(x)
n_tensors = len(list(layer.parameters()))
all_frozen = all(not p.requires_grad for p in layer.parameters())

y, n_tensors, all_frozen
```

### Answer check


```python
expected_y = torch.tensor([[1.0, 2.0, 3.0], [-4.0, 4.5, 5.0]])
assert torch.equal(y, expected_y)
assert tuple(layer.weight.shape) == (3, 2)
assert tuple(layer.bias.shape) == (3,)
assert n_tensors == 2 and all_frozen
```
