# C6-pytorch — Practice p04 — Solution


The three dense-layer counts are
$6(3+1)=24$, $4(6+1)=28$, and $2(4+1)=10$; gates contribute zero.
Hence $N=24+28+10=62=2^1\cdot31$, so $a+b=1+31=32$ and the answer
is **C**.

A (26) is the forget-the-biases trap — the most important one here:
weights alone give $18+24+8=50=2^1\cdot25$, so $a+b=26$.
B reports only the odd factor $b=31$.  D reports $N=62$ rather than
$a+b$, and E adds one to $N$ instead of adding the exponent to the odd
factor.


```python
import torch

torch.set_default_dtype(torch.float64)

layer_counts = (6 * (3 + 1), 4 * (6 + 1), 2 * (4 + 1))
N = sum(layer_counts)
a, b = 1, 31
decoded = a + b
answer_p04 = "C"
answer_p04, layer_counts, N, decoded
```

### Answer check


```python
assert answer_p04 == "C"
assert layer_counts == (24, 28, 10)
assert N == (2 ** a) * b == 62
assert b % 2 == 1 and decoded == 32
```
