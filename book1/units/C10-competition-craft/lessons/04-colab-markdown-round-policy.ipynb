# C10-competition-craft — Session 04: Colab, Markdown, and Round Policy

*One class session, roughly 85 minutes. Prerequisites: Sessions 1–3.*

This session turns a correct modeling notebook into a correct Colab submission: choose text versus code cells deliberately, render code and mathematics in Markdown, satisfy the coding/file contract, and state the Round 1 CPU versus Round 2 L4/GPU boundary exactly. Checkpoint answers are collected at the end.

## 1. Two cell types, two jobs

A **code cell** executes Python and creates names, files, figures, and printed evidence. A **text cell** renders Markdown: headings, prose, lists, code excerpts, and mathematics. A fenced excerpt inside a text cell is evidence for a reader; it does not execute and cannot define the required function.

A disciplined answer alternates roles: text states the claim, code produces the evidence, text interprets the evidence. In Colab, verify the cell-type selector before writing; changing cell type after pasting can silently turn prose into invalid Python or make required code non-executable.

## 2. A repeatable Colab authoring workflow

Use this sequence for every mixed-cell response:

1. outline the required deliverables as headings in text cells;
2. insert code cells directly below the claims they support;
3. run each code cell once while drafting, then remove dead experiments;
4. preview every text cell and repair malformed fences or mathematics;
5. Restart and Run All from a fresh runtime;
6. inspect required identifiers, outputs, and created files;
7. download the `.ipynb`, confirm the filename, and open the downloaded copy far enough to verify that the final edits are present.

The downloaded file—not the state of a browser tab—is the submission artifact.

### Checkpoint 2

1. Why should deliverable headings be outlined before code is written?

## 3. Fenced code snippets

A code excerpt in a text cell uses matching triple backticks and a language label:

```python
def predict_labels(X_test):
    return final_model.predict(X_test)
```

The opening fence, language label, body, and closing fence are all part of the authoring contract. Backticks placed on the same line as a multi-line function do not create a block; a missing closing fence can swallow everything below it. Render the cell and inspect the result.

**Worked repair 1.** The malformed text ````python def f(x): return x + 1`` has too many opening marks, no line break, and no matching close. Repair it as a three-line fenced block, then keep the executable definition in a separate code cell.

## 4. Inline and display mathematics

Use inline math for symbols inside a sentence, such as `$F_1$ balances precision and recall.` Use display math for a formula that deserves its own line:

$$
F_1 = \frac{2PR}{P+R}.
$$

Then define every symbol in prose and connect the formula to the scored quantity. Rendering a formula without interpretation is typesetting, not reasoning.

**Worked repair 2.** The text `Macro F1 = 1/K sum F1_k` is ambiguous. A complete repair renders `$K$` inline, displays `F_{1,\mathrm{macro}} = \frac{1}{K}\sum_{k=1}^{K} F_{1,k}`, and states that each class receives equal weight.

## 5. The mixed-cell submission contract

A submission is more than readable. Its executable surface must still satisfy the pinned contract: required imports, `SEED`, data load, frozen validation, bounded model selection, final refit, exact `predict_labels`, self-checks, and required output file or object. Text cells must explain the approach and show requested snippets/formulae without replacing executable cells.

Preflight in this order:

- Restart and Run All with no manual cell skipping;
- confirm required names exist and have the right types;
- run the mid-table prediction checks;
- confirm every claimed number appears in a visible output;
- confirm required files exist at the exact names and paths;
- download the `.ipynb` and verify the downloaded artifact.

## 6. The exact round boundary

**Round 1 is CPU-only.** Do not enable or claim a GPU accelerator for a Round 1 submission. A sentence such as “we trained on a GPU in Round 1” is a policy error even if the code would also run on a CPU.

**Round 2 allows the Colab L4/GPU environment.** That is the boundary to state; this session does not teach the later GPU workflow, device placement, or accelerator-specific optimization. “GPU is always allowed because Colab offers it” and “GPU is never allowed” are both wrong.

A correct declaration is short and auditable: `Round 1: CPU only. Round 2: Colab L4/GPU is allowed.`

## 7. Full worked audit

Suppose a response contains: one giant code cell, a prose paragraph pasted as Python comments, a malformed `predict_labels` fence, the literal text `F1 = 2PR/(P+R)`, a passing in-memory model, no fresh run, no downloaded file check, and “Round 1 GPU used.” Audit in six rows:

| Row | Finding | Repair | Evidence after repair |
|---|---|---|---|
| writeup | reasoning is hidden in comments | move approach and intuition to a text cell | rendered paragraph tied to printed score |
| authoring | cell roles are not deliberate | separate text and code by purpose | visible mixed-cell structure |
| fenced code | fence is malformed | use matched `python` fences | rendered excerpt plus executable function |
| Markdown math | formula is unrendered and unexplained | display the formula and define symbols | rendered equation plus interpretation |
| coding submission | interactive state is unproved | Restart and Run All, check names/files, download | clean run and verified `.ipynb` |
| round policy | Round 1 GPU claim is illegal | state CPU-only and contrast Round 2 L4/GPU | exact declaration |

This table is the scoring structure used in p15, p17, and p18. Each row needs its own artifact; a passing mention cannot earn another row's point.

### Checkpoint 7

2. Which row preserves the existing model-selection and prediction-function responsibilities?

## 8. Pitfalls, exam connections, and what comes next

Common failures are wrong cell type, unmatched fences, math delimiters that never render, required code shown only as prose, stale runtime state, wrong filename, and a round-policy slogan with no exact boundary. Repair them before modeling polish: a strong model cannot rescue a broken artifact.

Round 1's applied response grades the runnable notebook and the reasoning packaged around it. The exam connection is therefore literal: cell structure, exact identifiers, reproducible execution, visible evidence, file delivery, and policy compliance are scored work. Practice p15 isolates authoring repair; p17 asks for a complete mixed-cell campaign; p18 asks you to audit a flawed one.

Going deeper points forward to Round 2 material. Remember only the allowance boundary here; do not infer device-management rules that have not been taught.
