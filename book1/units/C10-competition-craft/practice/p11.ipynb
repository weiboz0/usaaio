Macro-F1 versus pooled F1 — why the averaging convention is the whole
story.

For a single-label classification task, define the **pooled** (micro)
F1 by summing the one-vs-rest counts over the classes before forming
the metric:
$TP_{\text{pool}} = \sum_k TP_k$,
$FP_{\text{pool}} = \sum_k FP_k$,
$FN_{\text{pool}} = \sum_k FN_k$, then
$\text{prec}_{\text{pool}} = \frac{TP_{\text{pool}}}{TP_{\text{pool}} + FP_{\text{pool}}}$,
$\text{rec}_{\text{pool}} = \frac{TP_{\text{pool}}}{TP_{\text{pool}} + FN_{\text{pool}}}$,
and pooled F1 their harmonic mean.
