# C10-competition-craft — Practice p19 — Solution

Both 187 and 323 contain the factor 17: $187=17\cdot11$ and $323=17\cdot19$. Therefore the normalized fraction is $-11/19$, with coprime numerator and positive denominator, and decoding only after that reduction gives $19-(-11)=30$.

```python
p = -11
q = 19
decoded = q - p
```

### Answer check

```python
assert (p, q) == (-11, 19)
assert decoded == 30
assert q > 0
```
