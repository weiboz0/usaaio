# C10-competition-craft — Practice p11 — Solution

### (a) Claim 1 — pooled F1 equals accuracy

In a single-label task, a correct row lies on exactly one class diagonal, so it contributes one (TP_k) and no pooled error. A wrong row has one predicted class and one different actual class: it contributes one (FP_k) to the predicted class and one (FN_k) to the actual class. Thus, if (c) rows are correct and (w) are wrong,

\[
TP_{pool}=c,\qquad FP_{pool}=FN_{pool}=w.
\]

Both pooled precision and recall are therefore (c/(c+w)), which is accuracy. The harmonic mean of two identical values is that same value, so pooled F1 is accuracy.

### (b) Claim 2 — exact per-class scores

For `struggles`, (TP=5), (FP=4), and (FN=15). Hence precision is (5/(5+4)=5/9), recall is (5/(5+15)=1/4), and

\[
F1_s=\frac{2(5/9)(1/4)}{5/9+1/4}
=\frac{5/18}{29/36}=\frac{10}{29}.
\]

For `thrives`, (TP=76), (FP=15), and (FN=4). Precision is (76/91), recall is (76/80=19/20), and

\[
F1_t=\frac{2(76/91)(19/20)}{76/91+19/20}
=\frac{8}{9}.
\]

Therefore

\[
F1_{macro}=\frac12\left(\frac{10}{29}+\frac89\right)
=\frac12\left(\frac{90+232}{261}\right)=\frac{161}{261}.
\]

### (c) Claim 3 — direction and mechanism

There are (5+76=81) correct rows, so pooled F1 equals accuracy (81/100). Cross-multiplication gives (16100<21141), hence (161/261<81/100). Pooled F1 weights all 100 rows equally, letting the 80 well-served majority rows dominate; macro-F1 instead gives each class half the score, so the weak minority score (10/29) cannot be diluted.

```python
import numpy as np

y_true = np.array(["struggles"] * 20 + ["thrives"] * 80)
y_pred = np.array(
    ["struggles"] * 5 + ["thrives"] * 15
    + ["struggles"] * 4 + ["thrives"] * 76
)

s_true = y_true == "struggles"
s_pred = y_pred == "struggles"
t_true = y_true == "thrives"
t_pred = y_pred == "thrives"

tp_s = (s_true & s_pred).sum()
fp_s = (~s_true & s_pred).sum()
fn_s = (s_true & ~s_pred).sum()
f1_s = 2 * tp_s / (2 * tp_s + fp_s + fn_s)

tp_t = (t_true & t_pred).sum()
fp_t = (~t_true & t_pred).sum()
fn_t = (t_true & ~t_pred).sum()
f1_t = 2 * tp_t / (2 * tp_t + fp_t + fn_t)
macro_f1 = float((f1_s + f1_t) / 2)

tp_pool = tp_s + tp_t
fp_pool = fp_s + fp_t
fn_pool = fn_s + fn_t
pool_prec = tp_pool / (tp_pool + fp_pool)
pool_rec = tp_pool / (tp_pool + fn_pool)
pooled_f1 = float(2 * pool_prec * pool_rec / (pool_prec + pool_rec))
(f1_s, f1_t, macro_f1, pooled_f1)
```

### Answer check

```python
assert (tp_s, fp_s, fn_s) == (5, 4, 15)
assert (tp_t, fp_t, fn_t) == (76, 15, 4)
assert np.isclose(f1_s, 10/29, atol=1e-12, rtol=0)
assert np.isclose(f1_t, 8/9, atol=1e-12, rtol=0)
assert np.isclose(macro_f1, 161/261, atol=1e-12, rtol=0)
assert np.isclose(pooled_f1, 81/100, atol=1e-12, rtol=0)
assert macro_f1 < pooled_f1
```
