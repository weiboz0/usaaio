# C10-competition-craft — Practice p07 — Solution

```python
import numpy as np

y_true = np.array(["struggles", "thrives", "thrives", "struggles", "thrives",
                   "struggles", "thrives", "thrives", "struggles", "thrives",
                   "thrives", "struggles", "thrives", "thrives", "struggles",
                   "thrives", "thrives", "thrives", "thrives", "thrives"])
y_pred = np.array(["struggles", "thrives", "struggles", "struggles", "thrives",
                   "thrives", "thrives", "thrives", "struggles", "struggles",
                   "thrives", "struggles", "thrives", "thrives", "thrives",
                   "thrives", "struggles", "thrives", "thrives", "thrives"])

labels = np.unique(y_true)
actual_masks = y_true[:, None] == labels[None, :]
predicted_masks = y_pred[:, None] == labels[None, :]
C = (actual_masks[:, :, None] & predicted_masks[:, None, :]).sum(axis=0).astype(int)
diag = np.diag(C).astype(float)
prec = diag / C.sum(axis=0)
rec = diag / C.sum(axis=1)
f1 = 2 * prec * rec / (prec + rec)
f1_macro = float(f1.mean())
acc = float((y_true == y_pred).mean())
(C, prec, rec, f1, f1_macro, acc)
```

Column totals count predictions and therefore form precision denominators; row totals count actual labels and form recall denominators. The two class F1 values are (8/13) and (22/27), whose mean is (251/351).

### Answer check

```python
assert np.array_equal(C, np.array([[4, 2], [3, 11]]))
assert np.allclose(prec, np.array([4/7, 11/13]), atol=1e-12, rtol=0)
assert np.allclose(rec, np.array([2/3, 11/14]), atol=1e-12, rtol=0)
assert np.allclose(f1, np.array([8/13, 22/27]), atol=1e-12, rtol=0)
assert np.isclose(f1_macro, 251/351, atol=1e-12, rtol=0)
assert acc == 0.75
```
