# C10-competition-craft — Practice p05 — Solution

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

SEED = 20260804

df = pd.read_csv("../data/train.csv")
FEATURES = [c for c in df.columns if c != "outcome"]
X = df[FEATURES]
y = df["outcome"].to_numpy()

pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5)),
]).fit(X, y)


def predict_labels(X_test):
    return pd.Series(pipe.predict(X_test), index=X_test.index)


probe = X.iloc[200:225]
probe_out = predict_labels(probe)
contract_ok = bool(
    isinstance(probe_out, pd.Series)
    and len(probe_out) == len(probe)
    and probe_out.index.equals(probe.index)
    and set(probe_out.unique()) <= set(np.unique(y))
)
contract_ok
```

The pipeline is fitted once on all labeled rows. The contract function only predicts and wraps the labels with the input index, so it is length-agnostic and preserves row identity.
