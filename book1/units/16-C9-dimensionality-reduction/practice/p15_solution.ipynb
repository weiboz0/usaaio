# C9-dimensionality-reduction — Practice p15 — Solution

```python
import numpy as np

SEGMENTS = ["A", "B", "C", "D"]
D_cent = np.array([
    [0.0, 14.2, 9.8, 21.5],
    [14.2, 0.0, 12.1, 16.4],
    [9.8, 12.1, 0.0, 18.9],
    [21.5, 16.4, 18.9, 0.0],
], dtype=np.float64)

verdicts = {"a": "unsupported", "b": "supported",
            "c": "unsupported", "d": "unsupported"}
candidate_distances = D_cent[0, 1:]
true_nearest = SEGMENTS[int(np.argmin(candidate_distances)) + 1]
claim_a_holds = bool(true_nearest == "B")
print("verdicts:", verdicts)
print("A distances to B/C/D:", candidate_distances)
print("true nearest / claim holds:", true_nearest, claim_a_holds)
```

Claim (a) is unsupported because global map distances are not quantitative; the original-space centroid distances are $14.2$ to B, $9.8$ to C, and $21.5$ to D, so C—not B—is actually nearest A.  Claim (c) is unsupported because apparent group sizes and densities can be distorted.  Claim (d) is unsupported because settings and random initialization can change the layout, so position-by-position comparison is not stable.  Claim (b) is the supported local-neighborhood inference.

### Answer check

```python
assert verdicts == {"a": "unsupported", "b": "supported", "c": "unsupported", "d": "unsupported"}
assert true_nearest == "C"
assert not claim_a_holds
assert np.isclose(D_cent[0, 2], 9.8, atol=1e-12, rtol=0)
```
