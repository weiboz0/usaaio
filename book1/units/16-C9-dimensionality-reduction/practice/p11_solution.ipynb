# C9-dimensionality-reduction — Practice p11 — Solution

**(a) Scores are centered.**  For feature $j$,
$$\frac1n\sum_{k=1}^n(X_{kj}-\mu_j)=\frac1n\sum_kX_{kj}-\mu_j=\mu_j-\mu_j=0.$$
Thus $\mathbf 1^{\mathsf T}X_c=0$.  For any $u$, linearity gives the mean of its scores as $\frac1n\mathbf 1^{\mathsf T}X_cu=0$.

**(b) Component variances.**  The scores $X_cv_i$ are centered by (a), so their sample variance is
$$\frac{\|X_cv_i\|^2}{n-1}=\frac{\|\sigma_i u_i\|^2}{n-1}=\frac{\sigma_i^2\|u_i\|^2}{n-1}=\frac{\sigma_i^2}{n-1}.$$
The second equality is the SVD action, and the last uses $\|u_i\|=1$.

**(c) Total decomposition.**  Feature $j$ contributes $\frac1{n-1}\sum_k(X_c)_{kj}^2$.  Summing over $j$ gives
$$\text{total} = \frac1{n-1}\sum_{k,j}(X_c)_{kj}^2=\frac{\|X_c\|_F^2}{n-1}.$$
The Frobenius–singular-value identity then yields $\text{total}=\sum_i\sigma_i^2/(n-1)$, exactly the sum of the component variances from (b).  Dividing each component by this total gives $\mathrm{evr}_i=\sigma_i^2/\sum_j\sigma_j^2$, hence $\sum_i\mathrm{evr}_i=1$.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
B = rng.normal(0.0, 1.0, (40, 4))
M = rng.normal(0.0, 0.8, (4, 6))
X = rng.normal(0.0, 2.0, 6) + B @ M + rng.normal(0.0, 0.3, (40, 6))
n = X.shape[0]
Xc = X - X.mean(axis=0)
_, s, Vt = np.linalg.svd(Xc, full_matrices=False)
total_var = float(X.var(axis=0, ddof=1).sum())
total_var_svd = float((s**2).sum() / (n - 1))
evr = s**2 / (s**2).sum()
evr_sum = float(evr.sum())
var0 = float(np.var(Xc @ Vt[0], ddof=1))

assert abs(total_var - total_var_svd) < 1e-9
assert np.isclose(evr_sum, 1.0, atol=1e-12, rtol=0)
assert abs(var0 - s[0]**2 / 39) < 1e-9
assert np.isclose(total_var, 17.7997, atol=1e-4, rtol=0)
assert np.isclose(evr[0], 0.5009, atol=1e-4, rtol=0)
print("total variance:", total_var)
print("EVR:", evr)
print("first score variance:", var0)
```
