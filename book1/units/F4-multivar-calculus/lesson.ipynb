## Unit goals

By the end of this unit you can:

- compute partial derivatives by hand as slice-derivatives — freeze every
  other input and differentiate with the ordinary Calc AB rules;
- assemble the gradient as the vector of all partials, and use its
  direction property: it points toward steepest increase, and its length
  is the rate of that increase;
- state and apply the multivariable chain rule as nudge accounting —
  rates multiply along a route, and every route from input to output
  contributes a term;
- derive $\tanh'(x) = 1 - \tanh^2(x)$ from the exponential definition,
  and use $0 < \tanh' \le 1$ (maximum at 0, saturation far away);
- hand-derive gradients of component-form sum-of-squares scores,
  $\partial Q/\partial w_j = -\frac{2}{N}\sum_n r_n X_{n,j}$, and
  implement them with broadcasting and axis sums only — no `np.dot`,
  no shortcut products, no loops;
- verify any hand-derived gradient numerically with central differences,
  choosing a sensible step size $h$.
