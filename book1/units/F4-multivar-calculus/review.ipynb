## Concept summary

| Concept | One-line summary | Key fact to retain |
|---|---|---|
| Partial derivatives | Freeze every other input, differentiate the slice with ordinary AB rules | $\dfrac{\partial}{\partial w_j}\sum_k w_k^2 = 2w_j$ — only terms containing $w_j$ survive |
| Gradient | The vector of all partials, one entry per input | points steepest uphill; uphill rate $= \lVert\nabla f\rVert$; rate along unit $u$ is $\nabla f \cdot u$ |
| Multivariable chain rule | Rates multiply along a route; routes add | $\dfrac{d}{dt}f(r(t)) = \sum_k \dfrac{\partial f}{\partial p_k}\, r_k'(t) = \nabla f \cdot r'(t)$ |
| tanh derivative | Quotient rule on the exponential definition | $\tanh' = 1 - \tanh^2 \in (0, 1]$; max $1$ at $x = 0$; $\to 0$ far away (saturation) |
| Sum-of-squares gradients | Chain through the square; inner derivative $-X_{n,j}$ | $\dfrac{\partial Q}{\partial w_j} = -\dfrac{2}{N}\sum_n r_n X_{n,j}$, $\;r_n = y_n - \sum_k X_{n,k}w_k$ |

## Derivative rules and finite-difference idiom sheet

**By hand (the forms the exam's normal forms expect):**

- Partials: differentiate one variable, others frozen — all AB rules
  apply per variable; inner derivatives are never optional
  ($\partial_b (a + 2b)^3 = 3(a+2b)^2 \cdot 2$).
- Scan-the-sum: $\partial/\partial w_j$ of $\sum_k c_k w_k$ is $c_j$;
  of $\sum_k w_k^2$ is $2w_j$.
- Gradient direction: steepest increase along
  $\nabla f / \lVert\nabla f\rVert$ at rate $\lVert\nabla f\rVert$;
  steepest decrease along $-\nabla f$; rate 0 orthogonal to $\nabla f$.
- Chain rule: one term per route,
  $\sum_k (\partial f/\partial p_k)\, r_k'(t)$.
- tanh: $\tanh' = 1 - \tanh^2 = 4/(e^x + e^{-x})^2$; $\tanh'(0) = 1$.
- Squashed weighted sum: $\partial_j \tanh(\sum_k a_k w_k)
  = (1 - \tanh^2 s)\, a_j$ — one shared slope factor.
- Sum of squares: $\partial Q/\partial w_j = -\frac{2}{N}\sum_n r_n
  X_{n,j}$; with an added penalty $\lambda\sum_k w_k^2$, add
  $2\lambda w_j$.
- Two-coefficient case: $\partial_{w_0} = -\frac{2}{N}\sum_n r_n$,
  $\;\partial_{w_1} = -\frac{2}{N}\sum_n r_n h_n$.

**NumPy idioms (central differences and banned-API-safe gradients):**

```python
(f(x + h, y) - f(x - h, y)) / (2 * h)      # central-difference partial
# full gradient checker: nudge ONE coordinate at a time (loop allowed
# in checkers): step = zeros; step[j] = h; (f(p+step)-f(p-step))/(2h)

pred = (X * w).sum(axis=1)                 # row-wise weighted sums  (N,)

grad = -(2 / N) * (r[:, None] * X).sum(axis=0)   # (d,) — no np.dot,
                                                 # no .T, no loops
np.tanh(s), 1 - np.tanh(s)**2              # value and slope factor
```

well-scaled inputs — bigger pays curvature error, smaller drowns in
float rounding.
