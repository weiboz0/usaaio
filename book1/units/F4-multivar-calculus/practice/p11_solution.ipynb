# F4-multivar-calculus — Practice p11 — Solution

**(a) Deriving the identity.** Write $N = e^x - e^{-x}$ and $D = e^x + e^{-x}$, so $\tanh x = N/D$. Note $N' = D$ and $D' = N$. By the quotient rule,

$$\tanh'(x) = \frac{N'D - N D'}{D^2} = \frac{D^2 - N^2}{D^2}.$$

Since $D^2 - N^2 = (D-N)(D+N) = (2e^{-x})(2e^{x}) = 4$, this equals $4/D^2$. On the other hand $1 - \tanh^2 x = (D^2 - N^2)/D^2 = 4/D^2$ as well, proving

$$\tanh'(x) = 1 - \tanh^2(x).$$

**(b) The bounds $0 < \tanh'(x) \le 1$.** From (a), $\tanh'(x) = 1 - \tanh^2(x)$. Since $\tanh^2(x) \ge 0$ for every real $x$, we get $\tanh'(x) \le 1$, with equality exactly when $\tanh(x) = 0$, i.e. only at $x = 0$. And since $|N| < D$ for every real $x$ (both $e^x$ and $e^{-x}$ are positive, so the difference is strictly smaller in magnitude than the sum), we have $\tanh^2(x) < 1$ and hence $\tanh'(x) > 0$. Together: $0 < \tanh'(x) \le 1$, with equality only at $x = 0$.

```python
import numpy as np

x = 0.7
h = 1e-6
direct = (np.tanh(x + h) - np.tanh(x - h)) / (2*h)
identity_val = 1.0 - np.tanh(x)**2
anchor_gap = abs(direct - identity_val)
```

At $x=0.7$, the identity predicts a positive derivative below one, consistent with the proof. The central difference provides an independent numerical sanity check of the algebra.
