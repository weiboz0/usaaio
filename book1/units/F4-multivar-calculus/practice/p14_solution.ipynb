# F4-multivar-calculus — Practice p14 — Solution

Let $z_n=\sum_kX_{n,k}w_k$, $\hat y_n=\tanh z_n$, and $r_n=y_n-\hat y_n$. Applying both chain rules gives $\partial L/\partial w_j=-(2/N)\sum_n r_n(1-\tanh^2z_n)X_{n,j}$.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (50, 3))
y = rng.normal(0, 1, 50)

def l_gradient(X, y, w):
    z = (X * w).sum(axis=1)
    prediction = np.tanh(z)
    weighted_residual = (y - prediction) * (1.0 - prediction**2)
    return (-2.0 / X.shape[0]) * (weighted_residual[:, None] * X).sum(axis=0)

def l_value(candidate):
    prediction = np.tanh((X * candidate).sum(axis=1))
    return ((y - prediction)**2).mean()

w = np.array([0.2, -0.3, 0.1])
grad = l_gradient(X, y, w)
h = 1e-6
numeric = np.array([
    (l_value(w + np.array([h, 0.0, 0.0])) - l_value(w - np.array([h, 0.0, 0.0]))) / (2*h),
    (l_value(w + np.array([0.0, h, 0.0])) - l_value(w - np.array([0.0, h, 0.0]))) / (2*h),
    (l_value(w + np.array([0.0, 0.0, h])) - l_value(w - np.array([0.0, 0.0, h]))) / (2*h)
])
check_gap = np.abs(grad - numeric).max()
```

The factor $1-\tanh^2z_n$ shrinks contributions when $|z_n|$ is large. This matches the p11(b) bound plus the saturation behaviour (tanh′ → 0 for large |input|) from lesson 02: the derivative approaches zero far from the origin.

### Answer check

```python
assert grad.shape == (3,)
assert np.allclose(grad, numeric, rtol=0.0, atol=1e-8)
assert check_gap < 1e-8
```
