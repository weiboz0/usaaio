# F4-multivar-calculus — Practice p04 — Solution

Differentiation gives $Q\'(w)=-2\sum_n x_n(y_n-wx_n)$, which at $w=2$ equals $-6$. Its normal-form encoding is $-1+6=5$, so the correct choice is B.

```python
import numpy as np

x = np.array([1, 2])
y = np.array([3, 5])
w = 2
q_prime = -2 * np.sum(x * (y - w * x))
normal_value = int(np.sign(q_prime) + abs(q_prime))
answer_p04 = "B"
```

### Answer check

```python
my_answer_letter = answer_p04
assert q_prime == -6
assert normal_value == 5
assert my_answer_letter == "B"
```
