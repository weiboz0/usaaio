# C9-dimensionality-reduction — Practice p06 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
A = rng.normal(0.0, 1.5, (9, 5)).astype(np.float64)

U, s, Vt = np.linalg.svd(A, full_matrices=False)
shapes_ok = bool(U.shape == (9, 5) and s.shape == (5,) and Vt.shape == (5, 5))
desc_ok = bool(np.all(np.diff(s) <= 0.0))
recon_gap = float(np.max(np.abs(U @ np.diag(s) @ Vt - A)))
fro_gap = float(abs((A * A).sum() - (s * s).sum()))
assert recon_gap < 1e-12
assert fro_gap < 1e-9

print("singular values:", s)
print("reconstruction / Frobenius gaps:", recon_gap, fro_gap)
```

The thin factor shapes are $(9,5)$, $(5,)$, and $(5,5)$.  The singular values already descend, reconstruction is accurate to $3.11\times10^{-15}$, and the two squared-Frobenius routes differ by only $1.42\times10^{-14}$.
