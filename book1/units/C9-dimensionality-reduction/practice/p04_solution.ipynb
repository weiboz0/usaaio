# C9-dimensionality-reduction — Practice p04 — Solution

```python
from fractions import Fraction

sigma2 = (25, 9, 1)
first_ratio = Fraction(sigma2[0], sum(sigma2))
p_plus_q = first_ratio.numerator + first_ratio.denominator
answer_p04 = "A"
```

The answer is **A**.  The first variance share is
$$\frac{5^2}{5^2+3^2+1^2}=\frac{25}{35}=\frac57,$$
so $p+q=12$.

- A is the reduced-fraction result.
- B can arise from retaining an unreduced numerator/denominator fragment.
- C uses the unreduced denominator plus the reduced numerator.
- D adds an unrelated sample-size adjustment even though it cancels in the ratio.
- E does not follow from either the squared or unsquared spectrum.
