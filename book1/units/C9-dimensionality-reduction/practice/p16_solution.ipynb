# C9-dimensionality-reduction — Practice p16 — Solution

```python
import numpy as np

evr_top6 = np.array([0.21, 0.13, 0.09, 0.07, 0.06, 0.05], dtype=np.float64)
choices = {"i": "neighbor-map", "ii": "original-space", "iii": "pca"}
kept2 = float(evr_top6[:2].sum())
lost2 = float(1.0 - kept2)
print("choices:", choices)
print("kept / lost in two components:", kept2, lost2)
```

For (i), the neighbor map is the best of the named views for flagging candidates because local neighbor relations are its preservation target; candidates should still be confirmed in the original measurements.  For (ii), original-space computation is best because one linear view shrinks distances and a neighbor map distorts global distances.  For (iii), PCA supplies explicit loading vectors and variance shares.  Its two-dimensional window keeps only $0.34$ of the variance and loses $0.66$, so clusters or separation carried by later directions can be hidden.

### Answer check

```python
assert choices == {"i": "neighbor-map", "ii": "original-space", "iii": "pca"}
assert np.isclose(kept2, 0.34, atol=1e-12, rtol=0)
assert np.isclose(lost2, 0.66, atol=1e-12, rtol=0)
```
