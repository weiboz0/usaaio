### Solution derivation

**Claim 1 — covariance structure.**  Because
$$
C^{\mathsf T}=\left(\frac{1}{n-1}X_c^{\mathsf T}X_c\right)^{\mathsf T}
=\frac{1}{n-1}X_c^{\mathsf T}X_c=C,
$$
$C$ is symmetric.  For every $a\in\mathbb R^d$,
$$
a^{\mathsf T}Ca=\frac{1}{n-1}a^{\mathsf T}X_c^{\mathsf T}X_ca
=\frac{\lVert X_ca\rVert_2^2}{n-1}\ge 0,
$$
so it is positive semidefinite.  Its $j$th diagonal entry is
$$
C_{jj}=\frac{1}{n-1}\sum_{i=1}^n (X_{ij}-\mu_j)^2,
$$
the sample variance of feature $j$.

**Claim 2 — directional variance.**  For a unit vector $u$, the score vector is $z=X_cu$.  It has mean zero because
$\mathbf 1^{\mathsf T}X_c=0$.  Hence its sample variance is
$$
\frac{1}{n-1}z^{\mathsf T}z
=\frac{1}{n-1}u^{\mathsf T}X_c^{\mathsf T}X_cu
=u^{\mathsf T}Cu.
$$

**Claim 3 — stationarity and the eigenproblem.**  Let $f(v)=v^{\mathsf T}Cv$ on the unit sphere and, for arbitrary $w\perp u$, set
$u(t)=(u+tw)/\lVert u+tw\rVert$.  Since $u^{\mathsf T}w=0$, $u'(0)=w$.  At a maximizing $u$,
$$
0=\left.\frac{d}{dt}f(u(t))\right|_{t=0}
=w^{\mathsf T}Cu+u^{\mathsf T}Cw=2w^{\mathsf T}Cu,
$$
where symmetry of $C$ was used.  Thus $Cu$ is orthogonal to every vector in $u^\perp$, so $Cu$ belongs to $\operatorname{span}\{u\}$: $Cu=\lambda u$.  Multiplying by $u^{\mathsf T}$ gives
$\lambda=u^{\mathsf T}Cu$, exactly the directional sample variance from Claim 2.

**Claim 4 — global maximum and repeated top eigenvalues.**  By the spectral theorem, take an orthonormal eigenbasis $q_1,\ldots,q_d$ with eigenvalues $\lambda_1,\ldots,\lambda_d$, and write a unit vector as $u=\sum_i a_iq_i$, where $\sum_i a_i^2=1$.  Then
$$
u^{\mathsf T}Cu=\sum_i\lambda_i a_i^2\le
\lambda_{\max}\sum_i a_i^2=\lambda_{\max}.
$$
Equality holds precisely when $a_i=0$ for every $i$ with $\lambda_i<\lambda_{\max}$.  Therefore the maximizers are exactly the unit vectors in the top-eigenvalue eigenspace.  If the top eigenvalue has multiplicity greater than one, every unit linear combination inside that whole eigenspace is a maximizer; there is no unique maximizing direction.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0
X_p20 = np.array([
    [4.0, 1.0, 2.0],
    [2.0, 3.0, 0.0],
    [0.0, 1.0, 2.0],
    [2.0, -1.0, 4.0],
    [5.0, 2.0, 1.0],
])

mu_p20 = X_p20.mean(axis=0)
_Xc_p20 = X_p20 - mu_p20
covariance_p20 = _Xc_p20.T @ _Xc_p20 / (X_p20.shape[0] - 1)
_eigenvalues_p20, _eigenvectors_p20 = np.linalg.eigh(covariance_p20)
direction_p20 = _eigenvectors_p20[:, -1]
variance_p20 = float(_eigenvalues_p20[-1])
projector_p20 = direction_p20[:, None] @ direction_p20[None, :]
```

```python
# Immutable contract check — do not edit.
_mu_ref_p20 = X_p20.mean(axis=0)
_Xc_ref_p20 = X_p20 - _mu_ref_p20
_C_ref_p20 = _Xc_ref_p20.T @ _Xc_ref_p20 / (X_p20.shape[0] - 1)
_evals_ref_p20, _evecs_ref_p20 = np.linalg.eigh(_C_ref_p20)
_lambda_ref_p20 = _evals_ref_p20[-1]
_v_ref_p20 = _evecs_ref_p20[:, -1]
_P_ref_p20 = np.outer(_v_ref_p20, _v_ref_p20)

assert isinstance(mu_p20, np.ndarray) and mu_p20.shape == (3,)
assert isinstance(covariance_p20, np.ndarray) and covariance_p20.shape == (3, 3)
assert isinstance(direction_p20, np.ndarray) and direction_p20.shape == (3,)
assert isinstance(projector_p20, np.ndarray) and projector_p20.shape == (3, 3)
assert np.allclose(mu_p20, _mu_ref_p20, atol=ATOL, rtol=RTOL)
assert np.allclose(covariance_p20, _C_ref_p20, atol=ATOL, rtol=RTOL)
assert np.allclose(covariance_p20, covariance_p20.T, atol=ATOL, rtol=RTOL)
assert np.isclose(np.linalg.norm(direction_p20), 1.0, atol=ATOL, rtol=RTOL)
assert np.isclose(float(variance_p20), _lambda_ref_p20, atol=ATOL, rtol=RTOL)
assert np.allclose(
    covariance_p20 @ direction_p20,
    float(variance_p20) * direction_p20,
    atol=ATOL, rtol=RTOL,
)
_score_var_p20 = ((X_p20 - mu_p20) @ direction_p20)
_score_var_p20 = (_score_var_p20 @ _score_var_p20) / (X_p20.shape[0] - 1)
assert np.isclose(_score_var_p20, variance_p20, atol=ATOL, rtol=RTOL)
assert np.allclose(
    projector_p20, np.outer(direction_p20, direction_p20),
    atol=ATOL, rtol=RTOL,
)
assert np.allclose(projector_p20, _P_ref_p20, atol=ATOL, rtol=RTOL)
assert np.allclose(projector_p20 @ projector_p20, projector_p20, atol=ATOL, rtol=RTOL)
```
