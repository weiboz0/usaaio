## Unit goals

By the end of this unit you can:

- run **PCA by two equivalent transparent routes** — center the data
  ($\mu$ = `X.mean(axis=0)`, $X_c = X - \mu$), take
  `np.linalg.svd(Xc, full_matrices=False)`, and read off principal
  directions (rows of $V^{\mathsf T}$) and component variances
  $\sigma_i^2/(n-1)$; also form the sample covariance
  $C=X_c^{\mathsf T}X_c/(n-1)$, derive $Cu=\lambda u$ from
  directional variance, and prove the covariance/SVD equivalence;
- state and use the **sign-flip ambiguity**: $v$ and $-v$ are the same
  axis; cross-route comparisons of individual directions use
  $|v \cdot w|$ or sign-fixing (largest-$|{\cdot}|$ entry made
  positive, F6-02's pin), while repeated-eigenvalue blocks are
  compared by subspace projectors rather than row order or signs;
- compute and read **variance-explained ratios**
  $\sigma_i^2 / \sum_j \sigma_j^2$, prove they sum to $1$ via the
  total-variance decomposition (p11), and plot 2-D projections of
  high-dimensional data;
- implement a reusable NumPy-only **`NumpyPCA`** whose `fit` stores
  `mean_`, `components_`, `explained_variance_`, and
  `explained_variance_ratio_`, and whose transform, fit-transform,
  inverse-transform, and reconstruction behavior is certified without
  sklearn/scipy PCA;
- **compress** C8's unit-row embedding stack `W` ($(N, 100)$, rows are
  tokens, $S = W W^{\mathsf T}$) by rank-$r$ truncation: build $W_r$,
  compute error curves *from the spectrum* (cumulative sums of
  $\sigma_i^2$ — the tail identity), choose the smallest $r$ meeting a
  stated **error budget** with a two-sided certificate, and measure
  what compression does to $S$ (the sagging diagonal, neighbor
  stability), including the proof-level identity
  $\lVert S - S_r \rVert_F^2 = \sum_{i>r} \sigma_i^4$ (p12);
- explain **umap-concept** in the stated-fact register: neighbor-graph
  methods prioritize local neighborhood structure over global
  distances; on-page distances, densities, sizes, and axes of a
  nonlinear map carry no quantitative meaning; results depend on
  settings and randomness — with **no** map-making library run
  anywhere in this unit;
- reason about **local-vs-global structure** with computable metrics:
  $k$-NN preservation (local) and worst-case stretch (global), the
  projection guarantee (a PCA view never exaggerates distances — but
  manufactures false neighbors on curved data), and the discipline of
  matching the question to the view.

**What this unit deliberately does not do:** run UMAP or any
map-making library (the syllabus item is the *concept*), and it uses
**no sklearn** — sklearn is outside this unit's prerequisite chain, so
all verification is by internal numerical self-checks across the
covariance, SVD, projector, and reconstruction routes.
