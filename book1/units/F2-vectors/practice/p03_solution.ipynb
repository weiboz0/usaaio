# F2-vectors — Practice p03 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
dot_1, angle_1 = 0.0, "right"     # 3*2 + (-1)*6 = 6 - 6
dot_2, angle_2 = 5.0, "acute"     # 1*3 + 2*1 = 5 > 0
dot_3, angle_3 = -5.0, "obtuse"   # 2*(-1) + (-3)*1 = -5 < 0
print(dot_1, angle_1, "|", dot_2, angle_2, "|", dot_3, angle_3)
```

Multiply matching entries and add.
The sign carries the geometry: zero means exactly perpendicular, positive
means the angle is under $90^\circ$, negative means it is over $90^\circ$ —
no norms or cosines needed for the classification.

## Part (b)

```python
def cosine_similarity(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))


print(cosine_similarity(np.array([1.0, 0.0]), np.array([1.0, 1.0])))
```

The dot product supplies the numerator, and the two norms divide the lengths
back out, leaving only $\cos\theta$.
The check pair sits at $45^\circ$, so the value is
$\cos 45^\circ = \sqrt2/2 \approx 0.7071$.

## Part (c)

```python
s1 = np.array([3.0, 2.0, 0.0, 0.0])
s2 = np.array([1.0, 2.0, 0.0, 0.0])
s3 = np.array([0.0, 0.0, 2.0, 3.0])

sim_12 = cosine_similarity(s1, s2)
sim_13 = cosine_similarity(s1, s3)
sim_23 = cosine_similarity(s2, s3)
most_alike = "s1-s2"

print(sim_12, sim_13, sim_23, "->", most_alike)
```

$s_1$ and $s_2$ both spend all their counts on the sports words, so their
similarity is $\dfrac{3 + 4}{\sqrt{13}\sqrt{5}} = \dfrac{7}{\sqrt{65}} \approx 0.868$.
$s_3$ uses only the cooking words, sharing no nonzero entry with the others,
so both of its similarities are exactly 0 — no direction agreement at all.
