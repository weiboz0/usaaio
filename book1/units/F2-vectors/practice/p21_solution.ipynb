# F2-vectors — Practice p21 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
def normalize(v):
    return v / np.sqrt(np.sum(v**2))


v1 = np.array([1.0, 2.0, 2.0])
v2 = np.array([1.0, 0.0, 0.0])
v3 = np.array([0.0, 1.0, 0.0])

u1 = normalize(v1)                      # (1, 2, 2) / 3

r2 = v2 - (v2 @ u1) * u1
u2 = normalize(r2)

r3 = v3 - (v3 @ u1) * u1 - (v3 @ u2) * u2
u3 = normalize(r3)

print(u1)
print(u2)
print(u3)
```

Each step removes, via a residual, every trace of the directions already
chosen, then rescales to unit length.
By hand the first two stages are clean: $\|v_1\| = 3$ gives
$u_1 = \tfrac13(1, 2, 2)$, and $v_2\cdot u_1 = \tfrac13$ leaves
$r_2 = \tfrac19(8, -2, -2)$, which normalizes to
$u_2 = \tfrac{1}{3\sqrt2}(4, -1, -1)$; the third stage is best left to
NumPy.

## Part (b)

```python
checks = [
    ("u1.u1", u1 @ u1), ("u2.u2", u2 @ u2), ("u3.u3", u3 @ u3),
    ("u1.u2", u1 @ u2), ("u1.u3", u1 @ u3), ("u2.u3", u2 @ u3),
]
print(checks)
```

All six conditions hold: self dot products 1, pair dot products 0 (to
float precision).
**Why $r_3$ is orthogonal to both:** expanding
$r_3\cdot u_1 = v_3\cdot u_1 - (v_3\cdot u_1)(u_1\cdot u_1) - (v_3\cdot u_2)(u_2\cdot u_1)$,
the first two terms cancel because $u_1\cdot u_1 = 1$ and the third vanishes
because $u_2\cdot u_1 = 0$ — and the same computation with the roles of
$u_1$ and $u_2$ swapped kills $r_3\cdot u_2$.
Subtracting the projection onto each axis removes exactly that axis's share,
and orthogonality of the earlier axes keeps the removals from interfering.

## Part (c)

```python
w = np.array([3.0, 0.0, 6.0])

t1 = w @ u1
t2 = w @ u2
t3 = w @ u3
rebuilt = t1 * u1 + t2 * u2 + t3 * u3
print(t1, t2, t3)
print(rebuilt)
```

Three dot products read off the coordinates, and the three pieces
reassemble $w = (3, 0, 6)$ exactly — the "good axes" property now holding
for a full set of three homemade perpendicular directions in 3-D.

### Answer check

```python
assert np.isclose(u1 @ u1, 1.0, atol=1e-9, rtol=0) and np.isclose(u2 @ u2, 1.0, atol=1e-9, rtol=0) and np.isclose(u3 @ u3, 1.0, atol=1e-9, rtol=0)
assert np.isclose(u1 @ u2, 0.0, atol=1e-9, rtol=0) and np.isclose(u1 @ u3, 0.0, atol=1e-9, rtol=0) and np.isclose(u2 @ u3, 0.0, atol=1e-9, rtol=0)
assert np.allclose(u1, np.array([1.0, 2.0, 2.0]) / 3.0, atol=1e-9, rtol=0)
assert np.allclose(u2, np.array([4.0, -1.0, -1.0]) / (3.0 * np.sqrt(2.0)), atol=1e-9, rtol=0)
assert np.allclose(t1 * u1 + t2 * u2 + t3 * u3, np.array([3.0, 0.0, 6.0]), atol=1e-9, rtol=0)
print("All checks passed.")
```
