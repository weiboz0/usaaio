# F2-vectors — Practice p14 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

**B** is the residual property in disguise: $r$ is orthogonal to $\hat u$,
and $p$ is a number times $\hat u$, so
$p\cdot r = (a\cdot\hat u)(\hat u\cdot r) = 0$ by the scalar-pull-out rule.

The others each fail:
**A** fails when $a\cdot\hat u < 0$ — a norm is never negative, so
$\|p\| = |a\cdot\hat u|$, with absolute value.
**C** confuses the Pythagorean relation: it is the *squared* norms that add,
$\|a\|^2 = \|p\|^2 + \|r\|^2$, and norms themselves do not subtract like
that.
**D** fails for any nonzero $a$ pointing along $\hat u$: then $p = a$ and
$r = 0$ with $a \neq 0$.
**E** is backwards — $p$ is parallel to $\hat u$, not orthogonal to it
(unless $p = 0$).

```python
answer_p14 = "B"

# Spot checks: B holds, A and D fail on these choices.
u = np.array([1.0, 0.0])
a = np.array([-3.0, 2.0])          # reaches backwards: a . u < 0
p = (a @ u) * u
r = a - p
print("p . r =", p @ r)                       # 0  (B)
print("||p|| =", np.linalg.norm(p), " a.u =", a @ u)   # 3 vs -3  (A fails)
a2 = np.array([4.0, 0.0])          # nonzero, along u
r2 = a2 - (a2 @ u) * u
print("r2 =", r2)                             # zero residual, a2 != 0  (D fails)
```

### Answer check

```python
assert answer_p14 == "B"
assert np.isclose(p @ r, 0.0, atol=1e-9, rtol=0)
assert not np.isclose(np.linalg.norm(p), a @ u, atol=1e-9, rtol=0)   # A's claim fails here
assert np.allclose(r2, 0.0, atol=1e-9, rtol=0) and not np.allclose(a2, 0.0, atol=1e-9, rtol=0)   # D's claim fails here
print("All checks passed.")
```
