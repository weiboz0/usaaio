# F2-vectors — Practice p16 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
X = rng.normal(size=(120, 5))
print(X.shape)
```

## Part (a)

```python
def unit_rows(X):
    norms = np.sqrt(np.sum(X**2, axis=1))   # (120,): one norm per row
    return X / norms[:, np.newaxis]         # (120, 5) / (120, 1) broadcasts


U = unit_rows(X)
print(U.shape)
print(U[0])
```

The `axis=1` sum of squares followed by `np.sqrt` computes all 120 norms in
one pass — `np.linalg.norm` is not needed.
The reshape to a `(120, 1)` column is the load-bearing step: it makes the
division divide each *row* by its own norm instead of mismatching shapes.

## Part (b)

```python
max_err = float(np.max(np.abs(np.sqrt(np.sum(U**2, axis=1)) - 1.0)))
print(max_err)
```

Recomputing the row norms of `U` and comparing against 1 checks the whole
batch at once; the maximum deviation is at the level of float rounding
(about $10^{-16}$), confirming every row is a unit vector.

### Answer check

```python
assert U.shape == (120, 5)
assert np.allclose(np.sqrt(np.sum(U**2, axis=1)), 1.0, atol=1e-9, rtol=0)
assert np.allclose(U[0], X[0] / np.sqrt(np.sum(X[0] ** 2)), atol=1e-9, rtol=0)
assert max_err < 1e-12
assert np.isclose(np.sum(U), -9.239197248426642, atol=1e-9, rtol=0)
print("All checks passed.")
```
