# F2-vectors — Practice p19 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
S = np.array([[4.0, 2.0, 0.0],
              [2.0, 1.0, 0.0],
              [0.0, 0.0, 3.0],
              [1.0, 1.0, 1.0]])
print(S.shape)
```

## Part (a)

```python
row_norms = np.sqrt(np.sum(S**2, axis=1))
print(row_norms)
```

One axis sum gives all four squared norms; the roots are
$(\sqrt{20},\ \sqrt5,\ 3,\ \sqrt3)$.
Note how different the lengths are — sentence 1 uses its words twice as
often as sentence 2.

## Part (b)

```python
U = S / row_norms[:, np.newaxis]
print(U)
print(np.sqrt(np.sum(U**2, axis=1)))   # all 1.0
```

Reshaping the norms to a `(4, 1)` column makes the division divide each row
by its own norm.
After this step every row carries only its direction — its mix of words —
with the amount of text divided out.

## Part (c)

```python
sim_12 = float(U[0] @ U[1])
sim_13 = float(U[0] @ U[2])
sim_14 = float(U[0] @ U[3])
most_alike = "s1-s2"
print(sim_12, sim_13, sim_14, "->", most_alike)
```

**Explanation:** sentence 2's counts $(2, 1, 0)$ are exactly half of
sentence 1's $(4, 2, 0)$ — the same *proportions* of guitar to drums — so
their unit rows are identical and the similarity is exactly $1.0$.
Cosine similarity compares directions only, so "the same mix, said more
briefly" counts as a perfect match, while sentence 3 (violin only) scores
$0$ and the balanced sentence 4 lands in between ($\approx 0.775$).

### Answer check

```python
assert np.allclose(row_norms, np.array([np.sqrt(20), np.sqrt(5), 3.0, np.sqrt(3)]), atol=1e-9, rtol=0)
assert np.allclose(np.sqrt(np.sum(U**2, axis=1)), 1.0, atol=1e-9, rtol=0)
assert np.isclose(sim_12, 1.0, atol=1e-9, rtol=0)
assert np.isclose(sim_13, 0.0, atol=1e-9, rtol=0)
assert np.isclose(sim_14, 6 / np.sqrt(60), atol=1e-9, rtol=0)
assert most_alike == "s1-s2"
print("All checks passed.")
```
