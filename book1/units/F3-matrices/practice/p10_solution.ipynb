# F3-matrices — Practice p10 — Solution

Normalizing the Gram matrix removes row scale and sign magnitude, so rows in the same direction family have absolute cosine near one. Rows 0, 1, and 2 form row 0's family, and the allowed set sweep finds three distinct families overall.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
base = rng.normal(0, 1, (3, 5))
mix = np.repeat(base, (3, 3, 2), axis=0)
scales = rng.uniform(0.5, 2.0, 8)[:, None] * np.where(rng.uniform(0, 1, 8)[:, None] < 0.3, -1, 1)
W = mix * scales + rng.normal(0, 1e-6, (8, 5))

def gram(W):
    return (W[:, None, :] * W[None, :, :]).sum(axis=2)

G = gram(W)
lengths = np.sqrt(np.diagonal(G))
C = G / (lengths[:, None] * lengths[None, :])
family_of_row0 = np.sort(np.where(np.abs(C[0]) > 0.99)[0])
unassigned = set(range(8))
n_families = 0
for row_index in range(8):
    if row_index in unassigned:
        family = set(np.where(np.abs(C[row_index]) > 0.99)[0].tolist())
        unassigned -= family
        n_families += 1
```

### Answer check

```python
assert np.array_equal(family_of_row0, np.array([0, 1, 2]))
assert n_families == 3
assert np.allclose(np.diagonal(C), np.ones(8), atol=1e-12, rtol=0)
```
