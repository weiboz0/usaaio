# F3-matrices — Practice p15 — Solution

For Style-K, linearity predicts the image of $(1,1)$ by adding the two basis images, yielding $(2,3)$ exactly, so $K=[[2,0],[0,3]]$. For Style-L, the basis images imply $L=[[1,-1],[1,1]]$, which predicts $(2,-1)\mapsto(3,1)$ rather than the documented $(3,0)$, so no linear map fits all three and `L` is `None`. Two basis vectors suffice because every 2-D input is their unique linear combination; the third example is therefore a consistency check rather than additional freedom.

```python
import numpy as np

K = np.array([[2., 0.], [0., 3.]])
k_input = np.array([1., 1.])
k_documented = np.array([2., 3.])
k_prediction = (K * k_input[None, :]).sum(axis=1)
check_K = np.abs(k_prediction - k_documented).max()

L_from_basis = np.array([[1., -1.], [1., 1.]])
l_input = np.array([2., -1.])
l_documented = np.array([3., 0.])
l_prediction = (L_from_basis * l_input[None, :]).sum(axis=1)
check_L = np.abs(l_prediction - l_documented).max()
L = None
```

### Answer check

```python
assert np.array_equal(K, np.array([[2., 0.], [0., 3.]]))
assert check_K < 1e-12
assert np.array_equal(l_prediction, np.array([3., 1.]))
assert check_L == 1.0
assert L is None
```
