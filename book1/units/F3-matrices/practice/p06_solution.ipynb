# F3-matrices — Practice p06 — Solution

Each output entry $C_{ij}$ is the sum of the elementwise products of row $i$ of $A$ and column $j$ of $B$. The two required loops enumerate output entries, while the one-dimensional sum implements that definition directly.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
A = rng.normal(0, 1, (4, 3))
B = rng.normal(0, 1, (3, 5))

def matmul_manual(A, B):
    C = np.empty((A.shape[0], B.shape[1]), dtype=np.result_type(A, B))
    for i in range(A.shape[0]):
        for j in range(B.shape[1]):
            C[i, j] = np.sum(A[i, :] * B[:, j])
    return C

C_manual = matmul_manual(A, B)
C_fast = A @ B
agreement_gap = np.abs(C_manual - C_fast).max()
```

### Answer check

```python
assert C_manual.shape == (4, 5)
assert np.allclose(C_manual, C_fast, rtol=1e-12, atol=1e-12)
assert agreement_gap < 1e-12
```
