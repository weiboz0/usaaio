# F3-matrices — Practice p03 — Solution

The second row is exactly twice the first, so it contributes no new direction. The third row is nonzero and cannot be a multiple of the first because it points along the third coordinate, leaving exactly two independent directions; the answer is **B**.

```python
import numpy as np

v1 = np.array([1, 2, 0])
v2 = np.array([2, 4, 0])
v3 = np.array([0, 0, 5])
V = np.stack((v1, v2, v3))
rank_V = 2
answer_p03 = "B"
```

### Answer check

```python
assert np.array_equal(v2, 2 * v1)
assert not np.array_equal(v3, np.zeros(3, dtype=int))
assert rank_V == 2
assert answer_p03 == "B"
```
