# C8-embeddings — Practice p15 — Solution

```python
from gensim.utils import simple_preprocess

NOTES = ("Day 12 on the river: two otters at dawn, then a heron. The heron waited; "
         "the otters did not. We logged the heron at 6 a.m., the otters at 6 a.m. "
         "and 8 a.m., and one badger at dusk.")

toks = simple_preprocess(NOTES)
n_tokens = len(toks)
n_types = len(dict.fromkeys(toks))
uniq_ordered = list(dict.fromkeys(toks))
n_otters = toks.count("otters")
n_heron = toks.count("heron")
first_animal = ("otters" if toks.index("otters") < toks.index("heron")
                else "heron")
```

### Diagnosis

Deduplication discarded both first-occurrence order and occurrence counts. Its iteration order is arbitrary and can happen to look plausible in one process, so the displayed ordering cannot support the heron-first claim; retaining one copy of each type likewise cannot support a once-each claim.

### Repaired report

The notes contain 31 tokens and 19 types, and `otters` first appears at position 5, before `heron` at position 9. Both `otters` and `heron` occur three times, not once.

### Answer check

```python
assert n_tokens == 31
assert n_types == 19
assert uniq_ordered == ["day", "on", "the", "river", "two", "otters", "at", "dawn",
                        "then", "heron", "waited", "did", "not", "we", "logged", "and",
                        "one", "badger", "dusk"]
assert (n_otters, n_heron) == (3, 3)
assert toks.index("otters") == 5 and toks.index("heron") == 9
assert first_animal == "otters"
```
