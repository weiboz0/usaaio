# C8-embeddings — Practice p05 — Solution

```python
import os, pathlib
_env_root = os.environ.get("USAAIO_BOOK_ROOT")
if _env_root:
    _root = pathlib.Path(_env_root).resolve()
else:
    _start = pathlib.Path.cwd().resolve()
    _root = next(
        p for p in [_start, *_start.parents]
        if (p / "syllabus.md").is_file() and (p / "curriculum").is_dir()
    )
os.environ["GENSIM_DATA_DIR"] = str(_root / "reference" / "cache" / "gensim")

import numpy as np
import gensim.downloader
```

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

vocab_size = int(len(kv.key_to_index))
dim = int(kv.vector_size)
has_glacier = bool("glacier" in kv.key_to_index)
has_quokkas = bool("quokkas" in kv.key_to_index)
vec = np.asarray(kv["glacier"], dtype=np.float64)
vec_first = float(vec[0])
vec_norm = float(np.sqrt((vec * vec).sum()))
```

The model reports 400,000 vocabulary entries and 100 coordinates per vector. `glacier` is present, while `quokkas` is absent. The lookup is converted immediately to the course's float64 arithmetic register.

### Answer check

```python
assert vocab_size == 400_000
assert dim == 100
assert has_glacier is True and has_quokkas is False
assert vec.shape == (100,) and vec.dtype == np.float64
assert np.isclose(vec_first, -0.7210000157356262, atol=1e-12, rtol=0)
assert np.isclose(vec_norm, 5.7544749113605835, atol=1e-12, rtol=0)
```
