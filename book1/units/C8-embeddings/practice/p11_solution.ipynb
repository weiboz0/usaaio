# C8-embeddings — Practice p11 — Solution

### (a) Entrywise identity

By the entrywise definition of matrix multiplication,
$$S_{ij}=(WW^{\mathsf T})_{ij}=\sum_{k=0}^{99}W_{ik}(W^{\mathsf T})_{kj}.$$
Transposition swaps row and column indices, so $(W^{\mathsf T})_{kj}=W_{jk}$. Therefore
$$S_{ij}=\sum_{k=0}^{99}W_{ik}W_{jk}=w_i\cdot w_j.$$

### (b) Structure

For symmetry, $S_{ji}=w_j\cdot w_i=w_i\cdot w_j=S_{ij}$ by commutativity of the real dot product. For the diagonal, $S_{ii}=w_i\cdot w_i=\lVert w_i\rVert^2=1$ by the dot-product norm identity and the unit-row hypothesis.

### (c) Cosine interpretation

Cosine similarity divides $w_i\cdot w_j$ by $\lVert w_i\rVert\lVert w_j\rVert$. Both norms equal 1, so the denominator is $1\cdot1$ and the cosine is exactly $S_{ij}$.

```python
kv = gensim.downloader.load("glove-wiki-gigaword-100")

WORDS = ["tide", "sailor", "violin"]
V = np.asarray(kv[WORDS], dtype=np.float64)
W = V / np.sqrt((V * V).sum(axis=1, keepdims=True))
S = W @ W.T
manual_01 = float((W[0] * W[1]).sum())
anchor_01 = float(S[0, 1])
```
