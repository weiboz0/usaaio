# C8-embeddings — Practice p01 — Solution

```python
from gensim.utils import simple_preprocess

s = "A 90-meter ferry, the Peregrine, left Pier 9 at dawn."
toks = simple_preprocess(s)
answer_p01 = "B"
```

The token list is `['meter', 'ferry', 'the', 'peregrine', 'left', 'pier', 'at', 'dawn']`, so its length is 8 and the answer is **B**.

- A undercounts by two; both `meter` and `at` survive.
- B is the exact count.
- C counts one extra item, commonly the one-letter `A` or a numeral.
- D counts two extra items, typically treating the two numerals as tokens.
- E also ignores the length and digit rules; punctuation does not create additional tokens.

### Answer check

```python
assert toks == ["meter", "ferry", "the", "peregrine", "left", "pier", "at", "dawn"]
assert len(toks) == 8
assert answer_p01 == "B"
```
