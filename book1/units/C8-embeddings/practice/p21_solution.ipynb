# C8-embeddings — Practice p21 — Solution

Negating a copied score array converts descending ranking into an ascending stable argsort without changing the supplied similarities. Stability preserves vocabulary indices 1, 2, and 5 in that order across their exact tie at 0.913; masking by index removes only the query before the top-four slice.

```python
import numpy as np

similarities = np.array([0.641, 0.913, 0.913, 1.0, 0.827, 0.913, 0.359])


def nearest_indices(similarities, query_index, k):
    scores = similarities.copy()
    order = np.argsort(-scores, kind="stable")
    without_self = order[order != query_index]
    return without_self[:k]


neighbors = nearest_indices(similarities, query_index=3, k=4)
self_excluded = bool(not np.any(neighbors == 3))
tie_order_ok = bool(np.array_equal(neighbors[:3], np.array([1, 2, 5])))
```

### Answer check

```python
assert neighbors.dtype.kind in "iu"
assert neighbors.shape == (4,)
assert np.array_equal(neighbors, np.array([1, 2, 5, 4]))
assert self_excluded is True
assert tie_order_ok is True
assert np.array_equal(
    similarities[[1, 2, 5]], np.array([0.913, 0.913, 0.913])
)
```
