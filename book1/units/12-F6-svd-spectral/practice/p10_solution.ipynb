# F6-svd-spectral — Practice p10 — Solution


The thin shape table fixes all dimensions. `Vt` is already $V^{\mathsf T}$, so its row orthonormality check is `Vt @ Vt.T`.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (7, 4))

U, s, Vt = np.linalg.svd(W, full_matrices=False)
shapes = (U.shape, s.shape, Vt.shape)
recon_gap = np.abs(U @ np.diag(s) @ Vt - W).max()
u_gap = np.abs(U.T @ U - np.eye(4)).max()
v_gap = np.abs(Vt @ Vt.T - np.eye(4)).max()
desc_ok = bool(np.all(np.diff(s) <= 0))
nonneg_ok = bool(np.all(s >= 0))
shapes, s, recon_gap, u_gap, v_gap
```

### Answer check


```python
assert shapes == ((7, 4), (4,), (4, 4))
assert recon_gap < 1e-10 and u_gap < 1e-10 and v_gap < 1e-10
assert desc_ok is True and nonneg_ok is True
```
