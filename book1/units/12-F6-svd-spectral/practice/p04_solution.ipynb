# F6-svd-spectral — Practice p04 — Solution


By hand,
$W^{\mathsf T}W=\begin{pmatrix}5&2\\2&8\end{pmatrix}$.
The dependent-rows equation is $(5-\lambda)(8-\lambda)-4=0$, or $\lambda^2-13\lambda+36=(\lambda-9)(\lambda-4)=0$. Thus $(\sigma_1,\sigma_2)=(3,2)$ and $s=9-4=5$. Since $s\ge0$, the decoded value is $2(5)+1=11$, so the answer is **C**.

A uses the singular-value gap $3-2=1$ as $m$ instead of the required squared-value gap. B finds the correct $m=5$ but uses the negative branch $2m$. D would require $m=6$, which is neither eigenvalue difference, and E would require $m=10$, likewise contradicting the two roots.


```python
import numpy as np

W = np.array([[2., 2.], [-1., 2.]])
C = W.T @ W
lam_desc = np.array([9., 4.])
sig = np.sqrt(lam_desc)
s_value = lam_desc[0] - lam_desc[1]
decoded = int(2 * s_value + 1)
answer_p04 = "C"
```

### Answer check


```python
assert np.array_equal(C, np.array([[5., 2.], [2., 8.]]))
assert np.array_equal(lam_desc, np.array([9., 4.]))
assert np.array_equal(sig, np.array([3., 2.]))
assert s_value == 5 and decoded == 11
assert answer_p04 == "C"
```
