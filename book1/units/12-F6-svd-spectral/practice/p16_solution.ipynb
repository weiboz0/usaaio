# F6-svd-spectral — Practice p16 — Solution


**(a) Tail form.** By definition $W_r=\sum_{k\le r}\sigma_k u_kv_k^{\mathsf T}$. Subtracting these kept atoms from the full atom sum leaves
$$W-W_r=\sum_{k>r}\sigma_k u_kv_k^{\mathsf T}.$$

**(b) The tail is SVD-shaped.** Any sub-family of an orthonormal family remains orthonormal, so the vectors with $k>r$ retain unit lengths and mutual perpendicularity on both sides. The coefficients $(\sigma_{r+1},\sigma_{r+2},\ldots)$ remain non-negative and descending because they are a suffix of the ordered singular-value list. Hence the displayed tail is a valid SVD of $W-W_r$.

**(c) Apply the master identity to the tail.** Applying p15 to the matrix $W-W_r$, whose singular values are precisely the tail coefficients, gives
$$\lVert W-W_r\rVert_F^2=\sum_{k>r}\sigma_k^2.$$
This identity computes the truncation's error; by itself it does not show that every other rank-$r$ matrix has at least this error. That optimality claim is the stated Eckart–Young fact.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (9, 6))
r = 2

U, s, Vt = np.linalg.svd(W, full_matrices=False)
Wr = U[:, :r] @ np.diag(s[:r]) @ Vt[:r, :]
err_sq = np.linalg.norm(W - Wr) ** 2
tail_sum = (s[r:]**2).sum()
anchor_gap = abs(err_sq - tail_sum)
err_sq, tail_sum, anchor_gap
```
