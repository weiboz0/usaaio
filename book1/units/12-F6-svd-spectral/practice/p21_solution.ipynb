# F6-svd-spectral — Practice p21 — Solution


The mirror bridge provides $V$ and $\sigma$ from $W^{\mathsf T}W$; the action recipe then supplies every column of $U$ at once by broadcasting.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (8, 4))

C = W.T @ W
lam_asc, V_asc = np.linalg.eigh(C)
lam = lam_asc[::-1]
V = V_asc[:, ::-1]
sig = np.sqrt(np.maximum(lam, 0))
U = (W @ V) / sig
recon_gap = np.abs(U @ np.diag(sig) @ V.T - W).max()
u_orth_gap = np.abs(U.T @ U - np.eye(4)).max()
desc_ok = bool(np.all(np.diff(sig) <= 0))
fro_gap = abs((sig**2).sum() - (W * W).sum())
sig, recon_gap, u_orth_gap, fro_gap
```

### Answer check


```python
assert sig.shape == (4,) and V.shape == (4, 4) and U.shape == (8, 4)
assert recon_gap < 1e-8 and u_orth_gap < 1e-8 and fro_gap < 1e-8
assert desc_ok is True
```
