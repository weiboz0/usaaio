Every nonzero vector is an eigenvector of $3I_3$, because $3I_3q=3q$ for every nonzero $q$. The quarter-turn has no real eigen-direction because it turns every nonzero real direction by $90^\circ$ rather than preserving or reversing it.


```python
import numpy as np

vals_a = np.array([4., -1.])
e_a = np.array([1., 0.])
lam_b = -2
vals_c = np.array([3., 3., 3.])
n_real_eigvecs = 0
vals_a, e_a, lam_b, vals_c, n_real_eigvecs
```

### Answer check


```python
assert np.array_equal(vals_a, np.array([4., -1.]))
assert np.array_equal(e_a, np.array([1., 0.]))
assert lam_b == -2
assert np.array_equal(vals_c, np.array([3., 3., 3.]))
assert n_real_eigvecs == 0
```
