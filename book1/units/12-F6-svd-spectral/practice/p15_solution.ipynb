# F6-svd-spectral — Practice p15 — Solution


**(a) Entry formula.** The atom $u_kv_k^{\mathsf T}$ has entry $u_k[i]v_k[j]$. Therefore the $(i,j)$ entry of $W=\sum_k\sigma_k u_kv_k^{\mathsf T}$ is
$$W_{ij}=\sum_k\sigma_k u_k[i]v_k[j].$$

**(b) Square and swap.** Substituting the entry formula twice gives
$$
\sum_{i,j}W_{ij}^2
=\sum_{i,j}\sum_{k,\ell}\sigma_k\sigma_\ell
u_k[i]u_\ell[i]v_k[j]v_\ell[j].
$$
All sums are finite, so distributivity and associativity allow their order to be exchanged. The factors depending on $i$ and $j$ then separate:
$$
\sum_{k,\ell}\sigma_k\sigma_\ell
\left(\sum_i u_k[i]u_\ell[i]\right)
\left(\sum_j v_k[j]v_\ell[j]\right)
=\sum_{k,\ell}\sigma_k\sigma_\ell
(u_k\mathbin{\cdot}u_\ell)(v_k\mathbin{\cdot}v_\ell).
$$
The entry sums are exactly the defining component sums for those dot products.

**(c) Orthonormality.** For both vector families, the dot product is $1$ when $k=\ell$ and $0$ when $k\ne\ell$. Thus every cross term vanishes and each surviving diagonal term is $\sigma_k^2$, proving
$$\lVert W\rVert_F^2=\sum_k\sigma_k^2.$$


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (5, 4))

frob_sq = (W * W).sum()
s = np.linalg.svd(W, full_matrices=False, compute_uv=False)
sig_sq_sum = (s * s).sum()
anchor_gap = abs(frob_sq - sig_sq_sum)
frob_sq, sig_sq_sum, anchor_gap
```
