# F6-svd-spectral — Practice p05 — Solution


Rank $2$ keeps $10$ and $6$, so the dropped squared content is $3^2+1^2=10$. The error is $\sqrt{10}$ and the answer is **B**.

A adds the dropped singular values before rooting; C reports the squared error; D and E use the wrong kept/dropped parcels.


```python
import numpy as np

sig = np.array([10., 6., 3., 1.])
err_sq = (sig[2:] ** 2).sum()
err = np.sqrt(err_sq)
answer_p05 = "B"
```

### Answer check


```python
assert err_sq == 10
assert np.isclose(err, np.sqrt(10.), rtol=0, atol=1e-12)
assert answer_p05 == "B"
```
