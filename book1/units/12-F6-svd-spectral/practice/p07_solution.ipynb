# F6-svd-spectral — Practice p07 — Solution


`eig` returns arbitrary order, so the values and matching columns are reordered together. The dependent-rows quadratic independently anchors both values.


```python
import numpy as np

M = np.array([[6., 2.], [1., 5.]])
vals_raw, vecs_raw = np.linalg.eig(M)
assert np.abs(vals_raw.imag).max() < 1e-12
vals_raw, vecs_raw = vals_raw.real, vecs_raw.real
order = np.argsort(vals_raw)[::-1]
vals_desc = vals_raw[order]
vecs_desc = vecs_raw[:, order]
resid_max = np.abs(M @ vecs_desc - vecs_desc * vals_desc).max()
quad_max = np.abs((6 - vals_desc) * (5 - vals_desc) - 2).max()
vals_desc, resid_max, quad_max
```

### Answer check


```python
assert np.allclose(vals_desc, np.array([7., 4.]), rtol=0, atol=1e-12)
assert resid_max < 1e-10
assert quad_max < 1e-10
```
