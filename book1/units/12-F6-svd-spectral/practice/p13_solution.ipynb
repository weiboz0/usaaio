# F6-svd-spectral — Practice p13 — Solution


The thin form supplies the three nonzero spectral directions. The full form supplies a complete seven-vector basis after four zero eigenvalues are padded.

Because the zero block is degenerate, no tail columns are compared. The cross-route verification uses the invariant trio: Frobenius reconstruction, eigen-equation residuals, and the separated top-2 projector.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (7, 3))
S = W @ W.T

U_thin, s_thin, Vt_thin = np.linalg.svd(W, full_matrices=False)
U_full, s_full, Vt_full = np.linalg.svd(W, full_matrices=True)
shapes_thin = (U_thin.shape, s_thin.shape, Vt_thin.shape)
shapes_full = (U_full.shape, s_full.shape, Vt_full.shape)
lam_pad = np.concatenate([s_full**2, np.zeros(4)])
recon_diff = U_full @ np.diag(lam_pad) @ U_full.T - S
pad_gap = np.abs(recon_diff).max()
recon_frob_gap = np.sqrt((recon_diff**2).sum())
eigen_resid_max = np.abs(S @ U_full - U_full * lam_pad).max()

lam_asc, Q_asc = np.linalg.eigh(S)
lam_desc, Q_desc = lam_asc[::-1], Q_asc[:, ::-1]
val_gap = np.abs(lam_desc - lam_pad).max()
P_eigh = Q_desc[:, :2] @ Q_desc[:, :2].T
P_svd = U_full[:, :2] @ U_full[:, :2].T
proj_gap = np.abs(P_eigh - P_svd).max()
shapes_thin, shapes_full, lam_pad, pad_gap, val_gap, proj_gap
```

### Answer check


```python
assert shapes_thin == ((7, 3), (3,), (3, 3))
assert shapes_full == ((7, 7), (3,), (3, 3))
assert pad_gap < 1e-8 and recon_frob_gap < 1e-8
assert eigen_resid_max < 1e-8
assert val_gap < 1e-8 and proj_gap < 1e-8
```
