# F6-svd-spectral — Practice p02 — Solution


The squared entries sum to $1+4+4+0+1+9=19$, so the norm is $\sqrt{19}$ and the answer is **C**.

A and B omit squared content; D forgets the square root; E is not the root of the entry-square sum.


```python
import numpy as np

B = np.array([[1., -2.], [2., 0.], [-1., 3.]])
frob_sq = (B * B).sum()
frob = np.sqrt(frob_sq)
answer_p02 = "C"
frob
```

### Answer check


```python
assert frob_sq == 19
assert np.isclose(frob, np.sqrt(19.), rtol=0, atol=1e-12)
assert answer_p02 == "C"
```
