### Review rubric

- activity is determined from $g_i(x^*)=0$, not from multiplier size;
- strict slackness implies a zero multiplier and a positive multiplier implies activity, but the converses are not asserted;
- all signs match $g_i ≤ 0$, $λ_i ≥ 0$, $L=f+λ^Tg$;
- weak duality has direction $d^* ≤ p^*$.

### Worked solution

**Activity and slackness.**  At $x^*=1$,
\[
g_1(x^*)=1-1=0,\qquad g_2(x^*)=-1-2=-3<0.
\]
Thus constraint 1 is active and constraint 2 is inactive, based on their constraint values.  Complementary slackness states $\lambda_i g_i(x^*)=0$.  Together with feasibility, it implies
\[
g_i(x^*)<0\ \Longrightarrow\ \lambda_i=0,
\qquad
\lambda_i>0\ \Longrightarrow\ g_i(x^*)=0.
\]
The converses need not hold: an active (for example redundant) constraint may have multiplier zero, and a zero multiplier does not imply strict slackness.

**Stationarity and complementarity.**  Since $f'(x)=2(x-3)$, $g_1'(x)=1$, and $g_2'(x)=-1$,
\[
\frac{\partial L}{\partial x}(1,4,0)
=2(1-3)+4(1)-0(-1)=-4+4=0.
\]
The two complementary products are $(4\cdot0,\,0\cdot(-3))=(0,0)$.  Primal feasibility and the supplied nonnegative multipliers also hold.

**Weak-duality chain.**  For any $\lambda_1,\lambda_2\ge0$, the dual function is $q(\lambda)=\inf_zL(z,\lambda)$, so $q(\lambda)\le L(x,\lambda)$ for every $x$.  If $x$ is feasible, each $g_i(x)\le0$, hence $\lambda_i g_i(x)\le0$ and
\[
q(\lambda)\le L(x,\lambda)
=f(x)+\lambda_1g_1(x)+\lambda_2g_2(x)
\le f(x).
\]
Taking the supremum over dual-feasible multipliers and the infimum over primal-feasible $x$ gives the correct direction $d^*\le p^*$.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

x_star_p20 = 1.0
lambda_star_p20 = np.array([4.0, 0.0])

g_values_p20 = np.array([x_star_p20 - 1.0, -x_star_p20 - 2.0], dtype=float)
active_p20 = np.isclose(g_values_p20, 0.0, atol=ATOL, rtol=RTOL)
stationarity_p20 = float(2.0 * (x_star_p20 - 3.0) + lambda_star_p20[0] - lambda_star_p20[1])
complementarity_p20 = lambda_star_p20 * g_values_p20
```
