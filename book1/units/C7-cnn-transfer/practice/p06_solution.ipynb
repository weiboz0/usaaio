# C7-cnn-transfer — Practice p06 — Solution

The implementation follows the definition literally: each output position
selects one full window, multiplies it elementwise by the unflipped kernel,
and sums in float64.


```python
import numpy as np

SEED = 20260804

def conv1d_valid(x, k):
    x = np.asarray(x, dtype=np.float64)
    k = np.asarray(k, dtype=np.float64)
    y = np.zeros(x.shape[0] - k.shape[0] + 1, dtype=np.float64)
    for i in range(y.shape[0]):
        y[i] = (x[i:i + k.shape[0]] * k).sum()
    return y

xa = np.array([3., 1., 4., 1., 5., 9., 2., 6.])
ka = np.array([1., -1., 2.])
y_a = conv1d_valid(xa, ka)
first_val = float(y_a[0])

rng = np.random.default_rng(SEED)
xc = rng.standard_normal(50)
kc = np.ones(7)
n_out_formula = 50 - 7 + 1
n_out_measured = int(conv1d_valid(xc, kc).shape[0])

k_step = np.array([-1., 1.])
sig = np.array([2., 2., 2., 5., 5., 1., 1.])
resp = conv1d_valid(sig, k_step)
jump_idx = int(np.argmax(np.abs(resp)))
```

### Answer check

```python
assert y_a.dtype == np.float64
assert np.array_equal(y_a, np.array([10., -1., 13., 14., 0., 19.]))
assert first_val == 10.0
assert n_out_formula == n_out_measured == 44
assert np.array_equal(resp, np.array([0., 0., 3., 0., -4., 0.]))
assert jump_idx == 4
```
