# F5-probability — Practice p09 — Solution

Summing each row converts 40 independent Bernoulli trials into one binomial count, and standardization centers and rescales those counts. Boolean masks then estimate the mass within one and two standard deviations.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(20260804)
flips = rng.integers(0, 2, size=(100_000, 40))
sums = flips.sum(axis=1)
z = (sums - sums.mean()) / sums.std()
inside_1sd = np.mean(np.abs(z) <= 1)
inside_2sd = np.mean(np.abs(z) <= 2)

plt.hist(z, bins=30)
plt.title("Standardized sums of 40 fair coin flips")
plt.xlabel("Standardized sum z")
plt.ylabel("Count")
plt.show()
```

The observed fractions, about 0.732 within one standard deviation and 0.962 within two, are reasonably close to the Gaussian benchmarks 0.68 and 0.95. Discreteness makes the one-standard-deviation comparison less exact.
