# F5-probability — Practice p14 — Solution

For one source, symmetry gives mean 0 and the weighted second moment is $(4+1+0+1+4)/5=2$, so its variance is 2. For 25 independent sources, means and variances add to $\mu_R=0$ and $\operatorname{Var}(R)=50$.

The two-standard-deviation interval is $0\pm2\sqrt{50}$, or approximately $[-14.142,14.142]$. A vectorized seeded simulation estimates the probability that a reading falls in this closed interval.

```python
import numpy as np

source_values = np.array([-2., -1., 0., 1., 2.])
source_probs = np.full(5, 0.2)
source_mu = (source_values * source_probs).sum()
source_second_moment = (source_values**2 * source_probs).sum()
source_variance = source_second_moment - source_mu**2
mu_R = 25 * source_mu
var_R = 25 * source_variance
lo = mu_R - 2 * np.sqrt(var_R)
hi = mu_R + 2 * np.sqrt(var_R)

SEED = 20260804
rng = np.random.default_rng(SEED)
readings = rng.integers(-2, 3, size=(200_000, 25)).sum(axis=1)
frac_inside = np.mean((readings >= lo) & (readings <= hi))
```

The simulated fraction 0.9611 is close to the Gaussian 0.95 benchmark, so the approximation held reasonably well. A sum of many independent, similarly scaled error sources tends toward a Gaussian shape, which licenses the stated approximation.

### Answer check

```python
assert abs(mu_R) < 1e-12
assert abs(var_R - 50) < 1e-12
assert abs(lo + 2 * np.sqrt(50)) < 1e-12
assert abs(hi - 2 * np.sqrt(50)) < 1e-12
assert abs(frac_inside - 0.961095) < 1e-12
assert abs(frac_inside - 0.95) < 0.02
```
