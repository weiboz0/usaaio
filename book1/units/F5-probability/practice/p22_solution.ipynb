**Type:** proof · **Difficulty:** core · **Concepts:** conditional-probability, bayes-rule

*Reasoning is required.* A condition affects 1 in 1,000 people. A test is positive for 99% of affected people and 2% of unaffected people.

**(a)** Let $D$ be “affected” and $+$ be “positive.” Derive $P(+)$ from the partition $D,D^c$, labeling both routes.

**(b)** Starting from $P(D\cap +)$ in two ways, derive Bayes' rule for $P(D\mid +)$ and substitute the rates. Express the posterior as reduced `posterior_num / posterior_den` with positive denominator.

**(c) Base-rate trap.** In 2–4 sentences, refute “99% sensitivity means $P(D\mid+)=0.99$.” Identify the reversed conditional, the false-positive route, and why sensitivity alone is insufficient.

Record `bayes_direction` as exactly `"P(D|+)"`. Merely quoting a final formula without the total-probability denominator earns no credit.


**Your derivation and explanation:**


The partition $D,D^c$ gives two disjoint routes to a positive test:

$$
\begin{aligned}
P(+)
&=P(+\cap D)+P(+\cap D^c)\\
&=P(+\mid D)P(D)+P(+\mid D^c)P(D^c)\\
&=\frac{99}{100}\frac{1}{1000}
 +\frac{2}{100}\frac{999}{1000}
=\frac{2097}{100000}.
\end{aligned}
$$

The same intersection can be factored in either conditional direction:

$$
P(D\cap +)=P(+\mid D)P(D)=P(D\mid +)P(+).
$$

Because $P(+)>0$,

$$
P(D\mid +)
=\frac{P(+\mid D)P(D)}{P(+\mid D)P(D)+P(+\mid D^c)P(D^c)}
=\frac{99/100000}{2097/100000}
=\frac{11}{233}.
$$

The 99% sensitivity is $P(+\mid D)$, which reverses the requested conditional $P(D\mid +)$. Unaffected people contribute the false-positive route $P(+\mid D^c)P(D^c)$, and their large base rate makes that route substantial. Sensitivity alone is insufficient because the posterior also depends on the prior prevalence and false-positive rate.

```python
from fractions import Fraction
from math import gcd

posterior_num: int = 11
posterior_den: int = 233
bayes_direction: str = "P(D|+)"

# Immutable structural and Bayes-identity checks; the written derivation is reviewed separately.
assert type(posterior_num) is int and type(posterior_den) is int
assert posterior_num >= 0 and posterior_den > 0
assert gcd(abs(posterior_num), posterior_den) == 1
assert bayes_direction == "P(D|+)"
_prior = Fraction(1, 1000)
_sensitivity = Fraction(99, 100)
_false_positive_rate = Fraction(2, 100)
_positive_route = _sensitivity * _prior
_all_positive_routes = _positive_route + _false_positive_rate * (1 - _prior)
_candidate = Fraction(posterior_num, posterior_den)
assert _candidate * _all_positive_routes == _positive_route
print("p22 checks passed")
```
