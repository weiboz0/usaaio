# F5-probability — Practice p07 — Solution

The weighted first moment is $E[X]=\sum_i p_i x_i$, and the weighted second moment is $E[X^2]=\sum_i p_i x_i^2$. Subtracting the square of the first from the second gives the requested definition-based variance.

```python
import numpy as np

def my_variance(values, probs):
    first_moment = (values * probs).sum()
    second_moment = (values**2 * probs).sum()
    return second_moment - first_moment**2

v = my_variance(np.array([1., 3., 7.]), np.array([0.25, 0.5, 0.25]))
```

### Answer check

```python
assert abs(v - 4.75) < 1e-12
direct_weighted = (0.25 * (1 - 3.5)**2 + 0.5 * (3 - 3.5)**2 + 0.25 * (7 - 3.5)**2)
assert abs(v - direct_weighted) < 1e-12
```
