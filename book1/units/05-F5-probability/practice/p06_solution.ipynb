# F5-probability — Practice p06 — Solution

Equal sector probabilities make the exact mean $(0+0+2+5+13)/5=4$. Integer sector indices select payouts without changing their equal probabilities, and the large seeded sample should put the estimate close to 4.

```python
import numpy as np

payouts = np.array([0, 0, 2, 5, 13])
exact_mean = (0 + 0 + 2 + 5 + 13) / 5

def estimate_mean(n_spins, seed):
    rng = np.random.default_rng(seed)
    sector_indices = rng.integers(0, payouts.size, size=n_spins)
    return np.mean(payouts[sector_indices])

est = estimate_mean(200_000, 20260804)
gap = abs(est - exact_mean)
```

### Answer check

```python
assert exact_mean == 4
assert abs(est - 3.98831) < 1e-12
assert gap < 0.03
```
