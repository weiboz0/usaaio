# F5-probability — Practice p08 — Solution

Separate draws from the same seeded generator create the two independent samples requested. Their empirical variances can therefore test variance additivity directly.

```python
import numpy as np

rng = np.random.default_rng(20260804)
x = rng.integers(1, 7, size=500_000)
y_values = np.array([0, 2, 5])
y = y_values[rng.integers(0, 3, size=500_000)]
var_x = np.var(x)
var_y = np.var(y)
var_sum = np.var(x + y)
additivity_gap = abs(var_sum - (var_x + var_y))
```

Because $x$ and $y$ are independent, their covariance is zero and the variance-of-sums rule gives $\operatorname{Var}(x+y)=\operatorname{Var}(x)+\operatorname{Var}(y)$, up to sampling error. The observed gap is consequently small relative to the component variances.

### Answer check

```python
assert abs(var_x - 2.9076472256) < 1e-12
assert abs(var_y - 4.222957918464) < 1e-12
assert abs(var_sum - 7.121121013504) < 1e-12
assert additivity_gap < 0.03
```
