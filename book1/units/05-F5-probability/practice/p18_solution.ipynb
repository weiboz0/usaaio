# F5-probability — Practice p18 — Solution

The original construction pairs each $a_i$ with a noisy value near $3a_i$, so its covariance should be strongly positive. Permuting only $b$ preserves its values but destroys those informative pairings.

```python
import numpy as np

rng = np.random.default_rng(20260804)
a = rng.integers(1, 11, 50_000).astype(float)
b = 3.0 * a + rng.normal(0, 2, 50_000)

def my_cov(a_values, b_values):
    return np.mean((a_values - np.mean(a_values)) * (b_values - np.mean(b_values)))

cov_ab = my_cov(a, b)
permutation = rng.permutation(50_000)
b_shuffled = b[permutation]
cov_ab_shuffled = my_cov(a, b_shuffled)
var_b = np.var(b)
var_b_shuffled = np.var(b_shuffled)
predicted_cov_same_perm = cov_ab
cov_same_perm = my_cov(a[permutation], b[permutation])
```

Shuffling only $b$ breaks its observation-by-observation alignment with $a$, so covariance collapses toward zero. The shuffle does not change the collection of $b$ values, hence its variance remains exactly the same. Covariance measures paired co-movement between two variables, information that either marginal variance alone cannot capture. If both arrays receive the same permutation, all original pairs remain together, so I predict the covariance will equal `cov_ab`; `cov_same_perm` verifies that prediction.

### Answer check

```python
assert cov_ab > 24
assert abs(cov_ab - 24.631052371789984) < 1e-10
assert abs(cov_ab_shuffled) < 0.3
assert abs(var_b - var_b_shuffled) < 1e-12
assert abs(cov_same_perm - predicted_cov_same_perm) < 1e-12
```
