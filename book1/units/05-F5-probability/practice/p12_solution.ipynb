# F5-probability — Practice p12 — Solution

Let $\mu=E[X]$. Expanding the definition and applying linearity gives $E[(X-\mu)^2]=E[X^2-2\mu X+\mu^2]=E[X^2]-2\mu E[X]+\mu^2$ because $\mu$ is constant; substituting $E[X]=\mu$ leaves $E[X^2]-\mu^2=E[X^2]-(E[X])^2$.

For the spinner, $\mu=2(0.5)+4(0.3)+9(0.2)=4$. The definition gives $0.5(2-4)^2+0.3(4-4)^2+0.2(9-4)^2=7$, while the shortcut gives $0.5(2^2)+0.3(4^2)+0.2(9^2)-4^2=23-16=7$.

```python
import numpy as np

values = np.array([2., 4., 9.])
probs = np.array([0.5, 0.3, 0.2])
mu = (values * probs).sum()
lhs = ((values - mu)**2 * probs).sum()
rhs = (values**2 * probs).sum() - mu**2
```

### Answer check

```python
assert abs(mu - 4) < 1e-12
assert abs(lhs - 7) < 1e-12
assert abs(rhs - 7) < 1e-12
assert abs(lhs - rhs) < 1e-12
```
