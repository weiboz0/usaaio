# F5-probability — Practice p10 — Solution

The first implementation averages products of centered observations; distributing that product yields the shortcut $\overline{ab}-\bar a\bar b$. The constructed $w$ contains twice $u$ plus small noise, whereas $q$ is drawn independently.

```python
import numpy as np

def my_cov(a, b):
    return np.mean((a - np.mean(a)) * (b - np.mean(b)))

def my_cov_shortcut(a, b):
    return np.mean(a * b) - np.mean(a) * np.mean(b)

rng = np.random.default_rng(20260804)
u = rng.integers(1, 7, 100_000)
w = 2 * u + rng.integers(0, 3, 100_000)
q = rng.integers(1, 7, 100_000)
cov_uw = my_cov(u, w)
cov_uq = my_cov(u, q)
shortcut_gap = abs(my_cov(u, w) - my_cov_shortcut(u, w))
```

The large positive $\operatorname{Cov}(u,w)\approx5.81$ reflects that $w$ rises with $u$. In contrast, independent $u$ and $q$ have covariance near zero, with the small residual caused by finite-sample noise.

### Answer check

```python
assert abs(cov_uw - 5.8110385808) < 1e-12
assert abs(cov_uq) < 0.03
assert shortcut_gap < 1e-12
```
