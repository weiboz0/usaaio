# F5-probability — Practice p15 — Solution

For four independent errors, the average has variance $(1/4)^2(4)(0.2^2)=0.01$. Its standard deviation is therefore exactly 0.1 g, which the seeded normal simulation checks.

```python
import numpy as np

sd_single = 0.2
sd_avg = sd_single / np.sqrt(4)

SEED = 20260804
rng = np.random.default_rng(SEED)
average_errors = rng.normal(0, 0.2, size=(100_000, 4)).mean(axis=1)
sim_sd = np.std(average_errors)
```

Averaging four independent readings halves the spread from 0.2 g to 0.1 g, so it improves precision by a factor of two, not four; the simulation gives about 0.10012 g. Because standard error falls as $1/\sqrt n$, a fourfold reduction in spread would require $n=16$ independent readings.

### Answer check

```python
assert abs(sd_avg - 0.1) < 1e-12
assert abs(sim_sd - 0.10011785788607384) < 1e-12
assert abs(sim_sd - sd_avg) < 0.001
```
