# F5-probability — Unit Overview

**Probability and concentration.** This double-length unit builds from random variables, expectation, variance, independence, Gaussian models, simulation, and covariance to conditional probability, Bayes' rule, and a finite-sample Hoeffding guarantee. Algebraic claims are proved in the discrete register; simulations are fixed-seed evidence, not substitutes for proof.

## Prerequisites

- **F1-scientific-python:** arrays, masks, elementwise operations, broadcasting, axis aggregation, seeded NumPy generation, and matplotlib basics.
- **Math:** Calculus AB baseline. No continuous-density integral is required.

## Concepts taught

`random-variables`, `expectation`, `variance`, `independence`, `variance-of-sums`, `gaussian-distribution`, `sampling-simulation`, `covariance`, `conditional-probability`, `bayes-rule`, `hoeffding-inequality`.

## Practice sets

Twenty-five problems are organized by the manifest.

**Set A — fundamentals:** `p01`, `p02`, `p05`–`p07`, and `p20`. The new `p20` drills both directions of a conditional from a joint table in reduced normal form.

**Set B — exam register:** `p03`, `p04`, `p08`–`p12`, `p19`, and `p21`–`p24`. The new arc implements empirical conditioning and Bayes (`p21`), derives Bayes through total probability and confronts the base-rate trap (`p22`), derives the two-sided Hoeffding constant in normal form (`p23`), and compares a seeded simulation with its theoretical envelope (`p24`).

**Set C — integration and challenge:** `p13`–`p18` and `p25`. The final problem combines a posterior, a two-sided concentration audit, and a two-condition deployment decision.

Every statement has a separately authored solution notebook. Solve the student statement first; do not use a solution as a prompt.

## How to study

1. Run every session top to bottom and attempt checkpoints before expanding answers.
2. Keep directions visible: the event after a conditional bar supplies the denominator.
3. For Hoeffding, write the assumptions, widths, tail count, and raw formula before arithmetic. A raw bound above 1 is valid but vacuous.
4. Work Set A, then B, then C. Run immutable checks only after filling every requested identifier and written derivation.
5. Finish with `review.ipynb`; use its redo table to target weak spots.

Two derivations deserve recall from a blank page: Session 02's variance of a scaled independent sum, and Session 05's union-bound route from the two one-sided Hoeffding events to the factor 2.
