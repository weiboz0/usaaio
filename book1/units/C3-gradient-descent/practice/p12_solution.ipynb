# C3-gradient-descent — Practice p12 — Solution

For one uniform index, the discrete expectation is $E[\ell'_I(w)]=\sum_{i=1}^n P(I=i)\ell'_i(w)=\sum_{i=1}^n (1/n)\ell'_i(w)=L'(w)$. For $g=m^{-1}\sum_{j=1}^m\ell'_{I_j}(w)$, linearity gives $E[g]=m^{-1}\sum_{j=1}^m E[\ell'_{I_j}(w)]=m^{-1}\sum_{j=1}^m L'(w)=L'(w)$, so stochastic descent points in the full-gradient direction on average. In the exact anchor the derivatives are $(-2,-4,-12,-24)$, whose full mean and the mean over all six unordered pair-means both equal $-10.5$. The seeded sample uses finitely many random draws, so sampling imbalance makes its gap small rather than exactly zero.

```python
import numpy as np

x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 3.0, 5.0, 7.0])
w = 1.0
SEED = 20260804
rng = np.random.default_rng(SEED)
draws = rng.integers(0, 4, size=(10000, 2))

derivatives = 2 * (x * w - y) * x
full_grad = derivatives.mean()
pair_means = np.array([
    (derivatives[0] + derivatives[1]) / 2,
    (derivatives[0] + derivatives[2]) / 2,
    (derivatives[0] + derivatives[3]) / 2,
    (derivatives[1] + derivatives[2]) / 2,
    (derivatives[1] + derivatives[3]) / 2,
    (derivatives[2] + derivatives[3]) / 2,
])
avg_batch_grad = pair_means.mean()
anchor_gap = abs(avg_batch_grad - full_grad)
sample_grad = derivatives[draws].mean()
sample_gap = abs(sample_grad - full_grad)
derivatives, full_grad, avg_batch_grad, anchor_gap, sample_grad, sample_gap
```

### Answer check

```python
assert np.allclose(derivatives, [-2.0, -4.0, -12.0, -24.0], atol=1e-12, rtol=0)
assert np.isclose(full_grad, -10.5, atol=1e-12, rtol=0)
assert np.isclose(avg_batch_grad, -10.5, atol=1e-12, rtol=0)
assert np.isclose(anchor_gap, 0.0, atol=1e-12, rtol=0)
assert np.isclose(sample_grad, -10.5828, atol=1e-12, rtol=0)
assert np.isclose(sample_gap, 0.0828, atol=1e-12, rtol=0)
```
