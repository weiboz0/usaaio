# C3-gradient-descent — Practice p15 — Solution

Switching from mean loss to sum loss multiplies every gradient by $n=500$, changing the error factor from $0.9$ to $-49$ at the unchanged learning rate, so the same minimizer does not prevent explosive alternating steps. Dividing the learning rate by $500$ gives $\eta_{fixed}=0.0001$ and restores exactly the mean-loss trace. Raw sum losses also scale with dataset size and therefore are not directly comparable across differently sized datasets, whereas mean MSE normalizes that size effect.

```python
import numpy as np

eta = 0.05
n = 500
scale = n
factor_mean = 1 - 2 * eta
factor_sum = 1 - 2 * n * eta

def trace(gradient_scale, learning_rate):
    w = 0.0
    values = []
    for _ in range(4):
        w = w - learning_rate * 2 * gradient_scale * (w - 4)
        values.append(w)
    return np.array(values)

trace_mean = trace(1, eta)
trace_sum = trace(n, eta)
eta_fixed = eta / n
trace_fixed = trace(n, eta_fixed)
scale, factor_mean, factor_sum, trace_mean, trace_sum, eta_fixed, trace_fixed
```

### Answer check

```python
assert scale == 500
assert np.isclose(factor_mean, 0.9, atol=1e-12, rtol=0)
assert np.isclose(factor_sum, -49.0, atol=1e-12, rtol=0)
assert np.allclose(trace_mean, [0.4, 0.76, 1.084, 1.3756], atol=1e-12, rtol=0)
assert np.allclose(trace_sum, [200.0, -9600.0, 470600.0, -23059200.0], atol=1e-9, rtol=0)
assert np.isclose(eta_fixed, 0.0001, atol=1e-15, rtol=0)
assert np.allclose(trace_fixed, trace_mean, atol=1e-12, rtol=0)
```
