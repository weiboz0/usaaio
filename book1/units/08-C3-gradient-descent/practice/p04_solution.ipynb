# C3-gradient-descent — Practice p04 — Solution

Since $L'(w)=2(w-4)$, the two updates from $w_0=0$ give $w_1=12/5$ and $w_2=84/25=3.36$. This fraction is already reduced, so $p+q=84+25=109$, corresponding to choice C.

```python
import numpy as np
from fractions import Fraction

eta = Fraction(3, 10)
w0 = Fraction(0, 1)
w1 = w0 - eta * 2 * (w0 - 4)
w2 = w1 - eta * 2 * (w1 - 4)
normal_form_value = w2.numerator + w2.denominator
answer_p04 = "C"
w1, w2, normal_form_value
```

### Answer check

```python
assert np.isclose(float(w2), 3.36, atol=1e-12, rtol=0)
assert w2 == Fraction(84, 25)
assert normal_form_value == 109
assert answer_p04 == "C"
```
