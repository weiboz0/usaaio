# C3-gradient-descent — Practice p11 — Solution

Since $L'(w_t)=2w_t$, substitution gives $w_{t+1}=w_t-2\eta w_t=(1-2\eta)w_t$; with $r=1-2\eta$, induction proves $w_t=r^t w_0$ because the formula holds at $t=0$ and $w_{t+1}=r(r^t w_0)=r^{t+1}w_0$. For every starting point to approach zero we need and suffice to have $|r|<1$, since then $r^t\to0$, while if $|r|\ge1$ any $w_0\ne0$ has nonvanishing or growing magnitude; solving $|1-2\eta|<1$ together with $\eta>0$ gives exactly $0<\eta<1$. In the regimes $0<\eta<1/2$, $\eta=1/2$, $1/2<\eta<1$, $\eta=1$, and $\eta>1$, the iterates respectively keep their sign and shrink, become zero after one step, alternate and shrink, alternate with constant magnitude, and alternate while growing (apart from the fixed start $w_0=0$). For the numeric anchor $\eta=0.8$ gives $r=-0.6$, so ten steps preserve a positive sign and yield $5(-0.6)^{10}=0.030233088$, exactly the shrinking-alternating behavior predicted above.

```python
import numpy as np

eta = 0.8
w0 = 5.0
w10_loop = w0
for _ in range(10):
    w10_loop = w10_loop - eta * 2 * w10_loop
w10_formula = (1 - 2 * eta) ** 10 * w0
anchor_gap = abs(w10_loop - w10_formula)
w10_loop, w10_formula, anchor_gap
```

### Answer check

```python
assert np.isclose(w10_loop, 0.030233088, atol=1e-15, rtol=0)
assert np.isclose(w10_formula, 0.030233088, atol=1e-15, rtol=0)
assert anchor_gap < 1e-15
assert w10_loop > 0 and abs(w10_loop) < abs(w0)
```
