# C3-gradient-descent — Practice p05 — Solution

The descent update is elementwise and returns a new array, leaving its input unchanged. For $L(w)=\sum_k w_k^2$ with $\eta=0.25$, each update multiplies every coordinate by $1-2\eta=0.5$, so two steps multiply the starting vector by $0.25$.

```python
import numpy as np

w = np.array([1.0, -2.0, 0.5])
grad = np.array([4.0, -2.0, 0.0])
eta = 0.1

def descent_step(w, grad, eta):
    return w - eta * grad

w_new = descent_step(w, grad, eta)
w_after_one = descent_step(w, 2 * w, 0.25)
w_after_two = descent_step(w_after_one, 2 * w_after_one, 0.25)
w_new, w_after_two
```

### Answer check

```python
assert np.allclose(w_new, [0.6, -1.8, 0.5], atol=1e-12, rtol=0)
assert np.allclose(w_after_two, [0.25, -0.5, 0.125], atol=1e-12, rtol=0)
assert np.allclose(w, [1.0, -2.0, 0.5], atol=1e-12, rtol=0)
```
