# F6-svd-spectral — Practice p14 — Solution


All five eigenvalues are distinct and separated, so sign-fixing both routes makes an individual-column comparison legitimate.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
A = rng.normal(0, 1, (5, 5))
Q0 = np.linalg.eigh(A + A.T)[1]
target = np.array([9., 6., 4., 2.5, 1.])
S = Q0 @ np.diag(target) @ Q0.T


def signfix(Q):
    rows = np.abs(Q).argmax(axis=0)
    signs = np.sign(Q[rows, np.arange(Q.shape[1])])
    return Q * signs


vals_asc, vecs_asc = np.linalg.eigh(S)
vals_h, vecs_h = vals_asc[::-1], vecs_asc[:, ::-1]
vals_raw, vecs_raw = np.linalg.eig(S)
assert np.abs(vals_raw.imag).max() < 1e-12
vals_raw, vecs_raw = vals_raw.real, vecs_raw.real
order = np.argsort(vals_raw)[::-1]
vals_e, vecs_e = vals_raw[order], vecs_raw[:, order]
val_gap = np.abs(vals_e - vals_h).max()
target_gap = np.abs(vals_h - target).max()
raw_gap = np.abs(vecs_e - vecs_h).max()
fixed_gap = np.abs(signfix(vecs_e) - signfix(vecs_h)).max()
resid_h = np.abs(S @ vecs_h - vecs_h * vals_h).max()
resid_e = np.abs(S @ vecs_e - vecs_e * vals_e).max()
resid_max = max(resid_h, resid_e)
val_gap, target_gap, raw_gap, fixed_gap, resid_max
```

### Answer check


```python
assert val_gap < 1e-8 and target_gap < 1e-8
assert fixed_gap < 1e-8 and resid_max < 1e-8
assert raw_gap >= 0
```
