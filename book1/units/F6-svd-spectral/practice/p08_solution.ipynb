# F6-svd-spectral — Practice p08 — Solution


The implementation follows the entry definition directly; the library call appears only in the permitted cross-check line.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
A = rng.normal(0, 1, (5, 7))


def frob(A):
    return np.sqrt((A * A).sum())


frob_val = frob(A)
check_gap = abs(frob_val - np.linalg.norm(A))
scale_gap = abs(frob(-3 * A) - 3 * frob_val)
frob_val, check_gap, scale_gap
```

### Answer check


```python
assert np.isclose(frob_val, 6.249335522882215, rtol=0, atol=1e-12)
assert check_gap < 1e-10
assert scale_gap < 1e-10
```
