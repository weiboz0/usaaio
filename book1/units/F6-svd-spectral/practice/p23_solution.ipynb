# F6-svd-spectral — Practice p23 — Solution


Rank counts nonzero singular values, the Frobenius norm roots their squared sum, scaling multiplies singular values by the absolute scale, and diagonal singular values are absolute diagonal entries sorted descending.


```python
import numpy as np

rank_a = 2
frob_a = np.sqrt(9**2 + 3**2)
top_stretch = 9
sig_scale = 27
sig_d = np.array([5., 2., 1.])
len_s = 3
rank_a, frob_a, top_stretch, sig_scale, sig_d, len_s
```

### Answer check


```python
assert rank_a == 2
assert np.isclose(frob_a, 3 * np.sqrt(10.), rtol=0, atol=1e-12)
assert top_stretch == 9 and sig_scale == 27
assert np.array_equal(sig_d, np.array([5., 2., 1.]))
assert len_s == 3
```
