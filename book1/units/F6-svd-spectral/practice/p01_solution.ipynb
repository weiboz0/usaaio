# F6-svd-spectral — Practice p01 — Solution


Multiplication gives $S(1,-1)=(3,-3)=3(1,-1)$, so the answer is **D**.

Trap by trap: A uses the other eigen-direction but assigns it $3$ instead of $7$; B is turned into $(5,2)$; C has the same direction as $(1,1)$ but its eigenvalue is still $7$, not $14$; and E contradicts the direct multiplication.


```python
import numpy as np

S = np.array([[5., 2.], [2., 5.]])
q = np.array([1., -1.])
mapped = S @ q
answer_p01 = "D"
mapped
```

### Answer check


```python
assert np.array_equal(mapped, 3 * q)
assert answer_p01 == "D"
```
