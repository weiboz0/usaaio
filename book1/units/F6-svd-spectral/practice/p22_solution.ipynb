# F6-svd-spectral — Practice p22 — Solution


Five iterations give the energy estimate $12.4412$, while 200 give $12.5686$; each application multiplies eigen-direction components by their eigenvalues, so repeated normalization leaves the component belonging to the largest eigenvalue dominant. Since this matrix is positive semidefinite, its eigenvalues are non-negative, so the dominant component does not alternate sign as it would for a negative dominant eigenvalue.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
B = rng.normal(0, 1, (4, 6))
S = B @ B.T
x0 = np.ones(4) / 2.0


def power_iterate(S, x0, iters):
    x = x0.copy()
    for _ in range(iters):
        y = S @ x
        x = y / np.sqrt((y**2).sum())
    return x


q_top = power_iterate(S, x0, 200)
lam_est = q_top @ (S @ q_top)
resid_max = np.abs(S @ q_top - lam_est * q_top).max()
q_short = power_iterate(S, x0, 5)
est_short = q_short @ (S @ q_short)
lam_est, est_short, resid_max
```

### Answer check


```python
assert resid_max < 1e-8
assert np.isclose(np.sqrt((q_top**2).sum()), 1.0, rtol=0, atol=1e-10)
assert np.isclose(lam_est, 12.568604606839548, rtol=0, atol=1e-10)
assert lam_est > est_short
```
