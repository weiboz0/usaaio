# F6-svd-spectral — Practice p19 — Solution


The verdicts follow directly from the numeric evidence below.


```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
W = rng.normal(0, 1, (5, 8))

s = np.linalg.svd(W, full_matrices=False, compute_uv=False)
s2 = np.linalg.svd(2 * W, full_matrices=False, compute_uv=False)
sT = np.linalg.svd(W.T, full_matrices=False, compute_uv=False)
scale_gap = np.abs(s2 - 2 * s).max()
claim1 = bool(scale_gap < 1e-10)
transpose_gap = np.abs(sT - s).max()
claim2 = bool(transpose_gap < 1e-10)
desc_ok = bool(np.all(np.diff(s) <= 0))
claim3 = not desc_ok
min_sigma = s.min()
claim4 = bool(min_sigma < 0)
scale_gap, transpose_gap, desc_ok, min_sigma
```

Claim 1 is **true**: `scale_gap` is at machine scale. Claim 2 is **true**: `transpose_gap` is at machine scale. Claim 3 is **false**: `desc_ok` is true, so `svd` returns descending values. Claim 4 is **false**: `min_sigma` is positive even though the matrix contains negative entries.


### Answer check


```python
assert claim1 is True and scale_gap < 1e-10
assert claim2 is True and transpose_gap < 1e-10
assert claim3 is False and desc_ok is True
assert claim4 is False and min_sigma >= 0
assert np.isclose(min_sigma, 0.9316583210804052, rtol=0, atol=1e-12)
```
