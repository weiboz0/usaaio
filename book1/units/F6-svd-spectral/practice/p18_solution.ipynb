# F6-svd-spectral — Practice p18 — Solution


The singular values already arrive in descending order. The error curve is read from the tail identity, while one directly assembled rank-3 matrix cross-checks that route.


```python
import numpy as np
import matplotlib.pyplot as plt

SEED = 20260804
rng = np.random.default_rng(SEED)
strengths = np.array([20., 12., 7., 4.])
Uraw = rng.normal(0, 1, (40, 4))
Vraw = rng.normal(0, 1, (12, 4))
Un = Uraw / np.sqrt((Uraw**2).sum(axis=0))
Vn = Vraw / np.sqrt((Vraw**2).sum(axis=0))
W = (Un * strengths) @ Vn.T + 0.15 * rng.normal(0, 1, (40, 12))

U, s, Vt = np.linalg.svd(W, full_matrices=False)
desc_ok = bool(np.all(np.diff(s) <= 0))
fro = np.linalg.norm(W)
rel = np.array([np.sqrt((s[r:]**2).sum()) / fro for r in range(13)])
W3 = U[:, :3] @ np.diag(s[:3]) @ Vt[:3, :]
cross_gap = abs(rel[3] - np.linalg.norm(W - W3) / fro)
qualifies = rel <= 0.10
assert qualifies.any()
r_star = int(np.argmax(qualifies))
floats_full = 40 * 12
floats_rstar = r_star * (40 + 12 + 1)
floats_saved = floats_full - floats_rstar
s, rel, cross_gap, r_star, floats_rstar, floats_saved
```

```python
rs = np.arange(13)
plt.plot(rs, rel, marker="o")
plt.axhline(0.10, linestyle="--")
plt.xlabel("r")
plt.ylabel("relative Frobenius error")
plt.grid(True, alpha=0.3)
plt.show()
```

The elbow is at about $r=4$, after the four planted directions have been retained. Past that point only the much smaller perturbation directions remain, so each additional atom removes little squared content and the curve flattens.


### Answer check


```python
assert desc_ok is True
assert cross_gap < 1e-10
assert np.isclose(rel[0], 1.0, rtol=0, atol=1e-12)
assert r_star == 5
assert floats_full == 480 and floats_rstar == 265 and floats_saved == 215
```
