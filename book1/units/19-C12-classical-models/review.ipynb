## Comparison checklist and common repairs

For every choice, state supervision, objective, geometry, scaling, output/probability semantics,
interpretability, nonlinear capacity, and validation. Repair common failures as follows:

- overflow in sigmoid/BCE: use sign-aware sigmoid and stable logit loss;
- wrong SVM active set: compute signed margins before hinge;
- wrong tree split: weight child impurities and apply the exact tie hierarchy;
- unstable single tree: consider depth/pruning or a validated bagging/forest alternative;
- fake boosting: update and normalize sample weights every round;
- inconsistent Lloyd trace: compute labels, centroids, and WCSS from the same state;
- misleading comparison: use identical leakage-safe folds and task-appropriate metrics.
