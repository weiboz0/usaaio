# C12-classical-models — Practice p02 — Solution


The signed margins are $t_if_i=(2,-1/2,1,-3)$.  Their hinge losses are
$(0,3/2,0,4)$.  Rows 2 and 4 are active; equality at row 3 is inactive under the stated
convention.  Activity at this snapshot alone does not certify a nonzero fitted dual coefficient,
so the unique choice is **A**.


```python
import numpy as np
ATOL = 1e-12
RTOL = 1e-10
answer_p02 = "A"
scores_p02 = np.array([-2.0, 0.5, 1.0, -3.0])
labels_p02 = np.array([-1.0, -1.0, 1.0, 1.0])
margins_p02 = labels_p02 * scores_p02
hinges_p02 = np.maximum(0.0, 1.0 - margins_p02)
active_p02 = margins_p02 < 1.0
```

### Answer check


```python
assert answer_p02 == "A"
assert np.allclose(margins_p02, [2.0, -0.5, 1.0, -3.0], atol=ATOL, rtol=RTOL)
assert np.allclose(hinges_p02, [0.0, 1.5, 0.0, 4.0], atol=ATOL, rtol=RTOL)
assert np.array_equal(active_p02, [False, True, False, True])
```
