# C12-classical-models — Practice p01 — Solution


From odds $p/(1-p)=3$, solving gives $p=3/4$.  Thus the thresholded class is
1 and $dp/dz=p(1-p)=(3/4)(1/4)=3/16$.  The unique choice is **A**.


```python
import numpy as np
ATOL = 1e-12
RTOL = 1e-10
answer_p01 = "A"
odds_p01 = 3.0
probability_p01 = odds_p01 / (1.0 + odds_p01)
derivative_p01 = probability_p01 * (1.0 - probability_p01)
prediction_p01 = int(probability_p01 >= 0.5)
```

### Answer check


```python
assert answer_p01 == "A"
assert np.isclose(probability_p01, 0.75, atol=ATOL, rtol=RTOL)
assert np.isclose(derivative_p01, 3.0 / 16.0, atol=ATOL, rtol=RTOL)
assert prediction_p01 == 1
```
