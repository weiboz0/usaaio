# C12-classical-models — Practice p05 — Solution


The assignment is $(0,0,1,1)$.  The updated centroids are $1/2$ and $13/2$.
The squared residuals are $1/4,1/4,9/4,9/4$, so WCSS is $5=5/1$.  Here $a=5,b=1$,
$b>0$, and $\gcd(5,1)=1$; therefore $a+b=6$ and the choice is **A**.


```python
from fractions import Fraction
from math import gcd
answer_p05 = "A"
assignments_p05 = (0, 0, 1, 1)
centroids_p05 = (Fraction(1, 2), Fraction(13, 2))
residuals_p05 = (Fraction(1, 4), Fraction(1, 4), Fraction(9, 4), Fraction(9, 4))
wcss_p05 = sum(residuals_p05, Fraction(0, 1))
a_p05, b_p05 = wcss_p05.numerator, wcss_p05.denominator
```

### Answer check


```python
assert answer_p05 == "A"
assert assignments_p05 == (0, 0, 1, 1)
assert centroids_p05 == (Fraction(1, 2), Fraction(13, 2))
assert wcss_p05 == Fraction(5, 1)
assert b_p05 > 0 and gcd(abs(a_p05), b_p05) == 1 and a_p05 + b_p05 == 6
```
