# C12-classical-models — Classical Supervised and Unsupervised Models (Unit Overview)

This is the index for six 90-minute sessions. The unit connects five model families through a
single question: **what objective, geometry, output, and validation method match the task?**
The examples are small enough to audit line by line and use fixed seed **20260804**.

## Unit goals

By the end of the unit, you can:

- derive binary logistic regression from odds through mean binary cross-entropy and its gradient;
- explain linear and kernel support-vector machines using signed margins, hinge loss, support
  vectors, and the roles of `C` and `gamma`;
- compute deterministic tree splits and implement a depth-limited classification tree;
- distinguish voting, bagging, random forests, and boosting, including an AdaBoost update;
- implement Lloyd's k-means algorithm with deterministic ties and empty-cluster handling; and
- compare models by supervision, objective, geometry, scaling, probability output,
  interpretability, nonlinear capacity, and validation method.

## Prerequisites recap

The declared prerequisites are **C1-ml-fundamentals**, **C2-linear-models**,
**C3-gradient-descent**, **C4-classical-ml-practice**, and
**F7-kernels-convex-optimization**. We reuse supervised versus unsupervised learning,
train/validation/test roles, bias and variance, affine scores, gradient descent, feature scaling,
pipelines, cross-validation, kernels, Lagrangians, and complementary slackness. Binary
cross-entropy is derived from binary-event probabilities here; no undeclared loss result is
assumed.
