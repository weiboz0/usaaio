## 1. The Dot Product, Algebraically

**Motivation.**
Norms measure one vector at a time.
The next question needs *two* vectors: do they point the same way, opposite
ways, or somewhere in between?
The dot product is the single number that answers this, and it is the
workhorse behind everything else in this unit.

**Definition (algebraic).**
For vectors $a$ and $b$ with the same number of entries,
$$a \cdot b = \textstyle\sum_i a_i b_i$$
— multiply matching entries, add everything up.

**The algebra rules** (used constantly from here on):

- symmetry: $a\cdot b = b\cdot a$;
- distributivity: $a\cdot(b + c) = a\cdot b + a\cdot c$;
- scalar pull-out: $(c\,a)\cdot b = c\,(a\cdot b)$ for any number $c$;
- self dot product: $a \cdot a = \sum_i a_i^2 = \|a\|^2$ — the squared norm
  from Session 1.

**Worked example.**
$(1, 2, 3)\cdot(4, -5, 6) = 4 - 10 + 18 = 12$.
NumPy offers three equivalent spellings:

```python
a = np.array([1.0, 2.0, 3.0])
b = np.array([4.0, -5.0, 6.0])

print(np.sum(a * b), np.dot(a, b), a @ b)     # three spellings, same number

# self dot product == squared norm
print(a @ a, np.linalg.norm(a) ** 2)
```

## 2. The Dot Product, Geometrically

**Motivation.**
The sum $\sum_i a_i b_i$ is easy to compute but opaque — what does the
number *mean*?
The answer is a second, geometric formula for the same quantity.

**Definition (geometric).**
$$a \cdot b = \|a\|\,\|b\|\cos\theta,$$
where $\theta$ is the angle between the two arrows.

**Why the two definitions agree.**
Apply the law of cosines to the triangle whose sides are $a$, $b$, and
$a - b$:
$$\|a-b\|^2 = \|a\|^2 + \|b\|^2 - 2\,\|a\|\|b\|\cos\theta .$$
Expand the left side with Session 1's rules
($\|a-b\|^2 = (a-b)\cdot(a-b) = a\cdot a - 2\,a\cdot b + b\cdot b$):
$$\|a\|^2 - 2\,a\cdot b + \|b\|^2 = \|a\|^2 + \|b\|^2 - 2\,\|a\|\|b\|\cos\theta ,$$
then cancel and divide by $-2$: $\;a \cdot b = \|a\|\|b\|\cos\theta$.
So the easy-to-compute sum *is* the angle formula in disguise — compute the
cheap version, get the geometry for free.

**Worked example (both definitions, same answer).**
Let $a = (2, 0)$ and $b = (1, \sqrt3)$.

- *Algebraic:* $a \cdot b = 2\cdot 1 + 0\cdot\sqrt3 = 2$.
- *Geometric:* $\|a\| = 2$ and $\|b\| = \sqrt{1+3} = 2$; $a$ points along
  the horizontal axis while $b$ rises $\sqrt3$ over a run of 1, so
  $\theta = 60^\circ$; hence
  $a\cdot b = 2 \cdot 2 \cdot \cos 60^\circ = 4 \cdot \tfrac12 = 2$. Same.

```python
a = np.array([2.0, 0.0])
b = np.array([1.0, np.sqrt(3.0)])

algebraic = np.sum(a * b)
geometric = np.linalg.norm(a) * np.linalg.norm(b) * np.cos(np.pi / 3)   # 60 degrees

print("algebraic:", algebraic, "  geometric:", geometric)
```

## 3. Reading the Sign

**Motivation.**
Often the *exact* value of $a\cdot b$ matters less than its sign — a
three-way classification of how two directions relate, available for the
price of one sum.

**The rule.**
For nonzero vectors, $\|a\|$ and $\|b\|$ are positive, so the sign of
$a \cdot b$ is the sign of $\cos\theta$:

| $a\cdot b$ | angle $\theta$ | meaning |
|---|---|---|
| positive | less than $90^\circ$ | broadly the same direction |
| zero | exactly $90^\circ$ | perpendicular |
| negative | more than $90^\circ$ | broadly opposite |

The "exactly zero means perpendicular" row is important enough to get its
own name and its own session — Session 3 builds on it heavily.

**Worked example.**
$(1, 4)\cdot(4, -1) = 4 - 4 = 0$: perpendicular, no protractor required.
$(2, 1)\cdot(1, 3) = 5 > 0$: an acute angle.
$(-1, 2)\cdot(3, 1) = -1 < 0$: an obtuse angle.

```python
pairs = np.array([[[1.0, 4.0], [4.0, -1.0]],
                  [[2.0, 1.0], [1.0, 3.0]],
                  [[-1.0, 2.0], [3.0, 1.0]]])

dots3 = np.sum(pairs[:, 0, :] * pairs[:, 1, :], axis=1)   # all three dots at once
print(dots3)   # [0, 5, -1] -> right, acute, obtuse
```

## 4. Cosine Similarity

**Motivation.**
The dot product mixes two things together: the lengths of the vectors and
the angle between them.
Often we only care about *direction agreement*.
Compare two pieces of text by their word counts: a long article and a short
paragraph about the same topic should register as similar, even though the
article's count vector is far longer.
The fix: take the dot product, then *divide the lengths back out*.

**Definition.**
$$\text{cosine similarity}(a, b) = \frac{a \cdot b}{\|a\|\,\|b\|} = \cos\theta$$

By the geometric formula, this is exactly the cosine of the angle between
$a$ and $b$, so it always lands in $[-1, 1]$:

- $+1$: same direction (angle $0^\circ$);
- $0$: perpendicular (angle $90^\circ$) — no direction agreement;
- $-1$: exactly opposite (angle $180^\circ$).

Rescaling either vector by a positive number does not change the value: the
factor shows up in the numerator and the denominator and cancels.
That is precisely what the word-count situation needs.

**Worked example 1 (small vectors, by hand).**
$a = (1, 0)$, $b = (1, 1)$:
$$\frac{a\cdot b}{\|a\|\,\|b\|} = \frac{1}{1\cdot\sqrt2} = \frac{\sqrt2}{2} \approx 0.707,$$
which is $\cos 45^\circ$ — correct, since $(1, 1)$ sits at $45^\circ$ from
$(1, 0)$.

**Worked example 2 (word-count vectors).**
Count how often the words ("cat", "dog", "piano") appear in three sentences:

- sentence 1: "the cat watched the cat watch the dog" → $c_1 = (2, 1, 0)$
- sentence 2: "a dog and another dog chased a cat" → $c_2 = (1, 2, 0)$
- sentence 3: "piano, piano, piano!" → $c_3 = (0, 0, 3)$

By hand:
$\cos(c_1, c_2) = \dfrac{2 + 2 + 0}{\sqrt5\cdot\sqrt5} = \dfrac45 = 0.8$
(both are pet sentences), while $\cos(c_1, c_3) = 0$: the count vectors
share no nonzero entries, so the piano sentence shows no direction agreement
with the cat-and-dog one.

```python
def cosine_similarity(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))


print(cosine_similarity(np.array([1.0, 0.0]), np.array([1.0, 1.0])))    # ~0.707
print(cosine_similarity(np.array([1.0, 0.0]), np.array([-2.0, 0.0])))   # -1: opposite
```

```python
# Word-count vectors over the words ("cat", "dog", "piano"):
c1 = np.array([2.0, 1.0, 0.0])
c2 = np.array([1.0, 2.0, 0.0])
c3 = np.array([0.0, 0.0, 3.0])

print("sentence 1 vs sentence 2:", cosine_similarity(c1, c2))           # 0.8
print("sentence 1 vs sentence 3:", cosine_similarity(c1, c3))           # 0.0
print("sentence 1 vs 10x sentence 2:", cosine_similarity(c1, 10 * c2))  # still 0.8
```

## 5. Unit Vectors and Normalization

**Motivation.**
Section 4 kept dividing by lengths.
It is often cleaner to divide *once, up front*: replace each vector by a
version of itself with length exactly 1.
Then every vector carries only its direction, and comparisons become plain
dot products.

**Definition.**
A **unit vector** is a vector whose Euclidean norm is 1.
**Normalizing** a nonzero vector $v$ means rescaling it to unit length:
$$\hat v = \frac{v}{\|v\|}.$$
(The hat marks unit vectors.)
Dividing every entry by the same positive number keeps the direction, and it
fixes the length: $\|\hat v\| = \frac{\|v\|}{\|v\|} = 1$.
The zero vector is the one vector you cannot normalize — it has no
direction, and dividing by $\|0\| = 0$ is undefined.

**Worked example.**
$v = (3, 4)$ has $\|v\| = 5$, so $\hat v = (3/5,\ 4/5) = (0.6, 0.8)$, and
indeed $\sqrt{0.36 + 0.64} = 1$.

**Payoff.**
Cosine similarity is just the dot product of the two unit vectors:
$$\frac{a\cdot b}{\|a\|\,\|b\|} = \frac{a}{\|a\|}\cdot\frac{b}{\|b\|} = \hat a \cdot \hat b .$$
Normalize once; every later comparison is a single cheap dot product.

```python
def normalize(v):
    return v / np.linalg.norm(v)


v = np.array([3.0, 4.0])
v_hat = normalize(v)
print("v_hat =", v_hat, "   norm:", np.linalg.norm(v_hat))

# cosine similarity == dot product of the unit vectors
a = np.array([2.0, 1.0])
b = np.array([1.0, 3.0])
print(cosine_similarity(a, b), normalize(a) @ normalize(b))
```

## 6. Worked Exam-Style Example

Unit vectors are a favorite target for **normal-form numeric MC** items:
the components are irrational, so the exam forces a unique integer encoding.
Here is one solved in full, in the exam's register.

---

**Problem (multiple choice; reasoning is required).**
Let $v = (6, 4)$.
Its unit vector can be written in the normal form
$$\hat v = \frac{1}{\sqrt{a}}\,(b,\ c),$$
where $a, b, c$ are positive integers with $\gcd(b, c) = 1$.
Compute $S = a + b + c$.

**(A)** 13  **(B)** 18  **(C)** 23  **(D)** 52  **(E)** 62

---

**Solution.**

*Step 1 — the norm.*
$\|v\| = \sqrt{36 + 16} = \sqrt{52} = 2\sqrt{13}$.

*Step 2 — divide and reduce.*
$$\hat v = \frac{(6, 4)}{2\sqrt{13}} = \frac{1}{\sqrt{13}}\,(3,\ 2).$$
The common factor 2 *must* come out of the entries: $\gcd(3, 2) = 1$
satisfies the condition, whereas $\frac{1}{\sqrt{52}}(6, 4)$ has
$\gcd(6, 4) = 2$ and is rejected by the normal form.

*Step 3 — decode the distractors before answering.*
$13 + 3 + 2 = 18$ → **(B)**.
Option **(E)** is the unreduced trap: $52 + 6 + 4 = 62$.
Option **(D)** is the bare squared norm.
Recognizing *which mistake each wrong option encodes* is a reliable
self-check: if your number matches a trap, you know exactly what to redo.

**Answer: (B).**

## 7. Common Pitfalls

**Pitfall 1 — `a * b` is not the dot product.**
In NumPy, `*` is *elementwise* (as covered in F1-scientific-python): it
returns an array, not a number.
The dot product needs the sum — `np.sum(a * b)` — or the dedicated
operator `@`:

```python
a = np.array([1.0, 2.0, 3.0])
b = np.array([4.0, -5.0, 6.0])

broken = a * b            # BROKEN as a "dot product": still an array
fixed = np.sum(a * b)     # fix: sum the products (or use a @ b)

print("broken:", broken, "  type: array of shape", broken.shape)
print("fixed :", fixed)
```

The symptom is a *shape* surprise: anything downstream expecting one number
receives three.
The same predict-the-shape habit from Session 1 catches it: a dot product of
two `(3,)` arrays must be a scalar.

**Pitfall 2 — cosine similarity of the zero vector.**
The zero vector has no direction, so "its angle with $b$" is meaningless —
and the formula agrees by dividing by zero and producing `nan`
(not-a-number), usually with a runtime warning:

```python
def cosine_similarity(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))


z = np.zeros(2)
b = np.array([1.0, 1.0])

with np.errstate(invalid="ignore"):        # silence the warning for the demo
    broken = cosine_similarity(z, b)       # BROKEN: 0/0
print("broken:", broken)


def cosine_similarity_safe(a, b):          # fix: refuse meaningless input
    na, nb = np.linalg.norm(a), np.linalg.norm(b)
    if na == 0.0 or nb == 0.0:
        raise ValueError("cosine similarity is undefined for the zero vector")
    return (a @ b) / (na * nb)


try:
    cosine_similarity_safe(z, b)
except ValueError as e:
    print("fixed :", e)
```

The fix is a *guard*, not a formula change: there is no correct value to
return, so the honest behavior is to refuse.
In practice this matters for word counts — an empty sentence produces the
zero vector, and a pipeline that feeds it into cosine similarity silently
fills with `nan`s.
