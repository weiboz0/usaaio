# F2-vectors — Practice p10 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
X = rng.normal(size=(150, 4))
q = rng.normal(size=4)
print(X.shape, q)
```

## Part (a)

```python
def euclidean_to_query(X, q):
    diffs = X - q                            # (150, 4) - (4,) broadcasts
    return np.sqrt(np.sum(diffs**2, axis=1))


def manhattan_to_query(X, q):
    diffs = X - q
    return np.sum(np.abs(diffs), axis=1)


euclid_all = euclidean_to_query(X, q)
manhattan_all = manhattan_to_query(X, q)
print(euclid_all[:3])
print(manhattan_all[:3])
```

Broadcasting subtracts `q` from all 150 rows in one step, and the `axis=1`
aggregation collapses each row of differences to a single distance.
The two functions differ only in the norm applied to the difference —
squares-then-root versus absolute-values-then-sum.

## Part (b)

```python
check_dominance = bool(np.all(manhattan_all >= euclid_all))
print(check_dominance)
```

For any difference vector, $\sum_i |d_i| \geq \sqrt{\sum_i d_i^2}$: squaring
the left side produces every $d_i^2$ term plus nonnegative cross terms, so it
is at least the right side squared.
`np.all` confirms the inequality across the whole batch at once — no loop
needed.

## Part (c)

```python
closest = int(np.argmin(euclid_all))
print(closest, euclid_all[closest])
```

`np.argmin` (an F1-scientific-python aggregation) returns the index of the
smallest distance, identifying the row of `X` nearest to the query under the
Euclidean ruler.
