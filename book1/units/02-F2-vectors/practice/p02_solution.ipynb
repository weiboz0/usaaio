# F2-vectors — Practice p02 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

## Part (a)

```python
# p - q = (-3, -4)
euclid_pq_by_hand = 5.0      # sqrt(9 + 16) = sqrt(25)
manhattan_pq_by_hand = 7.0   # 3 + 4
print(euclid_pq_by_hand, manhattan_pq_by_hand)
```

Both distances start from the same difference vector $(-3, -4)$.
The Euclidean ruler squares, sums, and takes a root (the 3-4-5 right
triangle); the Manhattan ruler just adds the absolute entries, so it is
always at least as large.

## Part (b)

```python
def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))


def manhattan_distance(a, b):
    return np.sum(np.abs(a - b))


p = np.array([1.0, 2.0])
q = np.array([4.0, 6.0])
print(euclidean_distance(p, q), manhattan_distance(p, q))
```

Each function is "norm of the difference": form `a - b` elementwise, then
apply the matching norm formula from the lesson.
Both reproduce the hand answers 5.0 and 7.0.

## Part (c)

```python
points = rng.integers(-9, 10, size=(6, 2)).astype(float)
q = np.array([1.0, 2.0])
print(points)
```

```python
diffs = points - q                              # (6, 2) - (2,) broadcasts to (6, 2)
euclid_all = np.sqrt(np.sum(diffs**2, axis=1))  # one distance per row
manhattan_all = np.sum(np.abs(diffs), axis=1)

closest_euclid = int(np.argmin(euclid_all))
closest_manhattan = int(np.argmin(manhattan_all))

print("Euclidean :", euclid_all, "-> closest:", closest_euclid)
print("Manhattan :", manhattan_all, "-> closest:", closest_manhattan)
```

Broadcasting subtracts `q` from every row in one step, and `axis=1` sums
across each row's entries, so all six distances come out of a single
expression per ruler.
With this seed both rulers agree: point 2, at $(3, 3)$, is closest to
$q = (1, 2)$ either way — the two rulers can order points differently in general,
but a point this near the query wins under both.

### Answer check

```python
assert euclid_pq_by_hand == 5.0
assert manhattan_pq_by_hand == 7.0
assert np.isclose(euclidean_distance(np.array([1.0, 2.0]), np.array([4.0, 6.0])), 5.0, atol=1e-9, rtol=0)
assert np.isclose(manhattan_distance(np.array([1.0, 2.0]), np.array([4.0, 6.0])), 7.0, atol=1e-9, rtol=0)
assert euclid_all.shape == (6,) and manhattan_all.shape == (6,)
assert np.allclose(euclid_all, np.linalg.norm(points - np.array([1.0, 2.0]), axis=1), atol=1e-9, rtol=0)
assert np.isclose(manhattan_all[2], 3.0, atol=1e-9, rtol=0)
assert closest_euclid == 2 and closest_manhattan == 2
print("All checks passed.")
```
