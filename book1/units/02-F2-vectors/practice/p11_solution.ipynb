# F2-vectors — Practice p11 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
d = np.array([3.0, 4.0])
Y = np.array([[4.0, 3.0],
              [8.0, 6.0],
              [-3.0, -4.0],
              [1.0, 7.0],
              [5.0, 0.0]])
print(Y.shape)
```

## Part (a)

```python
u = d / np.sqrt(np.sum(d**2))
print(u)
```

$\|d\| = \sqrt{9 + 16} = 5$, so $u = (0.6, 0.8)$ — the 3-4-5 triangle again.
Every later part measures along this unit direction, which is why
normalizing comes first.

## Part (b)

```python
reach = np.sum(Y * u, axis=1)        # (5,): y_i . u for every row
P = reach[:, np.newaxis] * u         # (5, 1) * (2,) -> (5, 2)
print(reach)
print(P)
```

The axis sum computes all five dot products with `u` in one expression, and
reshaping `reach` to a `(5, 1)` column lets broadcasting scale `u` by a
different amount per row.
The reaches are $(4.8,\ 9.6,\ -5.0,\ 6.2,\ 3.0)$ — note the negative one:
point 3 reaches *backwards* along the direction.

## Part (c)

```python
R = Y - P
res_norms = np.sqrt(np.sum(R**2, axis=1))
best = int(np.argmin(res_norms))
worst = int(np.argmax(res_norms))
print(res_norms)
print("best:", best, " worst:", worst)
```

The residual norms are $(1.4,\ 2.8,\ 0.0,\ 3.4,\ 4.0)$: `best` is row 2 and
`worst` is row 4.
**Interpretation:** a residual norm of exactly 0 means the point lies *on*
the line through the direction $u$ — its projection reproduces it perfectly.
Row 2 is $(-3, -4) = -d$, which points exactly opposite the direction but
still lies on the same line; row 4, $(5, 0)$, keeps the largest part of
itself outside the direction.

### Answer check

```python
assert np.allclose(u, np.array([0.6, 0.8]), atol=1e-9, rtol=0)
assert np.allclose(reach, np.array([4.8, 9.6, -5.0, 6.2, 3.0]), atol=1e-9, rtol=0)
assert np.allclose(res_norms, np.array([1.4, 2.8, 0.0, 3.4, 4.0]), atol=1e-9, rtol=0)
assert best == 2 and worst == 4
assert np.allclose(np.sum((Y - P) * u, axis=1), 0.0, atol=1e-9, rtol=0)   # residual rows orthogonal to u
print("All checks passed.")
```
