# C12-classical-models — Practice p22

**Type:** scenario analysis · **Difficulty:** intro · **Concepts:** logistic-regression, svm, classical-model-comparison

**Time budget:** 45 minutes

A hospital triage model receives six continuous measurements with very different units. A false
positive triggers an expensive test; a false negative is clinically worse. The deployed output
must be a calibrated risk probability, but a teammate proposes choosing between logistic
regression and a linear SVM by test accuracy after scaling the entire dataset.

Write an audit that:

1. identifies every leakage and objective/output mismatch in the proposal;
2. gives a train/validation/test or nested-CV pipeline with scaling fitted only on training folds;
3. compares BCE/probability output with hinge/margin output without treating an SVM score as a
   probability;
4. states how to choose regularization and a decision threshold from the asymmetric costs;
5. names at least one discrimination metric and two probability-quality/calibration checks;
6. explains when calibrated logistic regression is the natural first baseline and how a
   separately calibrated SVM could be evaluated fairly.

Use no numeric result not supplied by the scenario. A generic “cross-validate both” response
without the full data and decision protocol receives no credit.
