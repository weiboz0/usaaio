```python
import numpy as np
import matplotlib.pyplot as plt

SEED = 20260804
rng = np.random.default_rng(SEED)
```

**Part a — generate (random-seeding).** Set `SEED = 20260804`, build `rng`
with `np.random.default_rng`, and draw **`trials`** — a `(2000, 5)` array where
each row is one trial of rolling 5 dice (integer faces 1–6).

```python
SEED = 20260804
rng = np.random.default_rng(SEED)
trials = rng.integers(1, 7, size=(2000, 5))
print(trials.shape)
print(trials[:3])
```

One seeded generator, one draw. Asking for the whole `(2000, 5)` array in a single `integers` call keeps the experiment reproducible: same seed, same 10,000 faces, every run.

**Part b — summarize (aggregation-axis), using `trials` from part a.**
Compute:

- **`totals`** — the 5-dice total of each trial, shape `(2000,)`
- **`high_fraction`** — the fraction of trials with total ≥ 20
- **`per_die_mean`** — the average face of each die position across trials,
  shape `(5,)`

```python
totals = trials.sum(axis=1)
high_fraction = (totals >= 20).mean()
per_die_mean = trials.mean(axis=0)
print(totals[:10])
print("fraction with total >= 20:", high_fraction)
print("per-position averages:", per_die_mean.round(3))
```

Per-trial totals collapse the dice axis (`axis=1`); per-position averages collapse the trial axis (`axis=0`). The mask-`.mean()` idiom turns counting into one expression. All five per-position averages hover near 3.5, as they should for fair dice.

**Part c — picture (matplotlib-basics), using `totals` from part b.** Draw a
histogram of `totals` with bins covering every possible total (5 through 30 —
26 values), with a title and both axis labels. In one sentence: why do middle
totals dominate the ends?

```python
plt.figure(figsize=(7, 4))
plt.hist(totals, bins=26)
plt.title("Totals of 2000 five-dice trials")
plt.xlabel("five-dice total")
plt.ylabel("how many trials")
plt.show()
# Middle totals dominate because far more face combinations produce them than
# produce extreme totals like 5 (all ones) or 30 (all sixes).
```

The histogram takes the one array `totals` and shows where trials pile up: a big bulge in the middle, thin tails at 5 and 30 — many combinations make 17, exactly one makes 5.
