# F1-scientific-python — Practice p15 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
x = np.array([12.0, -7.0, 3.0, 40.0, -2.0, 55.0, 8.0])

# Drill loops (all correct, all to be replaced):
doubled_loop = []
for v in x:
    doubled_loop.append(2 * v)
doubled_loop = np.array(doubled_loop)

sq_total_loop = 0.0
for v in x:
    sq_total_loop += v ** 2

cleaned_loop = []
for v in x:
    cleaned_loop.append(0.0 if v < 0 else v)
cleaned_loop = np.array(cleaned_loop)

big_count_loop = 0
for v in x:
    if v > 10:
        big_count_loop += 1

max_jump_loop = 0.0
for k in range(1, len(x)):
    jump = abs(x[k] - x[k - 1])
    if jump > max_jump_loop:
        max_jump_loop = jump

print(doubled_loop, sq_total_loop, cleaned_loop, big_count_loop, max_jump_loop)
```

**Drill 1.** **`doubled`** — every element times 2, one expression.

```python
doubled = 2 * x
print(np.array_equal(doubled, doubled_loop))
```

**Drill 2.** **`sq_total`** — the total of the squares, one expression.

```python
sq_total = (x ** 2).sum()
print(np.isclose(sq_total, sq_total_loop, atol=1e-9, rtol=0))
```

**Drill 3.** **`cleaned`** — negatives replaced by 0, one expression.

```python
cleaned = np.where(x < 0, 0.0, x)
print(np.array_equal(cleaned, cleaned_loop))
```

**Drill 4.** **`big_count`** — how many elements exceed 10, one expression.

```python
big_count = (x > 10).sum()
print(big_count == big_count_loop)
```

**Drill 5.** **`max_jump`** — the largest absolute difference between
*consecutive* elements. Hint: `x[1:]` and `x[:-1]` line up each element with
its predecessor.

```python
max_jump = np.abs(x[1:] - x[:-1]).max()
print(np.isclose(max_jump, max_jump_loop, atol=1e-9, rtol=0))
```

The two shifted slices pair `x[k]` with `x[k-1]` for every `k` at once; elementwise subtraction, `np.abs`, and `.max()` replace the whole tracking loop. This shifted-slice trick is worth memorizing.
