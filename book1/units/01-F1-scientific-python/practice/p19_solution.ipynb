**Part a — shape and entry formula.** Using the shape-compatibility rules,
derive `T`'s shape for general $n$, and explain why entry $(i, j)$ equals
$x_i - x_j$ (which stretched copy meets which at position $(i, j)$?). Set
**`shape_T`** to the shape for $n = 5$.

*(Write your derivation here.)*

```python
shape_T = (5, 5)
```

`x[:, None]` is `(n, 1)` and `x[None, :]` is `(1, n)`; each stretches its size-1 axis, so the result is `(n, n)`. The column copy repeats `x[i]` across row `i`, the row copy repeats `x[j]` down column `j`; at `(i, j)` they meet as `x[i] - x[j]`.

**Part b — two properties, argued then verified.** From the entry formula,
argue in the markdown cell that (i) every diagonal entry `T[i, i]` is 0, and
(ii) `T[i, j]` is always the negative of `T[j, i]`. Then verify both in code
for `x = np.array([3.0, -1.0, 4.0, 1.0, 5.0])`: check the diagonal via
integer-array indexing with `T[np.arange(5), np.arange(5)]`, and check
property (ii) by comparing `T` with `-(x[None, :] - x[:, None])`.

*(Write your argument here.)*

```python
x = np.array([3.0, -1.0, 4.0, 1.0, 5.0])
T = x[:, None] - x[None, :]

diag = T[np.arange(5), np.arange(5)]          # pairwise picks (i, i)
print("diagonal:", diag)

swapped = x[None, :] - x[:, None]             # entry (i, j) = x[j] - x[i]
print("T[i,j] == -T[j,i] everywhere:", np.array_equal(T, -swapped))
```

(i) `T[i, i] = x_i - x_i = 0` for every `i`. (ii) `T[j, i] = x_j - x_i = -(x_i - x_j) = -T[i, j]`. In code, `x[None, :] - x[:, None]` has entry `(i, j)` equal to `x_j - x_i` — exactly `T[j, i]` — so comparing `T` with its negation checks the property at all $n^2$ positions at once.

**Part c — counting zeros.** Argue: if all $n$ values in `x` are distinct,
`T` contains **exactly** $n$ zeros. (Where can $x_i - x_j = 0$ happen?) Then
verify with a mask count on the `x` above, setting **`zero_count`**.

*(Write your argument here.)*

```python
zero_count = (T == 0).sum()
print(zero_count)
```

`x_i - x_j = 0` requires `x_i = x_j`; with all values distinct that forces `i = j`, so zeros occur exactly on the diagonal — $n$ of them, here 5.
