```python
import numpy as np
import matplotlib.pyplot as plt

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
# Given: seeded data for Tasks C and D.
SEED = 20260804
rng = np.random.default_rng(SEED)
study_hours = rng.random(40) * 10                              # 40 values in [0, 10)
quiz_score = 40 + 5 * study_hours + rng.integers(-8, 9, size=40)
draws = rng.random(2000)                                       # 2000 values in [0, 1)
```

**Task A — line plot.** Set **`x`** to 100 evenly spaced values from -1 to 5
and **`y`** to $x^2 - 4x + 3$. Draw a line plot of `y` against `x` with a
title and both axis labels.

```python
x = np.linspace(-1, 5, 100)
y = x**2 - 4*x + 3
plt.figure(figsize=(6, 4))
plt.plot(x, y)
plt.title("y = x^2 - 4x + 3")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

`linspace` gives smooth, evenly spaced x-values; `plot` connects the 100 points in order, which reads as a smooth curve.

**Task B — two lines and a legend.** Set **`y2`** to $2x - 3$ over the same
`x`. Draw both curves on ONE figure, give each a `label=`, add a legend, a
title, and both axis labels. Where the curves cross, the two formulas are
equal — read off roughly where that happens.

```python
y2 = 2 * x - 3
plt.figure(figsize=(6, 4))
plt.plot(x, y, label="y = x^2 - 4x + 3")
plt.plot(x, y2, label="y = 2x - 3", linestyle="--")
plt.title("Curve vs. line")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

Two `plot` calls before one `show` share the same axes. The legend maps each `label=` to its line; the curves touch where x^2 - 4x + 3 = 2x - 3, i.e. near x = 1.3 and x = 4.7.

**Task C — scatter plot.** Using the given `study_hours` and `quiz_score`
arrays (40 students), draw a scatter plot of score against hours studied, fully
labeled. Why is scatter the right choice here rather than a line plot?

```python
plt.figure(figsize=(6, 4))
plt.scatter(study_hours, quiz_score)
plt.title("Quiz score vs. hours studied")
plt.xlabel("hours studied")
plt.ylabel("quiz score")
plt.show()
```

Each student is one independent (x, y) observation with no order connecting one point to the next, so connecting them with a line would draw meaningless zig-zags.

**Task D — histogram.** Draw a histogram of the given `draws` array with 20
bins, fully labeled. Then set **`counts`** and **`edges`** from
`np.histogram(draws, bins=20)` — the same numbers `plt.hist` drew — and check
that `counts` sums to 2000.

```python
plt.figure(figsize=(6, 4))
plt.hist(draws, bins=20)
plt.title("2000 generator draws")
plt.xlabel("value")
plt.ylabel("how many draws")
plt.show()

counts, edges = np.histogram(draws, bins=20)
print(counts.sum(), edges[0], edges[-1])
```

A histogram takes ONE array, splits its range into 20 equal bins, and bars show how many values landed in each. `np.histogram` returns those bar heights (`counts`, 20 of them) and the 21 bin boundaries.
