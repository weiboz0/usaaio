# F1-scientific-python — Unit Overview

**The first unit of the course.** It teaches the two workhorse tools of
scientific Python — **NumPy** (fast math on whole collections of numbers) and
**matplotlib** (plots) — from zero. Every later unit builds on this one.

## What you are assumed to know

Only the course baseline:

- **Math:** Calculus AB (algebra, functions and graphs, basic trig, limits,
  derivatives and integrals in one variable, exponentials and logs).
- **Python:** variables and types, lists/dicts/sets, `if`/`for`/`while`,
  functions, basic classes, reading and writing files.

Nothing else. If you have never imported a library before: `import numpy as
np` at the top of a notebook makes NumPy available under the short name `np`,
and that line opens every notebook in this unit.

## Concepts taught

`numpy-arrays`, `array-indexing-slicing`, `elementwise-ops`, `broadcasting`,
`vectorization`, `aggregation-axis`, `random-seeding`, `matplotlib-basics`,
`seaborn-programming`.

session's answers are collected at the end of that session's notebook.

## Practice sets

Practice lives in `practice/` as 24 problems in three sets (the manifest tags
each problem with its set and difficulty):

- **Set A — fundamentals** (p01, p02, p04, p05, p06, p15, p22): drills, one or two

Each problem `pNN.ipynb` has a worked `pNN_solution.ipynb` — attempt honestly
before opening it.

## How to study this unit

1. Work each session notebook top to bottom, **running every cell yourself**
   and doing every checkpoint before reading its answer.
2. After the mapped sessions, do the Set A problems for that material.
3. After Session 04, work Sets B and C in order.
4. Finish with `review.ipynb` (summary tables, idiom sheet, self-quiz); use

## Code in this unit

A quick reference for the library calls in this unit's code. Each one is explained in a comment (or in the text beside the code) where it first appears.

**New in this unit:**

- `np.arange` — `np.arange(n)` gives the integers `0, 1, ..., n-1`; `np.arange(start, stop, step)` stops before `stop`
- `time.perf_counter` — a clock reading in seconds; subtract two readings to time code
- `np.array` — builds an array from a list (or a list of lists for a 2-D grid)
- `.sum` — adds up the entries; `axis=0` sums down the columns, `axis=1` along the rows
- `np.zeros` — an array of zeros with the given shape
- `np.ones` — an array of ones with the given shape, e.g. `(2, 3)`
- `np.full` — an array of the given shape with every entry equal to one value
- `np.linspace` — `np.linspace(a, b, n)`: `n` evenly spaced points from `a` to `b`, both ends included
- `np.pad` — a new array with extra entries (zeros by default) added around the edges
- `.astype` — a copy converted to another dtype (float to int drops the decimals)
- `.reshape` — the same entries read in a new shape, e.g. `(3, 4)`
- `.copy` — a new array with its own data, so changes do not reach the original
- `.mean` — the average of the entries (of a boolean array: the fraction that is True)
- `np.sqrt` — the square root of each entry
- `np.maximum` — the larger of two values, entry by entry
- `.max` — the largest entry (per column or row with `axis=`)
- `np.where` — `np.where(cond, a, b)`: entry from `a` where `cond` is True, else from `b`
- `.append` — list append: adds one item at the end of a Python list
- `np.abs` — the absolute value of each entry
- `np.array_equal` — True when two arrays have the same shape and the same entries
- `.argmax` — the position of the largest entry
- `.argmin` — the position of the smallest entry
- `.round` — each entry rounded to the given number of decimals
- `np.minimum` — the smaller of two values, entry by entry
- `np.random.default_rng` — a random generator `rng` started from a seed, so draws repeat exactly
- `.random` — `rng.random(n)`: `n` uniform decimals in `[0, 1)`
- `.integers` — `rng.integers(low, high, size=n)`: random whole numbers from `low` up to `high - 1`
- `.choice` — `rng.choice(a, size=n)`: `n` random picks from the entries of `a`
- `.shuffle` — `rng.shuffle(a)`: puts the entries of `a` in a random order, in place
- `plt.figure` — starts a new figure; `figsize=(w, h)` sets its size in inches
- `plt.plot` — draws a line through the points `(x, y)`
- `plt.title` — sets the plot's title
- `plt.xlabel` — labels the x-axis
- `plt.ylabel` — labels the y-axis
- `plt.legend` — shows a key naming each `label=` curve
- `plt.show` — displays the finished figure
- `plt.scatter` — draws one dot per `(x, y)` point
- `plt.hist` — a histogram: one bar per bin, its height the count of values in it
- `sns.set_theme` — sets seaborn's plot style, e.g. a light grid
- `plt.subplots` — makes a figure `fig` and axes `ax` to draw on
- `sns.histplot` — seaborn histogram of an array; `bins=` sets the bin edges, `ax=` the axes
- `.set` — `ax.set(title=..., xlabel=..., ...)`: sets several axes properties at once
- `np.histogram` — the histogram counts (and bin edges) as arrays, without drawing
- `.get_height` — one bar's height, i.e. its count
- `.get_xlabel` — the x-axis label string
- `.get_ylabel` — the y-axis label string
- `sns.scatterplot` — seaborn scatter plot; `hue=` and `style=` color and shape points by group
- `.get_text` — the string of one text item, such as one legend entry
- `.get_texts` — the text items of a legend
- `.get_legend` — the axes' legend box
- `.get_offsets` — the plotted `(x, y)` points of a scatter, shape `(n, 2)`
- `np.allclose` — True when every pair of entries differs by at most the tolerance (`atol`, `rtol`)
- `.get_title` — the axes' title string
- `.normal` — `rng.normal(loc, scale, size)`: bell-curve draws with mean `loc` and spread `scale`
- `.legend` — `ax.legend()`: shows a key naming each label on these axes
- `.axvline` — a vertical line across the axes at `x`
- `np.isclose` — True when two numbers differ by at most the tolerance (`atol`, `rtol`)
- `np.sort` — a sorted copy of the array, smallest first
- `.min` — the smallest entry (per column or row with `axis=`)
- `.set_title` — sets the axes' title
- `.set_xlabel` — labels the x-axis of these axes
- `.set_ylabel` — labels the y-axis of these axes
- `.set_xlim` — the x-range shown on these axes
