# F1-scientific-python — Unit Overview

**The first unit of the course.** It teaches the two workhorse tools of
scientific Python — **NumPy** (fast math on whole collections of numbers) and
**matplotlib** (plots) — from zero. Every later unit builds on this one.

## What you are assumed to know

Only the course baseline:

- **Math:** Calculus AB (algebra, functions and graphs, basic trig, limits,
  derivatives and integrals in one variable, exponentials and logs).
- **Python:** variables and types, lists/dicts/sets, `if`/`for`/`while`,
  functions, basic classes, reading and writing files.

Nothing else. If you have never imported a library before: `import numpy as
np` at the top of a notebook makes NumPy available under the short name `np`,
and that line opens every notebook in this unit.

## Concepts taught

`numpy-arrays`, `array-indexing-slicing`, `elementwise-ops`, `broadcasting`,
`vectorization`, `aggregation-axis`, `random-seeding`, `matplotlib-basics`,
`seaborn-programming`.

## Session map

The lesson runs as four class sessions, in order, in `lessons/`:

| Session | Notebook | ~Minutes | Covers |
| --- | --- | --- | --- |
| 01 | `lessons/01-arrays-and-indexing.ipynb` | 75 | why arrays; creating; dtype; shape/reshape; indexing, slicing, masks, integer-array indexing; pitfalls I; exam-style example |
| 02 | `lessons/02-broadcasting-and-vectorization.ipynb` | 90 | elementwise ops; `np.where`/`np.maximum`; broadcasting rules; grid building; vectorization; constrained-task example; axis aggregations; `keepdims`; pitfalls II |
| 03 | `lessons/03-randomness-and-plotting.ipynb` | 75 | `default_rng`; seeding discipline; seeded experiments; line/scatter/histogram; figure checklist; exam connections; going deeper |
| 04 | `lessons/04-seaborn-with-arrays.ipynb` | 70 | array-oriented `histplot` and `scatterplot`; explicit axes; fixed bins; semantic legends; deterministic checks; justified matplotlib boundary |

Every section ends with a **Checkpoint** (do it before moving on); each
session's answers are collected at the end of that session's notebook.

## Practice sets

Practice lives in `practice/` as 24 problems in three sets (the manifest tags
each problem with its set and difficulty):

- **Set A — fundamentals** (p01, p02, p04, p05, p06, p15, p22): drills, one or two
  concepts each. Do them alongside the sessions: p01 after Session 01; p02,
  p04, p15 after Session 02; p05, p06 after Session 03; p22 after Session 04.
- **Set B — exam register** (p03, p08, p09, p10, p14, p16, p17, p18, p19, p23):
  five-option multiple choice, constrained coding with API bans, and
  reasoning-required shape derivations. Do them after finishing all four
  sessions.
- **Set C — integration and challenge** (p07, p11, p12, p13, p20, p21, p24):
  multi-part arcs, scenario analyses, and two challenge problems. Do them
  last; budget the most time here.

Each problem `pNN.ipynb` has a worked `pNN_solution.ipynb` — attempt honestly
before opening it.

## How to study this unit

1. Work each session notebook top to bottom, **running every cell yourself**
   and doing every checkpoint before reading its answer.
2. After the mapped sessions, do the Set A problems for that material.
3. After Session 04, work Sets B and C in order.
4. Finish with `review.ipynb` (summary tables, idiom sheet, self-quiz); use
   its redo pointers on anything shaky.

Budget roughly: 5 hours of sessions + 8.5 hours of practice + 50 minutes of review.
