# C11-neural-training — Practice p12 — Solution


**Type:** proof / derivation · **Difficulty:** core · **Concepts:** manual-backpropagation


Reverse dependency order, beginning with \(dZ_2=(P-Y)/N\in\mathbb R^{N\times C}\), is
\[
dW_2=dZ_2^TH\;(C,M),\quad db_2=\sum_{i=1}^N dZ_{2,i}\;(C),
\]
\[
dH=dZ_2W_2\;(N,M),\quad dZ_1=dH\odot 1[Z_1>0]\;(N,M),
\]
\[
dW_1=dZ_1^TX\;(M,D),\quad db_1=\sum_{i=1}^N dZ_{1,i}\;(M).
\]
I use \(\operatorname{ReLU}'(0)=0\). Each bias is broadcast to every row, so
its gradient sums axis 0. Likewise, each shared weight receives the sum of all
example contributions through the matrix products. Every derivative must use
the same forward snapshot; updating either layer early would change values
still needed by upstream chain-rule paths.


```python
import numpy as np

N, D, M, C = 2, 2, 2, 2
X_p12 = np.array([[1.0, 2.0], [-1.0, 1.0]])
W1_p12 = np.eye(2)
b1_p12 = np.zeros(2)
W2_p12 = np.array([[1.0, 2.0], [-1.0, 1.0]])
Z1_p12 = X_p12 @ W1_p12.T + b1_p12
H_p12 = np.maximum(Z1_p12, 0.0)
dZ2_p12 = np.array([[0.2, -0.2], [-0.3, 0.3]])
dW2_p12 = dZ2_p12.T @ H_p12
db2_p12 = dZ2_p12.sum(axis=0)
dH_p12 = dZ2_p12 @ W2_p12
dZ1_p12 = dH_p12 * (Z1_p12 > 0.0)
dW1_p12 = dZ1_p12.T @ X_p12
db1_p12 = dZ1_p12.sum(axis=0)
```

### Answer check


```python
# Fixed nondegenerate probe values independently derived from the ledger formulas.
assert np.array_equal(Z1_p12, np.array([[1.0, 2.0], [-1.0, 1.0]]))
assert np.array_equal(H_p12, np.array([[1.0, 2.0], [0.0, 1.0]]))
assert np.allclose(dW2_p12, np.array([[0.2, 0.1], [-0.2, -0.1]]), atol=1e-15, rtol=1e-15)
assert np.allclose(db2_p12, np.array([-0.1, 0.1]), atol=1e-15, rtol=1e-15)
assert np.allclose(dH_p12, np.array([[0.4, 0.2], [-0.6, -0.3]]), atol=1e-15, rtol=1e-15)
assert np.allclose(dZ1_p12, np.array([[0.4, 0.2], [0.0, -0.3]]), atol=1e-15, rtol=1e-15)
assert np.allclose(dW1_p12, np.array([[0.4, 0.8], [0.5, 0.1]]), atol=1e-15, rtol=1e-15)
assert np.allclose(db1_p12, np.array([0.4, -0.1]), atol=1e-15, rtol=1e-15)
assert [a.shape for a in (dW2_p12, db2_p12, dH_p12, dZ1_p12, dW1_p12, db1_p12)] == [
    (C, M), (C,), (N, M), (N, M), (M, D), (M,)
]
```
