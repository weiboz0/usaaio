# C11-neural-training — Practice p04 — Solution


**Type:** multiple choice (numeric, normal form) · **Difficulty:** intro · **Concepts:** dropout


**B.** If \(M_i\sim\operatorname{Bernoulli}(q)\), inverted dropout gives
\(\widetilde h_i=M_i h_i/q\). Squaring the first coordinate gives
\(E[\widetilde h_1^2]=E[M_1^2]h_1^2/q^2=q\cdot4/q^2=4/q=16/3\).
The other coordinates retain their means, so
\(E[\widetilde h_2+\widetilde h_3]=1+4=5\). Therefore the ratio is
\((16/3)/5=16/15\). Its denominator is positive and
\(\gcd(16,15)=1\), so the fraction is reduced.


```python
from fractions import Fraction
from math import gcd

q_p04 = Fraction(3, 4)
h_p04 = [Fraction(-2), Fraction(1), Fraction(4)]
numerator_expectation_p04 = q_p04 * (h_p04[0] / q_p04) ** 2
denominator_expectation_p04 = q_p04 * h_p04[1] / q_p04 + q_p04 * h_p04[2] / q_p04
ratio_p04 = numerator_expectation_p04 / denominator_expectation_p04
answer_p04 = "B"
```

### Answer check


```python
assert answer_p04 == "B"
assert numerator_expectation_p04 == Fraction(16, 3)
assert denominator_expectation_p04 == Fraction(5, 1)
assert ratio_p04 == Fraction(16, 15)
assert ratio_p04.denominator > 0
assert gcd(abs(ratio_p04.numerator), ratio_p04.denominator) == 1
```
