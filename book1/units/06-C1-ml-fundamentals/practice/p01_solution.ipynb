- (iv) Given customers' purchase histories and nothing else, find groups of

(S = supervised, U = unsupervised)?

- **A.** i-S, ii-S, iii-S, iv-U
- **B.** i-S, ii-U, iii-S, iv-U

**Part A explanation.** Tasks (i) and (iii) are supervised: their training
examples carry the answer being predicted (the actual peak demand; the
cat/dog tag). Tasks (ii) and (iv) are unsupervised: there is no answer column,
and the goal is to discover groups already present in the data. Only option B
assigns all four correctly.

**Part B model answer.** Clustering can find the groups themselves: it splits
the unlabeled examples into clusters of mutually similar items, revealing
structure nobody annotated. It cannot provide the *meaning* or names of those
groups — interpreting a cluster as "sports articles" or "budget shoppers" is
left to a human. Example: grouping supermarket customers by purchase history
into segments with similar buying habits, with no customer-type labels given.
