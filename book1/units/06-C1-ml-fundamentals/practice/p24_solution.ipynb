A habitat classifier uses one rigid rule: predict “suitable” only when a single temperature threshold is crossed. On a balanced dataset it scores **63% training accuracy** and **60% test accuracy**; the always-majority baseline is **50%**. A teammate says, “The model is failing because it overfits, so make it even simpler.”

Set `diagnosis: str` to `"underfitting"` or `"overfitting"`, and set `recommended_change: str` to `"more-flexibility"` or `"less-flexibility"`. Then write 3–4 sentences that use both the absolute score level and the three-point train-test gap, recommend one concrete flexibility increase, and explicitly distinguish this pattern from overfitting's strong-training/weak-test pattern.


```python
diagnosis: str = "underfitting"
recommended_change: str = "more-flexibility"

```

Both accuracies are low—$63\%$ on training and $60\%$ on test, only modestly above the $50\%$ baseline—so the rigid threshold is underfitting. The train–test gap is only three percentage points, which says the rule generalizes about as well as it fits rather than memorizing the training set. I would add a second ecologically relevant feature, such as rainfall, and allow the rule a second threshold on that feature, so the boundary can express more than one cutoff instead of a single temperature split. Overfitting would instead show strong training performance paired with much weaker test performance and a large gap, which is not this pattern.
