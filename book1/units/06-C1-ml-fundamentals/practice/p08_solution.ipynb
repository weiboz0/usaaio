# C1-ml-fundamentals — Practice p08 — Solution

A screening program tests **20,000** people for a rare disease. In truth,
**40** of them have it. The screen flags **190** people as positive, and
**32** of the flagged people actually have the disease.

1. Compute the integers `TP`, `FP`, `FN`, `TN` (exact names) from the counts
   above.
2. Compute the screen's `accuracy`, `precision`, and `recall` (exact names).
3. Compute `accuracy_flag_nobody` (exact name): the accuracy a do-nothing
   rule would get by flagging no one at all.
4. **Scenario analysis** (markdown cell): compare the screen with the
   do-nothing rule. Which metric matters most for a disease screen, and why?
   What does precision ≈ 0.17 mean for the people who receive a positive
   result, and why might that trade-off still be acceptable?

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
total = 20000
TP = 32                    # flagged and actually sick
FP = 190 - 32              # flagged but healthy
FN = 40 - 32               # sick but not flagged
TN = total - TP - FP - FN  # healthy and not flagged
print("TP =", TP, " FP =", FP, " FN =", FN, " TN =", TN)

accuracy  = (TP + TN) / total
precision = TP / (TP + FP)
recall    = TP / (TP + FN)
print(f"accuracy = {accuracy:.4f}  precision = {precision:.4f}  recall = {recall:.4f}")

accuracy_flag_nobody = (total - 40) / total
print(f"do-nothing accuracy = {accuracy_flag_nobody:.4f}")
```

**Scenario analysis.** The do-nothing rule scores 99.8%
accuracy — *higher* than the screen's 99.17% — while catching zero sick
patients, so accuracy is the wrong yardstick on a dataset that is 99.8%
negative. **Recall matters most here**: the screen finds 80% of the
truly sick (32 of 40), and a missed patient (a false negative) goes
untreated, which is the costliest possible error. Precision ≈ 0.17
means only about 1 in 6 people who receive a positive result actually has the
disease — the other 158 face the worry and cost of a follow-up test. That
trade-off is usually acceptable because the follow-up confirms or clears them,
whereas the 8 missed patients get no second chance at detection.
