# C1-ml-fundamentals — Machine Learning Fundamentals (Unit Overview)

This unit introduces the core ideas of machine learning: what it means for a
computer to *learn a rule from examples*, how to measure honestly whether the
rule works, and the classic ways such measurements mislead. Everything is
built from plain NumPy arrays and a little matplotlib, as covered in
F1-scientific-python — no machine-learning library appears; seeing the ideas
built from raw arrays is the point.

## What you will learn

- telling **supervised** from **unsupervised** problems, and what
  **clustering** finds in unlabeled data;
- the **train/test split** discipline and why held-out data is the only
  honest yardstick;
- **overfitting** — recognizing it from the train-test gap;
- the **flexible-vs-rigid** spectrum of rules (an intuitive preview of a
  tradeoff made precise in later units);
- the error-type metrics **accuracy, precision, recall, F1**, computed from
  confusion-matrix counts;
- **macro-F1** for multi-class tasks;
- why **class imbalance** makes plain accuracy lie, and which metrics expose
  the truth.

## Prerequisites (from F1-scientific-python)

You should be comfortable with: creating NumPy arrays; indexing, slicing, and
boolean masks; elementwise operations; axis-wise aggregation
(`sum(axis=...)`, `mean`); seeded random generators
(`np.random.default_rng`); and basic matplotlib plots. Calculus AB algebra is
all the math required.

## Code in this unit

A quick reference for the library calls in this unit's code. Each one is explained in a comment (or in the text beside the code) where it first appears.

**New in this unit:**

- `np.diff` — differences of neighbors: `a[1] - a[0], a[2] - a[1], ...`
- `.scatter` — `ax.scatter(x, y)`: one marker per point on these axes
- `.set_yticks` — where the y-axis tick marks go
- `.set_yticklabels` — the text shown at the y-axis ticks
- `np.bincount` — counts how often each small non-negative integer occurs
- `np.diag` — the diagonal entries of a matrix (or a diagonal matrix from a vector)
- `enumerate` — gives `(index, item)` pairs while looping
- `np.trace` — the sum of the diagonal entries
- `all` — True when every item is True

**Also used here (met earlier):**

- `zip` — pair up items of sequences (first explained in F4)
- `np.ones` — array of ones (first explained in F1)
- `np.mean` — average of the entries (first explained in F3)
- `np.round` — round to given decimals (first explained in F5)
- `.copy` — independent copy of the array (first explained in F1)
- `np.all` — True if every entry is True (first explained in F2)
- `np.where` — entrywise if/else: `cond, a, b` (first explained in F1)
- `np.maximum` — entrywise larger of two values (first explained in F1)
- `.plot` — draw a line on axes (first explained in F4)
- `np.arange` — evenly stepped integers, stop excluded (first explained in F1)
