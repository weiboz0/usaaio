Recall the lesson's memorizer: it stores every training pair (input, label)
in a lookup table; on an exact input match it returns the stored label, and
otherwise it returns the most common training label.

**(a)** Show that whenever all training inputs are distinct, the memorizer's
training accuracy is exactly 1 — for *every* possible dataset. Explain why
this makes its training accuracy carry zero information about how good the
rule really is.

**(b)** Show that if no test input exactly equals any training input, the
memorizer's test predictions are identical to those of the always-majority
constant rule. Conclude what its test accuracy equals in that case.

**(c)** In two to four sentences, combine (a) and (b): explain why the
memorizer is simultaneously the most flexible possible rule on its training
set and effectively the most rigid possible rule on new data, and why the
gap between its two scores is pure overfitting.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

**(a)** When training inputs are distinct, each input is a key that appears
exactly once in the lookup table, paired with its own label. Asked about a
training input, the memorizer returns precisely the label that was stored for
it, so every one of its training answers is correct and training accuracy is
1. This holds no matter what the labels are — even labels assigned by coin
flips would be "predicted" perfectly. A score that comes out 1 for every
conceivable dataset cannot distinguish a good rule from a useless one, so it
carries zero information about quality.

**(b)** A test input that matches no training input misses the lookup table
entirely, so the memorizer falls back to its default: the most common
training label. That is true for *every* such test input, so its predictions
coincide, input by input, with the always-majority constant rule. Its test
accuracy therefore equals the fraction of test labels that match the
training-majority class — the constant rule's score.

**(c)** On the training set the memorizer is flexibility taken to the limit:
it bends around every single example and reproduces each label exactly, which
is what part (a) proves. On genuinely new data all that flexibility
evaporates and, by part (b), it behaves as the most rigid rule there is — a
constant. Nothing it "learned" transfers; its perfect training score and its
constant-rule test score differ only because it memorized specifics rather
than pattern. That train-to-test gap is overfitting in its purest,
provable-by-construction form.

Optional numeric spot-check of (a) and (b):

```python
def memorizer_predict(X_tr, y_tr, X_new):
    majority = int(np.argmax(np.bincount(y_tr)))
    table = {float(v): int(lab) for v, lab in zip(X_tr, y_tr)}
    return np.array([table.get(float(v), majority) for v in X_new])

# (a) random labels, distinct inputs -> training accuracy is still 1
X_tr = np.arange(30, dtype=float)
y_tr = rng.integers(0, 2, size=30)
train_acc = np.mean(memorizer_predict(X_tr, y_tr, X_tr) == y_tr)
print("training accuracy on coin-flip labels:", train_acc)

# (b) disjoint test inputs -> predictions equal the constant majority rule
X_te = np.arange(100, 130, dtype=float)      # no overlap with X_tr
mem_preds = memorizer_predict(X_tr, y_tr, X_te)
majority = int(np.argmax(np.bincount(y_tr)))
print("all predictions equal majority:", bool(np.all(mem_preds == majority)))
```

### Answer check

```python
assert train_acc == 1.0
assert np.all(mem_preds == majority)
print("p19 OK")
```
