A binary classifier is evaluated on 200 examples with confusion counts
TP = 30, FP = 10, FN = 20, TN = 140. Written as a fraction m/n in lowest
terms (gcd(m, n) = 1, n > 0), its **F1 score** is:

- **A.** 9/20
- **B.** 3/5
- **C.** 3/4
- **D.** 2/3
- **E.** 27/40

Record your choice as a single capital letter in the code cell below.

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
```

```python
answer_p09 = "D"
```

**Worked reasoning.** Precision = 30/(30+10) = 3/4; recall = 30/(30+20) =
3/5. F1 = 2 · (3/4)(3/5) / (3/4 + 3/5) = 2 · (9/20) / (27/20) = 18/27 = 2/3,
already in lowest terms (gcd(2, 3) = 1) — option **D**. The distractors are
the classic traps: B is the recall, C is the precision, E = 27/40 is the
arithmetic mean of precision and recall (the wrong average for F1), and
A = 9/20 is their product.

```python
TP, FP, FN, TN = 30, 10, 20, 140
p = TP / (TP + FP)
r = TP / (TP + FN)
f1 = 2 * p * r / (p + r)
print("precision:", p, " recall:", r, " F1:", f1, " (2/3 =", 2 / 3, ")")
```

### Answer check

```python
assert answer_p09 == "D"
assert np.isclose(f1, 2 / 3, atol=1e-9, rtol=0)
print("p09 OK")
```
