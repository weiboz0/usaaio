Which ONE of the following statements is **true**?

- **A.** Supervised learning requires a human to write the classification
  rule by hand before any data is collected.
- **B.** Unsupervised learning finds structure — such as groups of similar
  items — in data that carries no labels.
- **C.** A dataset counts as labeled whenever the analyst personally knows
  the right answers, even if no answer column exists in the data.
- **D.** Supervised learning never needs a test set, because the labels
  already reveal how good the learned rule is.
- **E.** Unsupervised learning is just supervised learning run a second time
  on the same data.

Record your choice as a single capital letter in the code cell below.

**Explanation.** B states exactly what unsupervised learning is: discovering
structure (clusters, patterns) in unlabeled data. A confuses learning from
examples with hand-written rules — the whole point of supervised learning is
that nobody writes the rule. C fails the operational test: supervision means
the answers are *in the data* as a label column, not in someone's head. D is
backwards — labels make a test set possible and necessary, since training
accuracy alone cannot expose overfitting. E is nonsense: the presence or
absence of labels is what separates the two settings.
