**85 minutes.** Convexity is the condition that turns local information into a global guarantee. We will work from segments, Jensen inequalities, one-dimensional line restrictions, and gradients. No Hessian test is assumed or needed.

## 1. Convex sets contain every segment

A set $C\subseteq\mathbb{R}^d$ is **convex** when, for every $x,y\in C$ and every $\theta\in[0,1]$,

$$ (1-\theta)x+\theta y\in C. $$

The endpoints $\theta=0,1$ are included. To prove convexity, start with arbitrary feasible $x,y$ and preserve every defining constraint along their convex combination. To refute convexity, one pair and one intermediate $\theta$ outside the set are enough.

A disk and a halfspace are convex. A circle boundary is not: opposite points lie on the circle, but their midpoint lies inside rather than on the boundary.

## 2. Affine equalities, halfspaces, and intersections

For the halfspace $H=\{x:a^Tx\le b\}$, take $x,y\in H$. Then

$$a^T((1-\theta)x+\theta y)=(1-\theta)a^Tx+\theta a^Ty\le b.$$

Thus every affine halfspace is convex. The same calculation with equality proves that $\{x:Ax=b\}$ is convex. An intersection of convex sets is convex because a segment whose endpoints lie in every set stays in every set. Unions do not generally preserve convexity.

**Worked exam-style example 1.** Consider

$$C=\{(u,v):u+2v\le4,\ u\ge0,\ v\ge0\}.$$

Each condition is an affine halfspace, so their intersection is convex. For $x=(0,2)$, $y=(4,0)$, and $\theta=1/4$, the convex combination is $(1,3/2)$ and still satisfies all three inequalities. The sample confirms the arithmetic; the arbitrary-point halfspace proof establishes the claim.

## 3. Convex functions and Jensen's inequality

A function $f:C\to\mathbb{R}$ on a convex domain $C$ is **convex** when, for all $x,y\in C$ and $\theta\in[0,1]$,

$$f((1-\theta)x+\theta y)\le(1-\theta)f(x)+\theta f(y).$$

The graph lies below its endpoint chord. The **Jensen gap**

$$(1-\theta)f(x)+\theta f(y)-f((1-\theta)x+\theta y)$$

is therefore nonnegative. For $f(t)=t^2$, direct expansion gives the gap $\theta(1-\theta)(x-y)^2\ge0$, which is a complete algebraic proof on $\mathbb{R}$.

Convex-set and convex-function statements differ: a set contains segments, while a function lies below chords on those segments. The domain must itself be convex for the definition to compare every pair.

### Checkpoint 3

1. Compute the midpoint Jensen gap for $f(t)=|t|$, $x=-2$, and $y=4$.
2. What single negative Jensen gap would establish about a claimed convex function?

```python
def jensen_gap(f, x, y, theta):
    return (1.0 - theta) * f(x) + theta * f(y) - f((1.0 - theta) * x + theta * y)

assert np.isclose(jensen_gap(abs, -2.0, 4.0, 0.5), 2.0, atol=ATOL, rtol=RTOL)
assert np.isclose(jensen_gap(lambda t: t**2, -3.0, 5.0, 0.25), 12.0, atol=ATOL, rtol=RTOL)
```

## 4. Line restrictions replace a Hessian test

Fix $x,y\in C$ and define the one-dimensional restriction

$$h(t)=f((1-t)x+ty),\qquad 0\le t\le1.$$

The function $f$ is convex on $C$ if and only if every such restriction is convex on its interval. One direction substitutes two points on the line into the definition. The other direction chooses the line through any requested $x,y$ and evaluates its convexity between $t=0$ and $t=1$.

This is especially useful when an algebraic expression becomes simple along a direction. For $f(x)=\|Bx-c\|_2^2$, put $r=Bx-c$ and $d=B(y-x)$. Then

$$h(t)=\|r+td\|_2^2=\|r\|_2^2+2t r^Td+t^2\|d\|_2^2.$$

Its Jensen gap is nonnegative because the quadratic coefficient $\|d\|_2^2$ is nonnegative. This proves convexity without introducing second derivatives.

## 5. Supporting planes: the first-order certificate

If $f$ is differentiable and convex on an open convex domain, then for every $x,y$ in the domain,

$$f(y)\ge f(x)+\nabla f(x)^T(y-x).$$

The affine expression on the right is the **supporting plane** at $x$. To see why it holds, restrict to $h(t)=f(x+t(y-x))$. Convexity of $h$ says its secant slopes cannot start below the right derivative at zero, so $h(1)-h(0)\ge h'(0)=\nabla f(x)^T(y-x)$.

Conversely, if this inequality holds for every pair, swapping the appropriate endpoints and combining the two inequalities recovers the chord inequality. Thus, for differentiable functions, the supporting-plane condition is a first-order characterization of convexity.

## 6. From local information to global optimality

Let $C$ be a convex feasible set and let $f$ be differentiable and convex on an open set containing $C$. A feasible point $x_*$ is globally optimal if

$$\nabla f(x_*)^T(y-x_*)\ge0\quad\text{for every }y\in C.$$

The supporting-plane inequality then gives $f(y)\ge f(x_*)$. For an unconstrained interior point, $\nabla f(x_*)=0$ is enough. On a boundary, a nonzero gradient can still be optimal because every feasible direction may have nonnegative directional change.

**Worked exam-style example 2.** Minimize $f(x)=(x-3)^2$ over $C=(-\infty,1]$. At $x_*=1$, $f'(1)=-4$. Every feasible $y$ satisfies $y-1\le0$, so $f'(1)(y-1)\ge0$. Convexity plus this first-order condition certifies that $x_*=1$ is the global constrained minimizer. Session 4 will encode the same boundary fact with a multiplier.

A local minimizer of a convex function over a convex set is global: if a better point existed, points sufficiently near the local minimizer on their connecting segment would also be better by Jensen, a contradiction. Without convexity, neither stationarity nor local minimality gives a global guarantee.

## 7. Common pitfalls, exam connections, and forward links

**Pitfall — a curved-looking set is nonconvex.** Curvature of a boundary is irrelevant; a disk is convex. Fix: test the complete segment condition.

**Pitfall — Jensen's inequality points the other way.** Broken claim: a convex graph lies above every chord. Fix: the graph lies below chords and above its supporting planes.

**Pitfall — zero gradient always means a global minimum.** The conclusion requires convexity and an appropriate domain. Fix: state the structural assumption before using the certificate.

**Pitfall — use an untaught Hessian test.** A matrix of second derivatives is unnecessary here. Fix: use Jensen, line restrictions, algebraic gaps, and the supporting-plane inequality.

**Exam connection.** Round 1 can ask for a segment witness, a Jensen-gap computation, or the reason a local condition is globally decisive. State whether you are proving a universal property or constructing one refutation.

**Going deeper.** Session 4 converts boundary optimality into Lagrangian and KKT certificates. Later SVM material uses a convex objective and affine constraints, but no SVM fact is needed in this unit.
