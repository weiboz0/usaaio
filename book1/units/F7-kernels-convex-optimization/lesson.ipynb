# F7-kernels-convex-optimization — Kernels, Convexity, and Certificates

This unit connects two ideas that often appear separately on a syllabus. A kernel is valid when every finite Gram matrix it produces is positive semidefinite. Convex optimization turns local gradient information and constraint multipliers into global conclusions. The bridge is the same disciplined habit: state the quantifier, build the algebraic certificate, and distinguish a proof from a numerical check.

**Exact scope:** four 85-minute sessions, 20 practice problems (640 minutes), and a 45-minute systemic review. Total: **1025 minutes**.


## Unit goals

By the end, you can:

1. certify a symmetric matrix as positive semidefinite (PSD) from quadratic forms or eigenvalues;
2. prove that Gram matrices and feature-map kernels are PSD;
3. use the finite-Gram quantifier to prove or refute kernel validity;
4. prove convex-set and convex-function claims from segments, Jensen inequalities, and line restrictions;
5. use the supporting-plane inequality to turn first-order information into global optimality;
6. form a Lagrangian with an explicit minimization sign convention; and
7. read weak-duality, complementary-slackness, and KKT certificates under their required assumptions.

The seven manifest concepts are `positive-semidefinite-matrices`, `kernel-validity`, `convex-sets`, `convex-functions`, `first-order-optimality`, `lagrangians`, and `optimization-duality`.


## Prerequisites — what is assumed, and no more

| Earlier unit | Live knowledge used here |
|---|---|
| F3-matrices | matrices as maps, matrix products, rank, Gram matrices, and outer products |
| F4-multivar-calculus | partial derivatives, gradients, and directional change |
| F6-svd-spectral | symmetric eigenvalues/eigenvectors, spectral decomposition, and the quadratic energy `x.T @ A @ x` |
| C3-gradient-descent | loss surfaces, local descent directions, and the fact that a local compass is not automatically a global guarantee |

F6 already proved that every ordinary Gram matrix has nonnegative eigenvalues. Session 1 names that property PSD, makes its equivalent tests precise, and extends the Gram idea to kernels. We do **not** introduce Hessians; every convexity argument stays within segments, one-dimensional restrictions, gradients, and algebra.

## How to study this unit

1. On every PSD or kernel claim, write the quantifier before computing: one matrix, or every finite input list?
2. Attempt each checkpoint before opening the collected answers at the session end.
3. For a claimed kernel proof, identify the nonnegative quadratic form or explicit feature map. A green numerical spectrum is evidence, never the universal proof.
4. For convexity, draw the segment between two points and write Jensen before reaching for derivatives.
5. For constrained optimization, rewrite every inequality as `g_i(x) <= 0` before choosing multiplier signs.
6. Read a KKT tuple as a certificate only after checking the convex/affine structure and the stated regularity assumption.

After all four sessions, spend 45 minutes on `review.ipynb`: answer all prompts first, then open its single collected answer section.
