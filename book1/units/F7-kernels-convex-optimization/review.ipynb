# F7-kernels-convex-optimization — Systemic Review

**45 minutes.** Answer all 12 prompts before opening the collected answer section. This review connects the seven unit concepts: `positive-semidefinite-matrices`, `kernel-validity`, `convex-sets`, `convex-functions`, `first-order-optimality`, `lagrangians`, and `optimization-duality`.

## Concept summary

| Concept | Universal statement or certificate | Fast refutation or audit |
|---|---|---|
| Positive-semidefinite matrices | $v^TAv\ge0$ for every $v$; equivalently a symmetric spectrum is nonnegative | one vector with negative quadratic energy |
| Kernel validity | separately prove $k(x,z)=k(z,x)$ for every pair and nonnegative Gram quadratic forms for every finite list | one asymmetric pair or one finite negative-energy witness |
| Convex sets | every segment between points in the set remains in the set | two points and an escaping convex combination |
| Convex functions | Jensen's inequality holds on every segment of a convex domain | one negative Jensen gap |
| First-order optimality | supporting plane lies below a differentiable convex function; feasible directional changes are nonnegative at an optimum | a violating point/direction, or missing convexity |
| Lagrangians | with minimization constraints $g_i\le0$, use $L=f+\sum\lambda_i g_i+\sum\nu_j h_j$ and $\lambda_i\ge0$ | wrong sign convention, infeasibility, or residual failure |
| Optimization duality | $q=\inf_xL$ gives lower bounds, so $d^*\le p^*$; KKT/zero gap need stated assumptions | reversed bound, open gap, or absent convexity/regularity |

## Formula and proof-idiom sheet

- PSD: $A=A^T$ and $v^TAv\ge0$ for all $v$; for symmetric $A$, equivalently $\lambda_{\min}(A)\ge0$.
- Gram certificate: $a^TWW^Ta=\|W^Ta\|_2^2\ge0$.
- Feature-map kernel: $k(x,z)=\phi(x)^T\phi(z)$, hence $K=\Phi\Phi^T$.
- Kernel symmetry: $k(x,z)=k(z,x)$ for every input pair; this must be proved separately.
- Kernel Gram quantifier: for every $n$, every finite $x_1,\ldots,x_n$, and every $a\in\mathbb{R}^n$, $\sum_{i,j}a_i a_j k(x_i,x_j)\ge0$. This quadratic condition alone does not imply symmetry.
- Convex set: $(1-\theta)x+\theta y\in C$.
- Convex function: $f((1-\theta)x+\theta y)\le(1-\theta)f(x)+\theta f(y)$.
- Line restriction: $h(t)=f((1-t)x+ty)$.
- Supporting plane: $f(y)\ge f(x)+\nabla f(x)^T(y-x)$.
- Feasible first-order certificate: $\nabla f(x_*)^T(y-x_*)\ge0$ for every feasible $y$.
- Lagrangian convention: minimize with $g_i\le0$, $h_j=0$, $\lambda_i\ge0$.
- Weak duality chain: $q(\lambda,\nu)\le L(x,\lambda,\nu)\le f(x)$, hence $d^*\le p^*$.
- KKT audit: primal feasibility, dual feasibility, stationarity, complementary slackness.

11. A constraint is inactive at $x_*$. What does complementary slackness force, and what does it not say about other constraints?
12. List the four KKT conditions and the convex/affine plus regularity assumptions required here before claiming multiplier existence and zero duality gap.

1. PSD but not PD. The eigenspace of eigenvalue zero contains nonzero vectors with zero quadratic energy.

2. Repeated inputs produce duplicate Gram rows, so singularity is unavoidable; validity requires PSD, which allows zero eigenvalues.

3. First, $k(x,z)=k(z,x)$ for every input pair. Separately, for every $n\ge1$, every finite input list $x_1,\ldots,x_n$, and every real coefficient vector $a$, the matrix $K_{ij}=k(x_i,x_j)$ satisfies $a^TKa\ge0$. The quadratic condition cannot replace the symmetry proof.

4. Those 500 sampled matrices were symmetric and PSD within the declared audit tolerance. Universal proofs of symmetry for every input pair and Gram nonnegativity for every finite list are still missing.

5. Invalid. The true smallest certificate uses one point `x_1=2`: $K=[1-2\cdot2]=[-3]$, and $a=[1]$ gives $a^TKa=-3<0$.

6. $(1,1)$ and $(-1,-1)$ are feasible, but their midpoint $(0,0)$ is not.

7. The chord value is $(1+9)/2=5$ and the midpoint value is $1$, so the gap is $4$.

8. For every pair $x,y$, prove the one-dimensional function $h(t)=f((1-t)x+ty)$ obeys the convex chord inequality on $[0,1]$; this is equivalent to convexity on the domain.

9. $f(y)\ge f(x)+\nabla f(x)^T(y-x)$. If $\nabla f(x_*)=0$, then $f(y)\ge f(x_*)$ for every $y$.

10. $L=f+\sum_i\lambda_i g_i+\sum_j\nu_jh_j$, with $\lambda_i\ge0$ and free $\nu_j$. Weak duality is $d^*\le p^*$.

11. Its multiplier must be zero. This says nothing about whether other constraints are active or what their multipliers are.

12. Primal feasibility, dual feasibility, stationarity, and complementary slackness. The setting taught here assumes a finite attained primal optimum, convex objective and inequality functions, affine equalities, and a Slater point; these support multiplier existence and strong duality.
