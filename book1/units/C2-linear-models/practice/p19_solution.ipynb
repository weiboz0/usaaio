**Type:** proof · **Difficulty:** core · **Concepts:** linear-regression-estimator-derivation

**Reasoning is required.**

The intercept is absorbed into $X$:
$X\in\mathbb R^{n\times p}$, $\beta\in\mathbb R^p$,
$y\in\mathbb R^n$, $r=X\beta-y$, and
\[
L(\beta)=\frac1n\lVert X\beta-y\rVert_2^2.
\]
Assume full column rank.

**(a)** Start from the indexed sum and derive every
$\partial L/\partial\beta_j$. Collect the components into
$\nabla_\beta L$, tracking $2/n$ and all shapes.

**(b)** Set the gradient to zero and derive the normal equations. Prove
full column rank makes $X^TX$ invertible. State the estimator in
solve-a-system notation, not as a numerical matrix-inverse algorithm.

**(c)** Prove $\hat r=X\hat\beta-y$ is orthogonal to every column.
Conversely, if $X^Tr_0=0$ at $\beta_0$, expand the squared residual at
$\beta_0+h$ and prove it cannot decrease.

**(d)** On the supplied data, compute \`gram_p19\`, \`rhs_p19\`,
\`beta_p19\` with \`np.linalg.solve\`, \`resid_p19\`, and the maximum
orthogonality gap. The final check must use
\`ATOL = 1e-10\`, \`RTOL = 0.0\`.

*Your derivation and proofs for (a)–(c) here.*

Write $r_i=\sum_{k=1}^p X_{ik}\beta_k-y_i$. Starting from the indexed loss,

$$
L(\beta)=\frac1n\sum_{i=1}^n r_i^2,
$$

its $j$-th partial derivative is

$$
\frac{\partial L}{\partial \beta_j}
=\frac1n\sum_{i=1}^n 2r_i\frac{\partial r_i}{\partial \beta_j}
=\frac2n\sum_{i=1}^n X_{ij}r_i.
$$

Collecting the $p$ scalar components gives

$$
\nabla_\beta L=\frac2nX^Tr,
$$

where $X^T$ has shape $p\times n$, $r$ has shape $n$, and the gradient has shape $p$. Setting it to zero gives $X^T(X\hat\beta-y)=0$, hence the normal equations

$$
(X^TX)\hat\beta=X^Ty.
$$

For any nonzero $v\in\mathbb R^p$, full column rank implies $Xv\ne0$, so

$$
v^TX^TXv=\lVert Xv\rVert_2^2>0.
$$

Thus $X^TX$ is positive definite and therefore invertible: if it had a nonzero null vector $v$, then $v^TX^TXv=0$, a contradiction. Computationally the estimator is written as

$$
\hat\beta=\operatorname{solve}(X^TX,X^Ty),
$$

not as an explicitly formed inverse.

The normal equations also give $X^T\hat r=0$, so $\hat r=X\hat\beta-y$ is orthogonal to every column of $X$. Conversely, suppose $X^Tr_0=0$ at $\beta_0$. For any $h\in\mathbb R^p$,

$$
\begin{aligned}
\lVert r(\beta_0+h)\rVert_2^2
&=\lVert r_0+Xh\rVert_2^2\\
&=\lVert r_0\rVert_2^2+2h^TX^Tr_0+\lVert Xh\rVert_2^2\\
&=\lVert r_0\rVert_2^2+\lVert Xh\rVert_2^2
\ge \lVert r_0\rVert_2^2.
\end{aligned}
$$

Therefore no perturbation can lower the loss. On the supplied fixture, $X^TX=\begin{pmatrix}4&3\\3&45/2\end{pmatrix}$, $X^Ty=(15/2,115/4)^T$, and $\hat\beta=(55/54,185/162)^T$.

```python
import numpy as np

ATOL = 1e-10
RTOL = 0.0

X = np.array([[1.0, -2.0],
              [1.0, -0.5],
              [1.0, 1.5],
              [1.0, 4.0]])
y = np.array([-1.0, 0.5, 2.0, 6.0])

gram_p19 = X.T @ X
rhs_p19 = X.T @ y
beta_p19 = np.linalg.solve(gram_p19, rhs_p19)
resid_p19 = X @ beta_p19 - y
orthogonality_gap_p19 = float(np.max(np.abs(X.T @ resid_p19)))
normal_check_p19 = np.allclose(X.T @ resid_p19, 0.0, atol=ATOL, rtol=RTOL)
```

```python
assert gram_p19.shape == (2, 2)
assert rhs_p19.shape == (2,)
assert beta_p19.shape == (2,)
assert resid_p19.shape == (4,)
assert np.allclose(gram_p19, np.array([[4.0, 3.0], [3.0, 22.5]]), atol=ATOL, rtol=RTOL)
assert np.allclose(rhs_p19, np.array([7.5, 28.75]), atol=ATOL, rtol=RTOL)
assert np.allclose(beta_p19, np.array([55 / 54, 185 / 162]), atol=ATOL, rtol=RTOL)
assert np.allclose(gram_p19 @ beta_p19, rhs_p19, atol=ATOL, rtol=RTOL)
assert normal_check_p19
assert np.allclose(X.T @ resid_p19, 0.0, atol=ATOL, rtol=RTOL)
assert orthogonality_gap_p19 <= ATOL
```
