# C2-linear-models — Practice p12 — Solution

Differentiating gives $g'(w)=2(w-a)+2\lambda w=2(1+\lambda)w-2a$. Setting this equal to zero yields $w^*=a/(1+\lambda)$. Since $g''(w)=2(1+\lambda)>0$ for every $\lambda\ge0$, the critical point is the unique global minimum.

If $a\ne0$, then $a/(1+\lambda)\ne0$ for every finite $\lambda\ge0$, so the closed form is never exactly zero. Separately, $g'(0)=-2a\ne0$, and a differentiable function cannot have an interior minimum where its derivative is nonzero. Thus L2 shrinks weights but does not create exact sparsity, unlike L1's kink.

```python
import numpy as np

a = 3.0
lam = 2.0
w_star_hand = a / (1 + lam)
grid = np.linspace(-1, 3, 400001)
values = (grid - a) ** 2 + lam * grid ** 2
w_star_grid = float(grid[np.argmin(values)])
anchor_gap = abs(w_star_hand - w_star_grid)
derivative_at_zero = -2 * a
```

At $(a,\lambda)=(3,2)$, the formula gives $w^*=1$. The grid search returns the same numeric anchor, while $g'(0)=-6$ confirms zero cannot minimize this instance.

### Answer check

```python
assert np.isclose(w_star_hand, 1.0, atol=1e-9, rtol=0)
assert anchor_gap <= 1e-5
assert np.isclose(derivative_at_zero, -6.0, atol=1e-9, rtol=0)
```
