# C5-neural-networks — Practice p13 — Solution

The constraints become $x_1-1\geq0$, $-x_1+4\geq0$, and $-x_1+x_2+2\geq0$.  The AND gate receives three binary detector values, so unit weights and bias $-(3-0.5)=-2.5$ fire exactly when all three are one.

Among the listed points, $(4,5)$ lies exactly on the boundary $x_1=4$.  It remains inside because the detector and the course step convention both use the inclusive comparison $z\geq0$.

```python
import numpy as np


def affine_layer(x, W, b):
    return (x[:, None, :] * W[None, :, :]).sum(axis=2) + b


def step_activation(z):
    return (z >= 0).astype(float)


SEED = 20260804
hand_pts = np.array([[2.0, 1.0], [0.0, 3.0], [3.0, 0.0], [4.0, 5.0]])
rng = np.random.default_rng(SEED)
pts = rng.uniform(-2, 6, (4000, 2))
inside_by_hand = [1.0, 0.0, 0.0, 1.0]
W1 = np.array([[1.0, 0.0], [-1.0, 0.0], [-1.0, 1.0]])
b1 = np.array([-1.0, 4.0, 2.0])
W2 = np.ones((1, 3))
b2 = np.array([-2.5])
labels_hand_pts = step_activation(affine_layer(step_activation(affine_layer(hand_pts, W1, b1)), W2, b2))[:, 0]
labels_pts = step_activation(affine_layer(step_activation(affine_layer(pts, W1, b1)), W2, b2))[:, 0]
frac_in = float(labels_pts.mean())
labels_hand_pts, frac_in
```

### Answer check

```python
assert np.allclose(labels_hand_pts, inside_by_hand, atol=1e-12, rtol=0)
assert np.allclose(W1, [[1.0, 0.0], [-1.0, 0.0], [-1.0, 1.0]], atol=1e-12, rtol=0)
assert np.allclose(b1, [-1.0, 4.0, 2.0], atol=1e-12, rtol=0)
assert np.isclose(frac_in, 0.25925, atol=1e-12, rtol=0)
assert list(map(float, inside_by_hand)) == [1.0, 0.0, 0.0, 1.0]  # by-hand answer pinned as a literal
```
