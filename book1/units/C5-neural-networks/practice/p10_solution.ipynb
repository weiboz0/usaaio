# C5-neural-networks — Practice p10 — Solution

The standard-deviation multiplier is $\sigma\sqrt{128}$.  It is about $0.0884$ for (a), exactly $1$ for (b), and about $3.394$ for (c).  Thus (a) vanishes, (c) explodes, and **b** is the order-one scaling.

```python
import numpy as np

SEED = 20260804


def propagate(sigma):
    rng = np.random.default_rng(SEED)
    h = rng.normal(0, 1, (256, 128))
    spreads = []
    for _ in range(15):
        W = rng.normal(0, sigma, (128, 128))
        h = (h[:, None, :] * W[None, :, :]).sum(axis=2)
        spreads.append(float(h.std()))
    return np.array(spreads)


stds_a = propagate(1 / 128)
stds_b = propagate(1 / np.sqrt(128))
stds_c = propagate(0.3)
final_a = float(stds_a[-1])
final_b = float(stds_b[-1])
final_c = float(stds_c[-1])
which_stable = "b"
final_a, final_b, final_c, which_stable
```
