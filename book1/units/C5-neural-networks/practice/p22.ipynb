# C5-neural-networks — Practice p22

**Type:** proof · **Difficulty:** advanced · **Concepts:** perceptron, decision-boundaries-geometric

*Reasoning is required. No coding is needed.*

The four corners of a rectangle are labeled in an XOR pattern:

- positive: \(P_1=(2,3)\), \(P_2=(19,17)\);
- negative: \(N_1=(2,17)\), \(N_2=(19,3)\).

Prove that no single threshold perceptron \(x\mapsto\mathbf 1[w\cdot x+b\ge0]\) realizes these labels. Your proof must:

1. compute the midpoint of the positive pair and the midpoint of the negative pair;
2. use linearity of \(z(x)=w\cdot x+b\) on averages;
3. derive the incompatible inequalities at that one midpoint.

A picture or the phrase “XOR is not linearly separable” earns no credit without the midpoint argument.
