# C5-neural-networks — Practice p08 — Solution

The singleton axes align examples, output units, and input features.  Summing only over the feature axis gives all example-unit pre-activations at once; the hidden step is then passed to the affine output.

```python
import numpy as np

SEED = 20260804
X = np.array([[1.0, 0.0, 2.0], [-1.0, 2.0, 0.0], [0.5, 1.0, 1.0]])
W1 = np.array([[1.0, -1.0, 2.0], [0.0, 3.0, -1.0]])
b1 = np.array([-1.0, 0.5])
W2 = np.array([[2.0, -3.0]])
b2 = np.array([0.5])


def affine_layer(x, W, b):
    return (x[:, None, :] * W[None, :, :]).sum(axis=2) + b


def step_activation(z):
    return (z >= 0).astype(float)


z1 = affine_layer(X, W1, b1)
h1 = step_activation(z1)
y = affine_layer(h1, W2, b2)
z1, h1, y
```

### Answer check

```python
assert np.allclose(z1, [[4.0, -1.5], [-4.0, 6.5], [0.5, 2.5]], atol=1e-12, rtol=0)
assert np.allclose(h1, [[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]], atol=1e-12, rtol=0)
assert np.allclose(y, [[2.5], [-2.5], [-0.5]], atol=1e-12, rtol=0)
```
