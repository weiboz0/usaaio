# C5-neural-networks — Practice p15 — Solution

The versions disagree exactly on the six diagonal points $(k,k)$, $k=0,\ldots,5$, where $z=x_1-x_2=0$.  Hand-designed integer boundaries hit zero frequently because grid points and integer coefficients often put samples exactly on the line.  The course and exam pin `>=`; changing the teammate's `>` to `>=` includes the boundary and fixes the discrepancy.

```python
import numpy as np

SEED = 20260804
g = np.arange(6)
X1, X2 = np.meshgrid(g, g, indexing="ij")
grid_pts = np.stack([X1.ravel(), X2.ravel()], axis=1).astype(float)
z_grid = grid_pts[:, 0] - grid_pts[:, 1]
labels_ge = (z_grid >= 0).astype(float)
labels_gt = (z_grid > 0).astype(float)
n_ge = int(labels_ge.sum())
n_gt = int(labels_gt.sum())
n_disagree = int((labels_ge != labels_gt).sum())
n_ge, n_gt, n_disagree
```

### Answer check

```python
assert n_ge == 21
assert n_gt == 15
assert n_disagree == 6
assert np.allclose(grid_pts[labels_ge != labels_gt], np.column_stack([g, g]), atol=1e-12, rtol=0)
```
