# C5-neural-networks — Practice p04 — Solution

For the first hidden unit, $z_1=2(1)-1(2)-1=-1$, so it outputs $0$.  For the second, $z_2=-1(1)+1(2)+1=2$, so it outputs $1$.  Thus $y=4(0)+5(1)-5/2=5/2$.  In lowest terms $p=5$, $q=2$, and $p+q=7$, giving **B**.

- A ($6$) is the drop-the-output-bias trap: without $b^{(2)}$,
  $y = 5 = 5/1$ and $p+q = 6$.
- C ($9$) is the skip-the-activation trap: feeding $z^{(1)} = (-1, 2)$
  straight to the readout gives $y = -4+10-5/2 = 7/2$ and $p+q = 9$.
- D ($15$) is the both-units-fire trap: $h = (1, 1)$ gives
  $y = 4+5-5/2 = 13/2$ and $p+q = 15$.
- E ($17$) is the ReLU-instead-of-step trap: $h = (\mathrm{relu}(-1),
  \mathrm{relu}(2)) = (0, 2)$ gives $y = 10-5/2 = 15/2$ and $p+q = 17$.

```python
import numpy as np

SEED = 20260804
answer_p04 = "B"
hidden_z = np.array([-1.0, 2.0])
hidden_h = (hidden_z >= 0).astype(float)
y = float((hidden_h * np.array([4.0, 5.0])).sum() - 2.5)
p, q = 5, 2
normal_form_sum = p + q
answer_p04, hidden_z, hidden_h, y, normal_form_sum
```
