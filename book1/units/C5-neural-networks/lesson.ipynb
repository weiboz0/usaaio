## Unit goals

By the end of this unit you can:

- define the **perceptron** — pre-activation
  $z = \sum_k w_k x_k + b$, output $1$ iff $z \ge 0$ — evaluate it by
  hand and in batch NumPy, and design its weights to realize a stated
  firing rule (boundary always included: the pinned $\ge$ convention);
- work with the three **activation functions** of this course — step,
  $\tanh$ (recap from F4), **ReLU** — knowing their values, ranges,
  derivatives, and characters, and build piecewise-linear shapes from
  scaled shifted ReLUs;
- state, prove, and *use* the collapse fact: an activation-free stack of
  affine layers is one affine layer
  ($W^{\mathrm{eff}}, b^{\mathrm{eff}}$ derived in component form) — the
  reason activations exist;
- describe an **MLP** as composed layers with the architecture notation
  $d_{\text{in}} \to h \to d_{\text{out}}$, keep the shape contract
  straight ($W$ is $(d_{\text{out}}, d_{\text{in}})$, **row = unit**),
  and run exact forward passes by hand and with the unit's two pinned
  NumPy helpers, `affine_layer(x, W, b)` and `step_activation(z)`;
- *design* **decision regions**: read and write half-plane detectors,
  build AND / OR / NOT / at-least-$r$ gates with half-integer biases,
  assemble convex-polygon membership networks by AND-ing edge
  detectors, and handle non-convex unions with a second hidden layer;
- run the **weight-initialization variance** argument end to end: the
  group-form independence fact (stated), the product variance
  $\operatorname{Var}[W_k x_k] = \sigma^2$, the layer formula
  $\operatorname{Var}[z] = C\sigma^2$, and the rule
  $\sigma = 1/\sqrt{C}$ — motivated purely by keeping forward-pass
  activation scales stable, and verified by seeded simulation.

**What this unit deliberately does not do: train.**
Training multi-layer networks needs backpropagation, which is beyond this
course's scope; the exam tests **inference engineering** — you will design
weights by hand.
No loss, no gradient step, no fitting appears anywhere in this unit
(contrast C3, which trained *linear* models); the initialization analysis
of Session 3 is likewise a forward-pass statement — choosing the scale
that keeps a deep stack numerically alive.
`C6-pytorch` keeps the same inference-only stance.
