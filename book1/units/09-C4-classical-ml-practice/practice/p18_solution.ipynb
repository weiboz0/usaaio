# C4-classical-ml-practice — Practice p18 — Solution

Let $\Delta_{jA}=|q_j-A_j|$ and $\Delta_{jB}=|q_j-B_j|$.  At a flip the squared distances are equal, so
$$\Delta_{1A}^2+\frac{\Delta_{2A}^2}{s^2}=\Delta_{1B}^2+\frac{\Delta_{2B}^2}{s^2},$$
which rearranges to
$$s^{*2}=\frac{\Delta_{2A}^2-\Delta_{2B}^2}{\Delta_{1B}^2-\Delta_{1A}^2}.$$
This positive crossing requires both numerator and denominator to have the same sign; in the stated orientation both are positive.  Then $d_A^2-d_B^2=-D+N/s^2$, so B is nearer for $s<s^*$, A is nearer for $s>s^*$, and they tie at equality.

For Manhattan distance, equality gives $\Delta_{1A}+\Delta_{2A}/s=\Delta_{1B}+\Delta_{2B}/s$, hence $s=(\Delta_{2A}-\Delta_{2B})/(\Delta_{1B}-\Delta_{1A})$ when both differences are positive.  Here this is $25/0.8=31.25$.

```python
import numpy as np

q = np.array([0.0, 0.0])
A = np.array([0.1, 30.0])
B = np.array([0.9, 5.0])
num = A[1] ** 2 - B[1] ** 2
den = B[0] ** 2 - A[0] ** 2
s_flip = float(np.sqrt(num / den))
s_grid = np.linspace(1, 100, 99001)
d2_A_grid = A[0] ** 2 + (A[1] / s_grid) ** 2
d2_B_grid = B[0] ** 2 + (B[1] / s_grid) ** 2
s_flip_grid = float(s_grid[np.argmax(d2_A_grid < d2_B_grid)])
s_flip_manhattan = float((A[1] - B[1]) / (B[0] - A[0]))
training = np.vstack([A, B])
mu = training.mean(axis=0)
sigma = training.std(axis=0)
sigma_ratio = float(sigma[1] / sigma[0])
training_z = (training - mu) / sigma
q_z = (q - mu) / sigma
d2_standardized = ((training_z - q_z) ** 2).sum(axis=1)
nn_after_standardization = np.array(["A", "B"])[np.argmin(d2_standardized)]

s_flip, s_flip_grid, s_flip_manhattan, sigma_ratio, nn_after_standardization
```

The exact Euclidean threshold is about 33.0719, and the 0.001-spaced grid first observes A as nearer at 33.072, providing a concrete computational anchor. Standardization supplies a relative shrink of only 31.25, below the Euclidean flip threshold, so it makes the feature scales comparable without forcing this particular neighbor to flip; B remains nearer.

### Answer check

```python
assert np.isclose(s_flip, 33.071891388307385, atol=1e-9, rtol=0)
assert np.isclose(s_flip_grid, 33.072, atol=1e-9, rtol=0) and abs(s_flip_grid - s_flip) < 0.01
assert np.isclose(s_flip_manhattan, 31.25, atol=1e-9, rtol=0)
assert np.isclose(sigma_ratio, 31.25, atol=1e-9, rtol=0)
assert np.allclose(d2_standardized, np.array([5.8225, 5.2225]), atol=1e-9, rtol=0)
assert nn_after_standardization == "B"
assert sigma_ratio < s_flip
```
