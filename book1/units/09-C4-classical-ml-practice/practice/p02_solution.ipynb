# C4-classical-ml-practice — Practice p02 — Solution

```python
import numpy as np

height_span = 0.5
income_span = 50_000.0
squared_span_ratio = (income_span / height_span) ** 2
answer_p02 = "B"
squared_span_ratio, answer_p02
```

Raw squared Euclidean distance is sensitive to numerical scale, and the typical income difference is about 100,000 times the typical height difference. Its squared contribution is therefore about ten billion times larger, so income almost entirely determines the nearest neighbor and B is correct.

### Answer check

```python
assert np.isclose(squared_span_ratio, 1.0e10, atol=1e-9, rtol=0)
expected_letter = "B"
assert answer_p02 == expected_letter
```
