# C2-linear-models — Practice p06 — Solution

```python
import numpy as np

SEED = 20260804
rng = np.random.default_rng(SEED)
X = rng.normal(0, 1, (25, 2))
y = 2 * X[:, 0] - X[:, 1] + 3 + rng.normal(0, 0.3, 25)

def mse(X, y, w, b):
    residuals = np.sum(X * w, axis=1) + b - y
    return float(np.mean(residuals ** 2))

mse_a = mse(X, y, np.array([2.0, -1.0]), 3.0)
mse_b = mse(X, y, np.array([1.5, -0.5]), 2.5)
better = str(np.array(["a", "b"])[np.argmin(np.array([mse_a, mse_b]))])
```

Candidate a has MSE about $0.0930$, while candidate b has MSE about $0.6103$. The smaller scoreboard value therefore belongs to candidate a.

### Answer check

```python
assert np.isclose(mse_a, 0.09300373725627907, atol=1e-9, rtol=0)
assert np.isclose(mse_b, 0.6103158861116948, atol=1e-9, rtol=0)
assert better == "a"
```
