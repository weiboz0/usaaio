# C2-linear-models — Practice p02 — Solution

```python
import numpy as np

predictions = np.array([2.0, 5.0, 3.0])
targets = np.array([1.0, 3.0, 3.0])
squared_errors = (predictions - targets) ** 2
mse_value = float(np.mean(squared_errors))
choice_values = np.array([1.0, 5.0 / 3.0, 5.0 / 2.0, 3.0, 5.0])
choice_letters = np.array(list("ABCDE"))
answer_p02 = str(choice_letters[np.flatnonzero(np.isclose(choice_values, mse_value, atol=1e-9, rtol=0))[0]])
```

The residuals are $1$, $2$, and $0$, so the squared errors sum to $5$. Dividing by the three examples gives $5/3$, which is choice B.

### Answer check

```python
assert np.isclose(mse_value, 5.0 / 3.0, atol=1e-9, rtol=0)
assert answer_p02 == "B"
```
