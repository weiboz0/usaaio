## Unit goals

By the end of C2 you can:

- compute $\hat y_i=\sum_k X_{ik}w_k+b$, residuals
  $r_i=\hat y_i-y_i$, MSE, and gradients at supplied weights;
- move between an explicit bias and the augmented design convention
  $X:(n,p)$, $\beta:(p,)$, with a ones column for the intercept;
- derive $\nabla_\beta L=(2/n)X^T(X\beta-y)$, the normal equations
  $X^TX\hat\beta=X^Ty$, and residual orthogonality $X^T\hat r=0$;
- distinguish exact consistency from least squares and distinguish
  existence from coefficient uniqueness/identifiability;
- prove full column rank iff $X^TX$ is positive definite/invertible,
  while rejecting the false shortcut $n>p$;
- implement full-rank OLS with \`np.linalg.solve\`, never a blind Gram
  inverse;
- use $X^+y$ for general-rank least squares, characterize the nullspace
  family, and prove the pseudoinverse solution is minimum norm;
- interpret $XX^+$ and $X^+X$ as projectors, using explicit numerical
  tolerances;
- evaluate L2/L1 penalties, derive their 1-D closed forms, and explain
  why L1 creates exact sparsity while L2 shrinks.

## Code in this unit

A quick reference for the library calls in this unit's code. Each one is explained in a comment (or in the text beside the code) where it first appears.

- `np.linalg.cond` — the condition number: largest over smallest singular value
- `np.meshgrid` — 2-D grids of x and y coordinates from two 1-D ranges
- `.contour` — `ax.contour(...)`: level curves on these axes
- `.clabel` — `ax.clabel(...)`: writes each level curve's value
- `.axhline` — a horizontal line across these axes at `y`
- `.set_xscale` — the x-axis scale, e.g. `"log"` or `"symlog"`
- `np.finfo` — facts about a float type; `np.finfo(float).eps` is the spacing at `1.0`
- `np.flatnonzero` — the indices of the nonzero (True) entries, as one flat array
- `np.append` — a new 1-D array: the entries of `a` followed by those of `b`
- `np.nonzero` — the indices of the nonzero entries, one array per axis
- `np.issubdtype` — True when a dtype belongs to a family, e.g. `np.floating`

- `np.isfinite` — True where an entry is a real number (not `inf` or `nan`)
