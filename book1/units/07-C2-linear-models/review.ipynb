## Concept summary

| Concept | Summary | Key fact |
|---|---|---|
| Linear model | explicit bias or augmented design | use a ones column or separate $b$, never both |
| MSE/gradient | $L=\lVert X\beta-y\rVert^2/n$ | for $r=X\beta-y$, $\nabla L=(2/n)X^Tr$ |
| Normal equations | $X^TX\hat\beta=X^Ty$ | equivalent to $X^T\hat r=0$ |
| Rank | independent columns | full rank iff Gram is positive definite/invertible; $n>p$ is insufficient |
| Consistency | exact fit versus closest fit | $X\beta=y$ may be inconsistent while least squares still exists |
| Pseudoinverse | $\beta^+=X^+y$ | minimum-norm minimizer; all minimizers are $\beta^++\mathcal N(X)$ |
| Projectors | $XX^+$ and $X^+X$ | project onto column and row spaces |
| L2 | squared-weight charge | shrinking pull $2\lambda w_k$ fades near zero |
| L1 | absolute-weight charge | constant one-sided pull creates exact zeros |
| Sparsity | exactly zero coefficients | rounded display is not evidence |

## Formula and idiom sheet

Explicit bias:
$\hat y_i=\sum_kX_{ik}w_k+b$.
Augmented design:
$X:(n,p)$, $\beta=(b,w_1,\ldots,w_d)^T:(p,)$,
$y:(n,)$. Residuals are predictions minus targets.

\[
L(\beta)=\frac1n\lVert X\beta-y\rVert^2,\qquad
\nabla_\beta L=\frac2nX^T(X\beta-y).
\]
\[
X^TX\hat\beta=X^Ty
\Longleftrightarrow X^T\hat r=0.
\]
Also,
$v^TX^TXv=\lVert Xv\rVert^2$.

Under a full-column-rank contract:

\`\`\`python
beta = np.linalg.solve(X.T @ X, X.T @ y)
\`\`\`

Do not form \`np.linalg.inv(X.T @ X)\`.
At general rank:

\`\`\`python
beta_plus = np.linalg.pinv(X) @ y
P_col = X @ np.linalg.pinv(X)
P_row = np.linalg.pinv(X) @ X
\`\`\`

Every minimizer is $\beta^++z$, $z\in\mathcal N(X)$, and
\[
\lVert\beta^++z\rVert^2
=\lVert\beta^+\rVert^2+\lVert z\rVert^2.
\]

Regularization:

- ridge adds $\lambda\sum_kw_k^2$ and gradient pull
  $2\lambda w_k$; 1-D minimizer $a/(1+\lambda)$;
- lasso adds $\lambda\sum_k|w_k|$; 1-D minimizer is zero when
  $|a|\le\lambda/2$, otherwise
  $a-\operatorname{sign}(a)\lambda/2$;
- this course excludes the bias from both penalties.

Numerical discipline:

\`\`\`python
ATOL = 1e-10
RTOL = 0.0
np.allclose(X.T @ (X @ beta - y), 0.0, atol=ATOL, rtol=RTOL)
\`\`\`
