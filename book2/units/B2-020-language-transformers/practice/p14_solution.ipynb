## Solution

**Base.** At layer 0, representation `h_i^0` is built from token `i` and position `i`, so it has no dependency on any `j>i`.

**Step.** Assume every `h_k^l` depends only on tokens at positions `<=k`. At query position `i`, the pre-softmax lower-triangular mask removes every key/value position `k>i`; therefore attention is a function only of `h_k^l` for `k<=i`, each of which depends only on original positions `<=k<=i`. Position-wise feed-forward, normalization, and residual operations do not mix positions, so `h_i^(l+1)` also depends only on positions `<=i`. Induction proves the final causal logit at `i` is future-independent.

```python
base_dependency = {i: set(range(i + 1)) for i in range(5)}
next_dependency = {
    i: set().union(*(base_dependency[k] for k in range(i + 1)))
    for i in range(5)
}
```

### Answer check

```python
assert all(next_dependency[i] <= set(range(i + 1)) for i in range(5))
assert all(j not in next_dependency[i] for i in range(5) for j in range(i + 1, 5))
```
