## Solution

Build the one-hot rows explicitly, then use matrix multiplication for the lookup.

```python
import torch

ATOL = RTOL = 1e-7

def one_hot_embedding_lookup(token_ids, table):
    one_hot = torch.zeros((token_ids.shape[0], table.shape[0]), dtype=torch.float32)
    one_hot[torch.arange(token_ids.shape[0]), token_ids] = 1.0
    return one_hot, one_hot @ table

token_ids = torch.tensor([4, 5, 4], dtype=torch.int64)
table = torch.arange(12 * 8, dtype=torch.float32).reshape(12, 8) / 10
one_hot, vectors = one_hot_embedding_lookup(token_ids, table)
```

### Answer check

```python
assert one_hot.shape == (3, 12) and one_hot.dtype == torch.float32
assert vectors.shape == (3, 8) and vectors.dtype == torch.float32
assert torch.allclose(one_hot.sum(dim=1), torch.ones(3), atol=ATOL, rtol=RTOL)
assert torch.allclose(vectors[0], table[4], atol=ATOL, rtol=RTOL)
assert torch.allclose(vectors[2], table[4], atol=ATOL, rtol=RTOL)
```
