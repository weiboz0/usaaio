## Solution

Let `Y=XE` and `G=dL/dY`. From `dY=X dE`,
`dL = tr(G^T X dE) = tr((X^T G)^T dE)`, hence `dL/dE=X^T G`.
Row `v` is `sum_b X[b,v] G[b,:]`. If ID `v` is absent, every coefficient is zero. Repeated IDs add their separate `G` rows. This is the direct lookup gradient; later shared computations can couple examples, but they still reach `E` through this same selection Jacobian.

```python
import torch
X = torch.tensor([[0.,1.,0.], [0.,1.,0.], [0.,0.,1.]])
G = torch.tensor([[1.,2.], [3.,4.], [5.,6.]])
gradient = X.T @ G
```

### Answer check

```python
assert torch.equal(gradient[0], torch.zeros(2))
assert torch.equal(gradient[1], G[0] + G[1])
assert torch.equal(gradient[2], G[2])
```
