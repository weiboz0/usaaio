## Solution

For non-padding tokens `x_1,...,x_T`, the chain rule gives
`p(x_1,...,x_T)=prod_{t=1}^T p(x_t | x_{<t})`. Thus
`-log p(x_1,...,x_T)=sum_t -log p(x_t|x_{<t})`.
Across a batch, if `M_bt` is 1 exactly at shifted, non-padding target positions, mean token cross-entropy is
`-[sum_bt M_bt log softmax(z_bt)[y_bt]] / [sum_bt M_bt]`.
The denominator counts valid target tokens, not padded positions, batch rows, or per-row means.

```python
import torch
mask = torch.tensor([[True, True, False], [True, False, False]])
token_losses = torch.tensor([[1.0, 2.0, 99.0], [4.0, 99.0, 99.0]])
mean_nll = token_losses[mask].sum() / mask.sum()
```

### Answer check

```python
assert mask.sum().item() == 3
assert torch.isclose(mean_nll, torch.tensor(7 / 3), atol=1e-7, rtol=1e-7)
```
