## Solution

Clone both outputs, overwrite selected inputs with `<mask>`, and fill only selected labels from the untouched original.

```python
import torch

ATOL = RTOL = 0

def apply_mlm_mask(token_ids, selected, mask_id=1):
    corrupted = token_ids.clone()
    labels = torch.full_like(token_ids, -100)
    corrupted[selected] = mask_id
    labels[selected] = token_ids[selected]
    return corrupted, labels

token_ids = torch.tensor([[4,5,6,0,0,0,0,0], [7,8,9,4,0,0,0,0]], dtype=torch.int64)
original = token_ids.clone()
selected = torch.tensor([[False,True,False,False,False,False,False,False], [True,False,False,True,False,False,False,False]])
corrupted, labels = apply_mlm_mask(token_ids, selected)
```

### Answer check

```python
assert torch.equal(token_ids, original)
assert corrupted.dtype == labels.dtype == torch.int64
assert corrupted.tolist() == [[4,1,6,0,0,0,0,0], [1,8,9,1,0,0,0,0]]
assert labels.tolist() == [[-100,5,-100,-100,-100,-100,-100,-100], [7,-100,-100,4,-100,-100,-100,-100]]
assert torch.equal(corrupted[selected], torch.ones(3, dtype=torch.int64))
```
