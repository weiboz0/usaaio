## Formula and protocol sheet

- One-hot lookup: `X @ E`; direct row gradient is `X.T @ dL/d(XE)`.
- Causal inputs/targets: `tokens[:,:-1]`, `tokens[:,1:]`.
- MLM: selected input becomes `<mask>`; selected label keeps truth; other labels are `-100`.
- Encoder: token + learned position → pre-norm attention residual → pre-norm GELU feed-forward residual.
- Pretraining: 40 causal then 40 MLM AdamW updates with one continued optimizer.
- Fine-tuning: verify state hash, attach head, freeze/audit, unfreeze/audit, evaluate disjoint held-out rows.
