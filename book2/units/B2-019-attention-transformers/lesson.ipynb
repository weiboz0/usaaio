Book 1 is a declared prerequisite program; this unit imports its skills through qualified contracts and does not duplicate their ownership.

## Learning goals

- Derive scaled dot-product attention and explain the query, key, and value roles.
- Apply padding and causal masks before softmax and audit forbidden attention weights.
- Split, concatenate, and cost multi-head attention with explicit dimensions.
- Construct sinusoidal positional encodings from pinned numeric positions.
- Train a small CPU attention predictor with stable cross-entropy.
- Trace pre-norm residual and position-wise feed-forward sublayers.
