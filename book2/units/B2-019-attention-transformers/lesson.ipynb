## Learning goals

- Derive scaled dot-product attention and explain the query, key, and value roles.
- Apply padding and causal masks before softmax and audit forbidden attention weights.
- Split, concatenate, and cost multi-head attention with explicit dimensions.
- Construct sinusoidal positional encodings from pinned numeric positions.
- Train a small CPU attention predictor with stable cross-entropy.
- Trace pre-norm residual and position-wise feed-forward sublayers.

## Code in this unit

A quick reference for the library calls in this unit's code. Each one is explained in a comment (or in the text beside the code) where it first appears.

**New in this unit:**

- `np.swapaxes` — the same data with two axes swapped, e.g. the last two (`-2, -1`)
- `np.tril` — a copy with the entries above the diagonal set to zero
- `.transpose` — reorders the axes, e.g. `.transpose(0, 2, 1, 3)` swaps axes 1 and 2
- `.is_floating_point` — True when the tensor has a float dtype
- `torch.matmul` — the matrix product `a @ b`, batched over leading axes
- `math.sqrt` — the square root of a Python number
- `torch.broadcast_to` — the tensor stretched to a bigger shape by broadcasting (error if impossible)
- `torch.softmax` — `exp` of the scores, scaled to sum to 1 along `dim`
- `.masked_fill` — a copy with a value written wherever the mask is True
- `torch.tril` — zeros the entries above the diagonal: the causal (lower-triangular) pattern
- `torch.count_nonzero` — the number of nonzero entries
- `.triu` — keeps the entries on and above a diagonal (`diagonal=1`: strictly above)
- `nn.functional.one_hot` — one row per integer id, with a 1 in that id's column (`num_classes` columns)
- `.unsqueeze` — inserts a length-1 axis at the given position
- `torch.sin` — the sine of each entry
- `torch.cos` — the cosine of each entry
- `nn.functional.cross_entropy` — mean softmax cross-entropy of logits `(n, K)` against integer targets `(n,)`
- `nn.ModuleList` — a list of layers that the module registers, so `.parameters()` sees them
- `torch.random.fork_rng` — a `with` block whose random draws leave the global generator unchanged
- `np.random.seed` — seeds NumPy's global random generator
- `math.exp` — `e` raised to a Python number
- `.swapaxes` — the same data with two axes swapped
- `TypeError` — the built-in error for a value of the wrong type
- `torch.any` — True if any entry is True (per row with `dim=-1`)
- `nn.LayerNorm` — normalizes each row's last axis to mean 0 and variance 1, then rescales
- `.zero_` — sets every entry to 0, in place
- `np.tile` — repeats an array end to end a given number of times
