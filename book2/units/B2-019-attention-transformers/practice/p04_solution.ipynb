## Independent solution

Option B is correct. A length-three causal allowed pattern is [[1,0,0],[1,1,0],[1,1,1]], because row i may use only keys j<=i. A padding mask instead removes placeholder key columns and need not be triangular. Both constraints are applied to scores before softmax so forbidden weights become exactly zero.

```python
import numpy as np
SEED = 20260808
ATOL = 1e-12
RTOL = 1e-12
ANSWER = "B"
causal_allowed = np.tril(np.ones((3, 3), dtype=bool))
scores = np.zeros((3, 3), dtype=np.float64)
masked = np.where(causal_allowed, scores, -np.inf)
weights = np.exp(masked - np.max(masked, axis=-1, keepdims=True))
weights = weights / np.sum(weights, axis=-1, keepdims=True)
expected_pattern = np.array([[True, False, False], [True, True, False], [True, True, True]])
```

### Answer check

```python
assert ANSWER == "B"
assert np.array_equal(causal_allowed, expected_pattern)
assert np.count_nonzero(weights[~causal_allowed]) == 0
np.testing.assert_allclose(weights.sum(axis=-1), np.ones(3), atol=ATOL, rtol=RTOL)
```
