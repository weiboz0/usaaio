## Independent solution

(a) Encoder self-attention: Q, K, and V all come from encoder source states. Use a source key-padding mask, broadcast from (B,1,1,7) to (B,h,7,7); it need not be causal. (b) Decoder self-attention: Q, K, and V all come from decoder target-prefix states. Use the logical AND of a lower-triangular causal mask and target key validity, broadcast to (B,h,4,4). (c) Cross-attention: Q comes from decoder states, while K and V come from encoder memory. Use the source key-padding mask broadcast to (B,h,4,7), with no target causal triangle over source columns. Score rows are four decoder destinations and columns are seven encoder sources, so the exact score shape is (B,h,4,7). Query-padding rows should separately be excluded from outputs or loss.

```python
SEED = 20260808
ATOL = 0.0
RTOL = 0.0
B, h, decoder_length, encoder_length = 2, 3, 4, 7
encoder_score_shape = (B, h, encoder_length, encoder_length)
decoder_score_shape = (B, h, decoder_length, decoder_length)
cross_score_shape = (B, h, decoder_length, encoder_length)
source_key_mask_shape = (B, 1, 1, encoder_length)
target_causal_mask_shape = (1, 1, decoder_length, decoder_length)
```
