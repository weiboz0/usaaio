## Formula and idiom sheet

- Scores: $S=QK^T/\sqrt{d_k}$; masks modify $S$ before row softmax.
- Weighted values: $O=\operatorname{softmax}(S)V$.
- Head shape: `(B,N,D) -> (B,h,N,D/h) -> (B,N,D)`.
- Standard score memory: exactly `B*h*N*N` scalars.
- Pre-norm: `y=x+MHA(LN(x))`; `z=y+FFN(LN(y))`.
- Cross-attention: decoder states supply Q; encoder states supply K and V.

## Self-quiz

1. Why is `K` transposed in the score product?
2. Which axis receives softmax?
3. Why divide scores by `sqrt(d_k)`?
4. What is the difference between padding and causal masks?
5. Why must masking happen before softmax?
6. Trace one head from `(B,N,D)` to `(B,h,N,D/h)`.
7. State the positional row at position zero.
8. Count score scalars for `(B,h,N)=(2,4,8)`.
9. Give Q/K/V sources for cross-attention.
10. Write both pre-norm residual equations.
11. Name the inputs cross-entropy expects.
12. State three reproducibility controls for a CPU training probe.

<details><summary>Self-quiz answers</summary>

1. To contract the shared feature dimension and expose key positions as columns.
2. The key/source axis.
3. To control dot-product variance as width grows.
4. Padding blocks placeholders; causal masking blocks future positions.
5. Allowed weights must renormalize to one while forbidden weights become zero.
6. Reshape width into `(h,D/h)` and transpose head before sequence.
7. Alternating zero sine and one cosine coordinates.
8. `2*4*8*8=512`.
9. Decoder Q, encoder K and V.
10. `y=x+MHA(LN(x))`; `z=y+FFN(LN(y))`.
11. Unnormalized logits and integer class targets.
12. Pinned data, fixed seed, CPU/dtype, exact APIs, and explicit tolerances.
