## Formula and protocol sheet

- **Gaussian log-density:** $\log N(x;\mu,\Sigma)=-\frac12\big(d\log2\pi+\log\det\Sigma+(x-\mu)^\top\Sigma^{-1}(x-\mu)\big)$; compute $\log\det$ with `torch.linalg.slogdet` and the quadratic form with `torch.linalg.solve`.
- **Determinant (symmetric):** product of eigenvalues; $2\times2$: $ac-b^2$.
- **Ellipse axes:** eigenvectors of $\Sigma$ (columns of `eigh`'s output), semi-axes $\sqrt{c\lambda_i}$.
- **Affine rule:** $E[Az+b]=AE[z]+b$, $\operatorname{Cov}(Az+b)=A\operatorname{Cov}(z)A^\top$; sample `x = mu + z @ L.T` with $LL^\top=\Sigma$.
- **MLE:** $\hat\mu=\bar x$; diagonal $\hat\sigma_j^2=\frac1n\sum_k(x_j^{(k)}-\bar x_j)^2$ (needs $n\ge2$ and unequal values per coordinate); full $\hat\Sigma$ with $1/n$ exists only if the centered rows span $\mathbb R^d$ ($n>d$).
- **Discrete KL:** $\sum_ip_i\log(p_i/q_i)$; $0\log0=0$; $q_i=0<p_i\Rightarrow\infty$; $\mathrm{KL}\ge0$ from $\log x\le x-1$.
- **Univariate Gaussian KL:** $\log\frac{s_2}{s_1}+\frac{s_1^2+(m_1-m_2)^2}{2s_2^2}-\frac12$; **to $N(0,I)$:** $\frac12\sum_j(\mu_j^2+e^{\ell_j}-\ell_j-1)$.
- **Projection distance:** $P=W(W^\top W)^{-1}W^\top$ via `solve`; compare subspaces by $\lVert P_A-P_B\rVert_F$.
- **ELBO:** $\log p(x)\ge E_q[\log p(x\mid z)]-\mathrm{KL}(q(z\mid x)\|p(z))$, with gap $\mathrm{KL}(q\|p(z\mid x))$.
- **Reparameterization:** $z=\mu+\exp(\frac12\ell)\odot\varepsilon$, `eps` passed in.
- **Negative ELBO:** per example $\frac12\sum_{\text{features}}(x-\hat x)^2+\frac12\sum_{\text{latents}}(\mu^2+e^\ell-\ell-1)$, then mean over the batch; constant $\frac d2\log2\pi$ dropped.
- **Training:** seed `20260927`, full-batch Adam, `zero_grad(set_to_none=True)` → forward → loss → `backward` → `step`; held-out rows evaluated only, with fixed `eps`.

## Common pitfalls, Exam connections, and forward boundary

Common pitfalls: inverting instead of solving, `z @ L` instead of `z @ L.T`, eigenvalue read as axis length, `nan` from $0\log0$, KL direction swaps, `sigma` for `sigma**2`, dropping the $-1$, output-layer ReLU in an autoencoder, comparing weights instead of subspaces, `.sample()` in the loss path, recon minus KL, mean-over-features reductions, and tuning on held-out rows.
Exam connections: Round 2 style tasks ask for an exact normal-form value, a guarded implementation matched to a tolerance, a proof with its conditions, or the first broken invariant in a trace.
Forward-only topics (Gaussian conditioning, importance-weighted bounds, hierarchical VAEs, GANs, and diffusion in `B2-023-generative-models-diffusion`) are not required or permitted shortcuts here.
