## Solution

Conventions: a term with $p_i=0$ contributes $0$ (since $t\log t\to0$); a term with $p_i>0$ and $q_i=0$ makes the divergence $+\infty$.

**$\mathrm{KL}(P\|Q)=\sum_ip_i\log(p_i/q_i)$:**

| $i$ | $p_i$ | $q_i$ | term |
|---|---|---|---|
| 1 | $1/2$ | $1/4$ | $\tfrac12\log2$ |
| 2 | $1/4$ | $1/4$ | $\tfrac14\log1=0$ |
| 3 | $1/4$ | $1/4$ | $0$ |
| 4 | $0$ | $1/4$ | $0$ (convention $p_i=0$) |

Total: $\mathrm{KL}(P\|Q)=\tfrac12\log2\approx0.34657$.

**$\mathrm{KL}(Q\|P)=\sum_iq_i\log(q_i/p_i)$:**

| $i$ | $q_i$ | $p_i$ | term |
|---|---|---|---|
| 1 | $1/4$ | $1/2$ | $\tfrac14\log\tfrac12=-\tfrac14\log2$ |
| 2 | $1/4$ | $1/4$ | $0$ |
| 3 | $1/4$ | $1/4$ | $0$ |
| 4 | $1/4$ | $0$ | $+\infty$ (convention $q_i>0$, $p_i=0$) |

Total: $\mathrm{KL}(Q\|P)=+\infty$.

**Symmetry.** The two directions differ (one finite, one infinite), so KL is not symmetric: it heavily penalizes the second argument for assigning zero mass where the first argument has mass.

**Answer: B.**

```python
import math

P = (0.5, 0.25, 0.25, 0.0)
Q = (0.25, 0.25, 0.25, 0.25)

def kl_terms(p, q):
    terms = []
    for pi, qi in zip(p, q):
        if pi == 0.0:
            terms.append(0.0)
        elif qi == 0.0:
            terms.append(math.inf)
        else:
            terms.append(pi * math.log(pi / qi))
    return terms

terms_pq = kl_terms(P, Q)
terms_qp = kl_terms(Q, P)
kl_pq, kl_qp = sum(terms_pq), sum(terms_qp)
ANSWER = "B"
print(terms_pq, kl_pq)
print(terms_qp, kl_qp)
```

### Answer check

```python
assert abs(kl_pq - 0.5 * math.log(2)) < 1e-12
assert kl_qp == math.inf
assert terms_pq[3] == 0.0 and terms_qp[3] == math.inf
assert kl_pq != kl_qp
assert ANSWER == "B"
```
