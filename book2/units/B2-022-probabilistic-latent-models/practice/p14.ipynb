## Task

Let $P=(p_1,\dots,p_m)$ and $Q=(q_1,\dots,q_m)$ be probability vectors, and define $\mathrm{KL}(P\|Q)=\sum_ip_i\log(p_i/q_i)$ with the conventions: terms with $p_i=0$ are $0$, and $\mathrm{KL}(P\|Q)=+\infty$ if some $p_i>0$ has $q_i=0$. Reasoning is required.

Using **only** the Session 2 lemma "$\log x\le x-1$ for all $x>0$, with equality only at $x=1$" (you may cite it without re-proving it; do not cite Jensen's inequality or Gibbs' inequality), prove:

(a) $\mathrm{KL}(P\|Q)\ge0$. Treat separately the case where some $p_i>0$ has $q_i=0$, and in the remaining case work over the support $S=\{i:p_i>0\}$; be explicit about why $\sum_{i\in S}q_i\le1$ is enough.

(b) $\mathrm{KL}(P\|Q)=0$ if and only if $P=Q$. Identify every inequality in your proof of (a) that must be tight, and conclude from each.

(c) Illustrate the role of the support with $P=(\tfrac12,\tfrac12,0)$ and $Q=(\tfrac13,\tfrac13,\tfrac13)$: compute $\mathrm{KL}(P\|Q)$ and the slack $1-\sum_{i\in S}q_i$ in your bound.
