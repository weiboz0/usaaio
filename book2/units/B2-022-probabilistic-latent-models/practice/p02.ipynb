## Task

A 2-D Gaussian has mean $\mu=(1,-1)$ and covariance $\Sigma=\begin{pmatrix}5&2\\2&2\end{pmatrix}$.
Consider the contour $\{x:(x-\mu)^\top\Sigma^{-1}(x-\mu)=1\}$.
Which statement about this ellipse is correct?

A. Its semi-axis lengths are $6$ and $1$, along $(2,1)/\sqrt5$ and $(1,-2)/\sqrt5$ respectively.

C. Its semi-axis lengths are $\sqrt5$ and $\sqrt2$, along the coordinate axes, because those are the diagonal entries.
D. Its semi-major axis has length $\sqrt6$ along $(2,1)/\sqrt5$, and its semi-minor axis has length $1$ along $(1,-2)/\sqrt5$.
E. Its semi-major axis has length $\sqrt6$ along the first row direction $(5,2)/\sqrt{29}$.

Justify with the eigenvalues (from the trace and the $2\times2$ determinant), one eigenvector check, and the square-root rule for semi-axes.
