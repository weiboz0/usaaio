(II) **Wide, short table.** $n=30$ rows in $d=50$. The goal is again a Gaussian log-density score. Choose diagonal or full covariance, and explain exactly what goes wrong with the other choice (cite the MLE existence condition).

(III) **Dashboard compression.** 8-D readings must be compressed to 2 numbers for display and reconstructed accurately; no new readings ever need to be generated. Choose between an autoencoder and a VAE.

(IV) **Synthetic data.** The same 8-D readings, but the goal is to generate new realistic readings and to show a smooth transition between two recorded operating states. Choose between an autoencoder and a VAE, and name the two VAE operations you would use.

(V) **Linear structure only.** The 8-D readings are known to lie close to a 2-D linear subspace, and the goal is the best linear 2-number compression. Choose between PCA, a linear autoencoder, and a VAE, and say what a trained linear autoencoder's subspace would be compared against.

For every case, state the decision in one line, the justification in at most four sentences, and the falsifying held-out check in one sentence.
