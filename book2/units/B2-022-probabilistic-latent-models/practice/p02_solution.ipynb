## Solution

**Eigenvalues.** $\operatorname{tr}\Sigma=5+2=7$ and $\det\Sigma=5\cdot2-2\cdot2=6$, so the eigenvalues solve $\lambda^2-7\lambda+6=0$, i.e. $\lambda_1=6$, $\lambda_2=1$.

**Eigenvector check.** $\Sigma\begin{pmatrix}2\\1\end{pmatrix}=\begin{pmatrix}10+2\\4+2\end{pmatrix}=6\begin{pmatrix}2\\1\end{pmatrix}$, so $(2,1)/\sqrt5$ is the unit eigenvector for $\lambda=6$.
Because $\Sigma$ is symmetric, the other eigenvector is orthogonal: $(1,-2)/\sqrt5$, and indeed $\Sigma(1,-2)=(5-4,\,2-4)=(1,-2)$, eigenvalue $1$.

**Square-root rule.** Writing $x-\mu=\sum_i c_iu_i$ in the eigenbasis, $(x-\mu)^\top\Sigma^{-1}(x-\mu)=\sum_i c_i^2/\lambda_i=1$, an ellipse with semi-axis $\sqrt{\lambda_i}$ along $u_i$.
So the semi-major axis is $\sqrt6$ along $(2,1)/\sqrt5$ and the semi-minor axis is $\sqrt1=1$ along $(1,-2)/\sqrt5$.

**Answer: D.**
A forgets the square root; B swaps the directions; C reads the diagonal, which is only valid for a diagonal covariance; E uses a row of $\Sigma$, which is not an eigenvector.

```python
import math
import torch

mu = torch.tensor([1.0, -1.0], dtype=torch.float64)
Sigma = torch.tensor([[5.0, 2.0], [2.0, 2.0]], dtype=torch.float64)
evals, evecs = torch.linalg.eigh(Sigma)          # ascending: [1, 6]
u_major = torch.tensor([2.0, 1.0], dtype=torch.float64) / math.sqrt(5)
u_minor = torch.tensor([1.0, -2.0], dtype=torch.float64) / math.sqrt(5)
semi_major, semi_minor = math.sqrt(6.0), 1.0
# Points at the end of each semi-axis lie on the contour
on_contour = [((p) @ torch.linalg.solve(Sigma, p)).item() for p in (semi_major * u_major, semi_minor * u_minor)]
ANSWER = "D"
print(evals, on_contour)
```

### Answer check

```python
assert torch.allclose(evals, torch.tensor([1.0, 6.0], dtype=torch.float64), atol=1e-12, rtol=0)
assert abs(Sigma.trace().item() - 7.0) < 1e-12
assert torch.allclose(Sigma @ u_major, 6.0 * u_major, atol=1e-12, rtol=0)
assert torch.allclose(Sigma @ u_minor, 1.0 * u_minor, atol=1e-12, rtol=0)
assert all(abs(value - 1.0) < 1e-12 for value in on_contour)
assert ANSWER == "D"
```
