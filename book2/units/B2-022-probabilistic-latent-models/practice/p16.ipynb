## Task

Let $p(x,z)=p(x\mid z)\,p(z)$ be a latent-variable model with $p(z)=N(0,I_k)$, and let $q(z\mid x)$ be any density with the same positive support as $p(x,z)$ in $z$ (each is positive exactly where the other is), so every log-ratio below is finite wherever $q>0$. Reasoning is required.

**Part (a): the ELBO identity.**
(a1) Using Bayes' rule $p(z\mid x)=p(x,z)/p(x)$, show that
$$\log p(x)=\underbrace{E_{z\sim q}\Big[\log\frac{p(x,z)}{q(z\mid x)}\Big]}_{\mathrm{ELBO}(x)}+\mathrm{KL}\big(q(z\mid x)\,\|\,p(z\mid x)\big).$$
(a2) Conclude $\mathrm{ELBO}(x)\le\log p(x)$, citing nonnegativity of KL from Session 2, and state exactly when equality holds.
(a3) Show $\mathrm{ELBO}(x)=E_{z\sim q}[\log p(x\mid z)]-\mathrm{KL}\big(q(z\mid x)\,\|\,p(z)\big)$.

**Part (b): the reparameterized gradient path.**
(b1) For one coordinate, $z=\mu+\exp(\tfrac12\ell)\,\varepsilon$ with $\varepsilon$ fixed. Derive $\partial z/\partial\mu=1$ and $\partial z/\partial\ell=\tfrac12\exp(\tfrac12\ell)\,\varepsilon$ with the chain rule.
(b2) For a downstream scalar loss $L(z)$, write $\partial L/\partial\mu$ and $\partial L/\partial\ell$ in terms of $L'(z)$, and evaluate them for $L(z)=z^2$, $\mu=0.3$, $\ell=\log4$, $\varepsilon=-0.5$.
(b3) In one paragraph, explain why drawing $z\sim q(z\mid x)$ directly with a sampler gives no pathwise gradient to the encoder outputs, and why moving the randomness into a parameter-free $\varepsilon$ fixes this without changing the distribution of $z$ (cite the Session 1 covariance rule).
