## Task

Let $x^{(1)},\dots,x^{(n)}\in\mathbb R^d$ be independent draws from $N\big(\mu,\operatorname{diag}(\sigma_1^2,\dots,\sigma_d^2)\big)$ with every $\sigma_j^2>0$. Reasoning is required.

(a) Write the log-likelihood $\ell(\mu,\sigma_1^2,\dots,\sigma_d^2)$ and show that it is a sum of $d$ terms, the $j$-th depending only on $(\mu_j,\sigma_j^2)$ and coordinate $j$ of the data.

(b) Prove that, for any fixed positive variances, $\hat\mu_j=\bar x_j=\frac1n\sum_kx^{(k)}_j$ is the unique maximizer over $\mu_j$.

(c) Let $S_j=\sum_k(x^{(k)}_j-\bar x_j)^2$. Using single-variable calculus (the sign of a derivative on $v>0$), prove that if $S_j>0$ the unique maximizer over $v=\sigma_j^2$ is $\hat\sigma_j^2=S_j/n$ (the $1/n$ convention). State exactly where $S_j>0$ is used, show that $S_j>0$ holds if and only if $n\ge2$ and the values $x^{(1)}_j,\dots,x^{(n)}_j$ are not all equal, and prove that if $S_j=0$ the log-likelihood is unbounded above, so no MLE exists.

(d) State (without proof) the full-covariance MLE and the condition under which it exists. Give an explicit data set in $\mathbb R^2$ with $n=3$ for which the diagonal MLE exists but the full-covariance MLE does not, and justify both claims.
