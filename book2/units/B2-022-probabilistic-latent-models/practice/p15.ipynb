## Task

Derive the closed-form KL divergence from $p=N(\mu,\sigma^2)$ to $q=N(0,1)$ in one dimension, with $\sigma>0$. Reasoning is required.

(a) Starting from the definition $\mathrm{KL}(p\|q)=E_{x\sim p}[\log p(x)-\log q(x)]$, write both log-densities explicitly and simplify the log-ratio as a function of $x$.

(b) Take the expectation term by term. State and justify each expectation identity you use (in particular $E_{x\sim p}[(x-\mu)^2]$ and $E_{x\sim p}[x^2]$), citing only Book 1 facts about expectation and variance.

(c) Conclude $\mathrm{KL}(p\|q)=\tfrac12(\mu^2+\sigma^2-\log\sigma^2-1)$, and rewrite it in terms of $\ell=\log\sigma^2$.

(d) Using the Session 2 lemma $\log x\le x-1$, prove the result is nonnegative and equals $0$ exactly when $\mu=0$ and $\sigma^2=1$.

(e) Evaluate it for $\mu=1$, $\ell=1$, both exactly and to six decimal places.
