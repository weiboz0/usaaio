## Solution

Write $v_j=\sigma_j^2>0$ and $x^{(k)}_j$ for coordinate $j$ of draw $k$.

### (a) Separation of the log-likelihood

For a diagonal covariance, $\det\Sigma=\prod_jv_j$ and $(x-\mu)^\top\Sigma^{-1}(x-\mu)=\sum_j(x_j-\mu_j)^2/v_j$, so the density factorizes: $N(x;\mu,\Sigma)=\prod_{j=1}^d\frac1{\sqrt{2\pi v_j}}\exp\!\big(-\frac{(x_j-\mu_j)^2}{2v_j}\big)$.
By independence the log-likelihood is the sum over draws of log-densities:
$$\ell(\mu,v_1,\dots,v_d)=\sum_{k=1}^n\sum_{j=1}^d\Big[-\tfrac12\log(2\pi v_j)-\frac{(x^{(k)}_j-\mu_j)^2}{2v_j}\Big]=\sum_{j=1}^d\ell_j(\mu_j,v_j),$$
$$\ell_j(\mu_j,v_j)=-\frac n2\log(2\pi v_j)-\frac1{2v_j}\sum_{k=1}^n(x^{(k)}_j-\mu_j)^2 .$$
The $j$-th term involves only $(\mu_j,v_j)$ and the $j$-th coordinates of the data, so $\ell$ is maximized by maximizing each $\ell_j$ separately.

### (b) The mean

Expanding around $\bar x_j$ (the cross term vanishes because $\sum_k(x^{(k)}_j-\bar x_j)=0$):
$$\sum_k(x^{(k)}_j-\mu_j)^2=\sum_k(x^{(k)}_j-\bar x_j)^2+n(\bar x_j-\mu_j)^2=S_j+n(\bar x_j-\mu_j)^2 .$$
Hence $\ell_j(\mu_j,v_j)=-\frac n2\log(2\pi v_j)-\frac{S_j}{2v_j}-\frac{n}{2v_j}(\bar x_j-\mu_j)^2$.
For fixed $v_j>0$ the last term is $\le0$ with equality iff $\mu_j=\bar x_j$, so $\hat\mu_j=\bar x_j$ is the unique maximizer over $\mu_j$.

### (c) The variance

Because $\mu_j=\bar x_j$ maximizes $\ell_j$ for every $v$, the joint maximum is found by maximizing the profile $f(v)=\ell_j(\bar x_j,v)=-\frac n2\log(2\pi)-\frac n2\log v-\frac{S_j}{2v}$ on $v>0$.
Its derivative is
$$f'(v)=-\frac{n}{2v}+\frac{S_j}{2v^2}=\frac{S_j-nv}{2v^2}.$$
**Case $S_j>0$.** Then $S_j/n$ lies in the domain $v>0$ — *this is exactly where $S_j>0$ is used*. Since $2v^2>0$, $f'(v)>0$ for $0<v<S_j/n$ and $f'(v)<0$ for $v>S_j/n$, so $f$ strictly increases then strictly decreases, and $\hat v=\hat\sigma_j^2=S_j/n$ is the unique maximizer.

**When is $S_j>0$?** $S_j$ is a sum of squares, so $S_j=0$ iff every $x^{(k)}_j=\bar x_j$, i.e. iff all $n$ values are equal. If $n=1$ the single value always equals its own mean, so $S_j=0$. Hence $S_j>0$ iff $n\ge2$ and the values $x^{(1)}_j,\dots,x^{(n)}_j$ are not all equal.

**Case $S_j=0$.** Then with $\mu_j=\bar x_j$, $\ell_j(\bar x_j,v)=-\frac n2\log(2\pi v)\to+\infty$ as $v\to0^+$. Since the other coordinates contribute a finite value at any fixed parameters, $\ell$ is unbounded above, so no maximizer (no MLE) exists.

### (d) Full covariance

(Stated without proof.) The full-covariance MLE is $\hat\mu=\bar x$ and $\hat\Sigma=\frac1n\sum_k(x^{(k)}-\bar x)(x^{(k)}-\bar x)^\top$; it exists iff $\hat\Sigma$ is positive definite (nonsingular), i.e. iff the centered data span $\mathbb R^d$, which requires $n\ge d+1$.

**Example.** $x^{(1)}=(0,0)$, $x^{(2)}=(1,1)$, $x^{(3)}=(2,2)$ in $\mathbb R^2$, $n=3$.

- *Diagonal MLE exists:* each coordinate takes values $0,1,2$ (not all equal, $n\ge2$), so $S_1=S_2=2>0$ and $\hat\sigma_1^2=\hat\sigma_2^2=2/3$ by (c).
- *Full MLE does not exist:* $\bar x=(1,1)$, and $\hat\Sigma=\frac13\big[(1,1)^{\otimes2}+0+(1,1)^{\otimes2}\big]=\frac23\begin{pmatrix}1&1\\1&1\end{pmatrix}$, with $\det\hat\Sigma=0$: the points are collinear. Take $\Sigma_\varepsilon=\hat\Sigma+\varepsilon I$. Then the data term $-\frac12\sum_k(x^{(k)}-\bar x)^\top\Sigma_\varepsilon^{-1}(x^{(k)}-\bar x)=-\frac n2\operatorname{tr}(\Sigma_\varepsilon^{-1}\hat\Sigma)\ge-\frac n2\cdot2$ stays bounded (eigenvalues of $\Sigma_\varepsilon^{-1}\hat\Sigma$ are $\frac{4/3}{4/3+\varepsilon}$ and $0$), while $-\frac n2\log\det\Sigma_\varepsilon=-\frac n2\log\big(\varepsilon(\tfrac43+\varepsilon)\big)\to+\infty$ as $\varepsilon\to0^+$. So the log-likelihood is unbounded and no full-covariance MLE exists.

## Numeric sanity check

```python
import math
import numpy as np

rng = np.random.default_rng(20260927)
data = rng.normal(size=(7, 3)) * np.array([1.0, 2.0, 0.5]) + np.array([1.0, -1.0, 0.0])
n = data.shape[0]

def diag_loglik(mu, v):
    return float(np.sum(-0.5 * np.log(2 * math.pi * v) - (data - mu) ** 2 / (2 * v)))

mu_hat = data.mean(axis=0)
v_hat = ((data - mu_hat) ** 2).sum(axis=0) / n
best = diag_loglik(mu_hat, v_hat)
perturbations = [diag_loglik(mu_hat + d * np.eye(3)[j], v_hat) for j in range(3) for d in (-1e-2, 1e-2)]
perturbations += [diag_loglik(mu_hat, v_hat * (1 + d * np.eye(3)[j])) for j in range(3) for d in (-1e-2, 1e-2)]

# (d) example
ex = np.array([[0.0, 0.0], [1.0, 1.0], [2.0, 2.0]])
S_ex = ((ex - ex.mean(axis=0)) ** 2).sum(axis=0)
Sigma_hat = (ex - ex.mean(axis=0)).T @ (ex - ex.mean(axis=0)) / 3

def full_loglik(Sigma):
    c = ex - ex.mean(axis=0)
    sign, logdet = np.linalg.slogdet(Sigma)
    quad = np.einsum("ki,ij,kj->", c, np.linalg.solve(Sigma, np.eye(2)), c)
    return -0.5 * (3 * 2 * math.log(2 * math.pi) + 3 * logdet + quad)

loglik_path = [full_loglik(Sigma_hat + eps * np.eye(2)) for eps in (1e-1, 1e-3, 1e-6, 1e-9)]
print(best, max(perturbations), S_ex, np.linalg.det(Sigma_hat), loglik_path)
```

### Answer check

```python
assert all(value < best for value in perturbations)
assert np.allclose(S_ex, [2.0, 2.0], rtol=1e-05, atol=1e-08) and np.all(S_ex > 0)
assert abs(np.linalg.det(Sigma_hat)) < 1e-12
assert all(b > a for a, b in zip(loglik_path, loglik_path[1:]))   # grows without bound as eps -> 0
assert loglik_path[-1] > 10
```
