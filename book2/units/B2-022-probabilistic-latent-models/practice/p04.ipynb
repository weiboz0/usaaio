## Task

An autoencoder receives a float32 batch `x` of shape `(32,20)`.
Its encoder is `Linear(20,8)`, ReLU, `Linear(8,3)`; its decoder is `Linear(3,8)`, ReLU, `Linear(8,20)`.
Which statement gives the bottleneck output shape and the reconstruction target correctly?

A. The bottleneck is `(32,8)` and the target is a label vector of shape `(32,)`.
B. The bottleneck is `(32,3)` and the target is the input `x` itself, shape `(32,20)`.
C. The bottleneck is `(3,)` and the target is the code.
D. The bottleneck is `(32,3)` and the target is the code, shape `(32,3)`.
E. The bottleneck is `(32,20)` and the target is `x`.

Write the full shape ledger and say whether this autoencoder is undercomplete or overcomplete.
