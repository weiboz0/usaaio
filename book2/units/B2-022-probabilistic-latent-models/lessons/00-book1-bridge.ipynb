## 1. Arrays, broadcasting, and axis reductions

Every computation in this unit is batched: data tensors are `(n,d)` with one example per row.
`X.mean(dim=0)` reduces over rows and returns one value per feature, shape `(d,)`; `X.sum(dim=1)` reduces over features and returns one value per example, shape `(n,)`.
Broadcasting lets `X - m` subtract a `(d,)` mean from every row of an `(n,d)` tensor without a loop.
Remediation: [book1:F1-scientific-python](../../../../book1/units/F1-scientific-python/lesson.ipynb) for NumPy arrays, broadcasting, `aggregation-axis`, and `random-seeding`.

**Checkpoint 1A.** For `X` of shape `(60,8)`, what are the shapes of `X.mean(dim=0)` and `((X - X.mean(dim=0))**2).sum(dim=1)`?

**Checkpoint 1B.** Two runs of the same seeded script give different random numbers. Name one likely cause.

**Collected answers.** `(8,)` and `(60,)`. The seed was set after some random draws, or a different generator was used for part of the draws (for example the global RNG versus a dedicated `torch.Generator`).

## 2. Matrix multiplication and linear solves

`A @ B` contracts the last axis of `A` with the first axis of `B`: `(n,d) @ (d,k)` is `(n,k)`.
Book 1 C2 solved normal equations with `np.linalg.solve` instead of forming an inverse; Session 1 uses the Torch twin `torch.linalg.solve(A, b)` the same way.
Remediation: [book1:F3-matrices](../../../../book1/units/F3-matrices/lesson.ipynb) for `matrix-multiplication`, and [book1:C2-linear-models](../../../../book1/units/C2-linear-models/lesson.ipynb) for solving linear systems.

**Checkpoint 2A.** What is the shape of `(5,3) @ (3,3) @ (3,5)`?

**Checkpoint 2B.** Why is `solve(A, b)` preferred to `inv(A) @ b`?

**Collected answers.** `(5,5)`. It is cheaper and more accurate, and it never forms an inverse that may be badly conditioned.

## 3. Expectation, variance, covariance, and the Gaussian

Book 1 F5 defined $E[x]$, $\operatorname{Var}(x)=E[(x-E x)^2]$, the scalar $\operatorname{Cov}(x,y)=E[(x-Ex)(y-Ey)]$, the univariate Gaussian $N(\mu,\sigma^2)$ with density $\frac1{\sqrt{2\pi\sigma^2}}e^{-(x-\mu)^2/(2\sigma^2)}$, and Monte Carlo estimation by seeded sampling.
Session 1 builds the covariance **matrix** from the scalar covariance; Session 2 uses $\operatorname{Var}(x)=E[x^2]-(Ex)^2$.
Remediation: [book1:F5-probability](../../../../book1/units/F5-probability/lesson.ipynb) for `expectation`, `variance`, `covariance`, `gaussian-distribution`, and `sampling-simulation`.

**Checkpoint 3A.** If $x\sim N(2,9)$, what is $E[x^2]$?

**Checkpoint 3B.** A Monte Carlo mean from $10^4$ draws is off by $0.01$ from the truth. Is the sampler necessarily wrong?

**Collected answers.** $E[x^2]=\operatorname{Var}(x)+(Ex)^2=9+4=13$. No: Monte Carlo error has standard deviation of order $\sigma/\sqrt n$ (here $3/100=0.03$), so $0.01$ is well within normal fluctuation.

## 4. Conditional probability and Bayes' rule

$p(a\mid b)=p(a,b)/p(b)$ and Bayes' rule $p(z\mid x)=p(x\mid z)p(z)/p(x)$ with $p(x)=\sum_z p(x\mid z)p(z)$.
Session 4 calls $p(z)$ the prior, $p(x\mid z)$ the likelihood, $p(x)$ the evidence, and $p(z\mid x)$ the posterior of a latent-variable model.
Remediation: [book1:F5-probability](../../../../book1/units/F5-probability/lesson.ipynb) for `conditional-probability` and `bayes-rule`.

**Checkpoint 4A.** $p(z=1)=0.3$, $p(x\mid z=1)=0.5$, $p(x\mid z=0)=0.1$. Compute $p(x)$.

**Checkpoint 4B.** Continue: compute $p(z=1\mid x)$.

**Collected answers.** $p(x)=0.3\cdot0.5+0.7\cdot0.1=0.22$. $p(z=1\mid x)=0.15/0.22=15/22\approx0.682$.

## 5. Eigenvectors, spectral decomposition, SVD, and PCA

A symmetric matrix factors as $S=Q\Lambda Q^\top$ with orthonormal eigenvector columns in $Q$ (`torch.linalg.eigh` returns ascending eigenvalues and column eigenvectors).
PCA's top-$k$ directions are the leading right singular vectors of the centered data matrix, equivalently the leading eigenvectors of its covariance.
Session 1 reads covariance ellipses from eigenpairs and Session 3 compares an autoencoder's subspace with PCA's.
Remediation: [book1:F6-svd-spectral](../../../../book1/units/F6-svd-spectral/lesson.ipynb) for `eigenvalues-eigenvectors`, `spectral-decomposition`, and `svd`; [book1:C9-dimensionality-reduction](../../../../book1/units/C9-dimensionality-reduction/lesson.ipynb) for `pca`.

**Checkpoint 5A.** What are the eigenvalues of $\operatorname{diag}(4,1)$, and which eigenvector is first in `eigh`'s output order?

**Checkpoint 5B.** Must data be centered before PCA? Why?

**Collected answers.** $1$ and $4$; `eigh` sorts ascending, so the eigenvector $(0,1)$ for eigenvalue $1$ comes first. Yes: PCA directions describe variation around the mean; uncentered data mixes the mean's direction into the first component.

## 6. MLPs, MSE, gradients, and the PyTorch training loop

An MLP alternates `nn.Linear` layers with ReLU; an `nn.Module` registers its parameters; MSE is `F.mse_loss`; gradients are partial derivatives collected by autograd (`requires_grad`), and one update is `zero_grad(set_to_none=True)` → forward → loss → `backward()` → `optimizer.step()` with `torch.optim.Adam`.
Train/test discipline means only training rows ever enter `backward()`.

**Checkpoint 6A.** How many parameters does `nn.Linear(8,16)` have, and what is the shape of its `weight`?
