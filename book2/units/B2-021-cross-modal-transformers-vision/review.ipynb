## Formula and protocol sheet

- Patches: `N=(H/P)(W/P)`; tokens with class slot: `(B,N+1,D)`; per-head score entries: `B(N+1)^2`.
- Boxes: half-open xyxy; `IoU=intersection/(area_a+area_b-intersection)`.
- Detector: threshold, stable descending score, then same-class NMS.
- U-Net: upsample to the skip's exact spatial size, concatenate channels, emit pixel-aligned logits.
- Segmentation: mean valid-pixel CE; hard foreground `Dice=(2I+eps)/(P+T+eps)`.
- Graph mean: `A.float() @ X / degree`; permutation acts as `X'=PX`, `A'=PAP^T`.
- Cross-modal: the stream being updated owns Q; the source being read owns K and V.
- Training: seed 20260901, immutable stored split order, mean reduction, pinned AdamW, evaluate held-out without updates.
