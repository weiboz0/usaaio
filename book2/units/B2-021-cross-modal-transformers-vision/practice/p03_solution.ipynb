# Solution to p03

Each `2x2` patch in a one-channel image contains four scalars, and a `4x4` image has a `2x2` patch grid, hence four patches. Flattening produces `(B,4,4)`. A learned linear projection maps each four-scalar patch to width 8, giving `(B,4,8)`. Prepending one width-8 class token gives `(B,5,8)`. Selecting token 0 gives `(B,8)`, and the two-class head returns `(B,2)`. Thus **A** is the valid path.

The other choices either invent 16 patches, select a feature coordinate instead of the class token, concatenate along the wrong axis, or confuse a loss operation with a logit-producing model path.
