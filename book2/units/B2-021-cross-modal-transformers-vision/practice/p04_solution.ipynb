# Solution to p04

For half-open boxes, the intersection spans x from `max(0,2)=2` to `min(4,5)=4`, width 2, and y from `max(0,1)=1` to `min(4,3)=3`, height 2. Its area is 4. The box areas are 16 and 6, so the union is `16+6-4=18`. The IoU is `4/18=2/9`. Therefore **A** is correct.

### Answer check

```python
from fractions import Fraction

A = (0,0,4,4)
B = (2,1,5,3)
iw = max(0, min(A[2],B[2]) - max(A[0],B[0]))
ih = max(0, min(A[3],B[3]) - max(A[1],B[1]))
intersection = iw * ih
area_a = (A[2]-A[0]) * (A[3]-A[1])
area_b = (B[2]-B[0]) * (B[3]-B[1])
union = area_a + area_b - intersection
assert (intersection, union, Fraction(intersection, union)) == (4, 18, Fraction(2,9))
```
