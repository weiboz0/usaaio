# Solution to p05

The decoder and skip tensors agree in batch and both spatial dimensions. Concatenating on the channel axis preserves `(B,H,W)` and adds channels, producing `(B,4+4,8,8)=(B,8,8,8)`. A fusion layer can map those eight channels back to four, and a per-pixel head can emit two logits at every original-resolution pixel, `(B,2,8,8)`. Therefore **C** is correct.

Width concatenation destroys spatial correspondence, addition requires the same full shape, flattening destroys the dense output contract, and resizing the target changes labels instead of repairing the model.
