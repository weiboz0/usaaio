# Solution to p15

Tensor concatenation along axis 1 requires equality on every non-concatenated axis. Consequently, for decoder `(B,C_d,H_d,W_d)` and skip `(B,C_s,H_s,W_s)`, the necessary conditions are equal batch sizes, `H_d=H_s`, and `W_d=W_s`; channel counts may differ. These conditions are also sufficient: each output location `(b,h,w)` can place the decoder's `C_d` entries followed by the skip's `C_s` entries, so the result is `(B,C_d+C_s,H_d,W_d)`.

For `(2,4,8,8)` and `(2,6,8,8)`, all non-channel axes agree, producing `(2,10,8,8)`. With skip `(2,6,7,8)`, the height condition fails (`8 != 7`); no interpolation is implicit or permitted, so concatenation must be rejected.

### Answer check

```python
import torch

decoder=torch.zeros(2,4,8,8)
skip=torch.zeros(2,6,8,8)
combined=torch.cat((decoder,skip),dim=1)
assert combined.shape==(2,10,8,8)
bad_skip=torch.zeros(2,6,7,8)
try:
    torch.cat((decoder,bad_skip),dim=1)
except RuntimeError:
    pass
else:
    raise AssertionError("height mismatch must prevent concatenation")
```
