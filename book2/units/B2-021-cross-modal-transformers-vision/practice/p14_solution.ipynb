# Solution to p14

Let valid half-open boxes be `A=(x1,y1,x2,y2)` and `B=(u1,v1,u2,v2)`, with each maximum coordinate strictly greater than its minimum. Their intersection side lengths are

`w_I=max(0,min(x2,u2)-max(x1,u1))` and
`h_I=max(0,min(y2,v2)-max(y1,v1))`.

Thus `I=w_I h_I`, `area(A)=(x2-x1)(y2-y1)`, `area(B)=(u2-u1)(v2-v1)`, `U=area(A)+area(B)-I`, and `IoU=I/U`. The max-with-zero clamp is essential for disjoint boxes: a negative raw overlap becomes zero, so `I=0` and IoU is zero rather than negative.

Scale all eight coordinates about the origin by `s>0`. Positive scaling preserves coordinate order and commutes with min and max, so both intersection side lengths are multiplied by `s`. Every area is therefore multiplied by `s^2`: `I'=s^2 I` and `U'=s^2 U`. Hence `IoU'=s^2 I/(s^2 U)=IoU`. Positivity is necessary: `s=0` collapses valid boxes and makes the union zero, while `s<0` reverses min/max coordinate ordering unless boxes are renormalized, so the stated direct coordinate transformation no longer preserves valid box representation.

Adding the same translation `(tx,ty)` to both boxes shifts all competing x or y endpoints equally. Differences of the relevant min/max endpoints are unchanged, so intersection, individual areas, union, and IoU are unchanged.

### Answer check

```python
from fractions import Fraction

def exact_iou(a,b):
    iw=max(0,min(a[2],b[2])-max(a[0],b[0]))
    ih=max(0,min(a[3],b[3])-max(a[1],b[1]))
    inter=iw*ih
    area_a=(a[2]-a[0])*(a[3]-a[1])
    area_b=(b[2]-b[0])*(b[3]-b[1])
    return Fraction(inter, area_a+area_b-inter)
A=(0,0,4,4); B=(2,1,5,3)
s=3
scaled_A=tuple(s*x for x in A); scaled_B=tuple(s*x for x in B)
translated_A=(7,-2,11,2); translated_B=(9,-1,12,1)
assert exact_iou(A,B)==Fraction(2,9)
assert exact_iou(scaled_A,scaled_B)==exact_iou(A,B)
assert exact_iou(translated_A,translated_B)==exact_iou(A,B)
assert exact_iou((0,0,1,1),(2,2,3,3))==0
```
