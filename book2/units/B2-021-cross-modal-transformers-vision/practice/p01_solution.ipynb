# Solution to p01

The patch grid is `(H/4, W/4) = (12/4, 8/4) = (3, 2)`, so there are `N=3*2=6` patch tokens per image. Prepending one class token gives `N+1=7` total tokens. Thus

`a/b = 6/7`.

The fraction is already normalized because `gcd(6,7)=1`, and the stipulated positive-denominator sign convention gives `b=7>0`. Therefore `a+b=6+7=13`, so the answer is **C**.

### Answer check

```python
from math import gcd

batch, channels, height, width, patch_side = 3, 1, 12, 8, 4
grid = (height // patch_side, width // patch_side)
patch_tokens = grid[0] * grid[1]
total_tokens = patch_tokens + 1
g = gcd(patch_tokens, total_tokens)
a, b = patch_tokens // g, total_tokens // g
assert grid == (3, 2)
assert (patch_tokens, total_tokens) == (6, 7)
assert gcd(a, b) == 1 and b > 0
assert a + b == 13
```
