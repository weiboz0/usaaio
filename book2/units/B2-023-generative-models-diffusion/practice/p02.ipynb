## Task

In this unit's GAN the discriminator outputs **logits** $\ell$, and $D=\sigma(\ell)$ is its probability that an input is real.
Which statement correctly identifies the loss each player minimizes and the reason the generator uses the non-saturating loss?

A. The discriminator minimizes $\mathrm{BCE}(\ell_{\text{real}},1)+\mathrm{BCE}(\ell_{\text{fake}},0)$ (two batch means, summed); the generator minimizes $\mathrm{BCE}(\ell_{\text{fake}},1)=-\text{mean}\log D(G(z))$, because when the discriminator confidently rejects fakes ($\ell_{\text{fake}}\ll0$) the minimax loss $\log(1-D(G(z)))$ has logit gradient $-\sigma(\ell)\approx0$, while the non-saturating loss has logit gradient $-(1-\sigma(\ell))\approx-1$.
B. The discriminator minimizes $\mathrm{BCE}(\ell_{\text{real}},0)+\mathrm{BCE}(\ell_{\text{fake}},1)$; the generator minimizes $\mathrm{BCE}(\ell_{\text{fake}},1)$, because the generator should agree with the discriminator's labels.
C. The discriminator minimizes $\mathrm{BCE}(\ell_{\text{real}},1)+\mathrm{BCE}(\ell_{\text{fake}},0)$; the generator minimizes $\mathrm{BCE}(\ell_{\text{fake}},0)$, the same fake-row term as the discriminator, so that both players optimize one shared objective.
D. Both players minimize the value function $V(D,G)$; the non-saturating loss is used because it equals $-V$, which makes $C(G)=-\log4+2\,\mathrm{JS}$ the generator's exact objective.
E. The discriminator minimizes $\mathrm{BCE}(\ell_{\text{real}},1)+\mathrm{BCE}(\ell_{\text{fake}},0)$; the generator minimizes $\log(1-D(G(z)))$, because that loss has its largest gradient when $D(G(z))\approx0$.
