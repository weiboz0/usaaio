## Solution

Assume $p_{\text{data}}$ and $p_g$ are positive on a common set $S$ and zero outside it, so $D^*=p_{\text{data}}/(p_{\text{data}}+p_g)$ on $S$ (p13), and let $m=\tfrac12(p_{\text{data}}+p_g)$, which is also positive on $S$.

### (a) Substituting $D^*$

With $D^*$ and $1-D^*=p_g/(p_{\text{data}}+p_g)$, and all integrals over $S$ (outside $S$ both densities vanish and contribute nothing),
$$C(G)=\max_DV(D,G)=V(D^*,G)=E_{x\sim p_{\text{data}}}\Big[\log\frac{p_{\text{data}}(x)}{p_{\text{data}}(x)+p_g(x)}\Big]+E_{x\sim p_g}\Big[\log\frac{p_g(x)}{p_{\text{data}}(x)+p_g(x)}\Big].$$

### (b) Rewriting with the mixture

On $S$, $p_{\text{data}}+p_g=2m$ with $m>0$, so each ratio is a ratio of **positive** numbers and the logarithm may be split:
$$\log\frac{p_{\text{data}}}{2m}=\log\frac{p_{\text{data}}}{m}-\log2,\qquad\log\frac{p_g}{2m}=\log\frac{p_g}{m}-\log2 .$$
This step uses positivity on $S$: $\log(u/(2v))=\log(u/v)-\log2$ requires $u,v>0$, and on $S$ both $p_{\text{data}},p_g>0$ and $m>0$.
Each expectation then splits by **linearity of expectation**, which applies because each piece is integrable: $E_{p_{\text{data}}}[\log2]=\log2$ since $p_{\text{data}}$ integrates to $1$, and $E_{p_{\text{data}}}[\log(p_{\text{data}}/m)]=\mathrm{KL}(P_{\text{data}}\|M)$ is finite because $p_{\text{data}}/m\le2$ on $S$ (so the integrand is bounded above by $\log2$, and KL is $\ge0$).
Therefore
$$C(G)=\Big(\mathrm{KL}(P_{\text{data}}\|M)-\log2\Big)+\Big(\mathrm{KL}(P_g\|M)-\log2\Big)=-\log4+\mathrm{KL}(P_{\text{data}}\|M)+\mathrm{KL}(P_g\|M).$$

### (c) The JS form

Session 1 defines $\mathrm{JS}(P\|Q)=\tfrac12\mathrm{KL}(P\|M)+\tfrac12\mathrm{KL}(Q\|M)$ with $M=\tfrac12(P+Q)$. Hence $\mathrm{KL}(P_{\text{data}}\|M)+\mathrm{KL}(P_g\|M)=2\,\mathrm{JS}(p_{\text{data}}\|p_g)$ and
$$C(G)=-\log4+2\,\mathrm{JS}(p_{\text{data}}\|p_g).$$

### (d) The minimum and its equality case

*Lower bound.* $\mathrm{KL}\ge0$ for any pair of distributions (B2-022, Session 1), so both KL terms in (b) are $\ge0$ and $C(G)\ge-\log4$ for every generator.

*($\Leftarrow$) If $p_g=p_{\text{data}}$*, then $m=\tfrac12(p_{\text{data}}+p_{\text{data}})=p_{\text{data}}=p_g$, so $M=P_{\text{data}}=P_g$ and both KL terms are $\mathrm{KL}(P\|P)=0$; hence $C(G)=-\log4$.

*($\Rightarrow$) If $C(G)=-\log4$*, then $\mathrm{KL}(P_{\text{data}}\|M)+\mathrm{KL}(P_g\|M)=0$. Two nonnegative numbers summing to $0$ are both $0$, and the KL equality case gives $P_{\text{data}}=M$ and $P_g=M$. Therefore $P_{\text{data}}=P_g$, i.e. $p_g=p_{\text{data}}$.

So $\min_GC(G)=-\log4$, attained **if and only if** $p_g=p_{\text{data}}$ (and then $D^*\equiv\tfrac12$).

### (e) Theory versus training

The non-saturating generator loss $-E_z[\log D(G(z))]$ is not $C(G)$: $C(G)$ is the value of the minimax game with the **optimal** discriminator plugged in, whereas training descends a different function of the generator (a $-\log D$ rather than $\log(1-D)$ loss) against whatever **current** discriminator a few alternating gradient steps have produced.
The two objectives still share their direction and their fixed point: both push $D(G(z))$ upward, and at $p_g=p_{\text{data}}$ with $D\equiv\tfrac12$ neither has any incentive to move.
Consequently (d) characterizes the idealized optimum but certifies nothing about a finite run with limited networks, a non-optimal discriminator, and the non-saturating loss: a run must be judged by empirical checks such as mode-coverage counts and held-out comparisons, not by appeal to the theorem.

### (f) A discrete check

$P_{\text{data}}=(\tfrac12,\tfrac12)$, $P_g=(\tfrac14,\tfrac34)$, $M=(\tfrac38,\tfrac58)$.
Directly from (a):
$$C=\tfrac12\log\tfrac{1/2}{3/4}+\tfrac12\log\tfrac{1/2}{5/4}+\tfrac14\log\tfrac{1/4}{3/4}+\tfrac34\log\tfrac{3/4}{5/4}=-1.318650 .$$
From the JS form, $\mathrm{JS}=0.033822$ and $-\log4+2(0.033822)=-1.318650$. They agree (the cell checks to $10^{-12}$).

```python
import math

P, Q = (0.5, 0.5), (0.25, 0.75)
M = tuple((p + q) / 2 for p, q in zip(P, Q))


def kl(p, q):
    return sum(pi * math.log(pi / qi) for pi, qi in zip(p, q) if pi > 0)


c_direct = (sum(p * math.log(p / (p + q)) for p, q in zip(P, Q))
            + sum(q * math.log(q / (p + q)) for p, q in zip(P, Q)))
js = 0.5 * kl(P, M) + 0.5 * kl(Q, M)
c_js = -math.log(4) + 2 * js
c_kl = -math.log(4) + kl(P, M) + kl(Q, M)
c_equal = -math.log(4) + kl(P, ((P[0] + P[0]) / 2, (P[1] + P[1]) / 2)) * 2
print(f"direct {c_direct:.6f}  via JS {c_js:.6f}  JS {js:.6f}")
```

### Answer check

```python
assert round(c_direct, 6) == -1.318650 and round(c_js, 6) == -1.318650
assert math.isclose(c_direct, c_js, abs_tol=1e-12, rel_tol=1e-09) and math.isclose(c_direct, c_kl, abs_tol=1e-12, rel_tol=1e-09)
assert round(js, 6) == 0.033822
assert c_direct > -math.log(4)
assert math.isclose(c_equal, -math.log(4), abs_tol=1e-15, rel_tol=1e-09)
assert M == (0.375, 0.625)
```
