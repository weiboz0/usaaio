## Solution

Throughout, $D$ ranges over **all** functions $\mathbb R^d\to(0,1)$, and $a=p_{\text{data}}(x)$, $b=p_g(x)$ at a fixed point $x$.

### (a) $V$ as one integral

$$V(D,G)=E_{x\sim p_{\text{data}}}[\log D(x)]+E_{z}[\log(1-D(G(z)))].$$
The first term is already $\int p_{\text{data}}(x)\log D(x)\,dx$.
For the second, the fact needed is the **definition of $p_g$ as the distribution of $G(z)$** (the law of the pushed-forward random variable, i.e. the "law of the unconscious statistician"): for any function $h$, $E_{z\sim N(0,I)}[h(G(z))]=E_{x\sim p_g}[h(x)]=\int p_g(x)h(x)\,dx$.
Applying it with $h(x)=\log(1-D(x))$,
$$V(D,G)=\int\Big[p_{\text{data}}(x)\log D(x)+p_g(x)\log\big(1-D(x)\big)\Big]dx .$$

### (b) The pointwise maximizer

Let $a>0$, $b>0$ and $f(y)=a\log y+b\log(1-y)$ on $(0,1)$. It is differentiable there with
$$f'(y)=\frac ay-\frac b{1-y}=\frac{a(1-y)-by}{y(1-y)}=\frac{a-(a+b)y}{y(1-y)} .$$
The denominator is positive on $(0,1)$, so the sign of $f'$ is the sign of $a-(a+b)y$:
$f'(y)>0$ for $0<y<y^*$, $f'(y^*)=0$, and $f'(y)<0$ for $y^*<y<1$, where $y^*=\frac a{a+b}\in(0,1)$ (because $0<a<a+b$).
Hence $f$ is strictly increasing on $(0,y^*]$ and strictly decreasing on $[y^*,1)$, so $f(y)<f(y^*)$ for every $y\ne y^*$: $y^*$ is the **unique** maximizer.
(Equivalently, $f''(y)=-a/y^2-b/(1-y)^2<0$, so $f$ is strictly concave and its only critical point is the global maximum.)

**No larger value near the ends.** As $y\to0^+$, $a\log y\to-\infty$ while $b\log(1-y)\to0$, so $f(y)\to-\infty$; as $y\to1^-$, $b\log(1-y)\to-\infty$ while $a\log y\to0$, so $f(y)\to-\infty$.
So $f$ cannot approach a supremum larger than $f(y^*)$ near either endpoint; the monotonicity above already shows every point near $0$ or $1$ lies strictly below $f(y^*)$.

### (c) From pointwise to the whole integral

Let $D^*(x)=y^*(x)$ wherever $a,b>0$ (with the conventions of (d) elsewhere). For any admissible $D$ and every $x$, part (b) gives
$$p_{\text{data}}(x)\log D(x)+p_g(x)\log(1-D(x))\ \le\ p_{\text{data}}(x)\log D^*(x)+p_g(x)\log(1-D^*(x)),$$
and integrating the inequality over $x$ gives $V(D,G)\le V(D^*,G)$. So $D^*$ maximizes $V$ over the class.
The argument needs the function class to impose **no coupling between values at different points**: the value $D(x)$ must be choosable independently at every $x$, so that the pointwise maximizer $x\mapsto y^*(x)$ is itself a member of the class (in particular, it must map into $(0,1)$ and be an admissible, integrable function).
A restricted class (for example a fixed-size neural network) may not contain $D^*$, and then the pointwise argument gives only an upper bound.

### (d) Boundary cases

* **$a>0=b$:** $f(y)=a\log y$ is strictly increasing on $(0,1)$ with $\sup f=\lim_{y\to1^-}a\log y=0$, **not attained** on $(0,1)$. Convention: $D^*(x)=1$.
* **$a=0<b$:** $f(y)=b\log(1-y)$ is strictly decreasing with $\sup f=\lim_{y\to0^+}b\log(1-y)=0$, **not attained**. Convention: $D^*(x)=0$.
* **$a=b=0$:** $f\equiv0$, so $\sup f=0$ is **attained at every** $y\in(0,1)$; the point does not affect $V$, and $D^*(x)$ may be set to any value (we use $\tfrac12$ for definiteness).

In the first two cases the conventional value lies outside $(0,1)$; it is the limit of a maximizing sequence, which is why $D^*$ is defined by convention there.

### (e) A worked pair

For $p_{\text{data}}(x)=2x$ and $p_g(x)=1$ on $[0,1]$ (both integrate to $1$), at every $x\in(0,1]$ both densities are positive and
$$D^*(x)=\frac{2x}{2x+1}.$$
$D^*(\tfrac14)=\frac{1/2}{3/2}=\tfrac13$ and $D^*(1)=\tfrac23$.
At $x=0$, $p_{\text{data}}(0)=0<1=p_g(0)$: this is the case $a=0<b$ of (d), so $D^*(0)=0$ by convention (and the closed form happens to agree).

```python
import math
from fractions import Fraction

import torch

# (b) numeric check of the unique maximizer for a few (a, b) pairs on a fine grid
grid = torch.linspace(1e-6, 1 - 1e-6, 200_001, dtype=torch.float64)
checks = []
for a, b in ((0.25, 0.375), (2.0, 1.0), (0.1, 3.0)):
    f = a * torch.log(grid) + b * torch.log1p(-grid)
    y_grid = grid[f.argmax()].item()
    checks.append((a, b, y_grid, a / (a + b)))

# (e) the worked pair
d_star = lambda x: Fraction(2) * x / (Fraction(2) * x + 1)
d_quarter, d_one = d_star(Fraction(1, 4)), d_star(Fraction(1))

# (c) pointwise optimum beats random discriminators on the (e) pair (midpoint rule on [0,1])
n = 20_000
xs = (torch.arange(n, dtype=torch.float64) + 0.5) / n
p_data, p_g = 2 * xs, torch.ones_like(xs)
V = lambda D: ((p_data * torch.log(D) + p_g * torch.log1p(-D)) / n).sum().item()
v_star = V(2 * xs / (2 * xs + 1))
gen = torch.Generator().manual_seed(20261015)
v_random = [V(torch.rand(n, generator=gen, dtype=torch.float64).clamp(1e-6, 1 - 1e-6)) for _ in range(5)]
v_half = V(torch.full_like(xs, 0.5))
print(checks)
print(d_quarter, d_one, round(v_star, 6), [round(v, 4) for v in v_random], round(v_half, 6))
```

### Answer check

```python
for a, b, y_grid, y_star in checks:
    assert abs(y_grid - y_star) < 1e-5
assert d_quarter == Fraction(1, 3) and d_one == Fraction(2, 3)
assert d_star(Fraction(0)) == 0
assert all(v < v_star for v in v_random) and v_half < v_star
```
