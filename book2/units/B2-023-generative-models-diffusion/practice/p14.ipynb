## Task

Show that the minimax value at the optimal discriminator is $C(G)=\max_DV(D,G)=-\log4+2\,\mathrm{JS}(p_{\text{data}}\|p_g)$ and characterize its minimum. Reasoning is required.

Assume $p_{\text{data}}$ and $p_g$ are positive on a common set $S$ and zero outside it, so p13's formula $D^*=p_{\text{data}}/(p_{\text{data}}+p_g)$ holds on $S$. Let $m=\frac12(p_{\text{data}}+p_g)$, the density of $M=\frac12(P_{\text{data}}+P_g)$.

(a) Substitute $D^*$ into $V$ and write $C(G)$ as two expectations.

(b) Rewrite each denominator with $m$ and show $C(G)=-\log4+\mathrm{KL}(P_{\text{data}}\|M)+\mathrm{KL}(P_g\|M)$. Justify every step that splits a logarithm or an expectation, and say where positivity on $S$ is used.

(c) Conclude $C(G)=-\log4+2\,\mathrm{JS}(p_{\text{data}}\|p_g)$ from the definition of JS in Session 1.

(d) Prove that $C(G)\ge-\log4$ for every generator, with equality **if and only if** $p_g=p_{\text{data}}$. Cite only the nonnegativity of KL and its equality case (B2-022 and Session 1), and prove both directions of the "iff".

(e) In three or four sentences, explain why the non-saturating generator loss used in training is not $C(G)$, what the two objectives still share, and what that means for applying (d) to real training runs.

(f) For the discrete analogue $P_{\text{data}}=(\tfrac12,\tfrac12)$ and $P_g=(\tfrac14,\tfrac34)$, compute $C(G)$ directly from (a) and from $-\log4+2\,\mathrm{JS}$, each to six decimals, and confirm that they agree.
