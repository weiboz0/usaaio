## Task

For each case, choose among a **GAN**, a **VAE** (B2-022), and a **diffusion model** (a DDPM, or latent diffusion where stated), and justify the choice against the stated sample-quality, likelihood, speed, and coverage constraints. Count speed in network evaluations per generated sample: one generator pass for a GAN, one decoder pass for a VAE, $T$ denoiser passes for a DDPM (and $2T$ with classifier-free guidance). Then give **one held-out check** that would falsify your recommendation.

(I) **Real-time variations.** A game must produce thousands of new 2-D sprite layouts per second on a phone CPU. Artists judge sharpness; no likelihood is needed; abundant training data exists.

(II) **Calibrated anomaly ranking.** A sensor team must rank new 8-D readings by how unlikely they are under the training distribution, using a quantity with a stated relation to $\log p(x)$, and must score $10^5$ readings per second. Generating samples is rarely needed.

(III) **Rare modes matter.** A simulator's 2-D output distribution has several well-separated modes, two of them rare, and every mode must appear among generated samples; generation is offline with no latency limit. A previous GAN attempt covered two of four modes.

(IV) **Text-conditioned images.** A team must generate $512\times512$ images from text prompts on a single GPU with limited memory, with adjustable adherence to the prompt.

(V) **CPU prototype with a certificate.** A course project must run in seconds on CPU on a known four-component 2-D mixture and must certify which components it reproduces and how sample statistics compare with held-out data.

For every case, state the decision in one line, the justification in at most four sentences (naming the specific property: a single-pass sampler, an ELBO, stable training and coverage, latent space plus cross-attention plus guidance, or the coverage/statistics certificates of p17–p19), and the falsifying held-out check in one sentence.
