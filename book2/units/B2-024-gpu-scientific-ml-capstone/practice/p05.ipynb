## Task

A network predicts mixture parameters $\hat P$ of shape `(N, 3, 3)`; row $k$ of each example is a component $(\hat w_k, \hat c_k, \hat s_k)$.
The targets $T$ have the same layout.
Which loss is **permutation-invariant** (unchanged when the components of $\hat P$ or of $T$ are relabelled) **and** equal to zero only when $\hat P$ equals $T$ up to such a relabelling?

A. The mean over all nine entries of $(\hat P-T)^2$, with $\hat P$ in the network's output order and $T$ in the generator's draw order.
B. For each example, $\min_{\pi}\frac19\sum_{k=1}^3\sum_{j=1}^3|\hat P_{\pi(k),j}-T_{k,j}|$ over the six permutations $\pi$ of the components, averaged over examples.
C. The mean absolute difference after sorting $\hat P$'s components by weight and $T$'s components by center.
D. $\big(\sum_k\hat w_k-\sum_kw_k\big)^2+\big(\sum_k\hat c_k-\sum_kc_k\big)^2+\big(\sum_k\hat s_k-\sum_ks_k\big)^2$, averaged over examples.
E. The mean absolute difference between the nine entries of $\hat P$ sorted as one list and the nine entries of $T$ sorted as one list.

For each option say whether it is permutation-invariant and whether its zero set is exactly "equal up to relabelling", with a two-component counterexample wherever one of the two properties fails.
