## Task

| index | class 0 | class 1 | class 2 |
|---:|---:|---:|---:|
| 0 | 0.91 | 0.05 | 0.04 |
| 1 | 0.10 | 0.85 | 0.05 |
| 2 | 0.82 | 0.10 | 0.08 |
| 3 | 0.05 | 0.05 | 0.90 |
| 4 | 0.95 | 0.03 | 0.02 |
| 5 | 0.30 | 0.65 | 0.05 |
| 6 | 0.80 | 0.15 | 0.05 |
| 7 | 0.06 | 0.14 | 0.80 |

Which output is correct?

A. indices `[0, 1, 3, 4]`, labels `[0, 1, 2, 0]`
B. indices `[0, 1, 3, 4, 7]`, labels `[0, 1, 2, 0, 2]`
C. indices `[0, 1, 2, 3, 4, 6, 7]`, labels `[0, 1, 0, 2, 0, 0, 2]`
D. indices `[0, 1, 2, 3, 7]`, labels `[0, 1, 0, 2, 2]`
E. indices `[0, 3, 4]`, labels `[0, 2, 0]`

List the candidates per class with their confidences, apply the cap, and name the rule each wrong option breaks.
