## Task

Let $A\in\mathbb R^{m\times n}$, $y\in\mathbb R^m$, and $\lambda>0$, and define
$$J(u)=\lVert Au-y\rVert_2^2+\lambda\lVert u\rVert_2^2,\qquad u\in\mathbb R^n .$$

(a) Expand $J(u)$ as a quadratic form in $u$ and compute $\nabla J(u)$, citing the sum-of-squares gradient rule.

(b) Prove that $M=A^\top A+\lambda I$ is invertible for every $\lambda>0$, **without** assuming anything about the rank of $A$. (Hint: compute $v^\top Mv$ for $v\ne0$.)

(c) Show that $u_\lambda=M^{-1}A^\top y$ is the only stationary point of $J$.

(d) Prove that $u_\lambda$ is the **unique global minimizer**: show that for every $u$, $J(u)-J(u_\lambda)=(u-u_\lambda)^\top M(u-u_\lambda)$, and conclude.

(e) For the $8\times81$ grid operator of Session 5, explain in at most four sentences why the statement fails for $\lambda=0$: what is the rank of $A^\top A$, and what does the set of minimizers of $\lVert Au-y\rVert^2$ look like?

Write complete sentences; every step must say which fact it uses.
